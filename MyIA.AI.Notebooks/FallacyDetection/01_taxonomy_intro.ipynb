## Méthodologie

La taxonomie Argumentum est organisée en **3 strates hiérarchiques** :

1. **Sophismes logiques formels** (Aristote, *Sophistici Elenchi*) : contradiction, cercle vicieux, affirmation du conséquent, etc.
2. **Sophismes rhétoriques** (Walton, *Argumentation Schemes and Fallacies*, 1996): appel à l'autorité, à la popularité, à la peur, etc.
3. **Sophismes pragmatiques** (Hamblin, *Fallacies*, 1970) : déplacement de la charge de la preuve, pétition de principe, etc.

Chaque sophisme est une **feuille** (≠ nœud interne de regroupement). Les nœuds internes (e.g., *sophismes ad hominem*) regroupent plusieurs feuilles (*ad hominem abusive*, *tu quoque*, *circumstantial*).

```python
# Charger la taxonomie Argumentum depuis le repo (CSV canonique, repo-local)
from pathlib import Path
from collections import Counter
import csv

# Robuste au cwd (papermill peut lancer depuis repo-root ou le dossier notebook) :
# on cherche le CSV en essayant plusieurs ancres relatives (meme pattern que 03).
_CANDIDATES = [
    Path("MyIA.AI.Notebooks/SymbolicAI/Argument_Analysis/data/argumentum_fallacies_taxonomy.csv"),
    Path("../SymbolicAI/Argument_Analysis/data/argumentum_fallacies_taxonomy.csv"),
    Path("../../SymbolicAI/Argument_Analysis/data/argumentum_fallacies_taxonomy.csv"),
]
CSV_PATH = next((c for c in _CANDIDATES if c.is_file()), None)
assert CSV_PATH is not None, "Taxonomie introuvable depuis cwd=" + str(Path.cwd())

with open(CSV_PATH, encoding="utf-8-sig", newline="") as f:
    rows = list(csv.DictReader(f))

total_nodes = len(rows)
# Feuille = noeud dont le 'path' n'est prefix d'aucun autre path.
all_paths = {r["path"].strip() for r in rows if r.get("path")}
leaves = [r for r in rows
          if r["path"].strip() and r["path"].strip() != "0"
          and not any(p.startswith(r["path"].strip() + ".") for p in all_paths
                      if p != r["path"].strip())]
by_famille = Counter(r["Famille"].strip() for r in leaves if r.get("Famille", "").strip())

print(f"Taxonomie chargée (CSV canonique Argumentum, {CSV_PATH.name})")
print(f"  noeuds total   : {total_nodes}")
print(f"  feuilles       : {len(leaves)}")
print(f"  familles       : {len(by_famille)}")
print("Répartition par famille :")
for famille, n in sorted(by_famille.items(), key=lambda x: -x[1]):
    print(f"  - {famille}: {n}")
```

Les 3 strates d'Argumentum s'appuient sur **3 corpus académiques de référence** :

| Strate | Source canonique | URL |

| Pragmatique | Hamblin, *Fallacies*, 1970 | [Reed College](https://www.reed.edu/hamblin/) |

**Argumentum** complète ces corpus par :

- Une **structuration hiérarchique** (3 strates × N sous-catégories × N feuilles).
- Un **encodage FR/EN bilingue** aligné terme à terme.
- Une **cartographie applicative** : pour chaque sophisme, des cas d'usage en détection automatique (NLP) et en analyse argumentative.
