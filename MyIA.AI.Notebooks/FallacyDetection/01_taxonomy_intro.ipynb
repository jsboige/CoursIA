## Méthodologie

La taxonomie Argumentum est organisée en **3 strates hiérarchiques** :

1. **Sophismes logiques formels** (Aristote, *Sophistici Elenchi*) : contradiction, cercle vicieux, affirmation du conséquent, etc.
2. **Sophismes rhétoriques** (Walton, *Argumentation Schemes and Fallacies*, 1996): appel à l'autorité, à la popularité, à la peur, etc.
3. **Sophismes pragmatiques** (Hamblin, *Fallacies*, 1970) : déplacement de la charge de la preuve, pétition de principe, etc.

Chaque sophisme est une **feuille** (≠ nœud interne de regroupement). Les nœuds internes (e.g., *sophismes ad hominem*) regroupent plusieurs feuilles (*ad hominem abusive*, *tu quoque*, *circumstantial*).

```python
# Charger la taxonomie Argumentum depuis le repo
from pathlib import Path
from collections import Counter

TAXONOMY_PATH = Path('data/argumentum_taxonomy.json')

import json
try:
    taxonomy = json.load(open(TAXONOMY_PATH))
    leaves = taxonomy.get('leaves', [])
    nodes = taxonomy.get('internal_nodes', [])
    print(f'Taxonomie chargée: {len(leaves)} feuilles, {len(nodes)} nœuds internes')
    # Répartition par strate
    by_strate = Counter(l.get('strate', '?') for l in leaves)
    print('Répartition par strate :')
    for strate, n in sorted(by_strate.items(), key=lambda x: -x[1]):
        print(f'  - {strate}: {n}')
except FileNotFoundError:
    print(f'[INFO] Taxonomie non trouvée à {TAXONOMY_PATH}.')
    print('Fallback : voir les sources canoniques en cellule suivante.')
    taxonomy = {'leaves': [], 'internal_nodes': []}
```

| Pragmatique | Hamblin, *Fallacies*, 1970 | [Reed College](https://www.reed.edu/hamblin/) |

**Argumentum** complète ces corpus par :

- Une **structuration hiérarchique** (3 strates × N sous-catégories × N feuilles).
- Un **encodage FR/EN bilingue** aligné terme à terme.
- Une **cartographie applicative** : pour chaque sophisme, des cas d'usage en détection automatique (NLP) et en analyse argumentative.
