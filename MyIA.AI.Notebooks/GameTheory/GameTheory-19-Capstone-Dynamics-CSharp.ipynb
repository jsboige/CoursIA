## 3. Expérience 1 — L'effondrement et son remède (Dynamics-01 + 02)

En horizon **fini**, la coopération s'effondre par induction à rebours : au dernier tour dévier domine, donc à l'avant-dernier aussi, et toute la chaîne remonte. En horizon **infini** escompté, une menace crédible — le grim trigger — la soutient dès que $\delta$ dépasse un seuil. Et ce seuil, le lake l'énonce analytiquement : `Discounting.lean` le force en nombres. L'expérience : le retrouver numériquement, et vérifier que **le théorème vit déjà dans les nombres**.


```
// Port .NET natif du noyau de l'experience 1 : pas de numpy, tableaux .NET et System.Console.
Console.WriteLine($".NET {Environment.Version} | runtime {System.Runtime.InteropServices.RuntimeInformation.FrameworkDescription}");

// Le PD canonique de Dynamics-01 (T > R > P > S et 2R > T + S)
double T = 5.0, R = 3.0, P = 1.0, S = 0.0;
if (!(T > R && R > P && P > S && 2 * R > T + S))
{
    throw new InvalidOperationException("parametres PD non canoniques");
}

// (a) Horizon fini N tours, delta = 1 : l'induction a rebours rend (D, D) unique SPNE
int N = 10;
Console.WriteLine($"Horizon fini N={N} : paiement SPNE (D,D)xN = {P * N:0.0} ; "
                + $"la cooperation fictive ({R * N:0.0}) n'y est pas credible.");

// (b) Horizon infini escompte : cooperation soutenable si V_C >= V_D
//     V_C = R/(1-d) ; deviation : T une fois, puis la punition P a jamais
Func<double, double> VC = d => R / (1 - d);
Func<double, double> VD_grim = d => T + d * P / (1 - d);

double delta_star = (T - R) / (T - P);  // seuil analytique -- celui que Discounting.lean enonce
Console.WriteLine($"\nSeuil analytique delta* = (T-R)/(T-P) = ({T:0.#}-{R:0.#})/({T:0.#}-{P:0.#}) = {delta_star:0.0000}");
Console.WriteLine(string.Format("{0,7} {1,11} {2,12} {3,11}", "delta", "V_C(coop)", "V_D(deviat)", "soutenable"));
foreach (var d in new[] { 0.30, 0.40, 0.50, 0.60, 0.70 })
{
    string soutenable = VC(d) >= VD_grim(d) ? "OUI" : "NON";
    Console.WriteLine(string.Format("{0,7:0.00} {1,11:0.0000} {2,12:0.0000} {3,11}", d, VC(d), VD_grim(d), soutenable));
}
Console.WriteLine($"\nAu seuil exact delta* = {delta_star:0.0000} : V_C = {VC(delta_star):0.0000}, "
                + $"V_D = {VD_grim(delta_star):0.0000} (indifference)");
```

```
// Punition plus dure => cooperation plus facile a soutenir (l'intuition du folk theorem,
// version numerique de la comparaison grim vs tit-for-tat de Dynamics-01, section seuils)
// TFT punit UN tour (le deviant subit S en jouant C dans la punition) puis pardonne
Func<double, double> VD_tft = d => T + d * S + d * d * R / (1 - d);

int nbDeltas = 20001;
double[] deltas = new double[nbDeltas];
for (int i = 0; i < nbDeltas; i++)
{
    deltas[i] = 0.01 + i * (0.99 - 0.01) / (nbDeltas - 1);
}

double delta_tft = double.NaN;
foreach (var d in deltas)
{
    if (VC(d) >= VD_tft(d))
    {
        delta_tft = d;   // premiere valeur du balayage ou la cooperation tient (argmax de ok_tft)
        break;
    }
}

Console.WriteLine($"Seuil grim trigger : delta* = {delta_star:0.0000}  (punition P a jamais)");
Console.WriteLine($"Seuil tit-for-tat  : delta* = {delta_tft:0.0000}  (une punition, puis pardon)");
Console.WriteLine();
Console.WriteLine("Le grim trigger soutient la cooperation PLUS TOT : une menace plus dure est");
Console.WriteLine("plus facile a rendre credible. Ce sont les nombres du lemme du lake :");
Console.WriteLine("grim_trigger_sustains_iff (RepeatedGames/GrimTrigger.lean, 0 sorry).");
```

### Lecture du résultat

Le seuil numérique `(T-R)/(T-P) = 0,5` coïncide avec la condition analytique que [`Discounting.lean`](game_theory_lean/RepeatedGames/Discounting.lean) énonce sur les mêmes paramètres : la coopération est soutenable si et seulement si $\delta \ge (T-R)/(T-P)$. La menace éternelle soutient la coopération dès δ ≥ 0,5 ; la punition d'un seul tour exige un joueur bien plus patient. **C'est la crédibilité de la menace — non sa douceur — qui fait la coopération** : le théorème Folk vu d'en bas, avant sa forme générale (le stretch `folk_theorem_discounted`, cf. §6).


## 4. Expérience 2 — Coopérer par preuve (Dynamics-04 + 04b)

La répétition a besoin d'un futur. Sans futur — one-shot —, la **transparence des programmes** offre une autre voie : si chacun peut lire le code de l'autre avant d'agir, la coopération peut venir d'une prédiction mutuelle **prouvée**, pas d'une menace. Deux niveaux de lecture s'y distinguent (la faille L1/L2 de Dynamics-04b) :

- **L1, la preuve de surface** : chercher dans le source la déclaration de coopération — sondable par une branche morte ;
- **L2, la preuve par Löb** : coopérer seulement si la coopération de l'autre est **prouvable** dans un système de prouvabilité — le point fixe FairBot×FairBot, prouvé dans le lake.

L'expérience : la grille mesurée des bots en L1, le témoin qui la fait tomber, et ce que le L2 garantit.


```
using System.Collections.Generic;

// Port .NET du noyau Dynamics-04 : les bots sont des delegates, leurs "sources" un registre explicite.
const int STEP_BUDGET = 1000;  // borne de simulation, comme Dynamics-04 (verdict CALCULE, jamais coupe)

int steps = 0;
Action _tick = () =>
{
    steps++;
    if (steps > STEP_BUDGET)
    {
        throw new TimeoutException("budget de pas depasse");
    }
};

// La sonde de surface : la chaine que FairBot_surface cherche chez l'autre.
const string SONDE_COOP = "return \"C\";";

// Les sources des bots : en Python on les lit par inspect.getsource ; ici elles sont declarees
// dans un registre explicite -- la transparence des programmes reste le principe, seule la
// mecanique d'introspection change.
var SRC = new Dictionary<string, string>
{
    ["CooperateBot"] = @"string CooperateBot(string otherSrc)
{
    return ""C"";
}",
    ["DefectBot"] = @"string DefectBot(string otherSrc)
{
    return ""D"";
}",
    ["FairBot_surface"] = @"string FairBot_surface(string otherSrc)
{
    _tick();
    if (otherSrc.Contains(SONDE_COOP)) return ""C"";
    return ""D"";
}",
    ["ProbeBot"] = @"string ProbeBot(string otherSrc)
{
    _tick();
    string sonde = ""return ""C"";"";
    return ""D"";
}",
};

// Les bots : memes regles que la source Python, en delegates.
var BOTS = new Dictionary<string, Func<string, string>>
{
    ["CooperateBot"] = other => "C",
    ["DefectBot"] = other => "D",
    ["FairBot_surface"] = other => { _tick(); return other.Contains(SONDE_COOP) ? "C" : "D"; },
    ["ProbeBot"] = other => { _tick(); string sonde = "return \"C\";"; _ = sonde; return "D"; },
};

string[] BOT_NAMES = { "CooperateBot", "DefectBot", "FairBot_surface", "ProbeBot" };

// Les paiements du PD canonique (T, R, P, S poses en cellule 4).
var PAYOFF = new Dictionary<string, double>
{
    ["C|C"] = R, ["C|D"] = S, ["D|C"] = T, ["D|D"] = P,
};

Func<string, string, (string a, string b, double pa, double pb)> duel = (na, nb) =>
{
    string a = BOTS[na](SRC[nb]);
    string b = BOTS[nb](SRC[na]);
    return (a, b, PAYOFF[$"{a}|{b}"], PAYOFF[$"{b}|{a}"]);
};

string entete = "A \\ B";
string ligneEntete = entete.PadLeft(18);
foreach (var n in BOT_NAMES)
{
    ligneEntete += n.PadLeft(18);
}
Console.WriteLine(ligneEntete);
foreach (var na in BOT_NAMES)
{
    string row = na.PadLeft(18);
    foreach (var nb in BOT_NAMES)
    {
        steps = 0;
        var (a, b, pa, pb) = duel(na, nb);
        row += string.Format("{0}/{1} ({2:0},{3:0})", a, b, pa, pb).PadLeft(18);
    }
    Console.WriteLine(row);
}
Console.WriteLine();
Console.WriteLine("16/16 duels rendent un verdict calcule dans la borne (step_cap=1000, bots deterministes).");
Console.WriteLine("La grille complete (5 bots, verificateur de coherence, CUPOD/PrudentBot) vit dans Dynamics-04.");
```

```
// Le temoin L1/L2 de Dynamics-04b : la preuve de surface est sondable
steps = 0;
var (a9, b9, pa9, pb9) = duel("FairBot_surface", "ProbeBot");
Console.WriteLine($"FairBot_surface x ProbeBot : {a9}/{b9} -> paiements ({pa9:0}, {pb9:0})");
Console.WriteLine("La sonde (branche morte cooperative) fait tomber la preuve de surface : FairBot_surface est exploite.");
Console.WriteLine();
Console.WriteLine("Le FairBot de Dynamics-04b exige une PREUVE (modalite de provabilite, theoreme de Lob) :");
Console.WriteLine("  - face a un cooperateur prouvable, il coopere      : fairBot_cooperates_with_provable_cooperator");
Console.WriteLine("  - sans preuve de cooperation, il defectionne        : fairBot_defects_without_proof");
Console.WriteLine("  - FairBot x FairBot coopere (point fixe de Lob)     : fairBot_fairBot_cooperation, loeb_mutual_fixedpoint");
Console.WriteLine("Les quatre vivent dans ProgramGames/FairBot.lean -- 0 sorry (inventaire mesure en section 6).");
```

### Lecture du résultat

La coopération one-shot **existe** — FairBot×FairBot se coordonne en (C,C) sans aucune menace — mais sa source change tout : en L1 (surface), elle est **sondable** et un ProbeBot l'exploite (paiement S = 0) ; en L2 (Löb), la coopération n'est accordée que sur preuve, et `fairBot_defects_without_proof` garantit qu'aucun affichage mensonger ne la déclenche. **La preuve mutuelle remplace la menace — à condition d'être une preuve, pas une lecture.** C'est la réponse de la sous-série à la question du capstone : la menace dégradée (finie) échoue, la preuve dégradée (de surface) est trompée ; la preuve réelle tient.


## 5. Expérience 3 — Le budget change l'issue (Dynamics-05 + 06 + 07)

Troisième voie de dégradation : le coût du raisonnement lui-même. Dynamics-05 construit des agents à **budget explicite** et prouve des certificats par familles (coopération mutuelle, non-exploitation) ; Dynamics-06 en tire le vocabulaire honnête des preuves bornées ; Dynamics-07 sépare la preuve **syntaxique** de la simulation **comportementale**. L'expérience : rejouer le même duel FairBot×FairBot à **budget de lecture croissant** — sous le seuil, la preuve ne tient plus dans la fenêtre, et la coopération tombe.


```
// Le duel de l'experience 2, rejoue a budget de lecture borne (le compagnon .NET de Dynamics-05)

// FairBot a budget de lecture borne : ne voit que les `budget` premiers caracteres.
// La borne est la version lisible du budget de preuve de Dynamics-05 : sous le seuil,
// la preuve ne tient plus dans la fenetre de lecture.
Func<string, int, string> FairBot_borne = (otherSrc, budget) =>
{
    int n = Math.Min(budget, otherSrc.Length);
    string lu = otherSrc.Substring(0, n);
    return lu.Contains(SONDE_COOP) ? "C" : "D";
};

string src_fb = SRC["FairBot_surface"];
int pos_preuve = src_fb.IndexOf(SONDE_COOP, StringComparison.Ordinal);
Console.WriteLine($"Source de FairBot_surface : {src_fb.Length} caracteres ; "
                + $"la declaration de cooperation vit au caractere {pos_preuve}.");
Console.WriteLine();
Console.WriteLine(string.Format("{0,8} {1,18}  lecture", "budget", "issue du duel"));
foreach (var budget in new[] { 10, 20, pos_preuve + 2, src_fb.Length })
{
    string a = FairBot_borne(src_fb, budget);
    string issue = string.Format("{0}/{0} {1}", a, a == "C" ? "(C,C) cooperation" : "(D,D) effondrement");
    string fenetre = src_fb.Substring(0, Math.Min(budget, src_fb.Length));
    string lecture = "'" + fenetre.Substring(Math.Max(0, fenetre.Length - 24)) + "'";
    Console.WriteLine(string.Format("{0,8} {1,18}  ...{2}", budget, issue, lecture));
}
Console.WriteLine();
Console.WriteLine("Sous le seuil de lecture, la preuve ne tient pas dans la fenetre : la cooperation tombe.");
Console.WriteLine("Le budget est un PARAMETRE du jeu, pas un detail d'implementation (Dynamics-06).");
```

```
using System.Collections.Generic;

// Les familles de certificats de Dynamics-05, rejouees cote .NET sur le banc borne ci-dessus
var certificats = new List<string>();

Action<string, bool, string, string> verifier = (nom, verdict, theoreme, source) =>
{
    certificats.Add(nom);
    Console.WriteLine(string.Format("{0}  {1,-46} {2}", verdict ? "OK" : "KO", nom, theoreme));
    if (!verdict)
    {
        Console.WriteLine($"    attendu depuis : {source}");
    }
};

// Famille 1 : cooperation mutuelle (a budget suffisant, le miroir coopere avec lui-meme)
verifier("mirror coopere avec lui-meme (budget plein)",
         FairBot_borne(src_fb, src_fb.Length) == "C",
         "mirror_mirror", "Dynamics-05 + ProgramGames/Bounded.lean");

// Famille 2 : non-exploitation (ProbeBot n'obtient jamais plus que T d'un agent borne)
steps = 0;
var d2 = duel("FairBot_surface", "ProbeBot");
verifier("ProbeBot n'extrait pas plus que T",
         d2.pb <= T,
         "defectBotBounded_unexploitable", "Dynamics-05 + ProgramGames/Bounded.lean");

// La question de Dynamics-07, posee sur une instance : la grille mesure des issues CALCULEES ;
// les theoremes du lake garantissent des equilibres. Les deux familles ci-dessus sont la
// traduction .NET de garanties Lean -- le certificat .NET est un TEMOIN, jamais une preuve.
Console.WriteLine();
Console.WriteLine($"{certificats.Count}/{certificats.Count} temoins rejoues : la question syntaxique vs");
Console.WriteLine("comportementale (Dynamics-07) reste ouverte -- un temoin vert n'est pas un theoreme.");
```

### Lecture du résultat

Le même FairBot, le même adversaire, la même règle — seul le **budget** change, et l'issue bascule de (C,C) à (D,D). La coopération n'est pas une propriété des programmes : c'est une propriété du couple (programmes, budget). Dynamics-06 en tire le vocabulaire honnête — une preuve bornée est une preuve **à condition de dire sa borne** — et Dynamics-07 la frontière : la simulation mesure des issues calculées, le lake garantit des équilibres ; les certificats .NET ci-dessus sont des **témoins**, jamais des preuves.


## 6. Ce que le lake Lean prouve

La sous-série ne simule pas seulement : elle prouve. Le lake [`game_theory_lean`](game_theory_lean/) porte deux modules sur le fil du capstone — `RepeatedGames` (la répétition : Stage, Discounting, GrimTrigger, Folk, Panel) et `ProgramGames` (la transparence : Basic, Bounded, FairBot). L'inventaire ci-dessous est **mesuré sur les fichiers du lake au moment de l'exécution** — nombres falsifiables, pas déclarés.


```
using System.IO;
using System.Linq;
using System.Text.RegularExpressions;

// Port manuel de pathlib + re : System.IO pour le parcours du lake, System.Text.RegularExpressions
// pour l'extraction des theoremes et des sorry.
string lake = string.Empty;
foreach (string p in new[] { "game_theory_lean", Path.Combine("MyIA.AI.Notebooks", "GameTheory", "game_theory_lean") })
{
    string full = Path.GetFullPath(p);
    if (Directory.Exists(full))
    {
        lake = full;
        break;
    }
}

if (lake.Length == 0)
{
    // le kernel peut demarrer ailleurs que le dossier du carnet : on remonte l'arborescence
    var dir = new DirectoryInfo(Directory.GetCurrentDirectory());
    while (dir is not null && lake.Length == 0)
    {
        string cand = Path.Combine(dir.FullName, "MyIA.AI.Notebooks", "GameTheory", "game_theory_lean");
        if (Directory.Exists(cand))
        {
            lake = cand;
        }
        dir = dir.Parent;
    }
}

if (lake.Length == 0)
{
    throw new InvalidOperationException("lake introuvable -- executer depuis le dossier GameTheory/");
}

// Chemin affiche relatif au depot : jamais de chemin absolu de machine dans la sortie.
string lakeRel = Path.GetFileName(lake);
int idx = lake.LastIndexOf("MyIA.AI.Notebooks", StringComparison.Ordinal);
if (idx >= 0)
{
    lakeRel = lake.Substring(idx).Replace(Path.DirectorySeparatorChar, '/');
}

Console.WriteLine($"lake : {lakeRel} \n");
foreach (var mod in new[] { "ProgramGames", "RepeatedGames" })
{
    string dirMod = Path.Combine(lake, mod);
    var fr = Directory.Exists(dirMod)
        ? Directory.GetFiles(dirMod, "*.lean")
            .Where(p => !p.EndsWith("_en.lean"))
            .OrderBy(p => p, StringComparer.Ordinal)
            .ToList()
        : new List<string>();

    var thm = new HashSet<string>();
    int sry = 0;
    foreach (var p in fr)
    {
        string texte = File.ReadAllText(p);
        foreach (Match m in Regex.Matches(texte, @"^(?:theorem|lemma)\s+(\w+)", RegexOptions.Multiline))
        {
            thm.Add(m.Groups[1].Value);
        }
        sry += Regex.Matches(texte, @"\bsorry\b").Count;
    }
    Console.WriteLine($"{mod,-15} {fr.Count} fichiers FR | {thm.Count} theoremes/lemmes distincts | {sry} sorry (code + prose)");
}

Console.WriteLine();
Console.WriteLine("Les quatre piliers du fil, localises par lecture des sources :");
var tousLean = Directory.GetFiles(lake, "*.lean", SearchOption.AllDirectories)
    .Where(p => !p.EndsWith("_en.lean"))
    .OrderBy(p => p, StringComparer.Ordinal)
    .ToList();
foreach (var nom in new[] { "grim_trigger_sustains_iff", "folk_theorem_discounted",
                            "fairBot_fairBot_cooperation", "loeb_mutual_fixedpoint" })
{
    string lieu = "?";
    foreach (var p in tousLean)
    {
        string texte = File.ReadAllText(p);
        if (Regex.IsMatch(texte, $"^(?:theorem|lemma)\\s+{nom}\\b", RegexOptions.Multiline))
        {
            string? parent = Path.GetDirectoryName(p);
            lieu = $"{(parent is null ? "?" : Path.GetFileName(parent))}/{Path.GetFileName(p)}";
            break;
        }
    }
    Console.WriteLine($"  {nom,-38} {lieu}");
}
```
