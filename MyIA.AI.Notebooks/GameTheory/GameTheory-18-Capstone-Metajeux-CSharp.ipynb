- **Composer** puis **casser** : composer des reponses locales ne survit pas a l'ambiguite

## Experience 1 — Composer, puis casser

**Ce que fait l'experience.** Deux jeux ouverts en serie : J1 choisit `x`, J2 repond `y`,
et le couplage est bilatere (`u2` depend de `x`, `u1` depend de `y`). On compose les
meilleures reponses locales, puis on demande au **jeu compose** — par enumeration des
deviations — si le profil obtenu est un equilibre.

**Le protocole tient en trois ingredients**, et l'experience en isole un par un :

- l'**univoquite** : la reponse locale de J1 a un unique maximum ;
- la **multivocite** : le maximum local est un plateau (ici, deux actions ex aequo par
  symetrie exacte au milieu de l'intervalle) ;
- le **couplage** : le choix de J1 revient a J1 par la reponse de J2.

Le carnet source [Metajeux-02](Metajeux/Metajeux-02-Casser-la-Composition-Python.ipynb)
montre que les deux premiers sont **inoffensifs separement** et que c'est leur conjonction
qui brise la selection. On rejoue les trois situations ici.


````
// === Experience 1 : composer des reponses locales, puis demander au jeu compose ===
// Protocole condense de Metajeux-02 (cellules 3 a 8) : memes formules, memes instances.
// Les fonctions locales `static` sont pures (aucun etat capture) ; les strategies sont des
// delegates de niveau superieur, donc reutilisables par la cellule de coherence finale.

int[] ACTIONS_X = { 0, 1, 2 };
int[] ACTIONS_Y = { 0, 1, 2 };
double C_PLATEAU = 0.5;
double R_COUPLAGE = 0.3;

// Formatage invariant (virgule decimale point, comme la sortie du twin Python).
static string F3s(double v) => v.ToString("+0.000;-0.000;+0.000", System.Globalization.CultureInfo.InvariantCulture);
static string F3(double v) => v.ToString("0.000", System.Globalization.CultureInfo.InvariantCulture);

// u1 : plateau exact en x=0 et x=1 (c=0.5), couple en retour par r*y.
Func<int, int, double> u1 = (x, y) => -(x - C_PLATEAU) * (x - C_PLATEAU) + R_COUPLAGE * y;
// u2 : reponse de J2, maximum unique en y = x (couplage dans l'autre sens).
Func<int, int, double> u2 = (x, y) => -(y - x) * (y - x);
// Contre-fait 1 : reponse locale univoque (sans plateau, sans couplage).
Func<int, int, double> u1_univoque = (x, y) => -(x - 1.0) * (x - 1.0);
// Contre-fait 2 : plateau sans couplage (r = 0).
Func<int, int, double> u1_r0 = (x, y) => -(x - C_PLATEAU) * (x - C_PLATEAU);

// Meilleure reponse de J2 voyant x.
Func<int, int> br2 = x => ACTIONS_Y.OrderByDescending(y => u2(x, y)).First();

// Correspondance locale de J1 (y exogene) : l'ENSEMBLE des optima, et les valeurs.
Func<(int[] opt, double[] vals)> br1_ensemble = () =>
{
    var vals = ACTIONS_X.Select(x => u1(x, 0)).ToArray();
    double vmax = vals.Max();
    var opt = ACTIONS_X.Where((x, i) => Math.Abs(vals[i] - vmax) < 1e-12).ToArray();
    return (opt, vals);
};

// Nash du jeu compose : aucune deviation profitable, J1 anticipant y' = br2(x').
static bool est_nash_compose(int x, int y, Func<int, int, double> u1Fn, Func<int, int, double> u2Fn, Func<int, int> br2Fn)
{
    bool ok1 = Enumerable.Range(0, 3).All(xp => u1Fn(x, y) >= u1Fn(xp, br2Fn(xp)) - 1e-12);
    bool ok2 = Enumerable.Range(0, 3).All(yp => u2Fn(x, y) >= u2Fn(x, yp) - 1e-12);
    return ok1 && ok2;
}

// --- Contre-fait 1 : couplage SANS plateau (reponse locale univoque) ---
int x_u = 1, y_u = br2(1);
bool contre_fait_univoque_ok = est_nash_compose(x_u, y_u, u1_univoque, u2, br2);

Console.WriteLine("[Contre-fait 1 : couplage SANS plateau, reponse locale univoque]");
Console.WriteLine($"  profil compose : ({x_u}, {y_u}) | Nash du compose : {contre_fait_univoque_ok}");

// --- Le regime casse : plateau multivoque + couplage ---
var (br1_A, vals_locales) = br1_ensemble();
var profils_composes = br1_A.Select(x => (x: x, y: br2(x))).ToList();
int selection_naive = br1_A.Min();

Console.WriteLine();
Console.WriteLine("[Regime casse : plateau multivoque + couplage]");
Console.WriteLine($"  valeurs locales de u1 (y exogene) : [{string.Join(", ", vals_locales.Select(v => $"'{F3s(v)}'"))}]");
Console.WriteLine($"  BR locale de J1 = [{string.Join(", ", br1_A)}]  (MULTIVOQUE : plateau par symetrie exacte c=0.5)");
Console.WriteLine($"  composition naive (plus petit argmax) : ({selection_naive}, {br2(selection_naive)})");
Console.WriteLine($"  composition des correspondances : [{string.Join(", ", profils_composes.Select(p => $"({p.x}, {p.y})"))}]");
Console.WriteLine();
Console.WriteLine($"{"Profil prevu",-14} {"u1(x, y)",9} {"meilleure deviation J1",23} {"ecart",8}  Verdict");
Console.WriteLine(new string('-', 78));
foreach (var (x, y) in profils_composes)
{
    double ecart = Enumerable.Range(0, 3).Max(xp => u1(xp, br2(xp))) - u1(x, y);
    int x_best = Enumerable.Range(0, 3).OrderByDescending(xp => u1(xp, br2(xp))).First();
    string verdict = ecart > 1e-9 ? "NON-NASH du compose (J1 devie)" : "equilibre du compose";
    Console.WriteLine($"({x}, {y})".PadRight(13) + " " + F3(u1(x, y)).PadLeft(9) + " "
                    + $"({x_best}, {br2(x_best)})".PadRight(12) + " " + F3s(ecart).PadLeft(8) + "  " + verdict);
}

// --- Contre-fait 2 : plateau SANS couplage (r = 0) ---
var ecarts_r0 = profils_composes
    .Select(p => Enumerable.Range(0, 3).Max(xp => u1_r0(xp, br2(xp))) - u1_r0(p.x, br2(p.x)))
    .ToArray();
bool contre_fait_r0_ok = ecarts_r0.All(e => Math.Abs(e) < 1e-9);

Console.WriteLine();
Console.WriteLine($"[Contre-fait 2 : plateau SANS couplage (r=0)] ecarts = [{string.Join(", ", ecarts_r0.Select(e => F3s(e)))}] -> les deux profils survivent (equilibres faibles)");
````

### Lecture du resultat — la composition predit un ensemble, pas un point

Trois mesures, une seule conclusion.

- **Contre-fait 1** (couplage sans plateau) : le profil compose `(1, 1)` est bien un
  equilibre du jeu compose, verifie par enumeration des deviations. Le couplage, seul, ne
  casse rien.
- **Regime casse** : la reponse locale de J1 est le plateau `{0, 1}` — les valeurs locales
  sont `-0.250`, `-0.250`, `-2.250`, l'ex aequo est produit par la symetrie exacte
  `c = 0.5`. La composition **naive** (prendre le plus petit element du plateau) propose
  `(0, 0)`, et le jeu compose le **refuse** : J1, anticipant `y = x`, gagne `+0.300` en
  deviant vers `x = 1`. Le profil `(1, 1)` de la composition des correspondances, lui, est
  un equilibre.
- **Contre-fait 2** (plateau sans couplage, `r = 0`) : les deux ecarts tombent a `+0.000` —
  les deux profils survivent, indifferentiable.

**Ce qui survit au changement de representation, ici :** la *correspondance* — l'ensemble
des meilleures reponses, avec son contexte — et non la *selection*, le point qu'on en
extrait. Un carnet qui lit « la meilleure reponse de la composition est la composition des
meilleures reponses » comme une egalite de points a tort : c'est une egalite d'ensembles, et
la lecture fonctionnelle ne survit que si la reponse locale est univoque.


## Experience 2 — Abstraire, puis mesurer la dette

**Ce que fait l'experience.** Un jeu `G` trop grand pour etre lu d'un coup : six duels 2x2
independants. On le **reduit** a trois etats abstraits (un duel moyen par bloc), on
**resout** l'abstrait exactement, on **retransporte** la strategie, puis on mesure ce que
la strategie retransportee vaut **dans le jeu d'origine** — jamais dans l'abstrait.

C'est le retournement du carnet [Metajeux-03](Metajeux/Metajeux-03-Abstraction-a-Dette-Python.ipynb) :
la question n'est plus « cette representation compresse-t-elle ? » mais « qu'ai-je le droit
d'oublier, et combien cela me coute ? ». On rejoue ici la construction, la resolution
exacte, la retransportation, et la **courbe de dette** sur la chaine de raffinement.


````
// === Experience 2 : abstraire, resoudre, retransporter, mesurer la dette DANS le jeu d'origine ===
// Protocole condense de Metajeux-03 : les six matrices sont des DONNEES citees du source Python.
// La graine 42 de `random` n'est pas portable entre runtimes ; elle n'est donc pas rejouee ici.
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Linq;

public static class MetaJeuxExp2
{
    public const int N_STATES = 6;
    public const int N_ACTIONS = 2;

    public static readonly double[][][] MATRICES_SOURCE =
    {
        new[] { new[] {  2.0, -3.0 }, new[] { -3.0,  2.0 } },
        new[] { new[] { -1.0, -2.0 }, new[] { -2.0, -2.0 } },
        new[] { new[] {  2.0, -3.0 }, new[] {  2.0,  2.0 } },
        new[] { new[] {  1.0, -3.0 }, new[] {  1.0,  0.0 } },
        new[] { new[] { -3.0, -3.0 }, new[] { -3.0, -2.0 } },
        new[] { new[] { -2.0,  1.0 }, new[] {  1.0, -3.0 } }
    };

    public static double VALUE_G { get; private set; }
    public static double DETTE_P3 { get; private set; }
    public static double BR1_MAX { get; private set; }
    public static double BR2_MAX { get; private set; }
    public static bool PLATEAU { get; private set; }
    public static bool NON_DECROISSANTE { get; private set; }
    public static readonly int[][] PARTITION_3 =
    {
        new[] { 0, 1 }, new[] { 2, 3 }, new[] { 4, 5 }
    };

    private static string F4(double v) => v.ToString("F4", CultureInfo.InvariantCulture);
    private static string F4s(double v) => v.ToString("+0.0000;-0.0000;0.0000", CultureInfo.InvariantCulture);

    private static (double[] p, double[] q, double value) Solve2x2(double[][] m)
    {
        for (int i = 0; i < 2; i++)
        for (int j = 0; j < 2; j++)
        {
            bool maxColonne = Enumerable.Range(0, 2).All(x => m[i][j] >= m[x][j]);
            bool minLigne = Enumerable.Range(0, 2).All(y => m[i][j] <= m[i][y]);
            if (maxColonne && minLigne)
                return (
                    i == 0 ? new double[] { 1, 0 } : new double[] { 0, 1 },
                    j == 0 ? new double[] { 1, 0 } : new double[] { 0, 1 },
                    m[i][j]);
        }

        double a = m[0][0], b = m[0][1], c = m[1][0], d = m[1][1];
        double den = a - b - c + d;
        double p = (d - c) / den;
        double q = (d - b) / den;
        if (p < 0 || p > 1 || q < 0 || q > 1)
            throw new InvalidOperationException("Support mixte hors du simplexe.");
        return (new double[] { p, 1 - p }, new double[] { q, 1 - q }, (a * d - b * c) / den);
    }

    private static double[][] AbstractMatrix(int[] block)
    {
        var result = new double[2][];
        for (int a1 = 0; a1 < 2; a1++)
        {
            result[a1] = new double[2];
            for (int a2 = 0; a2 < 2; a2++)
                result[a1][a2] = block.Select(s => MATRICES_SOURCE[s][a1][a2]).Average();
        }
        return result;
    }

    public static (Func<int, int, double> sig1, Func<int, int, double> sig2, double value)
        SolveAndTransport(int[][] partition)
    {
        var solutions = partition.Select(block => Solve2x2(AbstractMatrix(block))).ToList();
        var where = new Dictionary<int, int>();
        for (int k = 0; k < partition.Length; k++)
            foreach (int state in partition[k]) where[state] = k;
        Func<int, int, double> sig1 = (state, action) => solutions[where[state]].p[action];
        Func<int, int, double> sig2 = (state, action) => solutions[where[state]].q[action];
        return (sig1, sig2, solutions.Sum(s => s.value));
    }

    public static (double exploitabilite, double br1, double br2)
        Exploitabilite(Func<int, int, double> sig1, Func<int, int, double> sig2)
    {
        double br1 = 0, br2 = 0;
        for (int state = 0; state < N_STATES; state++)
        {
            double best1 = double.NegativeInfinity;
            for (int a1 = 0; a1 < N_ACTIONS; a1++)
            {
                double gain = 0;
                for (int a2 = 0; a2 < N_ACTIONS; a2++)
                    gain += sig2(state, a2) * MATRICES_SOURCE[state][a1][a2];
                best1 = Math.Max(best1, gain);
            }
            br1 += best1;

            double best2 = double.NegativeInfinity;
            for (int a2 = 0; a2 < N_ACTIONS; a2++)
            {
                double gain = 0;
                for (int a1 = 0; a1 < N_ACTIONS; a1++)
                    gain += sig1(state, a1) * (-MATRICES_SOURCE[state][a1][a2]);
                best2 = Math.Max(best2, gain);
            }
            br2 += best2;
        }
        return (br1 + br2, br1, br2);
    }

    public static void Run()
    {
        Console.WriteLine($"Jeu G : somme de {N_STATES} duels 2x2 independants (zero-sum), donnees citees du carnet source (graine 42 cote Python).");
        for (int s = 0; s < N_STATES; s++)
            Console.WriteLine($"  M_{s} = [[{MATRICES_SOURCE[s][0][0]}, {MATRICES_SOURCE[s][0][1]}], [{MATRICES_SOURCE[s][1][0]}, {MATRICES_SOURCE[s][1][1]}]]");

        VALUE_G = Enumerable.Range(0, N_STATES).Sum(s => Solve2x2(MATRICES_SOURCE[s]).value);
        var (sig1, sig2, v_tilde) = SolveAndTransport(PARTITION_3);
        var (exploitabilite, br1, br2) = Exploitabilite(sig1, sig2);
        DETTE_P3 = exploitabilite;
        BR1_MAX = br1;
        BR2_MAX = br2;

        Console.WriteLine();
        Console.WriteLine($"Valeur exacte de G (somme des {N_STATES} duels) : v(G) = {F4(VALUE_G)}");
        Console.WriteLine($"Abstraction P3 (3 blocs de 2 etats) : facteur de reduction = {(N_STATES / 3.0).ToString("F1", CultureInfo.InvariantCulture)}");
        Console.WriteLine($"  v(G_tilde) = {F4(v_tilde)}");
        Console.WriteLine("Mesure DANS G de la strategie retransportee :");
        Console.WriteLine($"  BR_1 = max_pi1 u_1(pi_1, sigma_2) = {F4(br1)}   (v(G) = {F4(VALUE_G)}, ecart = {F4s(br1 - VALUE_G)})");
        Console.WriteLine($"  BR_2 = max_pi2 u_2(sigma_1, pi_2) = {F4(br2)}   (-v(G) = {F4(-VALUE_G)}, ecart = {F4s(br2 + VALUE_G)})");
        Console.WriteLine($"  exploitabilite = {F4(exploitabilite)} -> la strategie retransportee n'est PAS un equilibre de G");

        var partitions = new (int taille, int[][] partition)[]
        {
            (6, new[] { new[] { 0 }, new[] { 1 }, new[] { 2 }, new[] { 3 }, new[] { 4 }, new[] { 5 } }),
            (4, new[] { new[] { 0, 1 }, new[] { 2 }, new[] { 3 }, new[] { 4, 5 } }),
            (3, new[] { new[] { 0, 1 }, new[] { 2, 3 }, new[] { 4, 5 } }),
            (2, new[] { new[] { 0, 1, 2, 3 }, new[] { 4, 5 } })
        };

        Console.WriteLine();
        Console.WriteLine("Courbe de dette (solve exact + retransport + exploitabilite dans G) :");
        Console.WriteLine($"  {"|G_tilde|",9}  {"v(G_tilde)",10}  {"exploitabilite",14}");
        var points = new List<(int taille, double dette)>();
        foreach (var (taille, partition) in partitions)
        {
            var (s1, s2, value) = SolveAndTransport(partition);
            var (dette, _, _) = Exploitabilite(s1, s2);
            points.Add((taille, dette));
            Console.WriteLine($"  {taille,9}  {F4(value),10}  {F4(dette),14}");
        }

        if (Math.Abs(points[0].dette) >= 1e-12)
            throw new InvalidOperationException("P6 doit rendre 0.");
        NON_DECROISSANTE = Enumerable.Range(0, points.Count - 1)
            .All(i => points[i].dette <= points[i + 1].dette + 1e-9);
        PLATEAU = Math.Abs(points[1].dette - points[2].dette) < 1e-9;
        string forme = (NON_DECROISSANTE ? "monotone non-decroissante" : "NON monotone")
                     + (PLATEAU ? " avec un palier P4 = P3" : "");
        Console.WriteLine($"  forme mesuree : {forme}");
    }
}

MetaJeuxExp2.Run();
````

### Lecture du resultat — la dette se lit des deux cotes

- **La reproduction est exacte** : les six duels repris comme **donnees** du carnet source sont
  identiques, au signe pres, a ceux du carnet Python ; la valeur exacte du jeu est
  `v(G) = -4.2143`, calculee par enumeration de supports, sans aucun processus iteratif.
- **L'abstraction ne survit pas** : la strategie resolue dans les trois etats abstraits,
  retransportee dans `G`, se fait exploiter des deux cotes (`BR_1 - v = +1.3961`,
  `BR_2 + v = +0.5130`). L'exploitabilite vaut `1.9091` : la dette n'est pas un phenomene
  symetrique, elle est **plus grande du cote du joueur qui perd le plus**.
- **La courbe de dette est lisible** : `0.0000` a six etats (sanity check — resoudre `G`
  puis retransporter ne change rien), `1.9091` a quatre etats, `1.9091` a trois etats, et
  plus encore a deux. La forme mesuree est **monotone non decroissante avec un palier
  P4 = P3** : a taille egale, la dette ne depend pas que de la taille mais de **quels**
  etats on fusionne (le carnet source montre qu'une fusion particuliere est gratuite).

**Ce qui survit au changement de representation, ici :** pas l'equilibre — **la dette**,
qui est une propriete transportable, chiffree, et lisible des deux cotes. C'est exactement
ce que le carnet [Metajeux-03](Metajeux/Metajeux-03-Abstraction-a-Dette-Python.ipynb) fait
recouper par une preuve Lean (voir la section « Ce que le lake Lean prouve » plus bas).

## Experience 3a — Temoigner : certifier l'impossibilite par la recherche

**Ce que fait l'experience.** Le dual du chemin minimal : prouver qu'**aucun** chemin plus
court n'existe. Sur les jeux ordinaux 2x2 de Robinson-Goforth, une chambre est une paire
d'ordres stricts sur 4 cases ; les generateurs echangeent deux cases adjacentes **d'un seul
cote** a la fois. Le graphe des chambres est donc le **produit cartesien** de deux copies
du permutoedre `S4`, et la distance s'y **decompose** :

```
d_chambre(G, H) = d_perm(row_G, row_H) + d_perm(col_G, col_H)
```

Le carnet [Metajeux-05](Metajeux/Metajeux-05-Temoins-Impossibilite-Python.ipynb) en tire un
**certificat** : si `d_row + d_col > k`, alors aucun chemin de longueur `<= k` n'existe —
l'impossibilite est un **verdict**, porte par une preuve, et verifiable par une recherche
independante. On rejoue ici la structure produit, le certificat sur trois cas, et le
contre-controle par BFS tronque.

````
// === Experience 3a : certifier un temoin d'impossibilite (structure produit + BFS tronque) ===
// Protocole condense de Metajeux-05 : meme graphe, meme structure.
using System;
using System.Collections.Generic;
using System.Linq;

public static class MetaJeuxExp3a
{
    public record Certificat(string Verdict, int DistanceMinimale, string Preuve);

    public static int NombreOrdres { get; private set; }
    public static int NombreChambres { get; private set; }
    public static int NombreEcarts { get; private set; }
    public static int DiametrePermutoedre { get; private set; }
    public static int DiametreChambres { get; private set; }
    public static int Atteints5 { get; private set; }
    public static int Atteints3 { get; private set; }
    public static int Atteints1 { get; private set; }
    public static int Seuil { get; private set; }
    public static List<string> Verdicts { get; } = new List<string>();

    private static IEnumerable<int[]> Permutations(int n)
    {
        int[] a = Enumerable.Range(1, n).ToArray();
        IEnumerable<int[]> Generate(int k)
        {
            if (k == 1) { yield return (int[])a.Clone(); yield break; }
            for (int i = 0; i < k; i++)
            {
                foreach (var p in Generate(k - 1)) yield return p;
                if (k % 2 == 0) (a[i], a[k - 1]) = (a[k - 1], a[i]);
                else (a[0], a[k - 1]) = (a[k - 1], a[0]);
            }
        }
        return Generate(n);
    }

    private static string Canonique(int[] t)
    {
        var values = t.Distinct().OrderBy(x => x).ToArray();
        return string.Join(",", t.Select(v => Array.IndexOf(values, v) + 1));
    }

    private static string SwapValeursAdjacentes(string ordre, int k)
    {
        var t = ordre.Split(',').Select(int.Parse).ToArray();
        int pk = Array.IndexOf(t, k), pk1 = Array.IndexOf(t, k + 1);
        (t[pk], t[pk1]) = (t[pk1], t[pk]);
        return string.Join(",", t);
    }

    private static IEnumerable<string> AdjPermutoedre(string ordre)
    {
        for (int k = 1; k <= 3; k++) yield return SwapValeursAdjacentes(ordre, k);
    }

    private static IEnumerable<string> AdjChambre(string chambre)
    {
        var sides = chambre.Split('|');
        for (int cote = 0; cote < 2; cote++)
            for (int k = 1; k <= 3; k++)
            {
                string[] t = sides.ToArray();
                t[cote] = SwapValeursAdjacentes(t[cote], k);
                yield return string.Join("|", t);
            }
    }

    private static Dictionary<string, int> Bfs(string depart, Func<string, IEnumerable<string>> voisins)
    {
        var dist = new Dictionary<string, int> { [depart] = 0 };
        var queue = new Queue<string>();
        queue.Enqueue(depart);
        while (queue.Count > 0)
        {
            string u = queue.Dequeue();
            foreach (string v in voisins(u))
                if (!dist.ContainsKey(v))
                {
                    dist[v] = dist[u] + 1;
                    queue.Enqueue(v);
                }
        }
        return dist;
    }

    private static HashSet<string> BfsTronque(string depart, Func<string, IEnumerable<string>> voisins, int profondeur)
    {
        var atteints = new HashSet<string> { depart };
        var frontiere = new HashSet<string> { depart };
        for (int step = 0; step < profondeur; step++)
        {
            var nouvelle = new HashSet<string>();
            foreach (string u in frontiere)
                foreach (string v in voisins(u))
                    if (atteints.Add(v)) nouvelle.Add(v);
            frontiere = nouvelle;
            if (frontiere.Count == 0) break;
        }
        return atteints;
    }

    private static Dictionary<string, Dictionary<string, int>> _distPerm;

    private static Certificat Certifier(string g, string h, int kMax)
    {
        string[] gs = g.Split('|'), hs = h.Split('|');
        int dRow = _distPerm[gs[0]][hs[0]];
        int dCol = _distPerm[gs[1]][hs[1]];
        int dMin = dRow + dCol;
        bool impossible = dMin > kMax;
        return new Certificat(
            impossible ? "IMPOSSIBLE" : "POSSIBLE",
            dMin,
            $"d_perm(row) + d_perm(col) = {dRow} + {dCol} = {dMin} {(impossible ? ">" : "<=")} {kMax}");
    }

    public static void Run()
    {
        var stricts = Permutations(4).Select(Canonique).Distinct().OrderBy(x => x).ToList();
        NombreOrdres = stricts.Count;
        var chambres = (from r in stricts from c in stricts select r + "|" + c).ToList();
        NombreChambres = chambres.Count;

        _distPerm = stricts.ToDictionary(o => o, o => Bfs(o, AdjPermutoedre));
        var distChambres = Bfs("1,2,3,4|1,2,3,4", AdjChambre);

        const string identite = "1,2,3,4";
        NombreEcarts = 0;
        foreach (string g in chambres)
        {
            var s = g.Split('|');
            int dRow = _distPerm[identite][s[0]];
            int dCol = _distPerm[identite][s[1]];
            if (distChambres[g] != dRow + dCol) NombreEcarts++;
        }

        DiametrePermutoedre = _distPerm.Values.Max(d => d.Values.Max());
        DiametreChambres = distChambres.Values.Max();

        Console.WriteLine($"Permutoedre : {NombreOrdres} ordres stricts | Chambres : {NombreChambres} sommets");
        Console.WriteLine($"Verification exhaustive sur {NombreChambres} chambres : {NombreEcarts} ecarts");
        Console.WriteLine($"Structure produit : {(NombreEcarts == 0 ? "EXACTE" : "NON")}");
        Console.WriteLine($"Diametre permutoedre : {DiametrePermutoedre} | Diametre chambres : {DiametreChambres}");

        int nonSym = 0;
        foreach (string o1 in stricts)
            foreach (string o2 in stricts)
                if (_distPerm[o1][o2] != _distPerm[o2][o1]) nonSym++;
        int violations = 0;
        foreach (string o1 in stricts)
            foreach (string o2 in stricts)
                foreach (string o3 in stricts)
                    if (_distPerm[o1][o3] > _distPerm[o1][o2] + _distPerm[o2][o3]) violations++;
        Console.WriteLine($"Table 24x24 : {nonSym} asymetries | inegalite triangulaire : {violations} violations sur {24 * 24 * 24} triplets");

        string chambreIdentite = "1,2,3,4|1,2,3,4";
        string renversement = "4,3,2,1|4,3,2,1";
        string hB = "1,2,3,4|3,4,1,2";
        string hC = "1,2,4,3|1,2,3,4";
        var cas = new (string nom, string g, string h, int k)[]
        {
            ("CAS A (antipode)", chambreIdentite, renversement, 5),
            ("CAS B (ligne identique)", chambreIdentite, hB, 3),
            ("CAS C (frontiere)", chambreIdentite, hC, 1)
        };
        foreach (var (nom, g, h, k) in cas)
        {
            var cert = Certifier(g, h, k);
            Verdicts.Add(cert.Verdict);
            Console.WriteLine($"{nom}, k_max={k} : {cert.Verdict} - {cert.Preuve}");
        }

        var atte5 = BfsTronque(chambreIdentite, AdjChambre, 5);
        var atte3 = BfsTronque(chambreIdentite, AdjChambre, 3);
        var atte1 = BfsTronque(chambreIdentite, AdjChambre, 1);
        Atteints5 = atte5.Count;
        Atteints3 = atte3.Count;
        Atteints1 = atte1.Count;
        Console.WriteLine($"CAS A : {Atteints5} sommets a <= 5 pas, antipode present ? {atte5.Contains(renversement)}");
        Console.WriteLine($"CAS B : {Atteints3} sommets a <= 3 pas, H_B present ? {atte3.Contains(hB)}");
        Console.WriteLine($"CAS C : {Atteints1} sommets a <= 1 pas, H_C present ? {atte1.Contains(hC)}");

        Seuil = Enumerable.Range(0, 13).First(k => Certifier(chambreIdentite, renversement, k).Verdict == "POSSIBLE");
        Console.WriteLine($"Verdict POSSIBLE pour l'antipode a partir de k = {Seuil} (distance reelle = 12) : bascule nette.");
    }
}

MetaJeuxExp3a.Run();
````

### Lecture du resultat — un verdict, et sa preuve

- **La structure produit est exacte** : sur les 576 chambres, la distance du produit
  egale la somme des distances des deux cotes — **0 ecart**. Les diametres mesures sont `6`
  (permutoedre) et `12` (chambres), et la table 24x24 est symetrique et triangulaire
  (0 violation sur 13 824 triplets).
- **Le certificat donne le verdict, et sa preuve** : le cas A (l'antipode) est `IMPOSSIBLE`
  a `k = 5` parce que `6 + 6 = 12 > 5` ; le cas B est `IMPOSSIBLE` a `k = 3` parce que
  `0 + 4 = 4 > 3` ; le cas C est `POSSIBLE` a `k = 1` parce que `1 + 0 = 1 <= 1`.
- **La recherche confirme le certificat** : le BFS tronque atteint 235 sommets a 5 pas
  (l'antipode est absent), 68 a 3 pas (`H_B` absent), 7 a 1 pas (`H_C` present). Les deux
  voies — le raisonnement analytique et la recherche exhaustive — **concordent**.
- **La bascule est nette** : le verdict de l'antipode bascule de `IMPOSSIBLE` a `POSSIBLE`
  exactement a `k = 12`, la distance reelle. Aucun palier, aucune zone grise.

**Ce qui survit au changement de representation, ici :** le **verdict**. Un temoin
d'impossibilite reste vrai quand on change de representation, parce que son contenu n'est
pas une mesure mais une **structure** — la decomposition du produit, qui se verifie.

## Experience 3b — Changer de substrat : ce qui ne survit pas

**La question falsifiable.** Sur 2x2, le verdict mesure par le carnet
[Metajeux-04](Metajeux/Metajeux-04-Chemin-Minimal-Robinson-Goforth-Python.ipynb) est que
traverser les murs (passer par des ex aequo) ne raccourcit **jamais** un chemin minimal.
Est-ce une propriete du jeu, ou une propriete du petit substrat ? Le carnet
[Metajeux-06](Metajeux/Metajeux-06-Chemin-Minimal-3x2-Ordinal-Python.ipynb) monte au
substrat **3x2 ordinal** — six cases par joueur, 720 tables strictes, 518 400 chambres
strictes — et **refute** la generalisation. On rejoue ici les deux recensements, dans le
meme modele ferme (fusion + scission des murs), pour comparer.

````
// === Experience 3b : changer de substrat (3x2 ordinal) -- le verdict survit-il ? ===
// Protocole condense de Metajeux-06 : meme modele ferme, recensements exhaustifs.
using System;
using System.Collections.Generic;
using System.Diagnostics;
using System.Globalization;
using System.Linq;

public static class MetaJeuxExp3b
{
    public record Exemple(string A, string B, int DistancePermutoedre, int DistanceTreillis);

    public static int Paires4 { get; private set; }
    public static int Raccourcis4 { get; private set; }
    public static int Egalites4 { get; private set; }
    public static int Paires6 { get; private set; }
    public static int Raccourcis6 { get; private set; }
    public static int Egalites6 { get; private set; }
    public static Exemple ExempleMesure { get; private set; }
    public static List<string> Tables3x2Strictes { get; } = new List<string>();
    public static int DiametrePermutoedre4 { get; private set; }
    public static int DiametreTreillis4 { get; private set; }
    public static int DiametrePermutoedre6 { get; private set; }
    public static int DiametreTreillis6 { get; private set; }

    private static Dictionary<string, List<string>> _adjPermutoedre6;
    private static Dictionary<string, List<string>> _adjTreillis6;

    private static IEnumerable<int[]> Tuples(int n)
    {
        int[] t = new int[n];
        IEnumerable<int[]> Generate(int k)
        {
            if (k == n) { yield return (int[])t.Clone(); yield break; }
            for (int v = 1; v <= n; v++)
            {
                t[k] = v;
                foreach (var x in Generate(k + 1)) yield return x;
            }
        }
        return Generate(0);
    }

    private static string Canonique(int[] t)
    {
        var values = t.Distinct().OrderBy(x => x).ToArray();
        return string.Join(",", t.Select(v => Array.IndexOf(values, v) + 1));
    }

    private static List<string> OrdresFaibles(int n) => Tuples(n).Select(Canonique).Distinct().OrderBy(x => x).ToList();

    private static string SwapAdjacent(string ordre, int k)
    {
        var t = ordre.Split(',').Select(int.Parse).ToArray();
        int pk = Array.IndexOf(t, k), pk1 = Array.IndexOf(t, k + 1);
        (t[pk], t[pk1]) = (t[pk1], t[pk]);
        return string.Join(",", t);
    }

    private static string Fusion(string ordre, int k)
    {
        var t = ordre.Split(',').Select(int.Parse).ToArray();
        return Canonique(t.Select(v => v == k + 1 ? k : v).ToArray());
    }

    private static IEnumerable<int[]> Combinaisons(int[] items, int k)
    {
        int[] indices = Enumerable.Range(0, k).ToArray();
        while (true)
        {
            yield return indices.Select(i => items[i]).ToArray();
            int i = k - 1;
            while (i >= 0 && indices[i] == items.Length - k + i) i--;
            if (i < 0) yield break;
            indices[i]++;
            for (int j = i + 1; j < k; j++) indices[j] = indices[j - 1] + 1;
        }
    }

    private static List<string> Scissions(string ordre)
    {
        var result = new List<string>();
        var t = ordre.Split(',').Select(int.Parse).ToArray();
        foreach (int v in t.Distinct().OrderBy(x => x))
        {
            var positions = Enumerable.Range(0, t.Length).Where(i => t[i] == v).ToArray();
            if (positions.Length < 2) continue;
            for (int j = 1; j < positions.Length; j++)
                foreach (int[] subset in Combinaisons(positions, j))
                {
                    var set = new HashSet<int>(subset);
                    var valeurs = new int[t.Length];
                    for (int i = 0; i < t.Length; i++)
                    {
                        if (t[i] == v) valeurs[i] = set.Contains(i) ? v : v + 1;
                        else valeurs[i] = t[i] > v ? t[i] + 1 : t[i];
                    }
                    result.Add(Canonique(valeurs));
                }
        }
        return result;
    }

    private static List<string> VoisinsPermutoedre(string ordre, int n)
    {
        var result = new List<string>();
        for (int k = 1; k < n; k++) result.Add(SwapAdjacent(ordre, k));
        return result;
    }

    private static List<string> VoisinsTreillis(string ordre, int n)
    {
        var result = new List<string>();
        var values = ordre.Split(',').Select(int.Parse).ToHashSet();
        for (int k = 1; k < n; k++)
            if (values.Contains(k) && values.Contains(k + 1)) result.Add(Fusion(ordre, k));
        result.AddRange(Scissions(ordre));
        return result;
    }

    private static Dictionary<string, int> Bfs(string source, Dictionary<string, List<string>> adjacency)
    {
        var distance = new Dictionary<string, int> { [source] = 0 };
        var queue = new Queue<string>();
        queue.Enqueue(source);
        while (queue.Count > 0)
        {
            string u = queue.Dequeue();
            foreach (string v in adjacency[u])
                if (!distance.ContainsKey(v))
                {
                    distance[v] = distance[u] + 1;
                    queue.Enqueue(v);
                }
        }
        return distance;
    }

    public static Dictionary<string, int> DistancesPermutoedre3x2(string source) => Bfs(source, _adjPermutoedre6);
    public static Dictionary<string, int> DistancesTreillis3x2(string source) => Bfs(source, _adjTreillis6);

    private static (int paires, int raccourcis, int egalites, Exemple exemple)
        Recenser(List<string> strictes, Dictionary<string, List<string>> adjPerm, Dictionary<string, List<string>> adjTreillis)
    {
        int raccourcis = 0, egalites = 0;
        Exemple exemple = null;
        const string cibleA = "1,2,3,4,5,6", cibleB = "1,5,6,4,3,2";
        foreach (string a in strictes)
        {
            var dPerm = Bfs(a, adjPerm);
            var dTreillis = Bfs(a, adjTreillis);
            foreach (string b in strictes)
            {
                if (dTreillis[b] < dPerm[b])
                {
                    raccourcis++;
                    if ((a == cibleA && b == cibleB) || exemple == null)
                        exemple = new Exemple(a, b, dPerm[b], dTreillis[b]);
                }
                else if (dTreillis[b] == dPerm[b]) egalites++;
            }
        }
        return (strictes.Count * strictes.Count, raccourcis, egalites, exemple);
    }

    public static void Run()
    {
        var of4 = OrdresFaibles(4);
        var st4 = of4.Where(x => x.Split(',').Distinct().Count() == 4).ToList();
        var of6 = OrdresFaibles(6);
        var st6 = of6.Where(x => x.Split(',').Distinct().Count() == 6).ToList();
        Tables3x2Strictes.Clear();
        Tables3x2Strictes.AddRange(st6);
        Console.WriteLine($"Substrat 2x2 : ordres faibles = {of4.Count} | stricts = {st4.Count} | chambres = {st4.Count * st4.Count}");
        Console.WriteLine($"Substrat 3x2 : ordres faibles = {of6.Count} | stricts = {st6.Count} | chambres = {st6.Count * st6.Count}");

        var adjP4 = st4.ToDictionary(x => x, x => VoisinsPermutoedre(x, 4));
        var adjC4 = of4.ToDictionary(x => x, x => VoisinsTreillis(x, 4));
        _adjPermutoedre6 = st6.ToDictionary(x => x, x => VoisinsPermutoedre(x, 6));
        _adjTreillis6 = of6.ToDictionary(x => x, x => VoisinsTreillis(x, 6));

        DiametrePermutoedre4 = Bfs(st4[0], adjP4).Values.Max();
        DiametreTreillis4 = 2 * Bfs(of4[0], adjC4).Values.Max();
        DiametrePermutoedre6 = Bfs(st6[0], _adjPermutoedre6).Values.Max();
        DiametreTreillis6 = 2 * Bfs(of6[0], _adjTreillis6).Values.Max();
        Console.WriteLine($"n=4 : permutoedre diametre = {DiametrePermutoedre4} | treillis diametre = {DiametreTreillis4}");
        Console.WriteLine($"n=6 : permutoedre diametre = {DiametrePermutoedre6} | treillis diametre = {DiametreTreillis6}");

        var chrono = Stopwatch.StartNew();
        var r4 = Recenser(st4, adjP4, adjC4);
        Paires4 = r4.paires; Raccourcis4 = r4.raccourcis; Egalites4 = r4.egalites;
        Console.WriteLine($"Recensement 2x2 (modele ferme) : {Paires4} paires | raccourcis = {Raccourcis4} | egalites = {Egalites4}");
        var r6 = Recenser(st6, _adjPermutoedre6, _adjTreillis6);
        Paires6 = r6.paires; Raccourcis6 = r6.raccourcis; Egalites6 = r6.egalites; ExempleMesure = r6.exemple;
        chrono.Stop();
        double pourcentage = 100.0 * Raccourcis6 / Paires6;
        Console.WriteLine($"Recensement 3x2 (modele ferme) : {Paires6} paires | raccourcis = {Raccourcis6} ({pourcentage.ToString("F1", CultureInfo.InvariantCulture)} %) | egalites = {Egalites6}");
        Console.WriteLine($"  exemple : ({ExempleMesure.A}) -> ({ExempleMesure.B}) : d_perm = {ExempleMesure.DistancePermutoedre}, d_treillis = {ExempleMesure.DistanceTreillis}");
        Console.WriteLine($"  ({chrono.Elapsed.TotalSeconds.ToString("F1", CultureInfo.InvariantCulture)}s -- exhaustif, pas d'echantillonnage)");
    }
}

MetaJeuxExp3b.Run();
````

### Lecture du resultat — 15,0 % de raccourcis la ou il n'y en avait aucun

**Le verdict 2x2 ne survit pas au changement de substrat.**

- A **4 cases**, le recensement exhaustif des 576 paires de chambres donne **0 raccourci** :
  traverser les murs ne rend jamais le chemin plus court. Les diametres sont egaux
  (`6 = 6`), et l'arithmetique explique pourquoi : le gain net d'un traversee vaut
  `(n - 1)(n - 4) / 2`, soit **0** a `n = 4`.
- A **6 cases**, le meme modele, le meme code, un seul parametre change : **77 760 paires
  sur 518 400 raccourcissent**, soit **15,0 %**. Le diametre par cote se contracte de
  `15` a `10`, et le gain net theorique vaut `(6 - 1)(6 - 4) / 2 = 5`. Un exemple
  explicite est affiche : deux tables dont la distance dans le permutoedre est `9` mais
  seulement `8` par les murs.

**Ce qui survit, et ce qui ne survit pas.** Le **verdict** « la stratification ne comprime
pas la metrique » etait une propriete du petit substrat : il **ne survit pas**. Ce qui
survit, en revanche, c'est la **methode** qui produit le verdict — un constructeur qui
produit le chemin, un verificateur qui le re-derive et re-cherche la minimalite, un
recensement **exhaustif** plutot qu'un echantillon, et un changement de substrat qui ne
touche qu'un parametre. C'est exactement ce que le carnet
[Metajeux-06](Metajeux/Metajeux-06-Chemin-Minimal-3x2-Ordinal-Python.ipynb) met en avant :
ce n'est pas le chiffre qui est nouveau, c'est la **frontiere de la loi**, mesuree.

## Ce que le lake Lean prouve

Le carnet [Metajeux-03](Metajeux/Metajeux-03-Abstraction-a-Dette-Python.ipynb) fait
elaborer le module `Abstraction` du lake `game_theory_lean` et en affiche le verdict :

- `raffinement_aggrave_la_dette` : raffiner une partition **aggrave** la dette — l'ecart
  mesure dans le carnet n'est donc pas une coincidence de la chaine de partitions
  choisie ;
- la theoreme ne depend que de `propext` (aucun `sorry`, aucun axiome fort : le carnet
  source affiche la liste des axiomes et le compte des `sorry` du module).

Ce carnet **lit** ce resultat, il ne le re-preuve pas : le lake n'est pas construit ici, et
un resultat formel recopie sans etre rejoue n'est pas une preuve. Ce qui est rejoue ici,
c'est la **mesure** ; le module Lean en est le **contre-triage**.

La lecon de la sous-serie tient dans cette articulation : une mesure seule est un fait
local, une preuve seule est un fait formel, les deux qui **se recoupent** sont une
connaissance.

## A retenir

- **Composer** ne survit qu'en **ensembles**. La correspondance des meilleures reponses se
  compose ; la selection d'un de ses elements ne se compose que si la reponse locale est
  univoque. Multivocite + couplage suffisent a briser la selection (`+0.300` d'ecart).
- **Abstraire** ne survit pas du tout a l'equilibre, mais la **dette** survit : elle est
  chiffree (`1.9091`), elle se lit des deux cotes (`+1.3961` / `+0.5130`), et elle est
  lisible des deux cotes *et* recoupeee par une preuve Lean.
- **Certifier** survit au changement de representation : le verdict d'impossibilite porte
  une structure (decomposition du produit), pas une mesure, et il se verifie par la
  recherche (`0` ecart sur 576 chambres, `13 824` triplets sans violation).
- **Une mesure** ne survit pas : `0` raccourci sur 576 paires a 4 cases, `15,0 %` sur
  518 400 paires a 6 cases. Le verdict « la stratification ne comprime pas la metrique »
  etait une propriete du petit substrat.

**Ce qui survit, en une phrase :** la **methode** (composer des correspondances, mesurer la
dette dans le jeu d'origine, certifier par la recherche, recenser exhaustivement) — et non
les nombres, qui sont des proprietes du substrat.
