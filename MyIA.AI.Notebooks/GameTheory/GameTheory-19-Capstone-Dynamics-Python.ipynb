## 3. Expérience 1 — L'effondrement et son remède (Dynamics-01 + 02)

En horizon **fini**, la coopération s'effondre par induction à rebours : au dernier tour dévier domine, donc à l'avant-dernier aussi, et toute la chaîne remonte. En horizon **infini** escompté, une menace crédible — le grim trigger — la soutient dès que $\delta$ dépasse un seuil. Et ce seuil, le lake l'énonce analytiquement : `Discounting.lean` le force en nombres. L'expérience : le retrouver numériquement, et vérifier que **le théorème vit déjà dans les nombres**.

```python
import sys
import numpy as np

print(f"numpy {np.__version__} | python {sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}")

# Le PD canonique de Dynamics-01 (T > R > P > S et 2R > T + S)
T, R, P, S = 5.0, 3.0, 1.0, 0.0
assert T > R > P > S and 2 * R > T + S, "parametres PD non canoniques"

# (a) Horizon fini N tours, delta = 1 : l'induction a rebours rend (D, D) unique SPNE
N = 10
print(f"Horizon fini N={N} : paiement SPNE (D,D)xN = {P * N:.1f} ; "
      f"la cooperation fictive ({R * N:.1f}) n'y est pas credible.")

# (b) Horizon infini escompte : cooperation soutenable si V_C >= V_D
#     V_C = R/(1-d) ; deviation : T une fois, puis la punition P a jamais
def VC(d):
    return R / (1 - d)

def VD_grim(d):
    return T + d * P / (1 - d)

delta_star = (T - R) / (T - P)  # seuil analytique -- celui que Discounting.lean enonce
print(f"\nSeuil analytique delta* = (T-R)/(T-P) = ({T:g}-{R:g})/({T:g}-{P:g}) = {delta_star:.4f}")
print(f"{'delta':>7} {'V_C(coop)':>11} {'V_D(deviat)':>12} {'soutenable':>11}")
for d in (0.30, 0.40, 0.50, 0.60, 0.70):
    print(f"{d:>7.2f} {VC(d):>11.4f} {VD_grim(d):>12.4f} {'OUI' if VC(d) >= VD_grim(d) else 'NON':>11}")
print(f"\nAu seuil exact delta* = {delta_star:.4f} : V_C = {VC(delta_star):.4f}, "
      f"V_D = {VD_grim(delta_star):.4f} (indifference)")
```

```python
# Punition plus dure => cooperation plus facile a soutenir (l'intuition du folk theorem,
# version numerique de la comparaison grim vs tit-for-tat de Dynamics-01, section seuils)
def VD_tft(d):
    # TFT punit UN tour (le deviant subit S en jouant C dans la punition) puis pardonne
    return T + d * S + d ** 2 * R / (1 - d)

deltas = np.linspace(0.01, 0.99, 20001)
ok_tft = VC(deltas) >= VD_tft(deltas)
delta_tft = float(deltas[np.argmax(ok_tft)]) if ok_tft.any() else float("nan")

print(f"Seuil grim trigger : delta* = {delta_star:.4f}  (punition P a jamais)")
print(f"Seuil tit-for-tat  : delta* = {delta_tft:.4f}  (une punition, puis pardon)")
print()
print("Le grim trigger soutient la cooperation PLUS TOT : une menace plus dure est")
print("plus facile a rendre credible. Ce sont les nombres du lemme du lake :")
print("grim_trigger_sustains_iff (RepeatedGames/GrimTrigger.lean, 0 sorry).")
```

### Lecture du résultat

Le seuil numérique `(T-R)/(T-P) = 0,5` coïncide avec la condition analytique que [`Discounting.lean`](game_theory_lean/RepeatedGames/Discounting.lean) énonce sur les mêmes paramètres : la coopération est soutenable si et seulement si $\delta \ge (T-R)/(T-P)$. La menace éternelle soutient la coopération dès δ ≥ 0,5 ; la punition d'un seul tour exige un joueur bien plus patient. **C'est la crédibilité de la menace — non sa douceur — qui fait la coopération** : le théorème Folk vu d'en bas, avant sa forme générale (le stretch `folk_theorem_discounted`, cf. §6).

## 4. Expérience 2 — Coopérer par preuve (Dynamics-04 + 04b)

La répétition a besoin d'un futur. Sans futur — one-shot —, la **transparence des programmes** offre une autre voie : si chacun peut lire le code de l'autre avant d'agir, la coopération peut venir d'une prédiction mutuelle **prouvée**, pas d'une menace. Deux niveaux de lecture s'y distinguent (la faille L1/L2 de Dynamics-04b) :

- **L1, la preuve de surface** : chercher dans le source la déclaration de coopération — sondable par une branche morte ;
- **L2, la preuve par Löb** : coopérer seulement si la coopération de l'autre est **prouvable** dans un système de prouvabilité — le point fixe FairBot×FairBot, prouvé dans le lake.

L'expérience : la grille mesurée des bots en L1, le témoin qui la fait tomber, et ce que le L2 garantit.

```python
import inspect
import itertools

STEP_BUDGET = 1000  # borne de simulation, comme Dynamics-04 (verdict CALCULE, jamais coupe)


class SimulationTimeout(Exception):
    """Levee quand le budget de pas est depasse (borne de Dynamics-04)."""


_state = {"steps": 0}


def _tick():
    _state["steps"] += 1
    if _state["steps"] > STEP_BUDGET:
        raise SimulationTimeout("budget de pas depasse")


def CooperateBot(_other_src):
    return "C"


def DefectBot(_other_src):
    return "D"


def FairBot_surface(other_src):
    # L1 : preuve de SURFACE -- la declaration de cooperation est-elle visible dans le source ?
    _tick()
    return "C" if 'return "C"' in other_src else "D"


def ProbeBot(other_src):
    # La sonde de Dynamics-04b : affiche une branche morte cooperative, joue D a l'execution
    _tick()
    _sonde = 'return "C"'  # cette ligne ne s'execute jamais -- elle n'existe que pour etre LUE
    return "D"


BOTS = {"CooperateBot": CooperateBot, "DefectBot": DefectBot,
        "FairBot_surface": FairBot_surface, "ProbeBot": ProbeBot}
SRC = {name: inspect.getsource(bot) for name, bot in BOTS.items()}  # les programmes se lisent MUTUELLEMENT

PAYOFF = {("C", "C"): R, ("C", "D"): S, ("D", "C"): T, ("D", "D"): P}


def duel(na, nb):
    a = BOTS[na](SRC[nb])
    b = BOTS[nb](SRC[na])
    return a, b, PAYOFF[(a, b)], PAYOFF[(b, a)]


entete = "A \\ B"
print(f"{entete:>18}" + "".join(f"{n:>18}" for n in BOTS))
for na in BOTS:
    row = f"{na:>18}"
    for nb in BOTS:
        _state["steps"] = 0
        a, b, pa, pb = duel(na, nb)
        row += f"{a}/{b} ({pa:.0f},{pb:.0f})".rjust(18)
    print(row)
print("\n16/16 duels rendent un verdict calcule dans la borne (step_cap=1000, bots deterministes).")
print("La grille complete (5 bots, verificateur de coherence, CUPOD/PrudentBot) vit dans Dynamics-04.")
```

```python
# Le temoin L1/L2 de Dynamics-04b : la preuve de surface est sondable
_state["steps"] = 0
a, b, pa, pb = duel("FairBot_surface", "ProbeBot")
print(f"FairBot_surface x ProbeBot : {a}/{b} -> paiements ({pa:.0f}, {pb:.0f})")
print("La sonde (branche morte cooperative) fait tomber la preuve de surface : FairBot_surface est exploite.")
print()
print("Le FairBot de Dynamics-04b exige une PREUVE (modalite de provabilite, theoreme de Lob) :")
print("  - face a un cooperateur prouvable, il coopere      : fairBot_cooperates_with_provable_cooperator")
print("  - sans preuve de cooperation, il defectionne        : fairBot_defects_without_proof")
print("  - FairBot x FairBot coopere (point fixe de Lob)     : fairBot_fairBot_cooperation, loeb_mutual_fixedpoint")
print("Les quatre vivent dans ProgramGames/FairBot.lean -- 0 sorry (inventaire mesure en section 6).")
```

### Lecture du résultat

La coopération one-shot **existe** — FairBot×FairBot se coordonne en (C,C) sans aucune menace — mais sa source change tout : en L1 (surface), elle est **sondable** et un ProbeBot l'exploite (paiement S = 0) ; en L2 (Löb), la coopération n'est accordée que sur preuve, et `fairBot_defects_without_proof` garantit qu'aucun affichage mensonger ne la déclenche. **La preuve mutuelle remplace la menace — à condition d'être une preuve, pas une lecture.** C'est la réponse de la sous-série à la question du capstone : la menace dégradée (finie) échoue, la preuve dégradée (de surface) est trompée ; la preuve réelle tient.

## 5. Expérience 3 — Le budget change l'issue (Dynamics-05 + 06 + 07)

Troisième voie de dégradation : le coût du raisonnement lui-même. Dynamics-05 construit des agents à **budget explicite** et prouve des certificats par familles (coopération mutuelle, non-exploitation) ; Dynamics-06 en tire le vocabulaire honnête des preuves bornées ; Dynamics-07 sépare la preuve **syntaxique** de la simulation **comportementale**. L'expérience : rejouer le même duel FairBot×FairBot à **budget de lecture croissant** — sous le seuil, la preuve ne tient plus dans la fenêtre, et la coopération tombe.

```python
# Le duel de l'experience 2, rejoue a budget de lecture borne (le compagnon Python de Dynamics-05)


def FairBot_borne(other_src, budget):
    """FairBot a budget de lecture borne : ne voit que les `budget` premiers caracteres.

    La borne est la version lisible du budget de preuve de Dynamics-05 : sous le seuil,
    la preuve ne tient plus dans la fenetre de lecture.
    """
    lu = other_src[:budget]
    return "C" if 'return "C"' in lu else "D"


src_fb = SRC["FairBot_surface"]
pos_preuve = src_fb.index('return "C"')
print(f"Source de FairBot_surface : {len(src_fb)} caracteres ; "
      f"la declaration de cooperation vit au caractere {pos_preuve}.")
print()
print(f"{'budget':>8} {'issue du duel':>18}  lecture")
for budget in (10, 20, pos_preuve + 2, len(src_fb)):
    a = FairBot_borne(src_fb, budget)
    issue = f"{a}/{a} {'(C,C) cooperation' if a == 'C' else '(D,D) effondrement'}"
    print(f"{budget:>8} {issue:>18}  ...{src_fb[:budget][-24:]!r}")
print()
print("Sous le seuil de lecture, la preuve ne tient pas dans la fenetre : la cooperation tombe.")
print("Le budget est un PARAMETRE du jeu, pas un detail d'implementation (Dynamics-06).")
```

```python
# Les familles de certificats de Dynamics-05, rejouees cote Python sur le banc borne ci-dessus
certificats = []


def verifier(nom, verdict_python, theoreme_lean, source):
    """Rejoue le geste du compagnon Python de Dynamics-05 : verifier, nommer, citer."""
    certificats.append((nom, verdict_python))
    print(f"{'OK' if verdict_python else 'KO'}  {nom:<46} {theoreme_lean}")
    if not verdict_python:
        print(f"    attendu depuis : {source}")


# Famille 1 : cooperation mutuelle (a budget suffisant, le miroir coopere avec lui-meme)
verifier("mirror coopere avec lui-meme (budget plein)",
         FairBot_borne(src_fb, len(src_fb)) == "C",
         "mirror_mirror", "Dynamics-05 + ProgramGames/Bounded.lean")

# Famille 2 : non-exploitation (ProbeBot n'obtient jamais plus que T d'un agent borne)
_state["steps"] = 0
_, _, pa, pb = duel("FairBot_surface", "ProbeBot")
verifier("ProbeBot n'extrait pas plus que T",
         pb <= T,
         "defectBotBounded_unexploitable", "Dynamics-05 + ProgramGames/Bounded.lean")

# La question de Dynamics-07, posee sur une instance : la grille mesure des issues CALCULEES ;
# les theoremes du lake garantissent des equilibres. Les deux familles ci-dessus sont la
# traduction Python de garanties Lean -- le certificat Python est un TEMOIN, jamais une preuve.
print()
print(f"{len(certificats)}/{len(certificats)} temoins rejoues : la question syntaxique vs")
print("comportementale (Dynamics-07) reste ouverte -- un temoin vert n'est pas un theoreme.")
```

### Lecture du résultat

Le même FairBot, le même adversaire, la même règle — seul le **budget** change, et l'issue bascule de (C,C) à (D,D). La coopération n'est pas une propriété des programmes : c'est une propriété du couple (programmes, budget). Dynamics-06 en tire le vocabulaire honnête — une preuve bornée est une preuve **à condition de dire sa borne** — et Dynamics-07 la frontière : la simulation mesure des issues calculées, le lake garantit des équilibres ; les certificats Python ci-dessus sont des **témoins**, jamais des preuves.

## 6. Ce que le lake Lean prouve

La sous-série ne simule pas seulement : elle prouve. Le lake [`game_theory_lean`](game_theory_lean/) porte deux modules sur le fil du capstone — `RepeatedGames` (la répétition : Stage, Discounting, GrimTrigger, Folk, Panel) et `ProgramGames` (la transparence : Basic, Bounded, FairBot). L'inventaire ci-dessous est **mesuré sur les fichiers du lake au moment de l'exécution** — nombres falsifiables, pas déclarés.

```python
from pathlib import Path
import re

lake = next((p for p in (Path("game_theory_lean"),
                         Path("MyIA.AI.Notebooks/GameTheory/game_theory_lean")) if p.is_dir()), None)
assert lake is not None, "lake introuvable -- executer depuis le dossier GameTheory/"

print(f"lake : {lake} \n")
for mod in ("ProgramGames", "RepeatedGames"):
    fr = sorted(p for p in (lake / mod).glob("*.lean") if not p.name.endswith("_en.lean"))
    thm = set()
    sry = 0
    for p in fr:
        texte = p.read_text(encoding="utf-8")
        thm.update(re.findall(r"^(?:theorem|lemma)\s+(\w+)", texte, re.M))
        sry += len(re.findall(r"\bsorry\b", texte))
    print(f"{mod:<15} {len(fr)} fichiers FR | {len(thm)} theoremes/lemmes distincts | {sry} sorry (code + prose)")

print()
print("Les quatre piliers du fil, localises par lecture des sources :")
for nom in ("grim_trigger_sustains_iff", "folk_theorem_discounted",
            "fairBot_fairBot_cooperation", "loeb_mutual_fixedpoint"):
    lieu = "?"
    for p in sorted(lake.rglob("*.lean")):
        if p.name.endswith("_en.lean"):
            continue
        if re.search(rf"^(?:theorem|lemma)\s+{nom}\b", p.read_text(encoding="utf-8"), re.M):
            lieu = f"{p.parent.name}/{p.name}"
            break
    print(f"  {nom:<38} {lieu}")
```

### Lecture du lake

- **Le seuil est un théorème** : `grim_trigger_sustains_iff` (δ ≥ seuil, **0 sorry**) — la condition numérique de l'expérience 1, prouvée dans les deux sens.
- **Le théorème Folk général est un stretch assumé** : `folk_theorem_discounted` porte ses sorry **documentés** (5+5, tolérance BG du prover, cf. `lakefile.lean`) — la sous-série distingue ce qui est prouvé de ce qui reste à prouver, `Panel.lean` ajoutant ses 3 ancrages déclarés. Tout le lake : **4 sorry distincts, tous localisés**.
- **L'arc Löb est entièrement prouvé** : `ProgramGames` — 38 théorèmes/lemmes, **0 sorry** — dont le point fixe `fairBot_fairBot_cooperation` et sa garde `fairBot_defects_without_proof`.
- **Ce que la simulation ne prouve pas** : les grilles des expériences 2-3 mesurent des issues calculées sous borne ; les théorèmes du lake garantissent des équilibres pour tous les budgets. La frontière est le sujet de Dynamics-07.
