## Introduction

Les **jeux combinatoires** sont une classe speciale de jeux :
- **2 joueurs** (Gauche et Droite)
- **Alternance** des coups
- **Information parfaite** (pas de hasard, pas de coups caches)
- **Terminaison** : le jeu finit toujours
- **Normal play** : le joueur qui ne peut plus jouer perd

Contrairement aux jeux stratégiques (Nash, equilibres mixtes), les jeux combinatoires n'ont pas d'utilites numériques - seulement **victoire ou defaite**.

### Jeux impartiaux vs partizans

| Type | Coups disponibles | Exemples |
|------|------------------|----------|
| **Impartial** | Mêmes pour les deux joueurs | Nim, Jeux de soustraction |
| **Partizan** | Différents selon le joueur | Echecs, Go, Domineering |

## 1. Positions P et N

Dans un jeu combinatoire, chaque position est soit :

- **P-position** (Previous player wins) : Le joueur qui vient de jouer gagne
- **N-position** (Next player wins) : Le joueur qui va jouer gagne

### Règles de classification

1. Une position **terminale** (sans coups possibles) est une **P-position**
2. Une position est une **N-position** si elle a au moins un coup vers une P-position
3. Une position est une **P-position** si tous ses coups menent a des N-positions

### Exemple : Jeu de 1-2-3

Règles : Un tas de n jetons, on peut retirer 1, 2 ou 3 jetons. Qui retire le dernier gagne.

```python
def classify_123_game(n_max: int) -> dict:
    """
    Classifie les positions du jeu 1-2-3.
    Retourne un dict {position: 'P' ou 'N'}
    """
    positions = {}
    
    for n in range(n_max + 1):
        if n == 0:
            # Position terminale : le joueur precedent a pris le dernier jeton
            positions[n] = 'P'
        else:
            # Verifier si on peut atteindre une P-position
            can_reach_P = False
            for take in [1, 2, 3]:
                if n >= take and positions.get(n - take) == 'P':
                    can_reach_P = True
                    break
            positions[n] = 'N' if can_reach_P else 'P'
    
    return positions

# Classifier les 16 premieres positions
positions = classify_123_game(16)

print("Position | Type | Motif")
print("-" * 25)
for n in range(17):
    t = positions[n]
    motif = "Perdante (n % 4 == 0)" if t == 'P' else "Gagnante"
    print(f"   {n:2d}    |  {t}   | {motif}")
```

**Lecture chiffree — pourquoi la période vaut 4 dans ce tableau.** La colonne Motif aligne un `P` exactement toutes les quatre lignes (`n % 4 == 0` aux lignes 0, 4, 8, 12). Le mecanisme se lit dans les coups disponibles {1, 2, 3} : depuis un multiple de 4, soustraire 1, 2 ou 3 fait sortir du multiple — aucun coup n'y reste ; depuis toute autre position n, le coup qui retranche exactement le reste n mod 4 y ramène. La période 4 n'est donc pas une propriete magique de ce jeu : c'est la longueur du plus grand coup plus un, et la couverture complete des restes qui va avec. La section 5 montrera le contre-exemple utile : pour S({1, 3, 4}) la plus grande valeur de Grundy monte a 3 et la période vaut 7, pas 5 — la periodicite des jeux de soustraction est frequente (observation de la section 5) mais sa longueur ne se deduit pas naivement de max(moves).

## 2. Le jeu de Nim

**Nim** est le jeu combinatoire le plus celebre :

- Plusieurs **tas** de jetons
- A chaque tour, retirer **au moins un jeton** d'un **seul tas**
- Le joueur qui retire le **dernier jeton** gagne

### Theoreme de Bouton (1901)

Une position de Nim avec des tas de tailles n1, n2, ..., nk est une :
- **P-position** si n1 XOR n2 XOR ... XOR nk = 0
- **N-position** sinon

```python
def nim_sum(*heaps) -> int:
    """Calcule la nim-sum (XOR) des tailles de tas."""
    result = 0
    for h in heaps:
        result ^= h
    return result

def nim_position_type(*heaps) -> str:
    """Determine si une position est P ou N."""
    return 'P' if nim_sum(*heaps) == 0 else 'N'

# Exemples
examples = [
    (1, 2, 3),
    (1, 2, 4),
    (3, 5, 6),
    (7, 7),
    (1, 1, 1),
]

print("Tas           | Nim-sum | Type")
print("-" * 35)
for heaps in examples:
    ns = nim_sum(*heaps)
    t = nim_position_type(*heaps)
    print(f"{str(heaps):13} | {ns:7} | {t}")
```

**Lecture chiffree — cinq lignes, trois mecanismes de la nim-sum.** Les exemples de la sortie se decodent un par un. `(7, 7) | 0 | P` : deux tas identiques s'annulent entierement — chaque bit present deux fois tombe, le second joueur copie et gagne. `(1, 1, 1) | 1 | N` : nombre impair de tas identiques, la parite residuelle laisse 1 — l'annulation par paires est le seul mecanisme, l'orphelin survit. `(1, 2, 3) | 0 | P` et `(3, 5, 6) | 0 | P` : zero sans aucune paire de tas egaux — en binaire, 01 ⊕ 10 ⊕ 11 = 00 et 011 ⊕ 101 ⊕ 110 = 000, chaque colonne de bits tombe pair. Face a `(1, 2, 4) | 7 | N`, aucune colonne ne s'equilibre : 001 ⊕ 010 ⊕ 100 = 111. La nim-sum generalise l'annulation par paires a l'annulation par parite de colonnes.


### Stratégie gagnante

Si vous etes en N-position (nim-sum != 0), vous pouvez toujours jouer pour atteindre une P-position.

```python
def find_winning_move(heaps: list) -> tuple:
    """
    Trouve un coup gagnant au Nim.
    Retourne (index_tas, nouvelle_taille) ou None si P-position.
    """
    ns = nim_sum(*heaps)
    if ns == 0:
        return None  # P-position, pas de coup gagnant
    
    for i, h in enumerate(heaps):
        # Nouvelle taille pour ce tas
        new_h = h ^ ns
        if new_h < h:  # On doit retirer, pas ajouter
            return (i, new_h)
    
    return None

# Exemple
heaps = [3, 5, 7]
print(f"Position: {heaps}")
print(f"Nim-sum: {nim_sum(*heaps)}")

move = find_winning_move(heaps)
if move:
    i, new_h = move
    print(f"\nCoup gagnant: reduire tas {i} de {heaps[i]} a {new_h}")
    new_heaps = heaps.copy()
    new_heaps[i] = new_h
    print(f"Nouvelle position: {new_heaps}")
    print(f"Nouvelle nim-sum: {nim_sum(*new_heaps)}")
```

**Lecture chiffree — le coup gagnant reconstruit la parite.** Pour `[3, 5, 7]`, la sortie donne `Nim-sum: 1`, puis `Coup gagnant: reduire tas 0 de 3 a 2` et `Nouvelle nim-sum: 0`. Le mecanisme : la nim-sum 1 signale qu'un nombre impair de tas portent le bit 1 ; le solveur cherche le tas dont la reduction peut equilibrer toutes les colonnes a la fois — ici 3 → 2, car 10 ⊕ 101 ⊕ 111 = 000. Le detail qui compte : le coup n'attaque pas le plus gros tas ni le tas le plus desequilibre, il attaque le tas ou la correction est realisable en une seule reduction. Toute position N en possede au moins une (theoreme de Bouton) ; aucune position P n'en possede — c'est exactement la definition.

## 3. Fonction Mex et valeur de Grundy

### Fonction Mex (Minimum Excludant)

Le **mex** d'un ensemble S est le plus petit entier naturel **absent** de S.

```python
def mex(s: set) -> int:
    """Minimum excludant : plus petit entier >= 0 absent de l'ensemble."""
    n = 0
    while n in s:
        n += 1
    return n

# Exemples
examples = [
    set(),           # mex = 0
    {0},             # mex = 1
    {0, 1, 2},       # mex = 3
    {1, 2, 3},       # mex = 0 (0 absent)
    {0, 2, 4},       # mex = 1
]

for s in examples:
    print(f"mex({s}) = {mex(s)}")
```

**Lecture chiffree — le mex complete par le bas.** Les cinq lignes de la sortie se lisent comme une seule regle : `mex(S)` est le plus petit entier absent de S. Les deux lignes piegeuses : `mex({1, 2, 3}) = 0` — l'ensemble ne contient pas zero, donc le mex tombe a zero meme si l'ensemble est « rempli » au-dela ; et `mex({0, 2, 4}) = 1` — le trou le plus bas commande, meme entoure de pairs. `mex(set()) = 0` fixe la base : la position terminale sans successeur recoit Grundy 0, qui est la definition meme d'une P-position. Ce choix de mex (et pas de max, ni de somme) garantit que deux positions aux suites de Grundy differentes ne collisionnent jamais — la propriete que la table de la section suivante exploite.


### Valeur de Grundy

La **valeur de Grundy** (ou nimber) d'une position G est définie recursivement :

- Position terminale : Grundy = 0
- Sinon : Grundy(G) = mex{Grundy(G') : G' accessible depuis G}

Propriete importante :
- P-position ssi Grundy = 0
- N-position ssi Grundy > 0

```python
def grundy_nim(n: int, memo: dict = None) -> int:
    """
    Valeur de Grundy de nim(n) - un seul tas de n jetons.
    (Spoiler: c'est toujours n)
    """
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    
    if n == 0:
        return 0
    
    # Positions accessibles : nim(0), nim(1), ..., nim(n-1)
    reachable = {grundy_nim(i, memo) for i in range(n)}
    result = mex(reachable)
    memo[n] = result
    return result

print("n  | grundy(nim(n)) | Verification")
print("-" * 35)
for n in range(10):
    g = grundy_nim(n)
    verify = "OK" if g == n else "ERREUR"
    print(f"{n:2} | {g:14} | {verify}")
```

**Lecture chiffree — grundy(nim(n)) = n, ligne par ligne.** La table aligne dix verifications `OK` : pour chaque n de 0 a 9, la valeur de Grundy calculee par la definition recursive (mex des successeurs) vaut exactement n. Autrement dit, un tas de Nim est son propre nimber : Grundy(n) = mex({0, 1, ..., n-1}) = n. La colonne Verification ne verifie pas un resultat numerique isole, elle boucle la definition — chaque ligne certifie que le mex des successeurs de n (qui sont exactement tous les entiers de 0 a n-1) retombe sur n. C'est ce pont identite qui rend le theoreme de Sprague-Grundy de la section suivante concret : chaque tas d'une position de Nim se comporte deja comme son nimber, et la somme de jeux s'annoncera comme un simple XOR de ces valeurs.

## 4. Theoreme de Sprague-Grundy

Le theoreme de Sprague-Grundy est le résultat central de la théorie des jeux combinatoires :

> **Theoreme** : Tout jeu impartial est equivalent a un tas de Nim.
> 
> Pour une **somme de jeux** G1 + G2 + ... + Gk :
> Grundy(G1 + G2 + ... + Gk) = Grundy(G1) XOR Grundy(G2) XOR ... XOR Grundy(Gk)

Cela signifie que pour analyser n'importe quel jeu impartial, il suffit de :
1. Calculer la valeur de Grundy de chaque composante
2. XOR les valeurs ensemble
3. Si le résultat est 0, c'est une P-position, sinon N-position

```python
def analyze_nim_game(heaps: list) -> dict:
    """
    Analyse complete d'une position de Nim.
    """
    grundy_values = heaps  # Pour Nim, grundy(n) = n
    total_grundy = nim_sum(*grundy_values)
    
    result = {
        'heaps': heaps,
        'grundy_values': grundy_values,
        'nim_sum': total_grundy,
        'position_type': 'P' if total_grundy == 0 else 'N',
        'winning_move': find_winning_move(heaps)
    }
    return result

# Exemple
game = analyze_nim_game([4, 7, 9])

print(f"Position: {game['heaps']}")
print(f"Valeurs de Grundy: {game['grundy_values']}")
print(f"Nim-sum: {game['nim_sum']}")
print(f"Type: {game['position_type']}-position")

if game['winning_move']:
    i, new_h = game['winning_move']
    print(f"\nCoup gagnant: tas {i}: {game['heaps'][i]} -> {new_h}")
```

**Lecture chiffree — l'analyse complete d'une position en une sortie.** Pour `[4, 7, 9]`, les cinq lignes racontent toute la theorie appliquee. `Valeurs de Grundy: [4, 7, 9]` — chaque tas est son nimber (section precedente). `Nim-sum: 10` — en binaire 100 ⊕ 111 ⊕ 1001 = 1010. `Type: N-position` — la nim-sum non nulle signe le tour gagnant. `Coup gagnant: tas 2: 9 -> 3` — verifier : 100 ⊕ 111 ⊕ 011 = 000, la nouvelle position est une P-position servie au prochain joueur. Le choix 9 → 3 n'est pas unique par principe, mais tout coup gagnant doit reduire un tas exactement a la valeur qui equilibre les deux autres : ici 4 ⊕ 7 = 3, la cible s'impose.

## 5. Jeux de soustraction

Un **jeu de soustraction** S(M) est défini par un ensemble de coups M :
- Un tas de n jetons
- On peut retirer m jetons si m est dans M et m <= n

Par exemple, S({1, 3, 4}) : on peut retirer 1, 3 ou 4 jetons.

```python
def grundy_subtraction(n: int, moves: set, memo: dict = None) -> int:
    """
    Valeur de Grundy pour un jeu de soustraction S(moves).
    """
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    
    if n == 0:
        return 0
    
    # Positions accessibles
    reachable = set()
    for m in moves:
        if m <= n:
            reachable.add(grundy_subtraction(n - m, moves, memo))
    
    result = mex(reachable)
    memo[n] = result
    return result

# Analyser S({1, 3, 4})
moves = {1, 3, 4}
memo = {}

print(f"Jeu de soustraction S({moves})")
print("\n n | Grundy | Type")
print("-" * 20)
for n in range(20):
    g = grundy_subtraction(n, moves, memo)
    t = 'P' if g == 0 else 'N'
    print(f"{n:2} | {g:6} | {t}")
```

**Observation** : Les valeurs de Grundy pour les jeux de soustraction sont souvent **periodiques** après un certain point.

### Le théorème en action : S({1,3,4}) et son jumeau de Nim

La section 4 énonce que tout jeu impartial est équivalent à un tas de Nim de **même valeur de Grundy**. Appliquons-le au jeu de soustraction de cette section : deux positions de S({1,3,4}) et le tas de Nim qui leur est indiscernable.

```python
# Le theoreme en action : une position de S({1,3,4}) equivalente a un tas de Nim.
# On reutilise grundy_subtraction (cellule ci-dessus) et le fait que, pour le Nim,
# grundy(n) = n : le jumeau de Nim d'une position est le tas de sa valeur de Grundy.

moves = {1, 3, 4}
print(f"Jeu de soustraction S({moves}) — jumeau de Nim (Sprague-Grundy)")
print("-" * 60)
for n in [5, 7]:
    g = grundy_subtraction(n, moves, {})
    kind = "P (perdue)" if g == 0 else "N (gagnante)"
    print(f"n = {n} jetons : Grundy = {g}  ->  jumeau = Nim de {g} jeton(s)  [{kind}]")
```

**Lecture du résultat** — la position `n = 5` (Grundy 3) est indiscernable, dans toute somme, d'un tas de Nim de 3 jetons ; la position `n = 7`, qui porte pourtant **7 jetons**, est indiscernable d'un tas de Nim **vide** (Grundy 0) : c'est une P-position, une perte. Le nombre de jetons n'est pas l'invariant de décision — la valeur de Grundy l'est, et elle désigne l'unique tas de Nim équivalent. Le théorème de Sprague-Grundy n'est pas seulement une formule : c'est un dictionnaire qui ramène n'importe quel jeu impartial au Nim.

Le theoreme de Bouton a fourni un critere elegant pour determiner la position gagnante au Nim (la nim-sum, ou XOR des tailles de tas), tandis que la fonction mex (minimum excludant) a permis de calculer recursivement les valeurs de Grundy pour des jeux de soustraction plus généraux. L'observation de la periodicite des valeurs de Grundy dans les jeux de soustraction ouvre la porte a l'analyse de jeux plus complexes, ou les structures periodiques simplifient considerablement la recherche de stratégies optimales.

### Concepts cles

| Concept | Description |
|---------|-------------|
| **P-position** | Position perdante pour le joueur a jouer |
| **N-position** | Position gagnante pour le joueur a jouer |

## References

Ce carnet explore la theorie des jeux combinatoires (combinatorial game theory, CGT) -- une branche de la theorie des jeux qui se distingue par la regle **de la derniere action gagnante** : un joueur qui ne peut plus jouer a perdu, et le resultat est une determination exacte (les positions sont etiquetables P ou N, perdantes ou gagnantes pour le prochain joueur). L'outil central est le **theoreme de Sprague-Grundy** : tout jeu combinatoire impartial (positions partagees, coups disponibles identiques pour les deux joueurs) est equivalent a un tas de Nim de taille determinee, ce qui rend la theorie entierement calculable. La formalisation canonique moderne est l'ouvrage de Berlekamp-Conway-Guy ; l'extension algebrique est l'ouvrage de Conway sur les nombres surreels. La section ci-dessous rassemble les references qui sous-tendent le propos.

1. **Bouton, C. L. (1902).** *Nim, a Game with a Complete Mathematical Theory.* Annals of Mathematics, Second Series 3(1/4): 35-39. -- la premiere analyse mathematique complete d'un jeu combinatoire : le jeu de Nim a plusieurs tas, dont Bouton demontre que la strategie gagnante repose sur la position du bit de parite XOR (la somme de Nim). Article fondateur du champ, qui etablit le paradigme de l'**analyse exacte** (par opposition a l'analyse strategique de Nash) : pour un jeu combinatoire impartial, il n'y a pas d'aleatoire, pas d'information cachee, et le resultat est calculable par recurrence.

2. **Sprague, R. (1935).** *Uber mathematische Kampfspiele.* Tohoku Mathematical Journal 41: 438-444. -- la formulation du theoreme qui porte son nom : tout jeu combinatoire impartial est equivalent a un etat unique determine par un nombre naturel appele **nimber** (aujourd'hui note `*n`). Sprague demontre que l'ensemble des jeux combinatoires impartiaux forme un semi-groupe commutatif, et que chaque jeu se decompose comme une somme de Nim de tailles determinees. Article pionnier, redige en allemand.

3. **Grundy, P. M. (1939).** *Mathematics and Games.* Eureka 2: 6-8. -- la redécouverte independante du meme theoreme, formulee pour un public etudiant. La recurrence porte le nom de **Sprague-Grundy** dans la litterature, les deux contributions etant parues a quatre ans d'intervalle sans contact. Grundy propose un algorithme direct (la recurrence classique) pour calculer le nimber d'une position donnee par parcours en arbre des options.

4. **Berlekamp, E. R., Conway, J. H. & Guy, R. K. (1982).** *Winning Ways for your Mathematical Plays.* Academic Press, 2 volumes (seconde edition en 4 volumes : A K Peters, 2001-2004). -- le traite fondateur de la CGT moderne, qui etend la theorie au-dela des jeux impartiaux pour couvrir les jeux partiaux (Hackenbush, Blue-Red Hackenbush) et le calcul des nombres surreels. L'ouvrage de reference sur le sujet depuis quarante ans ; l'approche pedagogique (par le jeu, puis par l'algebre) en fait un classique de la vulgarisation mathematique de haut niveau.

5. **Conway, J. H. (1976).** *On Numbers and Games.* Academic Press. -- la fondation algebrique de la CGT par l'invention des **nombres surreels** : une classe d'objets qui contient simultanement les reels, les ordinaux, et tous les jeux combinatoires, avec une structure de corps ordonne. L'ouvrage demontre que la theorie des jeux combinatoires et celle des nombres (transfinis) sont deux faces d'une meme construction algebrique, et fournit le langage formel (les classes de Dedekind transfines) qui sous-tend Winning Ways.

6. **Nowakowski, R. J. (dir., 1996, vol. 1 *Games of No Chance*, MSRI 29) ; Nowakowski, R. J. (dir., 2002, vol. 2 *More Games of No Chance*) ; Albert, M. H. & Nowakowski, R. J. (dir., 2009, vol. 3 *Games of No Chance 3*) ; Nowakowski, R. J. (dir., 2015, vol. 4 *Games of No Chance 4*) ; Larsson, U. (dir., 2019, vol. 5 *Games of No Chance 5*).** MSRI Publications / Cambridge University Press. -- l'anthologie de reference des resultats contemporains en CGT, issue des conferences *Games of No Chance* organisees au MSRI (Berkeley). Vol. 1 (1996) est l'ouvrage de base -- theoremes de base, extension au cas impartial et presentation des jeux partiaux ; le vol. 2 (2002) approfondit les jeux impartiaux et partiaux ; le vol. 3 (2009, double-edition Albert+Nowakowski) elargit l'eventail aux jeux a information imparfaite et aux algorithmes de recherche retrograde ; le vol. 4 (2015) recentre sur les resultats structurels et les applications algorithmiques recentes ; le vol. 5 (2019, Larsson seul editeur) ouvre vers les jeux a but optionnel (option-goals) et la theorie des jeux impartiaux composee.

7. **Bonanno, G. (2015).** *Game Theory* (open access textbook, 165 solved exercises). arXiv:1512.06808 [math.HO], **CC BY-NC-ND 4.0** (la page de copyright du PDF prime sur la metadata arXiv, qui annonce CC BY 4.0). -- manuel pedagogique general de theorie des jeux ; le mapping au carnet 08 est de **bord** (la CGT n'est pas un chapitre dedie du manuel). Bonanno n'est pas ancre sur ces sujets dans ce carnet ; les references titulaires sont les entrees 1 a 6 ci-dessus.
