## 2. Trois expériences de dix minutes

### 2.1 Le théorème d'Arrow par la simulation

Trois axiomes, trois règles, un verdict. Les axiomes :

- **Pareto (faible)** : si *tous* les électeurs préfèrent `x` à `y`, la règle doit classer `x` avant `y` ;
- **IIA** (indépendance des alternatives non pertinentes) : le classement social de `x` vs `y` ne doit dépendre que des classements individuels de `x` vs `y` — pas des positions d'une troisième alternative `z` ;
- **Non-dictature** : aucun électeur seul ne impose son classement.

Le théorème d'Arrow (1951) : avec **3 alternatives ou plus**, aucune fonction de bien-être social ne satisfait les trois. La cellule suivante définit le matériel (profils et règles), la suivante teste les axiomes.

```
// Materiau commun aux experiences : un profil = liste d'ordres individuels.
// Convention : chaque ordre est une liste des alternatives, de la meilleure a la pire.

// Les trois alternatives du parcours.
var ALTS = new List<string> { "A", "B", "C" };

// Score de Borda : chaque electeur donne m-1, m-2, ..., 0 points a ses alternatives.
Dictionary<string, int> Borda(List<List<string>> profil, List<string> alts)
{
    int m = alts.Count;
    var scores = alts.ToDictionary(a => a, a => 0);
    foreach (var ordre in profil)
    {
        for (int rang = 0; rang < ordre.Count; rang++)
        {
            scores[ordre[rang]] += m - 1 - rang;
        }
    }
    return scores;
}

// Score de pluralite : 1 point a l'alternative classee premiere par chaque electeur.
Dictionary<string, int> Pluralite(List<List<string>> profil, List<string> alts)
{
    var scores = alts.ToDictionary(a => a, a => 0);
    foreach (var ordre in profil)
    {
        scores[ordre[0]] += 1;
    }
    return scores;
}

// Score du dictateur k : son ordre personnel devient l'ordre social.
Dictionary<string, int> Dictateur(int k, List<List<string>> profil, List<string> alts)
{
    int m = alts.Count;
    var scores = new Dictionary<string, int>();
    var ordre = profil[k];
    for (int rang = 0; rang < ordre.Count; rang++)
    {
        scores[ordre[rang]] = m - 1 - rang;
    }
    return scores;
}

// Ordre social decroissant selon les scores (les egalites restent visibles).
List<string> OrdreSocial(Dictionary<string, int> scores)
{
    return scores.OrderByDescending(kv => kv.Value).Select(kv => kv.Key).ToList();
}

// Rendu d'un dictionnaire et d'une liste a la maniere du carnet Python,
// pour que les deux jumeaux affichent exactement la meme ligne.
string PyDict(Dictionary<string, int> scores)
{
    return "{" + string.Join(", ", scores.Select(kv => $"'{kv.Key}': {kv.Value}")) + "}";
}

string PyDictStr(Dictionary<string, string> appariement)
{
    return "{" + string.Join(", ", appariement.OrderBy(kv => kv.Key, StringComparer.Ordinal)
                                             .Select(kv => $"'{kv.Key}': '{kv.Value}'")) + "}";
}

string PyList(List<string> xs)
{
    return "[" + string.Join(", ", xs.Select(s => $"'{s}'")) + "]";
}

Console.WriteLine("Regles definies : borda, pluralite, dictateur.");
```

```
// Test des trois axiomes sur les trois regles.
// Pareto : pour toute paire unanime x > y, l'ordre social doit placer x avant y.
// IIA : deux profils P1/P2 ou CHAQUE electeur garde le meme ordre entre A et B,
//       mais ou l'ordre social A vs B bascule -> violation.
// Non-dictature : la regle ne doit pas coincider systematiquement avec l'ordre d'un electeur.

bool ParetoOk(Func<List<List<string>>, List<string>, Dictionary<string, int>> regle,
               List<List<string>> profil, List<string> alts)
{
    var ordre = OrdreSocial(regle(profil, alts));
    var pos = new Dictionary<string, int>();
    for (int i = 0; i < ordre.Count; i++)
    {
        pos[ordre[i]] = i;
    }
    for (int i = 0; i < alts.Count; i++)
    {
        for (int j = i + 1; j < alts.Count; j++)
        {
            string x = alts[i];
            string y = alts[j];
            // Unanimite x > y : l'ordre social doit placer x avant y...
            if (profil.All(o => o.IndexOf(x) < o.IndexOf(y)) && !(pos[x] < pos[y]))
            {
                return false;
            }
            // ...et unanimite y > x tout autant : chaque paire a DEUX orientations,
            // et l'unanimite inverse est un cas Pareto a part entiere.
            if (profil.All(o => o.IndexOf(y) < o.IndexOf(x)) && !(pos[y] < pos[x]))
            {
                return false;
            }
        }
    }
    return true;
}

// Meme verification, sur un ordre social deja calcule.
bool ParetoOkScores(Dictionary<string, int> scores, List<List<string>> profil, List<string> alts)
{
    return ParetoOk((p, a) => scores, profil, alts);
}

// Temoins IIA (mesures ci-dessous) :
//  pluralite : P1 -> social A>B ; P2 -> social B>A (preferences A-vs-B individuelles identiques)
//  borda     : P1' -> social B>A ; P2' -> social A>B (idem, 5 electeurs)
var P1_PLU = new List<List<string>>
{
    new List<string> { "A", "B", "C" },
    new List<string> { "B", "C", "A" },
    new List<string> { "A", "B", "C" },
};
var P2_PLU = new List<List<string>>
{
    new List<string> { "C", "A", "B" },
    new List<string> { "B", "C", "A" },
    new List<string> { "C", "A", "B" },
};
var P1_BOR = new List<List<string>>
{
    new List<string> { "A", "B", "C" },
    new List<string> { "A", "B", "C" },
    new List<string> { "A", "B", "C" },
    new List<string> { "B", "C", "A" },
    new List<string> { "B", "C", "A" },
};
var P2_BOR = new List<List<string>>
{
    new List<string> { "A", "C", "B" },
    new List<string> { "A", "C", "B" },
    new List<string> { "A", "C", "B" },
    new List<string> { "B", "A", "C" },
    new List<string> { "B", "A", "C" },
};

// true si le temoin ne bascule PAS (IIA respectee sur ce cas), false s'il
// bascule, null si la premisse est invalide : la premisse IIA exige que
// CHAQUE electeur garde son ordre x-vs-y entre P1 et P2 -- vecteur individuel
// par vecteur individuel, pas leur seul "tous". Un couple dont la premisse
// ne tient pas ne prouve rien sur la regle.
bool? IiaTemoin(Func<List<List<string>>, List<string>, Dictionary<string, int>> regle,
                List<List<string>> P1, List<List<string>> P2, List<string> alts,
                string x = "A", string y = "B")
{
    var prefs1 = P1.Select(o => o.IndexOf(x) < o.IndexOf(y)).ToList();
    var prefs2 = P2.Select(o => o.IndexOf(x) < o.IndexOf(y)).ToList();
    if (!prefs1.SequenceEqual(prefs2))
    {
        return null;
    }
    bool Soc(List<List<string>> P)
    {
        var s = regle(P, alts);
        return s[x] > s[y];
    }
    return Soc(P1) == Soc(P2);
}

var PROFIL_ARROW = new List<List<string>>
{
    new List<string> { "A", "B", "C" },
    new List<string> { "B", "A", "C" },
    new List<string> { "A", "C", "B" },
};

var lignes = new List<(string nom, bool p, bool i, bool nd)>();
lignes.Add(("Borda", ParetoOk(Borda, PROFIL_ARROW, ALTS), IiaTemoin(Borda, P1_BOR, P2_BOR, ALTS) == true, true));
lignes.Add(("Pluralite", ParetoOk(Pluralite, PROFIL_ARROW, ALTS), IiaTemoin(Pluralite, P1_PLU, P2_PLU, ALTS) == true, true));
// Dictateur 0 : Pareto et IIA tiennent par construction, la non-dictature tombe.
lignes.Add(("Dictateur(v1)", ParetoOk((P, a) => Dictateur(0, P, a), PROFIL_ARROW, ALTS), true, false));

Console.WriteLine($"{"Regle".PadRight(14)} {"Pareto".PadRight(8)} {"IIA".PadRight(6)} {"Non-dictature".PadRight(14)} Verdict");
foreach (var (nom, p, i, nd) in lignes)
{
    int verts = (p ? 1 : 0) + (i ? 1 : 0) + (nd ? 1 : 0);
    Console.WriteLine($"{nom.PadRight(14)} {(p ? "OK" : "VIOLE").PadRight(8)} {(i ? "OK" : "VIOLE").PadRight(6)} " +
                      $"{(nd ? "OK" : "VIOLE").PadRight(14)} {verts}/3 " + (verts < 3 ? "-> impossible" : ""));
}
Console.WriteLine();
Console.WriteLine("Details des temoins IIA :");
Console.WriteLine($"  Pluralite : P1 scores {PyDict(Pluralite(P1_PLU, ALTS))} -> ordre {PyList(OrdreSocial(Pluralite(P1_PLU, ALTS)))}");
Console.WriteLine($"              P2 scores {PyDict(Pluralite(P2_PLU, ALTS))} -> ordre {PyList(OrdreSocial(Pluralite(P2_PLU, ALTS)))}");
Console.WriteLine($"  Borda     : P1' scores {PyDict(Borda(P1_BOR, ALTS))} -> ordre {PyList(OrdreSocial(Borda(P1_BOR, ALTS)))}");
Console.WriteLine($"              P2' scores {PyDict(Borda(P2_BOR, ALTS))} -> ordre {PyList(OrdreSocial(Borda(P2_BOR, ALTS)))}");

// Affirmation des valeurs attendues : le jumeau doit retrouver exactement les
// valeurs affichees par la source Python, pas les recalculer autrement.
var bordaP2 = Borda(P2_BOR, ALTS);
if (bordaP2["A"] != 8 || bordaP2["B"] != 4 || bordaP2["C"] != 3
    || string.Join(",", OrdreSocial(bordaP2)) != "A,B,C")
{
    throw new InvalidOperationException("temoin Borda P2' : scores ou ordre social inattendus");
}
var pluraliteP1 = Pluralite(P1_PLU, ALTS);
if (pluraliteP1["A"] != 2 || pluraliteP1["B"] != 1 || pluraliteP1["C"] != 0)
{
    throw new InvalidOperationException("temoin pluralite P1 : scores inattendus");
}

// Controles de non-regression : les contre-exemples de la review (#20663),
// figes ici pour qu'aucune reecriture des predicats ne les reperde.
// 1. Unanimite INVERSE : tous les electeurs preferent C>B>A pendant que
//    l'ordre social affiche A>B>C -- Pareto doit tomber.
int controlesOk = 0;
var scoresInverse = new Dictionary<string, int> { { "A", 3 }, { "B", 2 }, { "C", 1 } };
var profilInverse = new List<List<string>>
{
    new List<string> { "C", "B", "A" },
    new List<string> { "C", "B", "A" },
};
if (ParetoOkScores(scoresInverse, profilInverse, ALTS))
{
    throw new InvalidOperationException("controle 1 : l'unanimite inverse doit faire tomber Pareto");
}
controlesOk += 1;
// 2. Controle positif : le meme ordre social face a l'unanimite directe tient.
var profilDirect = new List<List<string>>
{
    new List<string> { "A", "B", "C" },
    new List<string> { "A", "B", "C" },
};
if (!ParetoOkScores(scoresInverse, profilDirect, ALTS))
{
    throw new InvalidOperationException("controle 2 : l'unanimite directe doit faire tenir Pareto");
}
controlesOk += 1;
// 3. Premisse IIA invalide : les vecteurs individuels A-vs-B de P et Q
//    different electeur par electeur -- le couple ne prouve rien.
var P_INV = new List<List<string>>
{
    new List<string> { "A", "B", "C" },
    new List<string> { "B", "A", "C" },
};
var Q_INV = new List<List<string>>
{
    new List<string> { "B", "A", "C" },
    new List<string> { "A", "B", "C" },
};
if (IiaTemoin(Borda, P_INV, Q_INV, ALTS) != null)
{
    throw new InvalidOperationException("controle 3 : une premisse IIA invalide ne prouve rien");
}
controlesOk += 1;
// 4-5. Les temoins concrets du tableau gardent une premisse valide : chaque
//      electeur conserve bien son ordre A-vs-B entre P1 et P2.
bool MemeVecteurs(List<List<string>> P, List<List<string>> Q)
{
    return P.Select(o => o.IndexOf("A") < o.IndexOf("B"))
            .SequenceEqual(Q.Select(o => o.IndexOf("A") < o.IndexOf("B")));
}
if (!MemeVecteurs(P1_PLU, P2_PLU))
{
    throw new InvalidOperationException("controle 4 : le temoin de pluralite a une premisse valide");
}
controlesOk += 1;
if (!MemeVecteurs(P1_BOR, P2_BOR))
{
    throw new InvalidOperationException("controle 5 : le temoin de Borda a une premisse valide");
}
controlesOk += 1;
Console.WriteLine();
Console.WriteLine($"Controles de non-regression (contre-exemples review #20663) : {controlesOk}/5 OK");
```

**Lecture du résultat.** Aucune ligne n'est verte des trois colonnes : Borda et la pluralité violent l'IIA (déplacer **C** — une alternative perdante — suffit à inverser le classement social de A vs B), la dictature viole la non-dictature. C'est le théorème d'Arrow **vu par le code** : chaque axiome isolé est satisfaisable, la conjonction ne l'est pas — sur ces exemples.

Ce que la simulation **ne montre pas** : elle teste trois règles sur un profil. Arrow dit que **toutes** les règles (avec 3+ alternatives) échouent — une infinité de cas que seul un raisonnement couvre. SC-01 déroule la preuve de Geanakoplos pas à pas ; SC-04 la fait vérifier par un solveur SAT sur les profils à 3 électeurs ; le lake Lean (section 3 ci-dessous) la prouve une fois pour toutes.

Le témoin qui suit ferme cette porte : sur **3 alternatives**, la famille des règles IIA + Pareto est **finie**, et son recensement est exhaustif.

### 2.1.1 — Recenser la famille : combien de règles IIA + Pareto existe-t-il ?

La simulation ci-dessus teste **trois** règles sur **quelques** profils. La question qu'elle laisse ouverte — « une *autre* règle aurait-elle survécu ? » — porte sur une **famille**, pas sur un échantillon.

Pour **3 alternatives** et **3 électeurs**, cette famille est **finie**, donc énumérable :

- l'**IIA** impose que l'ordre social sur une paire `{x, y}` ne dépende que du **motif** des préférences individuelles sur cette paire (le masque des électeurs qui préfèrent `x` à `y`) ;
- le **Pareto** impose le verdict sur les deux motifs unanimes (tout le monde préfère `x` : `x` l'emporte ; personne ne préfère `x` : `y` l'emporte) ;
- il reste donc **6 motifs libres** par paire — les six motifs non unanimes sur trois électeurs — soit `2^6 = 64` fonctions de décision par paire, et `64^3 = 262 144` règles candidates au total.

Le témoin qui suit énumère ces **262 144** candidates et ne retient que celles qui produisent un **ordre total strict** sur les `6^3 = 216` profils possibles. Le résultat est **déterministe** : aucune donnée aléatoire n'intervient.

```
// Recensement exhaustif de la famille des regles IIA + Pareto, sur 3 alternatives
// et 3 electeurs. Le resultat est deterministe : aucune donnee aleatoire.

var ALTS_C = new List<string> { "A", "B", "C" };
const int N_VOT = 3;
var PAIRES = new List<(string x, string y)> { ("A", "B"), ("A", "C"), ("B", "C") };

// Les six permutations de trois alternatives, dans l'ordre lexicographique.
List<List<string>> Permutations(List<string> xs)
{
    var res = new List<List<string>>();
    void Perm(List<string> prefix, List<string> rest)
    {
        if (rest.Count == 0)
        {
            res.Add(new List<string>(prefix));
            return;
        }
        for (int i = 0; i < rest.Count; i++)
        {
            var r = new List<string>(rest);
            string c = r[i];
            r.RemoveAt(i);
            var p = new List<string>(prefix) { c };
            Perm(p, r);
        }
    }
    Perm(new List<string>(), new List<string>(xs));
    return res;
}

var perms = Permutations(ALTS_C);

// Les 6^3 = 216 profils : le dernier electeur varie le plus vite.
var PROFILS = new List<List<List<string>>>();
foreach (var p1 in perms)
{
    foreach (var p2 in perms)
    {
        foreach (var p3 in perms)
        {
            PROFILS.Add(new List<List<string>>
            {
                new List<string>(p1),
                new List<string>(p2),
                new List<string>(p3),
            });
        }
    }
}

// Pour chaque profil, le motif des electeurs qui preference x a y, pour chaque paire.
var MOTIFS = new List<int[]>();
foreach (var prof in PROFILS)
{
    var m = new int[PAIRES.Count];
    for (int i = 0; i < PAIRES.Count; i++)
    {
        var (x, y) = PAIRES[i];
        int masque = 0;
        for (int v = 0; v < prof.Count; v++)
        {
            if (prof[v].IndexOf(x) < prof[v].IndexOf(y))
            {
                masque += 1 << v;
            }
        }
        m[i] = masque;
    }
    MOTIFS.Add(m);
}

// Les 64 fonctions de decision d'une paire : motif -> vainqueur.
// Pareto fixe les deux motifs unanimes (masques 0 et 7) ; les six autres sont
// libres, d ou 2^6 = 64 fonctions.
List<string[]> FonctionsDePaire(string x, string y)
{
    var fonctions = new List<string[]>();
    for (int bits = 0; bits < 64; bits++)
    {
        var table = new string[8];
        for (int masque = 0; masque < 8; masque++)
        {
            if (masque == 0)
            {
                table[masque] = y;
            }
            else if (masque == 7)
            {
                table[masque] = x;
            }
            else
            {
                table[masque] = (((bits >> (masque - 1)) & 1) == 0) ? x : y;
            }
        }
        fonctions.Add(table);
    }
    return fonctions;
}

// Les vainqueurs des trois paires, dans un tampon reutilise (264 144 candidats).
var arcW = new string[3];
var arcL = new string[3];

void SetArcs(string[][] fonctions, int[] motifs)
{
    for (int i = 0; i < 3; i++)
    {
        var (x, y) = PAIRES[i];
        string vainqueur = fonctions[i][motifs[i]];
        arcW[i] = vainqueur;
        arcL[i] = vainqueur == x ? y : x;
    }
}

// Vrai si les vainqueurs des trois paires forment un ordre strict (pas de cycle).
bool EstOrdreTotal()
{
    for (int i = 0; i < 3; i++)
    {
        for (int j = 0; j < 3; j++)
        {
            for (int k = 0; k < 3; k++)
            {
                if (arcL[i] == arcW[j] && arcL[j] == arcW[k] && arcL[k] == arcW[i])
                {
                    return false;
                }
            }
        }
    }
    return true;
}

// Index de l'electeur dont l'ordre personnel coincide toujours avec l'ordre social.
int? ElecteurDictateur(string[][] fonctions)
{
    for (int v = 0; v < N_VOT; v++)
    {
        bool dicte = true;
        for (int pi = 0; pi < PROFILS.Count && dicte; pi++)
        {
            var prof = PROFILS[pi];
            var motifs = MOTIFS[pi];
            for (int i = 0; i < PAIRES.Count; i++)
            {
                var (x, y) = PAIRES[i];
                string vainqueur = fonctions[i][motifs[i]];
                string attendu = prof[v].IndexOf(x) < prof[v].IndexOf(y) ? x : y;
                if (vainqueur != attendu)
                {
                    dicte = false;
                    break;
                }
            }
        }
        if (dicte)
        {
            return v;
        }
    }
    return null;
}

// Les milliers : separateur d'espace, comme dans le carnet Python.
string Espace(int n)
{
    return n.ToString("N0", System.Globalization.CultureInfo.InvariantCulture).Replace(",", " ");
}

int CANDIDATS = 64 * 64 * 64;
var SURVIVANTS = new List<string[][]>();
var fcts = PAIRES.Select(p => FonctionsDePaire(p.x, p.y)).ToList();
foreach (var f1 in fcts[0])
{
    foreach (var f2 in fcts[1])
    {
        foreach (var f3 in fcts[2])
        {
            var fonctions = new[] { f1, f2, f3 };
            bool total = true;
            foreach (var m in MOTIFS)
            {
                SetArcs(fonctions, m);
                if (!EstOrdreTotal())
                {
                    total = false;
                    break;
                }
            }
            if (total)
            {
                SURVIVANTS.Add(fonctions);
            }
        }
    }
}

// Affirmation des valeurs attendues : les memes que la source Python.
int ecartes = CANDIDATS - SURVIVANTS.Count;
if (CANDIDATS != 262144 || PROFILS.Count != 216 || SURVIVANTS.Count != 3 || ecartes != 262141)
{
    throw new InvalidOperationException("census d'Arrow : nombre de candidats, de profils ou de survivants inattendu");
}

Console.WriteLine($"Candidats IIA + Pareto enumeres : {Espace(CANDIDATS)} (64 fonctions x 3 paires)");
Console.WriteLine($"Profils de 3 electeurs sur 3 alternatives : {PROFILS.Count}");
Console.WriteLine($"Survivants (ordre total strict sur les {PROFILS.Count} profils) : {SURVIVANTS.Count}");

for (int k = 0; k < SURVIVANTS.Count; k++)
{
    int? dictateur = ElecteurDictateur(SURVIVANTS[k]);
    Console.WriteLine(dictateur.HasValue
        ? $"  survivant {k + 1} : electeur dictateur n° {dictateur.Value + 1}"
        : $"  survivant {k + 1} : aucun electeur dictateur");
}

Console.WriteLine($"Regles ecartees (ordre social non total sur un profil) : {Espace(ecartes)}");
Console.WriteLine("Verdict : les survivants sont exactement les dictatures -- aucune des regles");
Console.WriteLine("recensees ne satisfait la non-dictature. L'impossibilite d'Arrow est donc ENUMERABLE");
Console.WriteLine("sur 3 alternatives ; le census est la version executable du raisonnement de SC-01.");
```

**Lecture du résultat.** Le recensement est **exhaustif** : sur les **262 144** règles candidates (64 fonctions de décision par paire, sur trois paires), **exactement 3** produisent un ordre social total sur les **216** profils — et ce sont **les trois dictatures**, une par électeur. Les **262 141** autres sont écartées parce qu'elles cyclent sur au moins un profil.

Ce que cela ajoute à la simulation du début : parmi les règles recensées, **aucune ne satisfait la non-dictature** — et pas seulement « les trois règles testées échouent ». Sur **3 alternatives et 3 électeurs**, l'impossibilité d'Arrow (*Social Choice and Individual Values*, Cowles Commission Monograph No. 12, John Wiley & Sons, 1951) est **finie et énumérable** : la contradiction se lit dans une liste de 262 144 candidats dont 3 survivent, tous dictatoriaux.

Ce que le témoin **ne** montre pas : il vaut pour **3 alternatives et 3 électeurs**. Dès **4 alternatives**, le nombre de profils (`24^m`) rend l'énumération directe impraticable et la famille des règles trop grande pour un recensement à la main — c'est là que la **preuve** de SC-01 (argument du pivot) et le **solveur** de SC-04 prennent le relais. Le census est le **cas fini, vérifiable à la main** ; il ne remplace ni SC-01 ni SC-04, il en montre le point d'entrée.

### 2.2 Le cycle de Condorcet

Le marquis de Condorcet a proposé en 1785 un critère qui semble inattaquable : si une alternative bat toutes les autres en duel majoritaire, c'est elle qu'il faut élire. Le profil ci-dessous — le plus petit profil paradoxaux — montre pourquoi ce critère peut être **vide** : personne ne bat tout le monde.

```
// Le profil cyclique minimal : 3 electeurs, 3 alternatives.
var CYCLE = new List<List<string>>
{
    new List<string> { "A", "B", "C" },
    new List<string> { "B", "C", "A" },
    new List<string> { "C", "A", "B" },
};

// Pour chaque paire, qui gagne au vote majoritaire (et de combien) ?
void Duels(List<List<string>> profil, List<string> alts)
{
    for (int i = 0; i < alts.Count; i++)
    {
        for (int j = i + 1; j < alts.Count; j++)
        {
            string x = alts[i];
            string y = alts[j];
            int pour_x = profil.Count(o => o.IndexOf(x) < o.IndexOf(y));
            int pour_y = profil.Count - pour_x;
            string vainqueur = pour_x > pour_y ? x : (pour_y > pour_x ? y : "egalite");
            Console.WriteLine($"  {x} vs {y} : {pour_x}-{pour_y} -> {vainqueur}");
        }
    }
}

Console.WriteLine("Duels majoritaires sur le profil cyclique :");
Duels(CYCLE, ALTS);
Console.WriteLine();

// Affirmation des valeurs attendues : les memes que la source Python.
var bordaCycle = Borda(CYCLE, ALTS);
var pluraliteCycle = Pluralite(CYCLE, ALTS);
if (bordaCycle["A"] != 3 || bordaCycle["B"] != 3 || bordaCycle["C"] != 3
    || pluraliteCycle["A"] != 1 || pluraliteCycle["B"] != 1 || pluraliteCycle["C"] != 1)
{
    throw new InvalidOperationException("cycle de Condorcet : scores de Borda ou de pluralite inattendus");
}

Console.WriteLine($"Scores de Borda sur le meme profil : {PyDict(bordaCycle)}");
Console.WriteLine($"Scores de pluralite : {PyDict(pluraliteCycle)}");
Console.WriteLine("Scores de Copeland (1 victoire, 0.5 nul, 0 defaite) : a vous -- Exercice 1.");
```

**Lecture du résultat.** A bat B (2-1), B bat C (2-1), C bat A (2-1) : le duel majoritaire **cycle**, il n'existe pas de gagnant de Condorcet. Pire : Borda et la pluralité rendent toutes deux une **égalité parfaite** (Borda 3/3/3, pluralité 1/1/1 — une voix chacune) — les deux règles sont muettes exactement là où le critère de Condorcet est vide.

Ce n'est pas un accident de fabrication : le théorème de birds-of-a-feather n'existe pas, mais le **paradoxe d'Arrow** (ci-dessus) et ce cycle sont les deux faces d'une même rigidité — agréger des ordres *sans jamais perdre d'information* est impossible, chaque règle choisit quelle information elle sacrifie. SC-03 déroule le bestiaire complet (Condorcet, Borda, Copeland, Dowd, paradoxe de l'échange, électeur médian) et montre quel paradoxe chaque règle accepte.

### 2.3 Mariage stable : Gale-Shapley

Troisième changement d'objet : on n'élit plus, on **apparie**. Deux populations (par convention des « proposants » et des « disposantes ») expriment chacune un ordre de préférence sur l'autre. Un appariement est **stable** si aucune paire (un proposant, une disposante) ne préfère mutuellement quitter leurs partenaires respectifs pour se mettre ensemble — une telle paire est dite **bloquante**.

L'algorithme de Gale-Shapley (1962) : les proposants demandent, les disposantes gardent provisoirement le meilleur et refusent le reste, jusqu'à ce que plus personne ne demande. Il produit toujours un appariement stable — et il est **optimal pour les proposants** : aucun proposant ne peut faire mieux dans aucun appariement stable.

```
// Gale-Shapley sur 4x4 preferences fixes (deterministe).
var PROPOSANTS = new Dictionary<string, List<string>>
{
    { "h1", new List<string> { "f1", "f2", "f3", "f4" } },
    { "h2", new List<string> { "f2", "f1", "f4", "f3" } },
    { "h3", new List<string> { "f3", "f4", "f1", "f2" } },
    { "h4", new List<string> { "f4", "f3", "f2", "f1" } },
};
var DISPOSANTES = new Dictionary<string, List<string>>
{
    { "f1", new List<string> { "h2", "h1", "h4", "h3" } },
    { "f2", new List<string> { "h1", "h2", "h3", "h4" } },
    { "f3", new List<string> { "h4", "h3", "h2", "h1" } },
    { "f4", new List<string> { "h3", "h4", "h1", "h2" } },
};

// Deferred acceptance : les proposants demandent, les disposantes choisissent.
(Dictionary<string, string>, int) GaleShapley(
    Dictionary<string, List<string>> proposants,
    Dictionary<string, List<string>> disposantes)
{
    var libres = new Queue<string>(proposants.Keys);
    var prochainChoix = proposants.Keys.ToDictionary(k => k, k => 0);
    var fiance = disposantes.Keys.ToDictionary(k => k, k => (string)null);
    int demandes = 0;
    while (libres.Count > 0)
    {
        string p = libres.Dequeue();
        string d = proposants[p][prochainChoix[p]];
        prochainChoix[p] += 1;
        demandes += 1;
        if (fiance[d] is null)
        {
            fiance[d] = p;
        }
        else if (disposantes[d].IndexOf(p) < disposantes[d].IndexOf(fiance[d]!))
        {
            libres.Enqueue(fiance[d]!);
            fiance[d] = p;
        }
        else
        {
            libres.Enqueue(p);
        }
    }
    var appariement = new Dictionary<string, string>();
    foreach (var kv in fiance)
    {
        if (kv.Value is not null)
        {
            appariement[kv.Value] = kv.Key;
        }
    }
    return (appariement, demandes);
}

var (M_hommes, nb_demandes) = GaleShapley(PROPOSANTS, DISPOSANTES);
var (M_femmes_brut, _) = GaleShapley(DISPOSANTES, PROPOSANTS);   // {disposante: proposant}
var M_femmes = new Dictionary<string, string>();
foreach (var kv in M_femmes_brut)
{
    M_femmes[kv.Value] = kv.Key;                                  // rebase en {proposant: disposante}
}
// Affirmation des valeurs attendues : les memes que la source Python.
if (nb_demandes != 4
    || M_hommes.Count != 4 || M_hommes["h1"] != "f1" || M_hommes["h2"] != "f2"
    || M_hommes["h3"] != "f3" || M_hommes["h4"] != "f4"
    || M_femmes.Count != 4 || M_femmes["h1"] != "f2" || M_femmes["h2"] != "f1"
    || M_femmes["h3"] != "f4" || M_femmes["h4"] != "f3")
{
    throw new InvalidOperationException("Gale-Shapley : appariement ou nombre de demandes inattendu");
}

Console.WriteLine($"Appariement optimal pour les proposants : {PyDictStr(M_hommes)} ({nb_demandes} demandes)");
Console.WriteLine($"Appariement optimal pour les disposantes : {PyDictStr(M_femmes)}");
```

```
// Verification des deux promesses de Gale-Shapley, sur les deux appariements.

// Cherche une paire (p, d) qui prefererait mutuellement se quitter.
(string, string)? PaireBloquante(Dictionary<string, string> M,
                                 Dictionary<string, List<string>> proposants,
                                 Dictionary<string, List<string>> disposantes)
{
    var inverse = new Dictionary<string, string>();
    foreach (var kv in M)
    {
        inverse[kv.Value] = kv.Key;
    }
    foreach (var p in proposants.Keys)
    {
        foreach (var d in disposantes.Keys)
        {
            if (d == M[p])
            {
                continue;
            }
            string actuel_p = M[p];
            string actuel_d = inverse[d];
            if (proposants[p].IndexOf(d) < proposants[p].IndexOf(actuel_p)
                && disposantes[d].IndexOf(p) < disposantes[d].IndexOf(actuel_d))
            {
                return (p, d);
            }
        }
    }
    return null;
}

// Rang 1 = meilleur.
int Rang(List<string> pref, string x)
{
    return pref.IndexOf(x) + 1;
}

foreach (var entree in new[] { ("optimal proposants", M_hommes), ("optimal disposantes", M_femmes) })
{
    var bloq = PaireBloquante(entree.Item2, PROPOSANTS, DISPOSANTES);
    Console.WriteLine($"{entree.Item1.PadRight(20)} stable : " +
                      (bloq is null ? "OUI" : $"NON, paire bloquante ('{bloq.Value.Item1}', '{bloq.Value.Item2}')"));
    if (bloq is not null)
    {
        throw new InvalidOperationException($"appariement {entree.Item1} : paire bloquante ('{bloq.Value.Item1}', '{bloq.Value.Item2}')");
    }
}

Console.WriteLine();
Console.WriteLine("Rangs obtenus par les proposants (1 = meilleur) :");
Console.WriteLine($"{"".PadRight(18)} {"h1".PadLeft(4)} {"h2".PadLeft(4)} {"h3".PadLeft(4)} {"h4".PadLeft(4)}");
var r_h = new List<int>();
var r_f = new List<int>();
foreach (var h in new[] { "h1", "h2", "h3", "h4" })
{
    r_h.Add(Rang(PROPOSANTS[h], M_hommes[h]));
    r_f.Add(Rang(PROPOSANTS[h], M_femmes[h]));
}

// Affirmation des valeurs attendues : les memes que la source Python.
if (r_h[0] != 1 || r_h[1] != 1 || r_h[2] != 1 || r_h[3] != 1
    || r_f[0] != 2 || r_f[1] != 2 || r_f[2] != 2 || r_f[3] != 2)
{
    throw new InvalidOperationException("Gale-Shapley : rangs obtenus inattendus");
}
bool auMieuxPlace = r_h.Zip(r_f, (a, b) => a <= b).All(v => v);
if (!auMieuxPlace)
{
    throw new InvalidOperationException("Gale-Shapley : un proposant est moins bien place dans l'appariement qu'il initie");
}

Console.WriteLine($"{"opt. proposants".PadRight(18)} {r_h[0],4} {r_h[1],4} {r_h[2],4} {r_h[3],4}");
Console.WriteLine($"{"opt. disposantes".PadRight(18)} {r_f[0],4} {r_f[1],4} {r_f[2],4} {r_f[3],4}");
Console.WriteLine();
Console.WriteLine("Chaque proposant est au moins aussi bien place dans l'appariement qu'il initie : "
                  + auMieuxPlace);
```

**Lecture du résultat.** Les deux appariements sont **stables** (aucune paire bloquante — la vérification tourne sur toutes les paires possibles), mais ils **diffèrent** : dans celui que les proposants initient, chaque proposant est au rang qu'il a dans l'autre ou mieux. La stabilité n'est pas neutre — **celui qui demande est favorisé**. C'est exactement pourquoi le mécanisme d'admission aux internats américains a changé de côté en 1998 : laisser les étudiants proposer (et non les hôpitaux) leur donnait l'optimalité sans perdre la stabilité.

SC-08 pousse chaque affirmation jusqu'à la preuve : couplage parfait garanti, man-optimalité **vérifiée par énumération exhaustive** de tous les appariements stables, et — côté formel — le lake Lean `game_theory_lean/StableMarriage/` prouve la correction de l'algorithme sans un seul `sorry`.
