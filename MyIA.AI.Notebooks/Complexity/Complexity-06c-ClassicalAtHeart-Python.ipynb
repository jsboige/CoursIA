## 1. La question — préparer un état vs implémenter une opération

Le problème de la **synthèse unitaire** (Aaronson-Kuperberg 2007) tient en une asymétrie. On sait depuis longtemps : avec l'accès à la bonne fonction classique $f$ (un oracle), un ordinateur quantique peut **préparer efficacement n'importe quel état quantique** $|0\rangle \mapsto \sum_x \alpha_x |x\rangle$. La question posée demande **plus** : le même type d'accès permet-il aussi d'**implémenter efficacement n'importe quelle opération unitaire** $U$ — pas seulement préparer un état, mais *agir* sur un état arbitraire ?

Formellement : existe-t-il un circuit efficace qui, donné accès à un oracle classique $f$, implémente $U$ pour un $U$ cible — quand $f$ est choisie judicieusement ? La conjecture **négative** de Lombardi-Ma-Wright disait non (il faudrait toujours de la dureté intrinsèque) ; Khurana a cru l'inverse pendant trois ans — minoritaire, « unpopular » — et le manuscrit d'OpenAI (Cité : « manuscript *claiming* a positive solution », vérification en cours) lui donne enfin raison, *si* la preuve tient.

## 3. Conséquence pour la cryptographie quantique — la dureté doit vivre quelque part

Si la preuve d'OpenAI tient, ce qu'elle change pour le champ :

- **La dureté de la crypto quantique doit vivre dans un problème classique.** Si un oracle classique (gratuit) implémente toute unitaire, alors « l'unitaire difficile » n'existe plus comme ressource de dureté quantique autonome : la sécurité doit venir d'une dureté *classique*.
- **Mais la crypto quantique peut en exiger bien moins que la classique** (Cité : voir Kretschmer, Kretschmer-Qian-Sinha-Tal, Kretschmer-Qian-Tal) — « quantum cryptography may still need far less of that hardness than classical cryptography does ». Le champ ne s'effondre pas : il se *recale*.

## 5. Jouet exécutable — la même machine que le 06b, l'autre question

Le socle du [06b — Aaronson-Gottesman](Complexity-06b-AaronsonGottesman-Dequantification-Python.ipynb) exécute le formalisme stabilisateur : Gottesman-Knill montre qu'un circuit **Clifford** se *simule* classiquement en $O(n^2)$ — une *simulation*, pas une *synthèse* : on reconnaît l'état final d'un circuit donné, on ne *construit* pas un circuit pour une unitaire cible. Le jouet ci-dessous distille la différence sur un exemple mesuré : préparer un état cible par oracle classique est trivial (une permutation de base) ; implémenter une unitaire arbitraire l'exige d'une structure.

```python
import numpy as np

# Un oracle classique = une permutation des etats de base.
# Preparer un etat cible arbitraire : une SEULE permutation suffit (Mesure).
n = 3  # 3 qubits, 8 etats de base
cible = np.array([0, 0, 0, 0.5, 0, 0, 0, np.sqrt(0.75)])  # etat cible (norme 1)
print(f"Etat cible sur {n} qubits : {cible} (Mesure)")

# Preparation : trouver la permutation qui envoie |0...0> sur le bon support
# et ajuster les amplitudes : operation TRIVIALE par oracle classique.
support = np.nonzero(cible)[0]
print(f"Support de la cible : etats de base {support} (Mesure)")
print("-> Preparation : 1 permutation de base + ajustement d'amplitudes (1 requete) (Mesure)")

# Implementer une unitaire ARBITRAIRE : beaucoup plus exigeant.
# Une unitaire n x n complexe a ~ n^2 parametres reels libres.
U_taille = 2 ** n
param_libres = U_taille ** 2
print(f"Une unitaire {U_taille}x{U_taille} a ~{param_libres} parametres reels libres (Mesure)")
print(f"Une permutation n'en encode que {U_taille}! arrangements (Mesure)")
print("-> Synthese : la permutation seule NE SUFFIT PAS pour une unitaire arbitraire")
print("   -> il faut de la STRUCTURE dans l'oracle (borne Huang-Ramkumar-Wright) (Cite)")
```

## Lecture du résultat — un décompte n'est pas une dimension

Les deux dernières lignes de la sortie se lisent ensemble, et le contraste est un **piège de lecture** si on le prend pour un argument de cardinal : $8^2 = 64$ paramètres réels pour une unitaire $8\times 8$, contre $8! = 40\,320$ permutations — donc « la permutation aurait plus d'objets ». Ce n'est pas le propos, parce que les deux nombres ne comptent pas la même chose. $64$ est une **dimension** : le nombre de paramètres réels continus qu'il faut pour décrire l'objet. $8!$ est un **cardinal** : le nombre d'objets discrets disponibles. Une permutation est un point isolé de l'espace des unitaires ; aussi nombreux soient ces points, ils ne couvrent pas un espace de dimension $64$ — de même qu'une infinité de points ne remplit pas un segment. Ce qui manque à l'oracle classique est donc la **structure** (la borne Huang-Ramkumar-Wright citée à la ligne suivante), pas le nombre d'arrangements.
