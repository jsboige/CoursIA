## 4. Les barrières algorithmiques — un exposant rationnel, et treize 9

Deux résultats cassent des **barrières d'exposant** vieilles de décennies : le **produit matriciel en $n^{9/4+o(1)}$** — un exposant *rationnel* pour une fois, obtenu par une approche « complètement différente » de la lignée $O(n^{2.373})$ ; et la **multiplication d'entiers en $O(n \log^{0.9999999999999} n)$** — en dessous de la barrière $n \log n$ qui tenait depuis les années 1960 (Knuth-Schönhage-Strassen). Deux subtilités que le fil du billet rend vivantes : le résultat matriciel **n'est pas constructif** (commentaire de Bram Cohen : une existence de type Ramsey, pas un algorithme qu'on peut exécuter) ; et l'exposant $0.9999999999999$ — **treize neuf** — dit une victoire asymptotique totalement invisible en pratique. On mesure les deux :

```python
import math

# (a) Matmul : le facteur asymptotique n^2.373 -> n^2.25
for n in (10**6, 10**9, 10**12):
    facteur = n ** (2.373 - 2.25)
    print(f"matmul n={n:.0e} : facteur asymptotique gagne = n^0.123 = {facteur:,.0f}x (Mesure)")
# ... mais avec un surcout de constante 100x (plausible pour un schema recursif neuf) :
n_bascule = 10 ** (math.log10(100) / (2.373 - 2.25))
print(f"si le nouvel algo coute 100x plus par operation, il ne gagne qu'a partir de n ~ 10^{math.log10(n_bascule):.1f} (Mesure)")

# (b) Entiers : log^0.9999999999999 vs log - l'ecart sur des tailles reelles
expo = 0.9999999999999   # treize neuf (Cite, billet)
for bits in (64, 10**6):
    log2n = bits  # n = 2**bits, donc log2(n) = bits
    ancien = log2n ** 1.0
    nouveau = log2n ** expo
    print(f"entiers n=2^{bits}: ancien ~{ancien:.1f}, nouveau ~{nouveau:.15f}, gain {ancien/nouveau:.15f}x (Mesure)")
print("-> la victoire asymptotique est exacte mais invisible a toute taille realiste (Mesure)")
```

## 5. 3SUM et SETH — le contrepoint humain du même lundi soir

Le **même lundi soir**, Virginia Williams et Josh Alman ont posté sur arXiv un préprint qui résout **3SUM en $O(n^{1.9992})$** et **All-Pairs Shortest Paths en $O(n^{2.9995})$** — réfutant des conjectures vieux de cinquante ans selon lesquelles les bonnes bornes étaient $n^2$ et $n^3$. Deux leçons dans le billet : (a) **l'idée cruciale a été fournie par un modèle Anthropic**, pas OpenAI — mais Anthropic a choisi une voie de diffusion opposée : plutôt que publier les solutions non digérées au monde, il les a transmises à des mathématiciens humains (Williams et Alman) qui les ont travaillées, vérifiées et signées ; (b) **SETH reste intacte** : le papier « dit explicitement que SETH est unaffected and still open » — 3SUM sous-quadratique ne tue pas la Strong Exponential Time Hypothesis. On mesure ce que $n^2 \to n^{1.9992}$ vaut vraiment, et le scaling réel d'un 3SUM quadratique :

```python
import random, time

# (a) L'ecart asymptotique n^2 vs n^1.9992
for n in (10**6, 10**12):
    print(f"3SUM n={n:.0e} : facteur gagne = n^0.0008 = {n**0.0008:.4f}x (Mesure)")
print("-> casser la barriere THEORIQUE n^2 ne donne presque rien a taille realiste (Mesure)")

# (b) Le 3SUM quadratique (hash des complements), scaling mesure
import math
random.seed(42)

def trois_somme_hash(L):
    """O(n^2) : pour chaque paire (i,j), cherche -(L[i]+L[j]) dans la table
    valeur -> indices, en exigeant un indice k distinct de i et de j --
    le meme contrat d'indices distincts i < j < k que la triple boucle."""
    indices = {}
    for k, v in enumerate(L):
        indices.setdefault(v, []).append(k)
    n = len(L)
    for i in range(n):
        for j in range(i + 1, n):
            for k in indices.get(-(L[i] + L[j]), ()):
                if k != i and k != j:
                    return True
    return False

# Garde de correction : le hachage doit rendre la meme reponse que la triple
# boucle (indices distincts). Sur [-2, 1, 4], la paire (0,1) vise
# -(L[0]+L[1]) = 1 : la valeur est presente mais son seul indice (1) est deja
# pris par j -> aucun triplet i<j<k ne somme a zero -> False.
# Sur [5, -3, -2, 40] : 5-3-2 = 0 avec trois indices distincts -> True.
assert not trois_somme_hash([-2, 1, 4])
assert trois_somme_hash([5, -3, -2, 40])
print("Garde 3SUM : temoins [-2,1,4]->False, [5,-3,-2,40]->True -- indices distincts respectes (Controle)")

# Garde de mesure : entiers TOUS POSITIFS -> aucun triplet nul possible ->
# aucun retour premuture, la boucle explore bien les n(n-1)/2 paires (mesure du quadratique complet)
temps = []
for n in (400, 800, 1600, 3200):
    L = [random.randint(1, 10**6) for _ in range(n)]
    t0 = time.perf_counter()
    r = trois_somme_hash(L)
    dt = time.perf_counter() - t0
    temps.append((n, dt))
    print(f"n={n:5d} : {dt*1000:8.2f} ms, triplet={r} (Mesure, parcours complet garanti)")
(n1, t1), (n2, t2) = temps[1], temps[-1]
expo_emp = math.log(t2 / t1) / math.log(n2 / n1)
print(f"Exposant empirique mesure entre n={n1} et n={n2} : {expo_emp:.2f} (Mesure - attendu ~2)")
```
