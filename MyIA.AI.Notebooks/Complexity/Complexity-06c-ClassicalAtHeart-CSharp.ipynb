## 1. La question — préparer un état vs implémenter une opération

Le problème de la **synthèse unitaire** (Aaronson-Kuperberg 2007) tient en une asymétrie. On sait depuis longtemps : avec l'accès à la bonne fonction classique $f$ (un oracle), un ordinateur quantique peut **préparer efficacement n'importe quel état quantique** $|0\rangle \mapsto \sum_x \alpha_x |x\rangle$. La question posée demande **plus** : le même type d'accès permet-il aussi d'**implémenter efficacement n'importe quelle opération unitaire** $U$ — pas seulement préparer un état, mais *agir* sur un état arbitraire ?

Formellement : existe-t-il un circuit efficace qui, donné accès à un oracle classique $f$, implémente $U$ pour un $U$ cible ? La conjecture **négative** de Lombardi-Ma-Wright disait non (il faudrait toujours de la dureté intrinsèque) ; Khurana a cru l'inverse pendant trois ans, minoritaire, et le manuscrit d'OpenAI (Cité : « manuscript *claiming* a positive solution », vérification en cours) lui donne enfin raison, *si* la preuve tient.

## 2. Le récit — trois ans, une conviction, et les machines

L'article est le récit d'une obsession. Khurana écrit : « *For the past three years, the unitary synthesis problem has occupied a majority of my research time, a considerable part of the rest of my waking life, and more than a few sleepless nights.* »

Elle s'y est intéressée **par la cryptographie** : avec son étudiant **Kabir Tomer**, elle avait montré que plusieurs formes de *quantum one-wayness* suffisent à une crypto intéressante. La question naturelle : la crypto quantique peut-elle reposer sur quelque chose de plus faible — sans dureté classique du tout ? La conjecture Lombardi-Ma-Wright disait oui, et impliquait une réponse **négative** à la synthèse unitaire.

Khurana a cru l'inverse pour deux raisons, une mathématique et une esthétique :

- **Les permutations de base sont déjà puissantes** : l'accès classique permet déjà de préparer des états et d'implémenter des permutations de base arbitraires. Elle ne voyait pas pourquoi elles ne serviraient pas à construire des unitaires arbitraires.
- **Symétrie du monde** : une réponse positive mettrait les *opérations* sur le même pied que les *états* — « the world seemed more beautiful, more symmetric that way ».

Le point le plus frappant : « *Over the past year, I worked extensively with the robots too, asking agents to explore ideas and attack the obstacles we were hitting.* » — les machines explorent, l'humaine garde l'intuition et signe le récit. ## 3. Conséquence pour la cryptographie quantique — la dureté doit vivre quelque part

Si la preuve d'OpenAI tient :

- **La dureté de la crypto quantique doit vivre dans un problème classique.** Si un oracle classique (gratuit) implémente toute unitaire, « l'unitaire difficile » n'existe plus comme ressource de dureté autonome : la sécurité doit venir d'une dureté *classique*.
- **Mais la crypto quantique peut en exiger bien moins que la classique** (Cité : Kretschmer, Kretschmer-Qian-Sinha-Tal, Kretschmer-Qian-Tal) — le champ ne s'effondre pas : il se *recale*.

La conclusion de Khurana tient le titre : « *I trained as a classical cryptographer... Perhaps I am classical at heart. So, it seems, is the field I love.* »

## 4. Les deux préprints quasi simultanés — une evidence, une borne

| Préprint | Ce qu'il établit (Cité) |
|---|---|
| **Nehoran & Yuen** | la preuve la plus forte d'une résolution **positive** juste avant l'annonce : plus d'une requête peut être puissante et **réduire la taille du circuit quantique** |
| **Huang, Ramkumar & Wright** | une borne **négative structurelle** : une structure est nécessaire pour synthétiser des unitaires aléatoires — les oracles non structurés **ne suffisent pas** |

Ensemble : la voie du manuscrit OpenAI (s'il tient) devait passer par un oracle *structuré* — cohérent avec l'intuition de Khurana sur les permutations.

## 5. Jouet exécutable — ce qu'une permutation fait, et ne fait pas

Le socle du [06b](Complexity-06b-AaronsonGottesman-Dequantification-Python.ipynb) exécute le formalisme stabilisateur : un circuit **Clifford** se *simule* classiquement — on **reconnaît** l'état final d'un circuit donné, on ne **construit** pas un circuit pour une unitaire cible. Le jouet ci-dessous distille la différence, et **corrige une lecture trop rapide** du port Python :

- **Une permutation de base** est une bijection des états de base : elle envoie $|0\ldots0\rangle$ sur **un seul** état de base. Elle ne prépare **pas** une superposition arbitraire — l'objet que la synthèse unitaire vise.
- **Réarranger le support et fournir les amplitudes** est une **description classique** d'un état cible ; ce n'est **pas** une synthèse quantique mesurée (aucune unitaire n'est construite ni exécutée).
- **Implémenter une unitaire arbitraire** exige de la **structure** dans l'oracle — le propos de la borne Huang-Ramkumar-Wright.

Le banc mesure, avec de vraies matrices complexes, le décompte (dimension vs cardinal) qui fonde ce propos.

```
#nullable enable
using System;
using System.Linq;
using System.Numerics;
using System.Globalization;
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;
CultureInfo.CurrentUICulture = CultureInfo.InvariantCulture;

int n = 3;
int dim = 1 << n;
// Etat cible : superposition de DEUX etats de base (support de taille 2), amplitudes complexes.
Complex[] cible = new Complex[dim];
cible[3] = new Complex(0.5, 0);
cible[7] = new Complex(Math.Sqrt(0.75), 0);
double norme = cible.Sum(z => z.Magnitude * z.Magnitude);
var support = Enumerable.Range(0, dim).Where(i => cible[i].Magnitude > 1e-12).ToArray();
Console.WriteLine($"Etat cible sur {n} qubits : support [{string.Join(",", support)}], norme² = {norme:F6} (Mesure).");

// (a) Vraie operation matricielle complexe : une porte Hadamard 2x2 (unitaire) appliquee a un vecteur de 2 amplitudes.
Complex[,] H = new Complex[2, 2] {
    { new Complex(1 / Math.Sqrt(2), 0), new Complex(1 / Math.Sqrt(2), 0) },
    { new Complex(1 / Math.Sqrt(2), 0), new Complex(-1 / Math.Sqrt(2), 0) }
};
Complex[] v = { Complex.One, Complex.Zero };   // |0>
Complex[] Hv = new Complex[2];
for (int i = 0; i < 2; i++) { Hv[i] = Complex.Zero; for (int j = 0; j < 2; j++) Hv[i] += H[i, j] * v[j]; }
Console.WriteLine($"H|0> = [{Hv[0].Real:F4} + {Hv[0].Imaginary:F4}i, {Hv[1].Real:F4} + {Hv[1].Imaginary:F4}i] : vraie multiplication matrice-complexe x vecteur-complexe (Mesure).");

// (b) Une permutation de base prepare UN SEUL etat de base (support taille 1) : pas la superposition de support 2 ci-dessus.
Console.WriteLine("Une permutation de base envoie |0..0> sur UN etat de base (support 1) ; elle ne prepare PAS la superposition de support 2 ci-dessus (Mesure).");

// (c) Le decompote dimension vs cardinal.
int paramLibres = dim * dim;
long fact = 1; for (int i = 2; i <= dim; i++) fact *= i;
Console.WriteLine($"Une unitaire {dim}x{dim} a ~{paramLibres} parametres reels libres (une DIMENSION) ; une permutation n'en encode que {dim}! = {fact} arrangements (un CARDINAL) (Mesure).");
Console.WriteLine("-> La permutation seule ne prepare PAS une superposition arbitraire ; il faut de la STRUCTURE dans l'oracle (borne Huang-Ramkumar-Wright) (Cite).");

if (support.Length != 2) throw new Exception("Le banc suppose un support de taille 2 (superposition), hors de la porte d'une permutation.");
Console.WriteLine($"Contrôle : la cible porte {support.Length} amplitudes non nulles — hors de la portée d'une permutation (Mesure).");
```

## Lecture du résultat — un décompte n'est pas une dimension

Les deux lignes « 64 paramètres réels » et « 8! arrangements » se lisent ensemble, et le contraste est un **piège de lecture** si on le prend pour un argument de cardinal : $8^2 = 64$ paramètres réels pour une unitaire $8\times 8$, contre $8! = 40\,320$ permutations — « la permutation aurait plus d'objets ». Ce n'est pas le propos : les deux nombres ne comptent pas la même chose. $64$ est une **dimension** (des paramètres réels continus) ; $8!$ est un **cardinal** (des objets discrets). Une permutation est un point isolé de l'espace des unitaires ; aussi nombreux soient ces points, ils ne couvrent pas un espace de dimension $64$ — de même qu'une infinité de points ne remplit pas un segment. Ce qui manque à l'oracle classique est la **structure** (la borne Huang-Ramkumar-Wright), pas le nombre d'arrangements.

Le [06](Complexity-06-Simuler-Circuit-Quantique-Classiquement-CSharp.ipynb) posait le coût de *simuler* un circuit quantique par du classique ; le 06b exécutait le cas Clifford simulable. La réponse — revendiquée positive, en cours de vérification — est que le classique achète plus qu'on ne le croyait : il ne se contente pas de préparer des états, il *implémente des opérations*, pourvu que l'oracle porte de la **structure**.

Ce que le port C# a corrigé : la permutation de base prépare **un seul** état de base (support 1) ; elle ne prépare **pas** une superposition arbitraire, et « réarranger le support + fournir les amplitudes » est une description classique, pas une synthèse mesurée.
