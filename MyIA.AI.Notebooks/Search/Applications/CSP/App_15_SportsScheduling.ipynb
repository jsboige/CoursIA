## 1. Modelisation du Problème

### Definition du CSP

| Élément | Description |
|---------|-------------|
| **Variables** | `match[i,j,r]` = 1 si équipe i recoit équipe j a la ronde r |
| **Domaines** | {0, 1} (binaire) |
| **Contraintes** | Voir ci-dessous |

### Contraintes Hard (obligatoires)

1. **Round-robin simple** : chaque pair d'équipes joue exactement une fois
2. **Une équipe par match** : une équipe joue exactement un match par ronde
3. **Symetrie** : si A vs B, alors pas B vs A dans la même ronde
4. **Capacite stade** : une équipe ne peut pas recevoir deux fois même ronde

### Contraintes Soft (préférences)

1. **Equilibre D/E** : alternance domicile/exterieur
2. **Minimiser deplacements** : reduire les km parcourus
3. **TV slots** : matchs attractifs le samedi soir
4. **Rivalries** : derbys dans des rondes spécifiques
**Dans le modèle exécuté**, la limite de deux matchs consécutifs du même type
est une contrainte obligatoire. Les créneaux TV sont illustrés dans une
extension du support ; les règles de derbys font l’objet d’un autre exercice.


```python
@dataclass
class Team:
    """Representation d'une equipe de football."""
    id: int
    name: str
    city: str
    latitude: float
    longitude: float
    stadium_capacity: int
    tv_attractiveness: float  # 0-1, influence TV

    def __hash__(self):
        return self.id

    def __eq__(self, other):
        return self.id == other.id


@dataclass
class SportsLeague:
    """Ligue sportive avec equipes et contraintes."""
    name: str
    teams: List[Team]
    rounds: int

    def __post_init__(self):
        self.n_teams = len(self.teams)
        # En round-robin simple: n_teams - 1 rondes
        if self.rounds is None:
            self.rounds = self.n_teams - 1

    def distance(self, team1: Team, team2: Team) -> float:
        """Calcule la distance en km entre deux villes (approximation)."""
        # Formule haversine simplifiee
        lat1, lon1 = team1.latitude, team1.longitude
        lat2, lon2 = team2.latitude, team2.longitude

        R = 6371  # Rayon terre en km
        phi1, phi2 = np.radians([lat1, lat2])
        delta_phi = np.radians(lat2 - lat1)
        delta_lambda = np.radians(lon2 - lon1)

        a = np.sin(delta_phi/2)**2 + np.cos(phi1) * np.cos(phi2) * np.sin(delta_lambda/2)**2
        c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1-a))

        return R * c


def create_ligue1_sample(n_teams: int = 8) -> SportsLeague:
    """Cree un echantillon de la Ligue 1 (simplifie)."""
    # Coordonnees approximatives des villes francaises
    cities = [
        ("Paris", 48.8566, 2.3522),
        ("Marseille", 43.2965, 5.3698),
        ("Lyon", 45.7640, 4.8357),
        ("Lille", 50.6292, 3.0573),
        ("Bordeaux", 44.8378, -0.5792),
        ("Nantes", 47.2184, -1.5536),
        ("Nice", 43.7102, 7.2620),
        ("Toulouse", 43.6047, 1.4442),
        ("Rennes", 48.1173, -1.6778),
        ("Strasbourg", 48.5734, 7.7521),
        ("Montpellier", 43.6119, 3.8772),
        ("Reims", 49.2583, 4.0317),
    ]

    teams = []
    for i in range(min(n_teams, len(cities))):
        city_name, lat, lon = cities[i]
        teams.append(Team(
            id=i,
            name=f"{city_name} FC",
            city=city_name,
            latitude=lat,
            longitude=lon,
            stadium_capacity=random.randint(20000, 60000),
            tv_attractiveness=random.uniform(0.3, 1.0)
        ))

    return SportsLeague(
        name=f"Ligue 1 (Echantillon {n_teams} equipes)",
        teams=teams,
        rounds=n_teams - 1
    )


# Creation de la ligue
league = create_ligue1_sample(n_teams=6)
print(f"Ligue creee: {league.name}")
print(f"Equipes: {[t.name for t in league.teams]}")
print(f"Rondes: {league.rounds}")
```

## 2. Calcul de la Matrice de Distances

La matrice contient les distances géographiques entre les villes et permet notamment de repérer les trajets de plus de 500 km.


```python
def compute_distance_matrix(league: SportsLeague) -> np.ndarray:
    """Calcule la matrice des distances entre toutes les equipes."""
    n = league.n_teams
    distances = np.zeros((n, n))

    for i in range(n):
        for j in range(i+1, n):
            d = league.distance(league.teams[i], league.teams[j])
            distances[i, j] = d
            distances[j, i] = d

    return distances


# Calcul et affichage
distances = compute_distance_matrix(league)

plt.figure(figsize=(8, 6))
plt.imshow(distances, cmap='YlOrRd')
plt.colorbar(label='Distance (km)')
plt.xticks(range(league.n_teams), [t.city for t in league.teams], rotation=45)
plt.yticks(range(league.n_teams), [t.city for t in league.teams])
plt.title('Matrice des Distances')
plt.tight_layout()
plt.show()

print("\nDistances (km):")
df_distances = pd.DataFrame(
    distances.astype(int),
    index=[t.city for t in league.teams],
    columns=[t.city for t in league.teams]
)
display(df_distances)
```

calculées avec la formule de Haversine.

|--------|-------------|---------------|

| Plus proches | Paris–Lille | Environ 203 km |
| Plus éloignées | Marseille–Lille | Environ 834 km |

**Points clés** :
1. La distance est symétrique et la diagonale vaut zéro.
2. Le seuil de l'exercice porte sur les valeurs décimales : strictement plus de 500 km.
3. Le tableau affiché tronque les décimales ; le calcul du seuil utilise la matrice d'origine.
4. La proximité géographique seule ne suffit pas à définir une rivalité sportive réelle.

> **Note technique** : Haversine donne une distance à la surface d'une Terre
> sphérique. Les trajets routiers réels et les retours ne sont pas modélisés ici.


## 3. Modelisation CP-SAT

Nous utilisons OR-Tools CP-SAT pour resoudre le problème.

```python
class SportsScheduler:
    """Solveur CP-SAT pour la planification sportive."""

    def __init__(self, league: SportsLeague, distances: np.ndarray):
        self.league = league
        self.distances = distances
        self.model = cp_model.CpModel()
        self.solver = cp_model.CpSolver()

        # Variables de decision
        self.match_vars = {}  # (home, away, round) -> BoolVar
        self._create_variables()
        self._add_constraints()

    def _create_variables(self):
        """Cree les variables de decision."""
        n = self.league.n_teams
        rounds = self.league.rounds

        # match[i,j,r] = 1 si equipe i recoit equipe j a la ronde r
        for r in range(rounds):
            for i in range(n):
                for j in range(n):
                    if i != j:
                        self.match_vars[(i, j, r)] = self.model.NewBoolVar(
                            f'match_{i}_{j}_r{r}'
                        )

    def _add_constraints(self):
        """Ajoute les contraintes au modele."""
        n = self.league.n_teams
        rounds = self.league.rounds

        # C1: Chaque pair d'equipes joue exactement une fois (round-robin simple)
        for i in range(n):
            for j in range(i+1, n):
                # Soit i recoit j, soit j recoit i, exactement une fois
                self.model.Add(
                    sum(self.match_vars[(i, j, r)] + self.match_vars[(j, i, r)]
                        for r in range(rounds)) == 1
                )

        # C2: Chaque equipe joue exactement une fois par ronde
        for r in range(rounds):
            for i in range(n):
                # Soit l'equipe i joue a domicile, soit a l'exterieur
                self.model.Add(
                    sum(self.match_vars[(i, j, r)] + self.match_vars[(j, i, r)]
                        for j in range(n) if j != i) == 1
                )

        # C3: Au plus un match a domicile par ronde (implicitement satisfait par C2)
        # C4: Au plus un match a l'exterieur par ronde (implicitement satisfait par C2)

    def add_balance_constraint(self, max_consecutive: int = 2):
        """
        Ajoute une contrainte d'equilibre domicile/exterieur.
        Evite plus de max_consecutive matchs consecutifs a domicile ou exterieur.
        """
        n = self.league.n_teams
        rounds = self.league.rounds

        for i in range(n):
            for r in range(rounds - max_consecutive):
                # Somme des matchs domicile sur max_consecutive+1 rondes
                home_streak = sum(
                    sum(self.match_vars[(i, j, r+k)] for j in range(n) if j != i)
                    for k in range(max_consecutive + 1)
                )
                self.model.Add(home_streak <= max_consecutive)

                # Meme chose pour l'exterieur
                away_streak = sum(
                    sum(self.match_vars[(j, i, r+k)] for j in range(n) if j != i)
                    for k in range(max_consecutive + 1)
                )
                self.model.Add(away_streak <= max_consecutive)

    def minimize_travel(self, weight: int = 100):
        """
        Ajoute un objectif de minimisation des deplacements.
        """
        n = self.league.n_teams
        rounds = self.league.rounds

        travel_terms = []
        for r in range(rounds):
            for i in range(n):
                for j in range(n):
                    if i != j:
                        # Distance parcourue si j joue a l'exterieur chez i
                        travel_terms.append(
                            int(self.distances[i, j]) * self.match_vars[(i, j, r)]
                        )

        self.model.Minimize(sum(travel_terms))

    def solve(self, time_limit: int = 30) -> bool:
        """Resout le probleme."""
        self.solver.parameters.max_time_in_seconds = time_limit
        self.solver.parameters.num_search_workers = 1
        self.solver.parameters.random_seed = 42

        self._status = self.solver.Solve(self.model)
        return self._status == cp_model.OPTIMAL or self._status == cp_model.FEASIBLE

    def get_schedule(self) -> Dict[int, List[Tuple[int, int]]]:
        """Retourne le calendrier solution."""
        # Utiliser status_name() pour verifier le statut (correction API OR-Tools)
        if getattr(self, "_status", None) not in (cp_model.OPTIMAL, cp_model.FEASIBLE):
            return {}

        schedule = {r: [] for r in range(self.league.rounds)}

        for (i, j, r), var in self.match_vars.items():
            if self.solver.Value(var) == 1:
                schedule[r].append((i, j))  # i recoit j

        return schedule


# Creation et resolution du probleme
scheduler = SportsScheduler(league, distances)
scheduler.add_balance_constraint(max_consecutive=2)
scheduler.minimize_travel()

print("Resolution en cours...")
if scheduler.solve(time_limit=10):
    print(f"Solution trouvee ! (Statut : {scheduler.solver.StatusName(scheduler._status)})")
    print(f"Objectif du solveur (somme des distances, km) = {scheduler.solver.ObjectiveValue():.0f}")
    schedule = scheduler.get_schedule()
else:
    print("Pas de solution trouvee")
    schedule = {}
```

### Interpretation : Resolution du Problème CSP

**Sortie obtenue** : une solution optimale ou faisable est trouvée par le solveur CP-SAT.

| Aspect | Valeur | Signification |
|--------|--------|---------------|
| Statut | OPTIMAL ou FEASIBLE | OPTIMAL prouve l'optimalité ; FEASIBLE indique une solution valide sans preuve d'optimalité |
| Temps de résolution | Dépend de la machine et des paramètres | Mesuré lors de l'exécution, sans garantie de durée générale |
| Variables de match | n × (n − 1) × rondes, soit 150 ici | Une variable booléenne pour chaque match orienté et chaque ronde |

**Points clés** :
1. **Résolution CP-SAT** : le solveur cherche une affectation des variables compatible avec les contraintes.
2. **Contraintes satisfaites** : chaque paire se rencontre une fois, chaque équipe joue une fois par ronde, pas plus de deux matchs consécutifs à domicile ou à l'extérieur.
3. **Objectif de distance** : l'expression du support est conservée. Avec une rencontre par paire et des distances symétriques, sa somme vaut toujours 7 390 km pour cette instance.
4. **Extensibilité** : le modèle peut recevoir d'autres contraintes, dont l'interdiction des longs déplacements consécutifs dans l'exercice suivant.

> **Note technique** : CP-SAT combine notamment propagation et recherche.
> La limite de temps borne la recherche ; l'absence de solution trouvée dans
> ce délai ne prouve pas à elle seule que le problème est impossible.


#### Exercice 1 : Limiter les longs deplacements consecutifs

Le modèle actuel contient un objectif de distance, mais ne limite pas les longs déplacements consécutifs. Une équipe peut, par exemple, jouer à Nantes puis à Lille lors de deux rondes successives.

**Enonce** : ajoutez une contrainte qui interdit a une équipe de faire plus d'un deplacement superieur a 500 km lors de deux rondes consecutives.

**Consignes** :
1. Pour chaque équipe et chaque paire de rondes consecutives, comptez les matchs a l'exterieur impliquant un deplacement > 500 km
2. Ajoutez une contrainte : la somme de ces longs deplacements sur 2 rondes consecutives ne doit pas depasser 1
3. Resolvez et comparez les statistiques de deplacement avec le calendrier initial

**Indice** : créez une variable auxiliaire boolVar `long_trip[i,r]` qui vaut 1 si l'équipe i voyage a l'exterieur de plus de 500 km a la ronde r, puis contraignez `sum(long_trip[i,r+k] for k in range(2)) <= 1`.

```python
# Exercice EX-0373 — Solution commentée
# Objectif : interdire deux déplacements strictement supérieurs à 500 km lors de deux journées consécutives pour une même équipe.

# Cette fonction parcourt les journées et affiche les rencontres pour afficher le calendrier
def display_schedule(league: SportsLeague, schedule: Dict[int, List[Tuple[int, int]]]):
    """Affiche le calendrier de maniere lisible."""
    print(f"\n=== CALENDRIER {league.name} ===\n")

    for r in sorted(schedule.keys()):
        print(f"Ronde {r+1}:")
        for home, away in schedule[r]:
            home_team = league.teams[home]
            away_team = league.teams[away]
            print(f"  {home_team.name} vs {away_team.name}")
        print()



# On compte les matchs à domicile et à l'extérieur, puis les kilomètres parcourus par chaque équipe
def compute_schedule_stats(league: SportsLeague, schedule: Dict[int, List[Tuple[int, int]]],
                          distances: np.ndarray) -> Dict:
    """Calcule les statistiques du calendrier."""
    n = league.n_teams

    home_games = defaultdict(int, {i: 0 for i in range(n)})
    away_games = defaultdict(int, {i: 0 for i in range(n)})
    total_travel = defaultdict(float, {i: 0.0 for i in range(n)})

    for r, matches in schedule.items():
        for home, away in matches:
            home_games[home] += 1
            away_games[away] += 1
            total_travel[away] += int(distances[home, away])  # int-tronque, comme l'objectif

    return {
        'home_games': dict(home_games),
        'away_games': dict(away_games),
        'total_travel': dict(total_travel),
        'avg_travel': np.mean(list(total_travel.values())),
        'travel_std': np.std(list(total_travel.values()))
    }



SEUIL = 500  # km : un long trajet est STRICTEMENT supérieur à ce seuil



# on crée un indicateur de long déplacement, puis on interdit que cet indicateur vaille 1 deux journées de suite.
def ajouter_contrainte_longs_trajets(scheduler, seuil=500):
    """Ajoute la contrainte de l'exercice, sans changer les règles de base."""
    n = scheduler.league.n_teams
    rondes = scheduler.league.rounds
    long_trip = {}

    for i in range(n):
        for r in range(rondes):
            # 1 = long déplacement ; 0 = domicile ou déplacement <= 500 km.
            long_trip[i, r] = scheduler.model.NewBoolVar(f"long_trip_{i}_{r}")
            # (j, i, r) : j reçoit i ; c'est donc i qui se déplace.
            matchs_lointains = [
                scheduler.match_vars[j, i, r]
                for j in range(n)
                if j != i and scheduler.distances[j, i] > seuil
            ]
            # Une équipe joue une seule fois par ronde : la somme vaut 0 ou 1. Si l'un de ces matchs est choisi, long_trip vaut 1 ; sinon 0
            scheduler.model.Add(long_trip[i, r] == sum(matchs_lointains))

        for r in range(rondes - 1):
            scheduler.model.Add(long_trip[i, r] + long_trip[i, r + 1] <= 1)

    return long_trip


def analyser_calendrier(calendrier, league, distances, seuil=500):
    """Recalcule les déplacements à partir des matchs, sans lire les BoolVar."""
    n, rondes = league.n_teams, league.rounds
    trajets = np.zeros((n, rondes))
    domicile = np.zeros((n, rondes), dtype=int)
    exterieur = np.zeros((n, rondes), dtype=int)

    for r, matchs in calendrier.items():
        for home, away in matchs:
            domicile[home, r] += 1
            exterieur[away, r] += 1
            trajets[away, r] = distances[home, away]

    # Un zéro signifie domicile OU déplacement de 500 km maximum.
    longs = (trajets > seuil).astype(int)
    violations = (longs[:, :-1] + longs[:, 1:] > 1).sum(axis=1)
    km = trajets.astype(int).sum(axis=1)
    tableau = pd.DataFrame({
        "Équipe": [t.city for t in league.teams],
        "Domicile": domicile.sum(axis=1),
        "Extérieur": exterieur.sum(axis=1),
        "Distance (km)": km,
        "Longs trajets": longs.sum(axis=1),
        "Enchaînements interdits": violations,
    })
    return tableau, longs, domicile, exterieur


def verifier_calendrier(calendrier, league, distances, avec_contrainte=False):
    """Vérifie les rencontres, les séries D/E et les longs trajets."""
    n, rondes = league.n_teams, league.rounds
    erreurs = []
    paires = defaultdict(int)
    if set(calendrier) != set(range(rondes)):
        return ["Les rondes attendues ne sont pas toutes présentes."]
    for r, matchs in calendrier.items():
        if len(matchs) != n // 2:
            erreurs.append(f"Nombre incorrect de matchs à la ronde {r + 1}.")
        participations = [0] * n
        for home, away in matchs:
            if home == away or not (0 <= home < n and 0 <= away < n):
                return ["Un match contient une équipe invalide."]
            participations[home] += 1
            participations[away] += 1
            # Le tri permet de compter une même paire indépendamment
            # du domicile : (Paris, Marseille) et (Marseille, Paris).
            paires[tuple(sorted((home, away)))] += 1
        if participations != [1] * n:
            erreurs.append(f"Participation incorrecte à la ronde {r + 1}.")
    for i in range(n):
        for j in range(i + 1, n):
            if paires[i, j] != 1:
                erreurs.append(f"La paire {i}, {j} ne joue pas exactement une fois.")

    tableau, longs, domicile, exterieur = analyser_calendrier(
        calendrier, league, distances, SEUIL
    )
    # Une fenêtre de trois journées ne doit jamais contenir trois matchs à domicile ni trois matchs à l'extérieur.
    for positions in (domicile, exterieur):
        for r in range(rondes - 2):
            if np.any(positions[:, r:r + 3].sum(axis=1) > 2):
                erreurs.append("Plus de deux matchs D/E consécutifs.")
    if avec_contrainte and tableau["Enchaînements interdits"].sum() != 0:
        erreurs.append("Deux longs trajets consécutifs subsistent.")
    return erreurs



# Le premier calendrier produit automatiquement pouvait déjà respecter la nouvelle règle par hasard. On garde cette référence dans
# schedule_initial_auto et on utilise un exemple illustratif explicite.
# Les indices sont : Paris=0, Marseille=1, Lyon=2,
# Lille=3, Bordeaux=4, Nantes=5. Les journées commencent à l'indice 0.
# Marseille se déplace à Nantes en R2, puis à Lille en R3.
schedule_initial_auto = schedule
schedule_initial_exemple = {
    0: [(1, 0), (2, 5), (4, 3)],
    1: [(0, 4), (3, 2), (5, 1)],
    2: [(2, 4), (3, 1), (5, 0)],
    3: [(0, 3), (1, 2), (4, 5)],
    4: [(1, 4), (2, 0), (3, 5)],
}
schedule = schedule_initial_exemple

if not schedule:
    raise RuntimeError("Exécute d'abord les cellules du modèle initial.")
erreurs_avant = verifier_calendrier(schedule, league, distances)
assert not erreurs_avant, erreurs_avant


# On ajoute notre fonction avant de lancer la recherche du solveur.
scheduler_corrige = SportsScheduler(league, distances)
scheduler_corrige.add_balance_constraint(max_consecutive=2)
scheduler_corrige.minimize_travel()
long_trip = ajouter_contrainte_longs_trajets(scheduler_corrige, SEUIL)

# Le solveur dispose de dix secondes pour chercher une solution.
if not scheduler_corrige.solve(time_limit=10):
    statut = scheduler_corrige.solver.StatusName(scheduler_corrige._status)
    raise RuntimeError(f"Pas de calendrier trouvé : {statut}")
calendrier_corrige = scheduler_corrige.get_schedule()
# Cette fois, on exige aussi le respect de la nouvelle règle.
erreurs_apres = verifier_calendrier(
    calendrier_corrige, league, distances, avec_contrainte=True
)
assert not erreurs_apres, erreurs_apres


# On mesure les kilomètres, leur dispersion entre les équipes,le nombre de longs trajets et surtout les enchaînements interdits.
# L'indicateur principal doit passer de 1 à 0 pour cet exemple.
avant, longs_avant, _, _ = analyser_calendrier(schedule, league, distances, SEUIL)
apres, longs_apres, _, _ = analyser_calendrier(
    calendrier_corrige, league, distances, SEUIL
)
stats_avant = compute_schedule_stats(league, schedule, distances)
stats_apres = compute_schedule_stats(league, calendrier_corrige, distances)
# le calendrier initial et le calendrier corrigé.
comparaison = pd.DataFrame({
    "Calendrier": ["Initial illustratif", "Avec contrainte"],
    "Distance totale (km)": [sum(stats_avant["total_travel"].values()), sum(stats_apres["total_travel"].values())],
    "Écart-type des km par équipe": [stats_avant["travel_std"], stats_apres["travel_std"]],
    "Longs trajets": [avant["Longs trajets"].sum(), apres["Longs trajets"].sum()],
    "Enchaînements interdits": [avant["Enchaînements interdits"].sum(), apres["Enchaînements interdits"].sum()],
})
print("COMPARAISON AVANT / APRÈS")
display(comparaison.round(2))
print("DÉTAIL PAR ÉQUIPE — CALENDRIER CORRIGÉ")
display(apres)
print("Statut :", scheduler_corrige.solver.StatusName(scheduler_corrige._status))
print("Vérification indépendante : toutes les contraintes sont respectées.")

print("\nCALENDRIER CORRIGÉ — domicile reçoit extérieur")
for r, matchs in sorted(calendrier_corrige.items()):
    print(f"Ronde {r + 1} : " + " ; ".join(
        f"{league.teams[h].city} reçoit {league.teams[a].city}"
        for h, a in matchs
    ))

# Visualiser les longs déplacements
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)
for ax, matrice, titre in zip(
    axes, [longs_avant, longs_apres], ["Calendrier initial", "Avec contrainte"]
):
    ax.imshow(matrice, cmap="Blues", vmin=0, vmax=1, aspect="auto")
    ax.set_xticks(range(league.rounds), [f"R{r + 1}" for r in range(league.rounds)])
    ax.set_yticks(range(league.n_teams), [t.city for t in league.teams])
    ax.set_title(titre)
    for i in range(league.n_teams):
        for r in range(league.rounds):
            ax.text(r, i, str(matrice[i, r]), ha="center", va="center",
                    color="white" if matrice[i, r] else "black")
fig.suptitle("Longs déplacements : aucun 1 suivi d'un autre 1 après correction")
plt.tight_layout()
plt.show()

distance_attendue = sum(int(distances[i, j])
                       for i in range(league.n_teams)
                       for j in range(i + 1, league.n_teams))
assert int(comparaison.iloc[0]["Distance totale (km)"]) == distance_attendue
assert int(comparaison.iloc[1]["Distance totale (km)"]) == distance_attendue
print(f"Chaque paire joue une fois : le total reste {distance_attendue} km.")
if avant["Enchaînements interdits"].sum() == 0:
    print("Le calendrier initial respectait déjà la nouvelle règle par hasard.")
    print("La contrainte garantit maintenant cette propriété dans toute solution.")

schedule = calendrier_corrige
scheduler = scheduler_corrige
result_ex0373 = calendrier_corrige
result = result_ex0373

assert avant['Enchaînements interdits'].sum() == 1
assert apres['Enchaînements interdits'].sum() == 0
assert sum(stats_avant['total_travel'].values()) == avant['Distance (km)'].sum()
assert sum(stats_apres['total_travel'].values()) == apres['Distance (km)'].sum()
```

### Contraintes TV

Les chaînes de television veulent des matchs attractifs a certains creneaux.

```python
class TVScheduler(SportsScheduler):
    """Extension avec contraintes TV."""

    def __init__(self, league: SportsLeague, distances: np.ndarray):
        super().__init__(league, distances)
        self.tv_slots = {}  # ronde -> (slot, importance)

    def add_tv_slot(self, round_idx: int, slot_name: str, importance: float):
        """
        Definit un creneau TV important.
        importance: poids pour les matchs attractifs dans ce creneau.
        """
        self.tv_slots[round_idx] = (slot_name, importance)

    def maximize_tv_attractiveness(self):
        """
        Optimise l'attractivite TV des matchs dans les creneaux importants.
        """
        n = self.league.n_teams

        attractiveness_terms = []

        for r, (slot_name, importance) in self.tv_slots.items():
            for i in range(n):
                for j in range(n):
                    if i != j:
                        # Attractivite combinee des deux equipes
                        combined = (
                            self.league.teams[i].tv_attractiveness +
                            self.league.teams[j].tv_attractiveness
                        )
                        # On veut MAXIMISER l'attractivite
                        attractiveness_terms.append(
                            int(importance * combined * 100) * self.match_vars[(i, j, r)]
                        )

        # On maximise l'attractivite (donc on minimise le negatif)
        self.model.Minimize(-sum(attractiveness_terms))


# Exemple avec contraintes TV
tv_scheduler = TVScheduler(league, distances)
tv_scheduler.add_balance_constraint(max_consecutive=2)

# Creneaux TV importants: premiere et derniere ronde
tv_scheduler.add_tv_slot(0, "Ouverture", importance=2.0)
tv_scheduler.add_tv_slot(league.rounds - 1, "Cloture", importance=2.0)

tv_scheduler.maximize_tv_attractiveness()

print("Resolution avec contraintes TV...")
if tv_scheduler.solve(time_limit=10):
    print("Solution trouvee!")
    tv_schedule = tv_scheduler.get_schedule()
    display_schedule(league, tv_schedule)
```

### Interpretation : Optimisation TV

**Sortie obtenue** : exemple du support ajoutant des poids TV aux rondes d'ouverture et de clôture.

| Aspect | Modélisation | Limite |
|--------|--------------|--------|
| Créneaux TV | Une importance par ronde | Aucun horaire précis n'est attribué aux matchs |
| Attractivité | Somme des scores des deux équipes | Toutes les équipes jouent à chaque ronde |
| Objectif TV | Somme pondérée par ronde | Dans cette formulation additive, le total est constant pour chaque ronde |

**Points clés** :
1. Cette extension appartient au support ; elle est distincte de la solution EX-0373.
2. Elle conserve les contraintes de championnat et la limite des séries D/E.
3. La règle des longs trajets n'y est pas ajoutée automatiquement.
4. Pour différencier réellement les calendriers, il faudrait par exemple affecter certains matchs à des créneaux ou utiliser des scores propres aux paires d'équipes.

## 6. Contraintes de Derby

Les derbys (matchs entre équipes geographiquement proches) peuvent avoir des contraintes speciales pour la securite.

```python
def find_derbies(league: SportsLeague, distances: np.ndarray,
                threshold_km: float = 200) -> List[Tuple[int, int]]:
    """
    Identifie les derbys (equipes proches geographiquement).
    """
    derbies = []
    n = league.n_teams

    for i in range(n):
        for j in range(i+1, n):
            if distances[i, j] <= threshold_km:
                derbies.append((i, j))

    return derbies


def schedule_derbies_on_weekends(league: SportsLeague, derbies: List[Tuple[int, int]],
                                schedule: Dict[int, List[Tuple[int, int]]]) -> Dict[int, List[Tuple[int, int]]]:
    """
    Reorganise le calendrier pour placer les derbys en week-end (simplifie).
    Dans un vrai systeme, on ajouterait cette contrainte au modele CP-SAT.
    """
    # Identification des derbys dans le calendrier
    derby_rounds = {}

    for r, matches in schedule.items():
        for home, away in matches:
            if (home, away) in derbies or (away, home) in derbies:
                derby_rounds[(home, away)] = r

    return derby_rounds


# Identification des derbys
derbies = find_derbies(league, distances, threshold_km=300)

print(f"Derbys identifies (distance < 300km):")
for i, j in derbies:
    print(f"  {league.teams[i].name} vs {league.teams[j].name} "
          f"({distances[i, j]:.0f} km)")

if schedule:
    derby_rounds = schedule_derbies_on_weekends(league, derbies, schedule)
    print(f"\nDerbys dans le calendrier:")
    for (home, away), r in derby_rounds.items():
        print(f"  Ronde {r+1}: {league.teams[home].name} vs {league.teams[away].name}")
```

### Interpretation : Identification des Derbys

**Sortie obtenue** : Liste des derbys (équipes distantes de moins de 300 km) et leurs rondes dans le calendrier.

| Aspect | Observation | Signification |
|--------|-------------|---------------|
| **Nombre de derbys** | Depend de la distribution geographique | Plus la densite d'équipes est elevee, plus il y a de derbys |
| **Seuil de distance** | 200-300 km | Distance raisonnable pour un deplacement en bus |
| **Ronde de derby** | Variable | Le solveur place les derbys selon les contraintes globales |

**Points cles** :
1. **Securite et logistique** : Les derbys necessitent une planification speciale (securite renforcee, transports)
2. **Contrainte potentielle** : Dans un vrai système, on pourrait forcer les derbys a etre joues en week-end ou a des heures spécifiques
3. **Impact sur les deplacements** : Les derbis reduisent les distances de voyage pour les équipes impliquees
4. **Modelisation extensible** : Ce système permet d'identifier et de gerer les matchs a haut enjeu

> **Note technique** : Dans un championnat professionnel comme la Ligue 1, les derbys (PSG vs Marseille, Lyon vs Saint-Etienne, etc.) sont des événements majeurs qui necessitent une planification speciale. Les contraintes de securite peuvent etre ajoutees au modèle CSP pour forcer ces matchs a des creneaux spécifiques.

## 7. Extension: Round-Robin Double

Dans un championnat double, chaque équipe joue deux fois contre chaque autre (une fois a domicile, une fois a l'exterieur).

```python
class DoubleRoundRobinScheduler(SportsScheduler):
    """
    Solveur pour round-robin double.
    Chaque pair d'equipes joue deux fois (aller-retour).
    """

    def _add_constraints(self):
        """Surcharge pour round-robin double."""
        n = self.league.n_teams
        rounds = self.league.rounds

        # C1: Chaque pair d'equipes joue exactement deux fois
        # Une fois i recoit j, une fois j recoit i
        for i in range(n):
            for j in range(i+1, n):
                # i recoit j exactement une fois
                self.model.Add(
                    sum(self.match_vars[(i, j, r)] for r in range(rounds)) == 1
                )
                # j recoit i exactement une fois
                self.model.Add(
                    sum(self.match_vars[(j, i, r)] for r in range(rounds)) == 1
                )

        # C2: Chaque equipe joue exactement une fois par ronde
        for r in range(rounds):
            for i in range(n):
                self.model.Add(
                    sum(self.match_vars[(i, j, r)] + self.match_vars[(j, i, r)]
                        for j in range(n) if j != i) == 1
                )


# Exemple round-robin double (2*(n-1) rondes)
double_league = create_ligue1_sample(n_teams=4)
double_distances = compute_distance_matrix(double_league)

# Pour double round-robin: 2*(n-1) rondes
double_league.rounds = 2 * (double_league.n_teams - 1)

double_scheduler = DoubleRoundRobinScheduler(double_league, double_distances)
double_scheduler.minimize_travel()

print(f"Double Round-Robin: {double_league.n_teams} equipes, {double_league.rounds} rondes")

if double_scheduler.solve(time_limit=10):
    double_schedule = double_scheduler.get_schedule()
    display_schedule(double_league, double_schedule)
```

### Interpretation : Round-Robin Double

**Sortie obtenue** : Calendrier complet avec 6 rondes pour 4 équipes (chaque pair joue deux fois, aller-retour).

| Aspect | Valeur | Signification |
|--------|--------|---------------|
| **Rondes** | 2 × (n-1) | Chaque équipe joue 2 fois contre chaque adversaire |
| **Matchs par équipe** | 2 × (n-1) | Double confrontation : aller domicile, retour exterieur |
| **Equilibre domicile/extérieur** | Exact pour chaque paire | Chaque équipe joue exactement une fois a domicile et une fois a l'exterieur contre chaque adversaire |

**Points cles** :
1. **Standard professionnel** : La plupart des championnats europeens utilisent le round-robin double (Ligue 1, Premier League, Liga...)
2. **Symetrie parfaite** : L'avantage de jouer a domicile est compense par le match retour
3. **Complexite accrue** : Le nombre de rondes double, ce qui augmente la taille du problème CSP
4. **Contraintes spécifiques** : Le modèle garantit que chaque pair d'équipes joue exactement une fois a domicile et une fois a l'exterieur

> **Note technique** : Le round-robin double est le format standard pour les championnats professionnels. Le modèle CP-SAT s'adapte facilement en modifiant les contraintes C1 pour imposer un match domicile et un match exterieur pour chaque pair d'équipes.
Le format aller-retour n’impose pas des distances identiques entre équipes : elles dépendent de leur position géographique.

## 8. Visualisation Graphique du Calendrier

```python
def plot_schedule_matrix(league: SportsLeague, schedule: Dict[int, List[Tuple[int, int]]]):
    """
    Affiche le calendrier comme une matrice equipes x rondes.
    """
    n = league.n_teams
    rounds = league.rounds

    # Matrice: -1 = exterieur, 0 = bye, 1 = domicile
    matrix = np.zeros((n, rounds))

    for r, matches in schedule.items():
        for home, away in matches:
            matrix[home, r] = 1   # Domicile
            matrix[away, r] = -1  # Exterieur

    plt.figure(figsize=(12, 6))

    # Couleurs: rouge = domicile, bleu = exterieur
    cmap = plt.cm.RdBu_r

    plt.imshow(matrix, cmap=cmap, aspect='auto', vmin=-1, vmax=1)

    plt.yticks(range(n), [t.city for t in league.teams])
    plt.xticks(range(rounds), [f'R{r+1}' for r in range(rounds)])

    plt.xlabel('Ronde')
    plt.ylabel('Equipe')
    plt.title(f'Matrice Domicile/Exterieur - {league.name}')

    # Legende
    plt.colorbar(label='Domicile (rouge) / Exterieur (bleu)')

    plt.tight_layout()
    plt.show()


if schedule:
    plot_schedule_matrix(league, schedule)
```

## 9. Benchmark: Taille vs Temps de Resolution

```python
def benchmark_scaling(max_teams: int = 12, time_limit: int = 5):
    """
    Compare le temps de resolution selon le nombre d'equipes.
    """
    import time

    results = []

    for n in range(4, max_teams + 1, 2):
        league = create_ligue1_sample(n)
        distances = compute_distance_matrix(league)

        scheduler = SportsScheduler(league, distances)
        scheduler.minimize_travel()

        start = time.time()
        success = scheduler.solve(time_limit=time_limit)
        elapsed = time.time() - start

        results.append({
            'n_teams': n,
            'n_rounds': league.rounds,
            'n_vars': len(scheduler.match_vars),
            'solved': success,
            'time': elapsed
        })

        print(f"n={n}: {elapsed:.2f}s, {'OK' if success else 'TIMEOUT'}")

    return pd.DataFrame(results)


print("Benchmark de scalabilite...")
df_benchmark = benchmark_scaling(max_teams=10)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(df_benchmark['n_teams'], df_benchmark['time'], 'bo-')
plt.xlabel('Nombre d\'equipes')
plt.ylabel('Temps (s)')
plt.title('Temps de resolution')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(df_benchmark['n_teams'], df_benchmark['n_vars'], 'go-')
plt.xlabel('Nombre d\'equipes')
plt.ylabel('Variables')
plt.title('Complexite (variables)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Interpretation : Scalabilite du Solveur CP-SAT

**Sortie obtenue** : temps de résolution et nombre de variables pour les instances de 4 à 10 équipes.

| Métrique | Lecture |
|----------|---------|
| Temps | Mesuré sur la machine lors de cette exécution |
| Variables | n × (n − 1) × rondes |
| Limite d'interprétation | Ces mesures ne permettent pas de fixer une taille maximale générale |

Pour un championnat simple, rondes = n − 1 : le nombre de variables croît donc
comme n × (n − 1)², soit O(n³). Le benchmark du support n'ajoute ni la limite
des séries D/E ni la nouvelle règle des longs trajets ; ce n'est pas un benchmark
de la solution complète EX-0373.

## Conclusion

### Ce que nous avons appris

1. **Modelisation CSP** : Un calendrier sportif peut etre modelise avec des variables binaires representant chaque match

2. **Contraintes** : Les contraintes hard assurent la validite du round-robin, les contraintes soft optimisent la qualite

3. **Trade-offs** : Minimiser les deplacements vs maximiser l'attractivite TV

4. **Scalabilite** : Le nombre de variables vaut n × (n − 1) × r ; les performances dépendent des instances

### Extensions possibles

- **Contraintes de referees** : Disponibilite des arbitres
- **Contraintes de stade** : Partage de stade (ex: AC Milan / Inter)
- **Contraintes de coupes** : Eviter les conflits avec coupes europeennes
- **Contraintes de repos** : Jours minimums entre matchs
