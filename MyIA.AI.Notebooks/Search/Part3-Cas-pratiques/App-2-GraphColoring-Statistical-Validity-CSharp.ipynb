## 1. Mise en place — trois stratégies et les moteurs de référence

NetworkX tire $G(n,p)$. Le glouton suit l'ordre des sommets ; DSATUR choisit la saturation puis le degré. CP-SAT minimise les couleurs utilisées et distingue optimum prouvé, solution faisable et absence de solution. Son budget vaut **2 unités de temps déterministe**, avec un seul worker et une graine ; ce ne sont pas deux secondes murales.

```
PyModule python;
using(Py.GIL()) {
 python=Py.CreateScope();
 python.Exec(@"import json, networkx as nx, scipy, scipy.stats as stats

def graph_json(n,p,seed):
    g=nx.gnp_random_graph(n,p,seed=seed)
    return json.dumps(list(g.edges()))

def tests_json(a,b):
    w=stats.wilcoxon(a,b,alternative='greater',zero_method='wilcox')
    u=stats.mannwhitneyu(a,b,alternative='greater')
    return json.dumps([float(w.statistic),float(w.pvalue),float(u.statistic),float(u.pvalue)])

def versions():
    return 'networkx '+nx.__version__+' ; scipy '+scipy.__version__");
 Console.WriteLine(python.Get("versions").Invoke().As<string>());
}
class Graphe {public int N; public bool[,] Adj; public Graphe(int n){N=n;Adj=new bool[n,n];} public void Arete(int u,int v){Adj[u,v]=Adj[v,u]=true;} public int Degre(int v){int d=0;for(int w=0;w<N;w++)if(Adj[v,w])d++;return d;}}
Graphe Generer(int n,double p,int seed){string json;using(Py.GIL()){using var f=python.Get("graph_json");using var args=new PyTuple(new[]{n.ToPython(),p.ToPython(),seed.ToPython()});using var r=f.Invoke(args);json=r.As<string>();}var g=new Graphe(n);foreach(var edge in JsonSerializer.Deserialize<int[][]>(json))g.Arete(edge[0],edge[1]);return g;}
(int,int[]) Glouton(Graphe g,int[] ordre=null){ordre ??= Enumerable.Range(0,g.N).ToArray();var colors=Enumerable.Repeat(-1,g.N).ToArray();foreach(int v in ordre){var used=new HashSet<int>();for(int w=0;w<g.N;w++)if(g.Adj[v,w]&&colors[w]>=0)used.Add(colors[w]);int c=0;while(used.Contains(c))c++;colors[v]=c;}return (colors.Length==0?0:colors.Max()+1,colors);}
(int,int[]) Dsatur(Graphe g){var colors=Enumerable.Repeat(-1,g.N).ToArray();for(int i=0;i<g.N;i++){int v=Enumerable.Range(0,g.N).Where(v=>colors[v]<0).OrderByDescending(v=>Enumerable.Range(0,g.N).Where(w=>g.Adj[v,w]&&colors[w]>=0).Select(w=>colors[w]).Distinct().Count()).ThenByDescending(v=>g.Degre(v)).ThenBy(v=>v).First();var used=Enumerable.Range(0,g.N).Where(w=>g.Adj[v,w]&&colors[w]>=0).Select(w=>colors[w]).ToHashSet();int c=0;while(used.Contains(c))c++;colors[v]=c;}return(colors.Length==0?0:colors.Max()+1,colors);}
bool Valide(Graphe g,int[] c){if(c.Length!=g.N||c.Any(x=>x<0))return false;for(int i=0;i<g.N;i++)for(int j=i+1;j<g.N;j++)if(g.Adj[i,j]&&c[i]==c[j])return false;return true;}
(int?,string) CpColor(Graphe g,double limite=2.0){if(g.N==0)return(0,"Optimal");int ub=Glouton(g).Item1;var m=new CpModel();var x=new BoolVar[g.N,ub];var used=new BoolVar[ub];for(int v=0;v<g.N;v++)for(int c=0;c<ub;c++)x[v,c]=m.NewBoolVar($"x_{v}_{c}");for(int v=0;v<g.N;v++)m.AddExactlyOne(Enumerable.Range(0,ub).Select(c=>x[v,c]));for(int u=0;u<g.N;u++)for(int v=u+1;v<g.N;v++)if(g.Adj[u,v])for(int c=0;c<ub;c++)m.Add(x[u,c]+x[v,c]<=1);for(int c=0;c<ub;c++){used[c]=m.NewBoolVar($"used_{c}");m.AddMaxEquality(used[c],Enumerable.Range(0,g.N).Select(v=>x[v,c]));}m.Minimize(LinearExpr.Sum(used));var solver=new CpSolver{StringParameters=$"max_deterministic_time:{limite.ToString(CultureInfo.InvariantCulture)} num_search_workers:1 random_seed:42"};var status=solver.Solve(m);if(status!=CpSolverStatus.Optimal&&status!=CpSolverStatus.Feasible)return(null,status.ToString());var colors=Enumerable.Range(0,g.N).Select(v=>Enumerable.Range(0,ub).Single(c=>solver.Value(x[v,c])==1)).ToArray();if(!Valide(g,colors))throw new Exception("Coloration CP-SAT invalide");return(colors.Distinct().Count(),status.ToString());}
Console.WriteLine("NetworkX, glouton, DSATUR et CP-SAT prêts ; colorations vérifiées sur les arêtes.");
```

## 3. Distribution cross-instance

```
var rng=new Random(42);int N=20;var graphs=Enumerable.Range(0,N).Select(_=>Generer(50,0.4,rng.Next(1,1000000000))).ToArray();var greedy=graphs.Select(g=>Glouton(g).Item1).ToArray();var dsatur=graphs.Select(g=>Dsatur(g).Item1).ToArray();var cp=graphs.Select(g=>CpColor(g)).ToArray();var mask=Enumerable.Range(0,N).Where(i=>cp[i].Item1.HasValue).ToArray();if(mask.Length==0)throw new Exception("Aucune solution CP-SAT pour les statistiques");var cpvals=mask.Select(i=>cp[i].Item1.Value).ToArray();
double Median(int[] a){var b=a.OrderBy(v=>v).ToArray();return (b[(b.Length-1)/2]+b[b.Length/2])/2.0;}
void Resume(string nom,int[] a){double avg=a.Average();Console.WriteLine($"{nom,-9} moyenne={avg:F2} médiane={Median(a):F1} écart-type={Math.Sqrt(a.Select(v=>(v-avg)*(v-avg)).Average()):F2} min/max={a.Min()}/{a.Max()}");}
Resume("Glouton",greedy);Resume("DSATUR",dsatur);Resume("CP-SAT",cpvals);Console.WriteLine($"CP-SAT : {cp.Count(v=>v.Item2=="Optimal")} Optimal, {cp.Count(v=>v.Item2=="Feasible")} Feasible, {N-mask.Length} sans solution");foreach(int i in Enumerable.Range(0,N).Except(mask))Console.WriteLine($"Graphe {i} : {cp[i].Item2}, borne gloutonne séparée {greedy[i]}");
```

## 4. Intervalle de confiance bootstrap à 95 %

Rééchantillonnons $B=2000$ fois avec remise, puis prenons les quantiles 2,5 % et 97,5 % des moyennes. Les intervalles décrivent chaque moyenne ; le test apparié de la section suivante porte directement sur les différences par graphe.

```
(double,double) Bootstrap(int[] a,int B=2000,int seed=42){var r=new Random(seed);var means=new double[B];for(int b=0;b<B;b++){double sum=0;for(int i=0;i<a.Length;i++)sum+=a[r.Next(a.Length)];means[b]=sum/a.Length;}return(Quantile(means,.025),Quantile(means,.975));}
foreach(var (nom,a) in new[]{("Glouton",greedy),("DSATUR",dsatur),("CP-SAT",cpvals)}){var(lo,hi)=Bootstrap(a);Console.WriteLine($"{nom,-9} moyenne={a.Average():F2} IC95%=[{lo:F2}, {hi:F2}] largeur={hi-lo:F2}");}
```

### Lecture du résultat

Chaque intervalle est obtenu par rééchantillonnage des observations disponibles. L'amplitude reflète aussi la taille limitée de l'échantillon.

```
double[] Tests(int[] a,int[] b){using(Py.GIL()){using var f=python.Get("tests_json");using var args=new PyTuple(new[]{a.ToPython(),b.ToPython()});using var r=f.Invoke(args);return JsonSerializer.Deserialize<double[]>(r.As<string>());}}
var pairs=new[]{("Glouton > DSATUR",greedy,dsatur),("DSATUR > CP-SAT",mask.Select(i=>dsatur[i]).ToArray(),cpvals),("Glouton > CP-SAT",mask.Select(i=>greedy[i]).ToArray(),cpvals)};Console.WriteLine($"Seuil Bonferroni : {0.05/3:F4}");foreach(var (nom,a,b) in pairs){var test=Tests(a,b);Console.WriteLine($"{nom,-20} W={test[0]:F1} p_W={test[1]:E3} significatif={test[1]<0.05/3} ; p_MannWhitney={test[3]:E3} ; n={a.Length}");}
```

## 6. Taille d'effet rank-bisériale appariée

Pour les différences non nulles, $r=(R_+-R_-)/(R_++R_-)$ ; les rangs ex aequo sont moyennés. Les égalités sont exclues du calcul de $r$, mais comptées dans le bilan. Les trois p-valeurs appariées sont inférieures à 0,0167, le seuil corrigé.

```
double RangBiserial(int[] a,int[] b){var d=a.Zip(b,(x,y)=>x-y).Where(x=>x!=0).ToArray();if(d.Length==0)return 0;var ordre=Enumerable.Range(0,d.Length).OrderBy(i=>Math.Abs(d[i])).ToArray();double plus=0,moins=0;for(int i=0;i<ordre.Length;){int j=i+1;while(j<ordre.Length&&Math.Abs(d[ordre[j]])==Math.Abs(d[ordre[i]]))j++;double rang=(i+1+j)/2.0;for(int k=i;k<j;k++)if(d[ordre[k]]>0)plus+=rang;else moins+=rang;i=j;}return(plus-moins)/(plus+moins);}
foreach(var(nom,a,b) in pairs){var diff=a.Zip(b,(x,y)=>x-y).ToArray();Console.WriteLine($"{nom,-20} r={RangBiserial(a,b):F3}, victoires={diff.Count(v=>v>0)}, égalités={diff.Count(v=>v==0)}, défaites={diff.Count(v=>v<0)}");}
```

## 8. Synthèse et exercices

Les trois exercices sont répartis après les notions : Welsh–Powell, bootstrap de la médiane et balayage avec CP-SAT. Comparez moyenne et médiane sans déduire à elles seules la forme d'une distribution.

## À retenir

- Une comparaison sur un seul graphe peut être atypique ; apparier les stratégies par graphe conserve l'information commune.
- Le bootstrap encadre une moyenne, Wilcoxon teste les différences appariées et Bonferroni ajuste les comparaisons multiples.
- Une solution CP-SAT faisable donne une borne supérieure ; seul `Optimal` prouve l'optimum du modèle. Les échecs ne sont pas des mesures du solveur.
