```
#nullable enable
using System;
using System.Globalization;
using System.Numerics;

// Tout le carnet tourne en arithmetique exacte : System.Numerics.BigInteger pour les
// entiers, un rationnel exact reduit (numerateur / denominateur BigInteger) la ou la
// source emploie fractions.Fraction, et des entiers gaussiens pour les complexes. Aucune
// egalite flottante n'est testee : aucun double n'apparait dans les controles.

// Rationnel exact reduit : comme fractions.Fraction, jamais comme un double.
public readonly struct RationnelC03 : IEquatable<RationnelC03>
{
    public BigInteger Numerateur { get; }
    public BigInteger Denominateur { get; }

    public RationnelC03(BigInteger numerateur) : this(numerateur, BigInteger.One) { }

    public RationnelC03(BigInteger numerateur, BigInteger denominateur)
    {
        if (denominateur == BigInteger.Zero)
            throw new DivideByZeroException("Denominateur nul : rationnel indefini.");
        if (denominateur < BigInteger.Zero)
        {
            numerateur = -numerateur;
            denominateur = -denominateur;
        }
        BigInteger g = BigInteger.GreatestCommonDivisor(BigInteger.Abs(numerateur), denominateur);
        if (g == BigInteger.Zero) g = BigInteger.One;
        Numerateur = numerateur / g;
        Denominateur = denominateur / g;
    }

    public static RationnelC03 operator +(RationnelC03 a, RationnelC03 b) =>
        new RationnelC03(a.Numerateur * b.Denominateur + b.Numerateur * a.Denominateur,
                        a.Denominateur * b.Denominateur);

    public static RationnelC03 operator -(RationnelC03 a, RationnelC03 b) =>
        new RationnelC03(a.Numerateur * b.Denominateur - b.Numerateur * a.Denominateur,
                        a.Denominateur * b.Denominateur);

    public static RationnelC03 operator *(RationnelC03 a, RationnelC03 b) =>
        new RationnelC03(a.Numerateur * b.Numerateur, a.Denominateur * b.Denominateur);

    public static RationnelC03 operator *(RationnelC03 a, BigInteger k) =>
        new RationnelC03(k * a.Numerateur, a.Denominateur);

    public static RationnelC03 operator *(BigInteger k, RationnelC03 a) => a * k;

    public static RationnelC03 operator /(RationnelC03 a, RationnelC03 b) =>
        new RationnelC03(a.Numerateur * b.Denominateur, a.Denominateur * b.Numerateur);

    public static RationnelC03 operator -(RationnelC03 a) =>
        new RationnelC03(-a.Numerateur, a.Denominateur);

    // Puissance entiere exacte (C# n'a pas d'operateur **).
    public RationnelC03 Puissance(int exposant)
    {
        if (exposant == 0) return new RationnelC03(BigInteger.One);
        if (exposant < 0) return new RationnelC03(BigInteger.One) / Puissance(-exposant);
        RationnelC03 resultat = new RationnelC03(BigInteger.One);
        for (int i = 0; i < exposant; i++) resultat = resultat * this;
        return resultat;
    }

    public static bool operator ==(RationnelC03 a, RationnelC03 b) =>
        a.Numerateur == b.Numerateur && a.Denominateur == b.Denominateur;

    public static bool operator !=(RationnelC03 a, RationnelC03 b) => !(a == b);

    public bool Equals(RationnelC03 autre) => this == autre;

    public override bool Equals(object? obj) => obj is RationnelC03 r && Equals(r);

    public override int GetHashCode() => HashCode.Combine(Numerateur, Denominateur);

    public override string ToString() => Denominateur == BigInteger.One
        ? Numerateur.ToString(CultureInfo.InvariantCulture)
        : Numerateur.ToString(CultureInfo.InvariantCulture) + "/" + Denominateur.ToString(CultureInfo.InvariantCulture);
}

// Verdict executable : leve si la valeur calculee differe de l'attendu.
public static class VerifC03
{
    public static void Egale(bool condition, string message)
    {
        if (!condition) throw new InvalidOperationException("ECART : " + message);
    }
}

var unSurDeux = new RationnelC03(BigInteger.One, new BigInteger(2));
var moitie = unSurDeux + unSurDeux;
Console.WriteLine($"Rationnel exact : 1/2 + 1/2 = {moitie} ; 2 * 1/2 = {new RationnelC03(BigInteger.One, new BigInteger(2)) * new RationnelC03(new BigInteger(2))} ; 1/3 reduit = {new RationnelC03(BigInteger.One, new BigInteger(3))}");
VerifC03.Egale(moitie == new RationnelC03(BigInteger.One), "1/2 + 1/2 doit valoir 1 exactement");
VerifC03.Egale(new RationnelC03(BigInteger.One, new BigInteger(2)).Denominateur == new BigInteger(2), "le denominateur reste 2");
Console.WriteLine("Arithmetique exacte en place : BigInteger + rationnel reduit, aucune egalite flottante.");
```

## 1. Les quarante-huit cartes et l'appariement de Born (équation 73)

La source : « From the three derived directions at the qubit end of the read link, by the signed permutation maps of
the three axes, we obtain forty-eight maps falling in exactly two kinds by determinant, twenty-four proper and
twenty-four improper, every one preserving the Born pairing of the six axis states at all one thousand seven hundred
and twenty-eight pairs » — c'est le **check exact du postulat de projection**, le plus petit cas travaillé par le
chapitre (« the smallest case worked here is the first corollary, the projection postulate, it is the wash at one
act »).

Le rejeu : les trois axes donnent six états d'axe (deux par axe). L'appariement de Born entre deux états d'axe se lit
$|\langle i|j\rangle|^2$ — la lecture standard : $1$ pour un état avec lui-même, $0$ pour son opposé sur le même axe,
$1/2$ entre axes distincts (constaté : la source ne réimprime pas sa définition de l'appariement ; ce rejeu utilise
la lecture standard et retrouve les comptes imprimés). Une carte signée envoie l'axe $a$ sur l'axe $\pi(a)$ avec un
flip $\varepsilon_a$ ; il y a $3!\cdot 2^3 = 48$ cartes, et chacune agit sur les six états. Le comptage des paires
ordonnées préservées : $48 \times 36 = 1728$.

En C#, l'appariement est porté par un **rationnel exact** (`Rationnel`, numérateur et dénominateur `BigInteger`,
réduit comme `fractions.Fraction`) : les valeurs `1`, `0` et `1/2` sont des objets exacts, jamais des flottants.


```
#nullable enable
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Numerics;

public readonly struct RationnelC05 : IEquatable<RationnelC05>
{
    public BigInteger Numerateur { get; }
    public BigInteger Denominateur { get; }

    public RationnelC05(BigInteger numerateur) : this(numerateur, BigInteger.One) { }

    public RationnelC05(BigInteger numerateur, BigInteger denominateur)
    {
        if (denominateur == BigInteger.Zero) throw new DivideByZeroException();
        if (denominateur < BigInteger.Zero)
        {
            numerateur = -numerateur;
            denominateur = -denominateur;
        }
        BigInteger g = BigInteger.GreatestCommonDivisor(BigInteger.Abs(numerateur), denominateur);
        if (g == BigInteger.Zero) g = BigInteger.One;
        Numerateur = numerateur / g;
        Denominateur = denominateur / g;
    }

    public static bool operator ==(RationnelC05 a, RationnelC05 b) =>
        a.Numerateur == b.Numerateur && a.Denominateur == b.Denominateur;

    public static bool operator !=(RationnelC05 a, RationnelC05 b) => !(a == b);

    public bool Equals(RationnelC05 autre) => this == autre;

    public override bool Equals(object? obj) => obj is RationnelC05 r && Equals(r);

    public override int GetHashCode() => HashCode.Combine(Numerateur, Denominateur);

    public override string ToString() => Denominateur == BigInteger.One
        ? Numerateur.ToString(CultureInfo.InvariantCulture)
        : Numerateur.ToString(CultureInfo.InvariantCulture) + "/" + Denominateur.ToString(CultureInfo.InvariantCulture);
}

public static class VerifC05
{
    public static void Egale(bool condition, string message)
    {
        if (!condition) throw new InvalidOperationException("ECART : " + message);
    }
}

// Six etats d'axe (deux par axe) et appariement de Born, en arithmetique exacte.
public static class BornC05
{
    public static List<(int Axe, int Direction)> Etats(int[] axes)
    {
        var etats = new List<(int, int)>();
        foreach (int a in axes)
            foreach (int s in new[] { +1, -1 })
                etats.Add((a, s));
        return etats;
    }

    public static List<((int Axe, int Direction) E1, (int Axe, int Direction) E2)> Paires(List<(int Axe, int Direction)> etats)
    {
        var paires = new List<((int, int), (int, int))>();
        foreach (var e1 in etats)
            foreach (var e2 in etats)
                paires.Add((e1, e2));
        return paires;
    }

    public static RationnelC05 AppariementBorn((int Axe, int Direction) e1, (int Axe, int Direction) e2)
    {
        if (e1.Axe == e2.Axe)
            return e1.Direction == e2.Direction
                ? new RationnelC05(BigInteger.One)
                : new RationnelC05(BigInteger.Zero);
        return new RationnelC05(BigInteger.One, new BigInteger(2));
    }
}

var axes = new[] { 0, 1, 2 };
var etatsAxe = BornC05.Etats(axes);
var pairesOrdonnees = BornC05.Paires(etatsAxe);

Console.WriteLine($"Etats d'axe : {etatsAxe.Count} | paires ordonnees par etat du couple : {pairesOrdonnees.Count}");
Console.WriteLine($"Appariement d'un etat avec lui-meme      : {BornC05.AppariementBorn((0, +1), (0, +1))}");
Console.WriteLine($"Appariement avec l'oppose du meme axe    : {BornC05.AppariementBorn((0, +1), (0, -1))}");
Console.WriteLine($"Appariement entre axes distincts         : {BornC05.AppariementBorn((0, +1), (1, +1))}");

VerifC05.Egale(etatsAxe.Count == 6, $"Etats d'axe : 6 attendus, obtenu {etatsAxe.Count}");
VerifC05.Egale(pairesOrdonnees.Count == 36, $"Paires ordonnees : 36 attendues, obtenu {pairesOrdonnees.Count}");
VerifC05.Egale(BornC05.AppariementBorn((0, +1), (0, +1)) == new RationnelC05(BigInteger.One),
 "Appariement d'un etat avec lui-meme : 1 attendu");
VerifC05.Egale(BornC05.AppariementBorn((0, +1), (0, -1)) == new RationnelC05(BigInteger.Zero),
               "Appariement avec l'oppose du meme axe : 0 attendu");
VerifC05.Egale(BornC05.AppariementBorn((0, +1), (1, +1)) == new RationnelC05(BigInteger.One, new BigInteger(2)),
               "Appariement entre axes distincts : 1/2 attendu");
Console.WriteLine("Appariements de Born : CONFORME aux trois lectures (1, 0, 1/2), en rationnel exact.");
```

```
#nullable enable
using System;
using System.Collections.Generic;
using System.Globalization;

public static class VerifC06
{
    public static void Egale(bool condition, string message)
    {
        if (!condition) throw new InvalidOperationException("ECART : " + message);
    }
}

// Une carte signee : une permutation des axes et un flip par axe.
public readonly struct CarteC06
{
    public int[] Perm { get; }
    public int[] Flips { get; }

    public CarteC06(int[] perm, int[] flips)
    {
        Perm = perm;
        Flips = flips;
    }
}

public static class CartesC06
{
    // Signe d'une permutation par comptage d'inversions.
    public static int Parite(int[] perm)
    {
        int inversions = 0;
        for (int i = 0; i < perm.Length; i++)
            for (int j = i + 1; j < perm.Length; j++)
                if (perm[i] > perm[j]) inversions++;
        return inversions % 2 == 0 ? +1 : -1;
    }

    // Determinant de la carte = signe de la permutation x produit des flips.
    public static int Determinant(CarteC06 carte) =>
        Parite(carte.Perm) * carte.Flips[0] * carte.Flips[1] * carte.Flips[2];

    public static List<int[]> Permutations(int[] items)
    {
        var resultat = new List<int[]>();
        void Explorer(List<int> restants, List<int> courant)
        {
            if (restants.Count == 0)
            {
                resultat.Add(courant.ToArray());
                return;
            }
            for (int i = 0; i < restants.Count; i++)
            {
                var reste = new List<int>(restants);
                int choisi = reste[i];
                reste.RemoveAt(i);
                var nouveau = new List<int>(courant) { choisi };
                Explorer(reste, nouveau);
            }
        }
        Explorer(new List<int>(items), new List<int>());
        return resultat;
    }

    public static List<CarteC06> Toutes(int[] axes)
    {
        var cartes = new List<CarteC06>();
        foreach (var perm in Permutations(axes))
            for (int f0 = 0; f0 < 2; f0++)
                for (int f1 = 0; f1 < 2; f1++)
                    for (int f2 = 0; f2 < 2; f2++)
                        cartes.Add(new CarteC06(perm, new[] { f0 == 0 ? +1 : -1, f1 == 0 ? +1 : -1, f2 == 0 ? +1 : -1 }));
        return cartes;
    }
}

var axesCartes = new[] { 0, 1, 2 };
var cartes = CartesC06.Toutes(axesCartes);
var propres = new List<CarteC06>();
var impropres = new List<CarteC06>();
foreach (var carte in cartes)
{
    if (CartesC06.Determinant(carte) == +1) propres.Add(carte);
    else impropres.Add(carte);
}

Console.WriteLine($"3! x 2^3 = {6 * 8} cartes ; propres : {propres.Count} , impropres : {impropres.Count}");
Console.WriteLine($"48 = 24 + 24 -> {cartes.Count == 48 && propres.Count == 24 && impropres.Count == 24}");

VerifC06.Egale(cartes.Count == 48, $"Cartes signees : 48 attendues, obtenu {cartes.Count}");
VerifC06.Egale(propres.Count == 24, $"Cartes propres : 24 attendues, obtenu {propres.Count}");
VerifC06.Egale(impropres.Count == 24, $"Cartes impropres : 24 attendues, obtenu {impropres.Count}");
VerifC06.Egale(propres.Count + impropres.Count == cartes.Count, "les deux especes doivent couvrir les 48 cartes");
Console.WriteLine("Classement par determinant : CONFORME (24 propres, 24 impropres). determinant = signe de la permutation x produit des flips.");
```

```
#nullable enable
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Numerics;

public readonly struct RationnelC07 : IEquatable<RationnelC07>
{
    public BigInteger Numerateur { get; }
    public BigInteger Denominateur { get; }

    public RationnelC07(BigInteger numerateur) : this(numerateur, BigInteger.One) { }

    public RationnelC07(BigInteger numerateur, BigInteger denominateur)
    {
        if (denominateur < BigInteger.Zero)
        {
            numerateur = -numerateur;
            denominateur = -denominateur;
        }
        BigInteger g = BigInteger.GreatestCommonDivisor(BigInteger.Abs(numerateur), denominateur);
        if (g == BigInteger.Zero) g = BigInteger.One;
        Numerateur = numerateur / g;
        Denominateur = denominateur / g;
    }

    public static bool operator ==(RationnelC07 a, RationnelC07 b) =>
        a.Numerateur == b.Numerateur && a.Denominateur == b.Denominateur;

    public static bool operator !=(RationnelC07 a, RationnelC07 b) => !(a == b);

    public bool Equals(RationnelC07 autre) => this == autre;

    public override bool Equals(object? obj) => obj is RationnelC07 r && Equals(r);

    public override int GetHashCode() => HashCode.Combine(Numerateur, Denominateur);
}

public static class VerifC07
{
    public static void Egale(bool condition, string message)
    {
        if (!condition) throw new InvalidOperationException("ECART : " + message);
    }
}

public readonly struct CarteC07
{
    public int[] Perm { get; }
    public int[] Flips { get; }

    public CarteC07(int[] perm, int[] flips)
    {
        Perm = perm;
        Flips = flips;
    }
}

public static class CartesC07
{
    public static List<CarteC07> Toutes(int[] axes)
    {
        var cartes = new List<CarteC07>();
        var permutations = new List<int[]>();

        void Explorer(List<int> restants, List<int> courant)
        {
            if (restants.Count == 0)
            {
                permutations.Add(courant.ToArray());
                return;
            }
            for (int i = 0; i < restants.Count; i++)
            {
                var reste = new List<int>(restants);
                int choisi = reste[i];
                reste.RemoveAt(i);
                Explorer(reste, new List<int>(courant) { choisi });
            }
        }

        Explorer(new List<int>(axes), new List<int>());
        foreach (var perm in permutations)
            for (int f0 = 0; f0 < 2; f0++)
                for (int f1 = 0; f1 < 2; f1++)
                    for (int f2 = 0; f2 < 2; f2++)
                        cartes.Add(new CarteC07(perm, new[] { f0 == 0 ? +1 : -1, f1 == 0 ? +1 : -1, f2 == 0 ? +1 : -1 }));
        return cartes;
    }
}

public static class BornC07
{
    public static List<(int Axe, int Direction)> Etats(int[] axes)
    {
        var etats = new List<(int, int)>();
        foreach (int a in axes)
            foreach (int s in new[] { +1, -1 })
                etats.Add((a, s));
        return etats;
    }

    public static List<((int Axe, int Direction) E1, (int Axe, int Direction) E2)> Paires(List<(int Axe, int Direction)> etats)
    {
        var paires = new List<((int, int), (int, int))>();
        foreach (var e1 in etats)
            foreach (var e2 in etats)
                paires.Add((e1, e2));
        return paires;
    }

    public static RationnelC07 AppariementBorn((int Axe, int Direction) e1, (int Axe, int Direction) e2)
    {
        if (e1.Axe == e2.Axe)
            return e1.Direction == e2.Direction
                ? new RationnelC07(BigInteger.One)
                : new RationnelC07(BigInteger.Zero);
        return new RationnelC07(BigInteger.One, new BigInteger(2));
    }
}

// Action d'une carte signee sur un etat d'axe : (axe, direction) -> (pi(axe), direction x flip).
public static class ActionsC07
{
    public static (int Axe, int Direction) ImageEtat(CarteC07 carte, (int Axe, int Direction) etat)
    {
        int a = etat.Axe;
        int s = etat.Direction;
        return (carte.Perm[a], s * carte.Flips[a]);
    }
}

var axesPreservation = new[] { 0, 1, 2 };
var cartesPreservation = CartesC07.Toutes(axesPreservation);
var etatsPreservation = BornC07.Etats(axesPreservation);
var pairesPreservation = BornC07.Paires(etatsPreservation);

int pairesPreservees = 0;
int totalPaires = cartesPreservation.Count * pairesPreservation.Count;
foreach (var carte in cartesPreservation)
    foreach (var paire in pairesPreservation)
    {
        var e1 = paire.E1;
        var e2 = paire.E2;
        if (BornC07.AppariementBorn(ActionsC07.ImageEtat(carte, e1), ActionsC07.ImageEtat(carte, e2)) == BornC07.AppariementBorn(e1, e2))
            pairesPreservees++;
    }

Console.WriteLine($"#maps = 48 = 24 + 24, det = +/-1, #Born pairs preserved = {pairesPreservees} among {totalPaires}");
Console.WriteLine($"Eq. 73 (eq. 73 de la source) -> {pairesPreservees == 1728 && totalPaires == 1728}");

VerifC07.Egale(totalPaires == 1728, $"Total des paires : 1728 attendu (48 x 36), obtenu {totalPaires}");
VerifC07.Egale(pairesPreservees == 1728, $"Paires de Born preservees : 1728 attendues, obtenu {pairesPreservees}");
VerifC07.Egale(pairesPreservees == totalPaires, "toutes les paires doivent etre preservees");
Console.WriteLine("Eq. 73 : CONFORME — les 48 cartes preservees les 1728 paires de Born (48 x 36). Comparaison en rationnel exact.");
```

## 2. Le check CHSH « at twice the classical square »

Le théorème 47.1 compte, parmi ses neuf checks exacts, « the Clauser-Horne-Shimony-Holt expectation at twice the
classical square » ; le résultat ouvert du chapitre précise : « the Bell inequality at the rotated settings of the
derived rotation group with the Clauser-Horne-Shimony-Holt maximum twice the square root of two ». Autrement dit :
le maximum quantique $2\sqrt{2}$ contre la borne classique $2$, et le carré de l'espérance
$8 = 2 \cdot 2^2$ — **deux fois le carré classique**.

Le rejeu en entiers exacts : aux réglages optimaux $A_0 = \sigma_z$, $A_1 = \sigma_x$, $B_0 = (\sigma_z +
\sigma_x)/\sqrt{2}$, $B_1 = (\sigma_z - \sigma_x)/\sqrt{2}$, l'opérateur CHSH se factorise en $S = \sqrt{2}\,M$ où
$M = \sigma_z \otimes \sigma_z + \sigma_x \otimes \sigma_x$ est une **matrice à coefficients entiers**. Tout se
vérifie sans irrationnel : $M^2 = 2I - 2\,\sigma_y \otimes \sigma_y$ (entiers gaussiens), puis $M^4 = 4M^2$, ce qui
force le spectre de $M^2$ dans $\{0, 4\}$ — et $M^2 \neq 0$, $M^2 \neq 4I$ montre que les deux classes sont
atteintes : le spectre de $M$ est $\{0, 0, +2, -2\}$. Le maximum $2\sqrt{2}$ de $S$ suit, et son carré vaut
exactement $8$.

En C#, les matrices portent des **entiers gaussiens** (partie réelle et partie imaginaire en `BigInteger`) : la
comparaison des matrices se fait coefficient par coefficient, jamais par une égalité flottante.

```
#nullable enable
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Numerics;

// Tout le calcul tourne en arithmetique EXACTE : entiers BigInteger et entiers
// gaussiens (complexes a parties entieres). Aucune egalite flottante n'est
// testee — c'est le pendant C# des complexes entiers natifs de Python.

// Controle executable : leve sur ecart (le carnet ne finit pas vert sur une
// identite fausse).
public static class VerifC10
{
    public static void Egale(bool condition, string message)
    {
        if (!condition) throw new InvalidOperationException("ECART : " + message);
    }
}

// Entier gaussien a + bi avec a, b entiers exacts (BigInteger) : addition et
// multiplication closes sur l'anneau Z[i], i^2 = -1.
public readonly struct GaussienC10 : IEquatable<GaussienC10>
{
    public BigInteger Re { get; }
    public BigInteger Im { get; }

    public GaussienC10(BigInteger re) : this(re, BigInteger.Zero) { }

    public GaussienC10(BigInteger re, BigInteger im)
    {
        Re = re;
        Im = im;
    }

    public static GaussienC10 Zero => new GaussienC10(BigInteger.Zero, BigInteger.Zero);

    public static GaussienC10 operator +(GaussienC10 a, GaussienC10 b) => new GaussienC10(a.Re + b.Re, a.Im + b.Im);

    public static GaussienC10 operator -(GaussienC10 a, GaussienC10 b) => new GaussienC10(a.Re - b.Re, a.Im - b.Im);

    // (a+bi)(c+di) = (ac-bd) + (ad+bc)i — multiplication dans l'anneau.
    public static GaussienC10 operator *(GaussienC10 a, GaussienC10 b) =>
        new GaussienC10(a.Re * b.Re - a.Im * b.Im, a.Re * b.Im + a.Im * b.Re);

    // Entier pur : partie imaginaire nulle (teste plus bas pour M).
    public bool EstEntier => Im == BigInteger.Zero;

    public bool Equals(GaussienC10 autre) => Re == autre.Re && Im == autre.Im;

    public override bool Equals(object? obj) => obj is GaussienC10 g && Equals(g);

    public override int GetHashCode() => HashCode.Combine(Re, Im);

    public override string ToString() => Im == BigInteger.Zero
        ? Re.ToString(CultureInfo.InvariantCulture)
        : Re.ToString(CultureInfo.InvariantCulture) + (Im < BigInteger.Zero ? "-" : "+")
          + BigInteger.Abs(Im).ToString(CultureInfo.InvariantCulture) + "i";
}

// Matrice dense a coefficients gaussiens, avec l'egalite exacte case par case.
public sealed class MatriceC10
{
    public GaussienC10[][] Cases { get; }
    public int Lignes => Cases.Length;
    public int Colonnes => Cases[0].Length;

    public MatriceC10(GaussienC10[][] cases) { Cases = cases; }

    public GaussienC10 this[int i, int j] => Cases[i][j];

    public bool Egale(MatriceC10 autre)
    {
        if (Lignes != autre.Lignes || Colonnes != autre.Colonnes) return false;
        for (int i = 0; i < Lignes; i++)
            for (int j = 0; j < Colonnes; j++)
                if (!Cases[i][j].Equals(autre.Cases[i][j])) return false;
        return true;
    }
}

public static class OpsC10
{
    // Produit tensoriel de deux matrices, en arithmetique exacte — le
    // remplacement C# du kron() du jumeau Python.
    public static MatriceC10 Kron(MatriceC10 a, MatriceC10 b)
    {
        int n = a.Lignes, m = a.Colonnes, p = b.Lignes, q = b.Colonnes;
        var cases = new GaussienC10[n * p][];
        for (int i = 0; i < n * p; i++) cases[i] = new GaussienC10[m * q];
        for (int i = 0; i < n; i++)
            for (int j = 0; j < m; j++)
                for (int k = 0; k < p; k++)
                    for (int l = 0; l < q; l++)
                        cases[i * p + k][j * q + l] = a[i, j] * b[k, l];
        return new MatriceC10(cases);
    }

    // Produit matriciel exact (aucun arrondi : sommes d'entiers gaussiens).
    public static MatriceC10 Matmul(MatriceC10 a, MatriceC10 b)
    {
        var cases = new GaussienC10[a.Lignes][];
        for (int i = 0; i < a.Lignes; i++) cases[i] = new GaussienC10[b.Colonnes];
        for (int i = 0; i < a.Lignes; i++)
            for (int j = 0; j < b.Colonnes; j++)
            {
                GaussienC10 somme = GaussienC10.Zero;
                for (int k = 0; k < a.Colonnes; k++) somme = somme + a[i, k] * b[k, j];
                cases[i][j] = somme;
            }
        return new MatriceC10(cases);
    }

    public static MatriceC10 Scalaire(int c, MatriceC10 a)
    {
        var cases = new GaussienC10[a.Lignes][];
        for (int i = 0; i < a.Lignes; i++)
        {
            cases[i] = new GaussienC10[a.Colonnes];
            for (int j = 0; j < a.Colonnes; j++) cases[i][j] = a[i, j] * new GaussienC10(new BigInteger(c));
        }
        return new MatriceC10(cases);
    }

    public static MatriceC10 Somme(MatriceC10 a, MatriceC10 b)
    {
        var cases = new GaussienC10[a.Lignes][];
        for (int i = 0; i < a.Lignes; i++)
        {
            cases[i] = new GaussienC10[a.Colonnes];
            for (int j = 0; j < a.Colonnes; j++) cases[i][j] = a[i, j] + b[i, j];
        }
        return new MatriceC10(cases);
    }
}

// Pauli 2x2 : identite, sz, sx entiers purs ; sy porte +/-i — entiers
// gaussiens exacts (le pendant C# du commentaire Python « exacts en Python »).
var I2 = new MatriceC10(new[] { new[] { new GaussienC10(1), new GaussienC10(0) }, new[] { new GaussienC10(0), new GaussienC10(1) } });
var I4 = OpsC10.Kron(I2, I2);
var sz = new MatriceC10(new[] { new[] { new GaussienC10(1), new GaussienC10(0) }, new[] { new GaussienC10(0), new GaussienC10(-1) } });
var sx = new MatriceC10(new[] { new[] { new GaussienC10(0), new GaussienC10(1) }, new[] { new GaussienC10(1), new GaussienC10(0) } });
var sy = new MatriceC10(new[] { new[] { new GaussienC10(0), new GaussienC10(0, -1) }, new[] { new GaussienC10(0, 1), new GaussienC10(0) } });

// Observable de chaine M = sz (x) sz + sx (x) sx : somme exacte de produits
// tensoriels ; M^2 doit verifier l'identite 2I - 2 sy(x)sy.
var M = OpsC10.Somme(OpsC10.Kron(sz, sz), OpsC10.Kron(sx, sx));
var M2 = OpsC10.Matmul(M, M);
var cibleM2 = OpsC10.Somme(OpsC10.Scalaire(2, I4), OpsC10.Scalaire(-2, OpsC10.Kron(sy, sy)));

bool toutEntier = true;
for (int i = 0; i < M.Lignes; i++)
    for (int j = 0; j < M.Colonnes; j++)
        if (!M[i, j].EstEntier) toutEntier = false;
Console.WriteLine($"M = sz (x) sz + sx (x) sx, entiers purs : {toutEntier}");
Console.WriteLine($"M^2 == 2I - 2 sy(x)sy   -> {M2.Egale(cibleM2)}");

VerifC10.Egale(toutEntier, "M doit etre a coefficients entiers");
VerifC10.Egale(M2.Egale(cibleM2), "M^2 doit valoir 2I - 2 sy(x)sy");
Console.WriteLine("M : 4x4 entiers purs ; M^2 = 2I - 2 sy(x)sy : CONFORME (entiers gaussiens, exacts).");
```

```
#nullable enable
using System;
using System.Globalization;
using System.Numerics;

// Check CHSH en arithmetique exacte. La classe spectrale se lit sur le
// polynome minimal de M : M^4 = 4 M^2 force le spectre de M^2 dans {0, 4},
// et les exclusions (M^2 != 0, M^2 != 4I) prouvent que la classe 4 est
// ATTEINTE sans etre la seule. La comparaison de bornes se fait en
// rationnels exacts — le pendant C# de Fraction.

public static class VerifC11
{
    public static void Egale(bool condition, string message)
    {
        if (!condition) throw new InvalidOperationException("ECART : " + message);
    }
}

// Rationnel exact : denominateur normalise positif, fraction reduite par le
// PGCD — l'equivalent C# de fractions.Fraction cote Python.
public readonly struct RationnelC11 : IEquatable<RationnelC11>
{
    public BigInteger Numerateur { get; }
    public BigInteger Denominateur { get; }

    public RationnelC11(BigInteger numerateur) : this(numerateur, BigInteger.One) { }

    public RationnelC11(BigInteger numerateur, BigInteger denominateur)
    {
        if (denominateur < BigInteger.Zero)
        {
            numerateur = -numerateur;
            denominateur = -denominateur;
        }
        BigInteger g = BigInteger.GreatestCommonDivisor(BigInteger.Abs(numerateur), denominateur);
        if (g == BigInteger.Zero) g = BigInteger.One;
        Numerateur = numerateur / g;
        Denominateur = denominateur / g;
    }

    // Puissance par multiplication repetee (exposants petits : le carnet
    // n'a besoin que de carres).
    public RationnelC11 Puissance(int exposant)
    {
        RationnelC11 resultat = new RationnelC11(BigInteger.One);
        for (int i = 0; i < exposant; i++) resultat = resultat * this;
        return resultat;
    }

    public static RationnelC11 operator *(RationnelC11 a, RationnelC11 b) =>
        new RationnelC11(a.Numerateur * b.Numerateur, a.Denominateur * b.Denominateur);

    public static bool operator ==(RationnelC11 a, RationnelC11 b) =>
        a.Numerateur == b.Numerateur && a.Denominateur == b.Denominateur;

    public static bool operator !=(RationnelC11 a, RationnelC11 b) => !(a == b);

    public bool Equals(RationnelC11 autre) => this == autre;

    public override bool Equals(object? obj) => obj is RationnelC11 r && Equals(r);

    public override int GetHashCode() => HashCode.Combine(Numerateur, Denominateur);
}

// Entier gaussien exact (voir cellule 10) — recopie local pour garder la
// cellule autonome, meme moteur que GaussienC10.
public readonly struct GaussienC11 : IEquatable<GaussienC11>
{
    public BigInteger Re { get; }
    public BigInteger Im { get; }

    public GaussienC11(BigInteger re) : this(re, BigInteger.Zero) { }

    public GaussienC11(BigInteger re, BigInteger im)
    {
        Re = re;
        Im = im;
    }

    public static GaussienC11 Zero => new GaussienC11(BigInteger.Zero, BigInteger.Zero);

    public static GaussienC11 operator +(GaussienC11 a, GaussienC11 b) => new GaussienC11(a.Re + b.Re, a.Im + b.Im);

    public static GaussienC11 operator -(GaussienC11 a, GaussienC11 b) => new GaussienC11(a.Re - b.Re, a.Im - b.Im);

    public static GaussienC11 operator *(GaussienC11 a, GaussienC11 b) =>
        new GaussienC11(a.Re * b.Re - a.Im * b.Im, a.Re * b.Im + a.Im * b.Re);

    public bool Equals(GaussienC11 autre) => Re == autre.Re && Im == autre.Im;

    public override bool Equals(object? obj) => obj is GaussienC11 g && Equals(g);

    public override int GetHashCode() => HashCode.Combine(Re, Im);
}

public sealed class MatriceC11
{
    public GaussienC11[][] Cases { get; }
    public int Lignes => Cases.Length;
    public int Colonnes => Cases[0].Length;

    public MatriceC11(GaussienC11[][] cases) { Cases = cases; }

    public GaussienC11 this[int i, int j] => Cases[i][j];

    public bool Egale(MatriceC11 autre)
    {
        if (Lignes != autre.Lignes || Colonnes != autre.Colonnes) return false;
        for (int i = 0; i < Lignes; i++)
            for (int j = 0; j < Colonnes; j++)
                if (!Cases[i][j].Equals(autre.Cases[i][j])) return false;
        return true;
    }
}

public static class OpsC11
{
    // Produit tensoriel exact (voir cellule 10).
    public static MatriceC11 Kron(MatriceC11 a, MatriceC11 b)
    {
        var cases = new GaussienC11[a.Lignes * b.Lignes][];
        for (int i = 0; i < a.Lignes * b.Lignes; i++) cases[i] = new GaussienC11[a.Colonnes * b.Colonnes];
        for (int i = 0; i < a.Lignes; i++)
            for (int j = 0; j < a.Colonnes; j++)
                for (int k = 0; k < b.Lignes; k++)
                    for (int l = 0; l < b.Colonnes; l++)
                        cases[i * b.Lignes + k][j * b.Colonnes + l] = a[i, j] * b[k, l];
        return new MatriceC11(cases);
    }

    // Produit matriciel exact.
    public static MatriceC11 Matmul(MatriceC11 a, MatriceC11 b)
    {
        var cases = new GaussienC11[a.Lignes][];
        for (int i = 0; i < a.Lignes; i++) cases[i] = new GaussienC11[b.Colonnes];
        for (int i = 0; i < a.Lignes; i++)
            for (int j = 0; j < b.Colonnes; j++)
            {
                GaussienC11 somme = GaussienC11.Zero;
                for (int k = 0; k < a.Colonnes; k++) somme = somme + a[i, k] * b[k, j];
                cases[i][j] = somme;
            }
        return new MatriceC11(cases);
    }

    public static MatriceC11 Scalaire(int c, MatriceC11 a)
    {
        var cases = new GaussienC11[a.Lignes][];
        for (int i = 0; i < a.Lignes; i++)
        {
            cases[i] = new GaussienC11[a.Colonnes];
            for (int j = 0; j < a.Colonnes; j++) cases[i][j] = a[i, j] * new GaussienC11(new BigInteger(c));
        }
        return new MatriceC11(cases);
    }

    public static MatriceC11 Somme(MatriceC11 a, MatriceC11 b)
    {
        var cases = new GaussienC11[a.Lignes][];
        for (int i = 0; i < a.Lignes; i++)
        {
            cases[i] = new GaussienC11[a.Colonnes];
            for (int j = 0; j < a.Colonnes; j++) cases[i][j] = a[i, j] + b[i, j];
        }
        return new MatriceC11(cases);
    }
}

// Reconstruction des Pauli (autonomie de cellule) puis M, M^2, M^4.
var I2 = new MatriceC11(new[] { new[] { new GaussienC11(1), new GaussienC11(0) }, new[] { new GaussienC11(0), new GaussienC11(1) } });
var I4 = OpsC11.Kron(I2, I2);
var sz = new MatriceC11(new[] { new[] { new GaussienC11(1), new GaussienC11(0) }, new[] { new GaussienC11(0), new GaussienC11(-1) } });
var sx = new MatriceC11(new[] { new[] { new GaussienC11(0), new GaussienC11(1) }, new[] { new GaussienC11(1), new GaussienC11(0) } });
var sy = new MatriceC11(new[] { new[] { new GaussienC11(0), new GaussienC11(0, -1) }, new[] { new GaussienC11(0, 1), new GaussienC11(0) } });
var M = OpsC11.Somme(OpsC11.Kron(sz, sz), OpsC11.Kron(sx, sx));
var M2 = OpsC11.Matmul(M, M);
var M4 = OpsC11.Matmul(M2, M2);
var nul4 = new MatriceC11(new[] { new GaussienC11[4], new GaussienC11[4], new GaussienC11[4], new GaussienC11[4] });
for (int i = 0; i < 4; i++)
    for (int j = 0; j < 4; j++) nul4.Cases[i][j] = GaussienC11.Zero;

// M^4 = 4 M^2 : le spectre de M^2 est dans {0, 4} (polynome minimal x(x-4)).
Console.WriteLine($"M^4 == 4 M^2            -> {M4.Egale(OpsC11.Scalaire(4, M2))}");
// Les deux exclusions prouvent que la classe 4 est atteinte SANS etre seule :
// M^2 non nulle et pas proportionnelle a l'identite.
Console.WriteLine($"M^2 != 0 (classe 4 atteinte) -> {!M2.Egale(nul4)}");
Console.WriteLine($"M^2 != 4I (classe 0 atteinte) -> {!M2.Egale(OpsC11.Scalaire(4, I4))}");

// Determinant nul par deux arguments independants : trace nulle et deux
// lignes egales (0 et 3) — d'ou le 0 double du spectre.
BigInteger trace = BigInteger.Zero;
for (int i = 0; i < 4; i++) trace += M[i, i].Re;
bool deuxLignesEgales = true;
for (int j = 0; j < 4; j++) if (!M[0, j].Equals(M[3, j])) deuxLignesEgales = false;
Console.WriteLine($"trace(M) = 0, deux lignes egales -> det(M) = 0 : {trace == BigInteger.Zero && deuxLignesEgales}");
Console.WriteLine("=> spectre de M : {0, 0, +2, -2}, max |eig| = 2 ; S = sqrt(2) M, max = 2 sqrt(2)");

// Comparaison de bornes CHSH en rationnels exacts : borne classique 2 contre
// (2 sqrt(2))^2 = 8 = 2 x 2^2 — l'ecart factor sqrt(2) n'est jamais flottant.
var classique = new RationnelC11(new BigInteger(2));
var esperanceCarre = new RationnelC11(new BigInteger(8));
Console.WriteLine($"(2 sqrt(2))^2 == 8 == 2 x classique^2 -> {esperanceCarre == classique.Puissance(2) * new RationnelC11(new BigInteger(2))}");

VerifC11.Egale(M4.Egale(OpsC11.Scalaire(4, M2)), "M^4 doit valoir 4 M^2");
VerifC11.Egale(!M2.Egale(nul4), "M^2 ne doit pas etre nulle (classe 4 atteinte)");
VerifC11.Egale(!M2.Egale(OpsC11.Scalaire(4, I4)), "M^2 ne doit pas valoir 4I (classe 0 atteinte)");
VerifC11.Egale(trace == BigInteger.Zero, "trace(M) doit valoir 0");
VerifC11.Egale(deuxLignesEgales, "les lignes 0 et 3 de M doivent etre egales (det(M) = 0)");
VerifC11.Egale(esperanceCarre == classique.Puissance(2) * new RationnelC11(new BigInteger(2)),
               "l'esperance au carre doit valoir 2 x le carre de la borne classique");
Console.WriteLine("Check CHSH : CONFORME — spectre {0, 0, +2, -2}, M^4 = 4 M^2, (2 sqrt(2))^2 = 8 = 2 x 2^2 (rationnel exact).");
```

### Lecture du résultat (CHSH)

La glose **Bell** re-dérive aux réglages tournés du groupe de rotation des directions dérivées, et son check tient
**en arithmétique entière** : $M^2 = 2I - 2\,\sigma_y \otimes \sigma_y$ puis $M^4 = 4M^2$ forcent le spectre de
$M^2$ dans $\{0, 4\}$, les deux classes étant atteintes ; le spectre de $M$ est $\{0, 0, \pm 2\}$, l'opérateur CHSH
$S = \sqrt{2}M$ culmine à $2\sqrt{2}$, et le carré de l'espérance vaut **8 = 2 × 2² — deux fois le carré
classique**, exactement la formule imprimée. Aucun irrationnel n'entre dans la vérification : seul le facteur
$\sqrt{2}$ de la factorisation est extérieur aux entiers, et il ne sert qu'à nommer le maximum. La borne classique
