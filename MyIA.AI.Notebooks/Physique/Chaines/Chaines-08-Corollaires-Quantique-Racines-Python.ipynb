## 1. Les quarante-huit cartes et l'appariement de Born (équation 73)

La source : « From the three derived directions at the qubit end of the read link, by the signed permutation maps of
the three axes, we obtain forty-eight maps falling in exactly two kinds by determinant, twenty-four proper and
twenty-four improper, every one preserving the Born pairing of the six axis states at all one thousand seven hundred
and twenty-eight pairs » — c'est le **check exact du postulat de projection**, le plus petit cas travaillé par le
chapitre (« the smallest case worked here is the first corollary, the projection postulate, it is the wash at one
act »).

Le rejeu : les trois axes donnent six états d'axe (deux par axe). L'appariement de Born entre deux états d'axe se lit
$|\langle i|j\rangle|^2$ — la lecture standard : $1$ pour un état avec lui-même, $0$ pour son opposé sur le même axe,
$1/2$ entre axes distincts (constaté : la source ne réimprime pas sa définition de l'appariement ; ce rejeu utilise
la lecture standard et retrouve les comptes imprimés). Une carte signée envoie l'axe $a$ sur l'axe $\pi(a)$ avec un
flip $\varepsilon_a$ ; il y a $3!\cdot 2^3 = 48$ cartes, et chacune agit sur les six états. Le comptage des paires
ordonnées préservées : $48 \times 36 = 1728$.

```python
axes = [0, 1, 2]
etats_axe = [(a, s) for a in axes for s in (+1, -1)]   # six etats d'axe (axe, direction)

def appariement_born(e1, e2):
    """Appariement de Born |<i|j>|^2 entre deux etats d'axe, en Fraction exacte."""
    (a1, s1), (a2, s2) = e1, e2
    if a1 == a2:
        return Fraction(1) if s1 == s2 else Fraction(0)
    return Fraction(1, 2)

paires_ordonnees = [(e1, e2) for e1 in etats_axe for e2 in etats_axe]
print("Etats d'axe :", len(etats_axe), "| paires ordonnees par etat du couple :", len(paires_ordonnees))
print("Appariement d'un etat avec lui-meme      :", appariement_born((0, +1), (0, +1)))
print("Appariement avec l'oppose du meme axe    :", appariement_born((0, +1), (0, -1)))
print("Appariement entre axes distincts         :", appariement_born((0, +1), (1, +1)))
```

```python
def parite(perm):
    """Signe d'une permutation de 3 elements par comptage d'inversions."""
    inversions = sum(1 for i in range(len(perm)) for j in range(i + 1, len(perm)) if perm[i] > perm[j])
    return -1 if inversions % 2 else +1

cartes = []
for perm in permutations(axes):                 # 3! = 6 ordres des axes
    for flips in product((+1, -1), repeat=3):   # 2^3 = 8 flips d'axe
        cartes.append((perm, flips))

def determinant(carte):
    """Determinant de la carte = signe de la permutation x produit des flips."""
    perm, flips = carte
    return parite(perm) * flips[0] * flips[1] * flips[2]

propres = [c for c in cartes if determinant(c) == +1]
impropres = [c for c in cartes if determinant(c) == -1]
print("3! x 2^3 =", 6 * 8, "cartes ; propres :", len(propres), ", impropres :", len(impropres))
print("48 = 24 + 24 ->", len(cartes) == 48 and len(propres) == 24 and len(impropres) == 24)
```

```python
def image_etat(carte, etat):
    """Action d'une carte signee sur un etat d'axe : (axe, direction) -> (pi(axe), direction x flip)."""
    perm, flips = carte
    a, s = etat
    return (perm[a], s * flips[a])

paires_preservees = 0
total_paires = len(cartes) * len(paires_ordonnees)
for carte in cartes:
    for e1, e2 in paires_ordonnees:
        if appariement_born(image_etat(carte, e1), image_etat(carte, e2)) == appariement_born(e1, e2):
            paires_preservees += 1

print("#{maps} = 48 = 24 + 24, det = +/-1, #{Born pairs preserved} =", paires_preservees, "among", total_paires)
print("Eq. 73 (eq. 73 de la source) ->", paires_preservees == 1728 and total_paires == 1728)
```

### Lecture du résultat (équation 73)

Les quarante-huit cartes se partagent **exactement en deux espèces par déterminant** — vingt-quatre propres,
vingt-quatre impropres — et **chacune** préserve l'appariement de Born des six états d'axe aux **1728 paires
ordonnées** (48 cartes × 36 paires). La préservation tient structurellement : une carte signée envoie un couple
d'un même axe sur un couple d'un même axe (appariement $0$ ou $1$) et un couple d'axes distincts sur des axes
distincts ($1/2$) — l'énumération le constate mécaniquement, carte par carte, paire par paire. C'est le check exact
que la source attache au **postulat de projection** : « the wash at one act » laisse l'extrémité proche dans le
projecteur assorti, et sa vérification est cette symétrie de l'appariement. Deux des neuf checks exacts du chapitre
sont rejoués dans ce carnet : celui-ci et, au paragraphe suivant, l'espérance CHSH.

## 2. Le check CHSH « at twice the classical square »

Le théorème 47.1 compte, parmi ses neuf checks exacts, « the Clauser-Horne-Shimony-Holt expectation at twice the
classical square » ; le résultat ouvert du chapitre précise : « the Bell inequality at the rotated settings of the
derived rotation group with the Clauser-Horne-Shimony-Holt maximum twice the square root of two ». Autrement dit :
le maximum quantique $2\sqrt{2}$ contre la borne classique $2$, et le carré de l'espérance
$8 = 2 \cdot 2^2$ — **deux fois le carré classique**.

Le rejeu en entiers exacts : aux réglages optimaux $A_0 = \sigma_z$, $A_1 = \sigma_x$, $B_0 = (\sigma_z +
\sigma_x)/\sqrt{2}$, $B_1 = (\sigma_z - \sigma_x)/\sqrt{2}$, l'opérateur CHSH se factorise en $S = \sqrt{2}\,M$ où
$M = \sigma_z \otimes \sigma_z + \sigma_x \otimes \sigma_x$ est une **matrice à coefficients entiers**. Tout se
vérifie sans irrationnel : $M^2 = 2I - 2\,\sigma_y \otimes \sigma_y$ (entiers gaussiens), puis $M^4 = 4M^2$, ce qui
force le spectre de $M^2$ dans $\{0, 4\}$ — et $M^2 \neq 0$, $M^2 \neq 4I$ montre que les deux classes sont
atteintes : le spectre de $M$ est $\{0, 0, +2, -2\}$. Le maximum $2\sqrt{2}$ de $S$ suit, et son carré vaut
exactement $8$.

```python
def kron(A, B):
    """Produit tensoriel de deux matrices, en arithmetique exacte (entiers / entiers gaussiens)."""
    n, m, p, q = len(A), len(B), len(B[0]), len(B[0])
    return [[A[i // len(B)][j // len(B[0])] * B[i % len(B)][j % len(B[0])]
             for j in range(len(A[0]) * len(B[0]))] for i in range(n * m)]

def matmul(A, B):
    return [[sum(A[i][k] * B[k][j] for k in range(len(B))) for j in range(len(B[0]))] for i in range(len(A))]

def scalaire(c, A):
    return [[c * x for x in ligne] for ligne in A]

def somme(A, B):
    return [[x + y for x, y in zip(la, lb)] for la, lb in zip(A, B)]

I2 = [[1, 0], [0, 1]]
I4 = kron(I2, I2)
sz = [[1, 0], [0, -1]]
sx = [[0, 1], [1, 0]]
sy = [[0, -1j], [1j, 0]]          # entiers gaussiens : exacts en Python

M = somme(kron(sz, sz), kron(sx, sx))
M2 = matmul(M, M)
cible_M2 = somme(scalaire(2, I4), scalaire(-2, kron(sy, sy)))
print("M = sz (x) sz + sx (x) sx, entiers purs :", all(isinstance(x, int) for ligne in M for x in ligne))
print("M^2 == 2I - 2 sy(x)sy   ->", M2 == cible_M2)
```

```python
M4 = matmul(M2, M2)
nul4 = [[0] * 4 for _ in range(4)]
print("M^4 == 4 M^2            ->", M4 == scalaire(4, M2))   # spectre de M^2 dans {0, 4}
print("M^2 != 0 (classe 4 atteinte) ->", M2 != nul4)
print("M^2 != 4I (classe 0 atteinte) ->", M2 != scalaire(4, I4))
print("trace(M) = 0, deux lignes egales -> det(M) = 0 :", sum(M[i][i] for i in range(4)) == 0 and M[0] == M[3])
print("=> spectre de M : {0, 0, +2, -2}, max |eig| = 2 ; S = sqrt(2) M, max = 2 sqrt(2)")

classique = Fraction(2)             # borne classique CHSH
esperance_carre = Fraction(8)       # (2 sqrt(2))^2
print("(2 sqrt(2))^2 == 8 == 2 x classique^2 ->", esperance_carre == 2 * classique ** 2)
```

### Lecture du résultat (CHSH)

La glose **Bell** re-dérive aux réglages tournés du groupe de rotation des directions dérivées, et son check tient
**en arithmétique entière** : $M^2 = 2I - 2\,\sigma_y \otimes \sigma_y$ puis $M^4 = 4M^2$ forcent le spectre de
$M^2$ dans $\{0, 4\}$, les deux classes étant atteintes ; le spectre de $M$ est $\{0, 0, \pm 2\}$, l'opérateur CHSH
$S = \sqrt{2}M$ culmine à $2\sqrt{2}$, et le carré de l'espérance vaut **8 = 2 × 2² — deux fois le carré
classique**, exactement la formule imprimée. Aucun irrationnel n'entre dans la vérification : seul le facteur
$\sqrt{2}$ de la factorisation est extérieur aux entiers, et il ne sert qu'à nommer le maximum. La borne classique
$2$ reste $2$ : la glose ne prouve pas l'inégalité de Bell de la littérature, elle constate la correspondance par
nom à un check exact.

## 4. Le contre-test : frapper la racine mesure

Les contre-tests du compendium « alter the objects singly and read what moves ». Celui du chapitre 47 :
« Striking one root moves five entries from derived or gloss to open, all five of them, the projection postulate
the first. » La racine frappée est la **mesure comme lavage à un acte** (théorème 42.1) ; les entrées qui pendent à
cette racine dans la table : le **postulat de projection** et la **décohérence** (dérivées), la **forme de
Lindblad**, l'**effet Zeno** et la **mesure généralisée** (gloses au résultat du lavage). On rejoue le geste sur la
table : frapper la racine, retyper les cinq entrées en *open*, lire le compte et la première.

```python
# Contre-test imprime par le source : frapper la racine mesure (le lavage du ch. 42)
# deplace CINQ entrees vers open, le postulat de projection en tete.
# La table imprime cette racine de deux facons : "mesure : le lavage a un acte (Thm 42.1)"
# pour les deux derivees, "resultat du lavage" pour les trois gloses -- une seule racine,
# deux graphies, toutes deux frappees.
racines_frappees = {"mesure : le lavage a un acte (Thm 42.1)", "resultat du lavage"}
table_apres_frappe = []
deplacees = []
for nom, espece, racine, type_racine in entrees:
    if racine in racines_frappees:
        deplacees.append(nom)
        table_apres_frappe.append((nom, "open", racine, type_racine))
    else:
        table_apres_frappe.append((nom, espece, racine, type_racine))

print("Entrees deplacees vers open : 5 ->", len(deplacees) == 5)
print("En tete : postulat de projection ->", deplacees[0] == "postulat de projection")
print("Les cinq :", deplacees)
especes_apres = Counter(e for _, e, _, _ in table_apres_frappe)
print("Apres frappe : derivees", especes_apres["derivee"], "+ gloses", especes_apres["glose"], "+ gatees", especes_apres["gatee"], "+ open", especes_apres["open"])
```

### Lecture du résultat (contre-test)

Le geste imprimé se rejoue exactement : frapper la racine mesure déplace **cinq** entrées de *dérivée* ou *glose*
vers *open* — les deux dérivées du lavage à un acte et ses trois gloses — et la **première** est le **postulat de
projection**, comme l'imprime la phrase du census. C'est la lecture du typage qui s'en trouve éprouvée : la
dépendance n'est pas décorative, retirer une racine fait tomber d'un coup tout ce qui y pend, et le census a
compté les branches avant de les couper.

## 5. Le nommage Wigner — carrefour avec Chaînes-07

Le résultat ouvert qui se tient dans ce chapitre : « Four corollaries re-derive from named clauses at three of
four » — l'inégalité de Bell (check rejoué au §2), le couplage de Clebsch-Gordan aux deux extrémités d'un lien,
l'amplitude de tunnelling aux liens sombres $(3,3)$ à l'ordre deux — « and the Wigner function is the first
naming, its missing clause the swap face as a cell, the record distribution at the electron web of the construction
carrying no negative cell while the swap face of the many-body result is negative beyond two copies ».

Cette clause manquante est précisément ce que
[Chaînes-07](Chaines-07-Fonction-Wigner-Signe-N-Corps-Python.ipynb) a mesuré au chapitre 46 : la fonction de Wigner
comme cellule swap-face tient à **quatre des cinq vérifications**, et c'est au **recouvrement** qu'elle part. Le
chapitre 47 le reformule en dépendance : la distribution d'enregistrements de la toile électron ne porte **aucune
