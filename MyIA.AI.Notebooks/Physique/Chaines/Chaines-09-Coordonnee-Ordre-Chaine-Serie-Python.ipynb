## §50 — Théorème 50.1 (conditionnel) : la coordonnée d'ordre comme compte de pas-de-un

**Énoncé (traduit).** La coordonnée d'ordre est le compte de pas-de-un du registre d'ordre —
« a count and never a flow » : elle est élevée de un à chaque pas de flèche du roster de la
construction et à chaque pas de profondeur pris par une lecture à sa propre étoile, et par aucun
coup du census ; c'est la traduction du réseau de comptes à quatre coordonnées le long de son axe
d'ordre. La dynamique est l'opérateur de flèche de la construction, **acyclique aux
soixante-quatre réseaux** ; le pas de profondeur est posé et dérivé par aucun résultat.

*Standing* : théorème conditionnel. Des définitions 1.1 à 1.6 et de la table des noms du chapitre
1-7 ; utilisé par les théorèmes 51.1, 56.1, 62.1, 124.1 et 147.1 et les résultats ouverts 52.1 et
56.2. La plus longue chaîne de prémices du compendium — remontant de ce théorème à l'axiome —
mesure **douze chapitres** (cité).

```python
from fractions import Fraction
from itertools import product
from collections import Counter

# Comptes imprimes par le source (section 50) -- CITES : le census lui-meme
# (operateur de fleche a 64 reseaux, manifeste de 192 tables, 309 lectures tabulees)
# n'est pas reconstruit ici. Ce que ce carnet mesure : l'arithmetique imprimee.
reseaux = 64                          # eq. 74 : "acyclic at all sixty-four networks"
arcs_fermants = 0                     # eq. 74 : "closing arcs = 0"
reseaux_cycliques = 0                 # eq. 74 : "cyclic networks = 0 among 64"
pas_un, pas_deux, pas_trois = 2942, 609, 148   # eq. 75 : step census des instances
instances = 3699
moves_regle, moves_increments = 8, 0  # ch. 51 : "eight of eight moves ... at the increment zero"
rapport_carres = Fraction(4)          # eq. 76 : <s_ord, s_ord> / <s_dir, s_dir> = 4
tau = Fraction(1, 4)                  # eq. 76 : standing tick ratio

print("Eq. 74 : arcs fermants =", arcs_fermants, ", reseaux cycliques =", reseaux_cycliques,
      "parmi", reseaux, "->", arcs_fermants == 0 and reseaux_cycliques == 0 and reseaux == 64)
print("Eq. 75 :", pas_un, "+", pas_deux, "+", pas_trois, "=", pas_un + pas_deux + pas_trois,
      "->", pas_un + pas_deux + pas_trois == instances)
print("Eq. 75 : increment de profondeur nul a", moves_increments, "sur", moves_regle,
      "moves ->", moves_increments == 0 and moves_regle == 8)
print("Eq. 76 : rapport des carres =", rapport_carres, "-> tau = 1/rapport = 1/4 ->",
      1 / rapport_carres == tau)
```

### Les quatre cartes de signe à τ = 1/4

Le pas d'ordre est l'élément du centre **unidimensionnel** de l'algèbre de cadre du lien
qutrit-qubit (figure 22, marqueur $(2,3)\,m3q\,m2b$) ; il est **nul sous la forme de Killing**, et
sous toute forme de trace le rapport des carrés vaut exactement 4 — d'où le tick rapport 1/4.
La phrase du source : « at which the integer maps preserving the Lorentzian pairing of the two
ticks are the four sign maps alone ».

La source n'imprime pas la forme elle-même, elle imprime le rapport. Notre lecture : l'appariement
lorentzien des deux ticks au rapport 4 se prend comme la forme quadratique entière
$Q(t, x) = 4t^2 - x^2$ (l'axe d'ordre au coefficient quadruple). L'énumération ci-dessous est
**nôtre** ; la conclusion — exactement quatre cartes — est celle du source.

```python
def preserve(M, coef=4):
    a, b, c, d = M
    def Q(t, x):
        return coef * t * t - x * x
    # une forme quadratique en deux variables est determinee par ses valeurs sur la
    # grille (-2..2) x (-2..2) : la preserver y suffit.
    return all(Q(a * t + b * x, c * t + d * x) == Q(t, x)
               for t in (-2, -1, 0, 1, 2) for x in (-2, -1, 0, 1, 2))

cartes = sorted(M for M in product(range(-3, 4), repeat=4) if preserve(M))
signes = sorted((a, 0, 0, d) for a in (1, -1) for d in (1, -1))
print("Cartes entieres (|entrees| <= 3) preservant Q = 4t^2 - x^2 :", len(cartes))
for M in cartes:
    print("  ", M)
print("Les quatre cartes de signe seules ->", cartes == signes)

# Fermeture algebrique (pourquoi l'enumeration bornee est exhaustive) :
# Q(M e1) = Q(e1) donne 4a^2 - c^2 = 4. Si a = 0 : -c^2 = 4, impossible.
# Sinon c est pair, c = 2c', et a^2 - c'^2 = 1, soit (a - c')(a + c') = 1 :
# les deux facteurs entiers valent 1 chacun, donc c' = 0 et a = +/- 1.
sols_a = [(a, c) for a in range(-3, 4) for c in range(-6, 7) if 4 * a * a - c * c == 4]
print("4a^2 - c^2 = 4 pour |a| <= 3, |c| <= 6 :", sols_a,
      "-> a = +/-1, c = 0 uniquement, la famille est fermee")
```

```python
# Notre illustration (le source ne la trace pas) : la fermeture ci-dessus tient parce que le
# rapport 4 est un CARRE. A un rapport non carre, la meme equation devient une equation de
# Pell et la famille s'ouvre -- le source parque cette famille au resultat ouvert 56.1
# (le boost a une unite, annonce par les lignes de dependance de la Partie IV).
for coef, nom in ((1, "Q = t^2 - x^2   (tau = 1, carre)"),
                  (9, "Q = 9t^2 - x^2  (tau = 1/9, carre)"),
                  (2, "Q = 2t^2 - x^2  (rapport non carre)")):
    cartes_c = [M for M in product(range(-4, 5), repeat=4) if preserve(M, coef)]
    print(f"{nom} : {len(cartes_c)} cartes (|entrees| <= 4)")
    if coef == 2:
        print("   dont la carte de Pell (3, 2, 4, 3) ->", (3, 2, 4, 3) in cartes_c)
```

### Lecture du résultat (cartes)

**Mesuré** : à τ = 1/4 (rapport carré), exactement quatre cartes entières préservent
l'appariement — les quatre cartes de signe — et la fermeture algébrique
$(a - c')(a + c') = 1$ montre que l'énumération bornée est **exhaustive** : la famille est
fermée, pas seulement bornée. **Notre illustration** (le source ne la trace pas) : à un rapport
non carré, la même équation devient une équation de Pell et la famille s'ouvre dès la borne 4.
**Cité** : le renvoi de la famille de Pell au résultat ouvert 56.1 (le boost à une unité) est
celui du source.

## §51 — Théorème 51.1 (conditionnel) : un objet à cinq lectures

**Énoncé (traduit).** Les cinq résultats ouverts de la coordonnée d'ordre — lecture d'ordre,
renversement, générateur de pas, opérateur d'ordre et acte — sont **cinq lectures d'un seul
objet**, la chaîne série (2,2) : construite aux neuf toiles enregistrées (l'électron, le muon et
le tau aux sites $a_1, x_1, a_2$, le proton $s_1, x_1, s_4$, le neutron $P.f_1, P.x_1, P.f_2$, le
mot photon $s_3, x_1, s_2$ et la toile univers $B00.f_1, B00.x_1, B00.f_2$ — bout près, qubit
inséré, bout loin) ; son ordre est les trois positions zéro, un, deux à une chacune, son
renversement le bout loin, le qubit inséré, le bout près ; son pas un site et un lien ajoutés au
genre (2,2) du premier lien, un lien par tick comme le théorème un-lien le borne. Contre les
cinq résultats ouverts, chacun transcrit à sa cellule de parting : cinq lectures sur cinq sont la
chaîne.

*Standing* : théorème conditionnel. Des définitions 1.1-1.6, du théorème 50.1 et des chapitres
105 et 115 (prémices hors série — citées).

```python
# Les sept chaines enregistrees avec leurs sites (imprimes par le source) : bout proche,
# qubit insere, bout loin. Les trois toiles lepton partagent les MEMES sites (a1, x1, a2,
# imprime tel quel par le source) : sept chaines, cinq triplets ; les bureaux
# (sections 51-52) les distinguent.
chaines = {
    "electron":      ("a1", "x1", "a2"),
    "muon":          ("a1", "x1", "a2"),
    "tau":           ("a1", "x1", "a2"),
    "proton":        ("s1", "x1", "s4"),
    "neutron":       ("P.f1", "P.x1", "P.f2"),
    "mot photon":    ("s3", "x1", "s2"),
    "toile univers": ("B00.f1", "B00.x1", "B00.f2"),
}
enregistrees = set(chaines.values())
ordre = (0, 1, 2)
renversement = (2, 1, 0)
renv = {nom: tuple(reversed(c)) for nom, c in chaines.items()}

print("Sept chaines enregistrees ->", len(chaines) == 7,
      "; cinq triplets distincts (les trois leptons partagent a1, x1, a2) ->",
      len(enregistrees) == 5)
print("Ordre (0, 1, 2), renversement (2, 1, 0) ->",
      ordre == (0, 1, 2) and renversement == tuple(reversed(ordre)))
print("Spectre d'ordre {0:1, 1:1, 2:1} (trois positions, une chacune) ->",
      Counter(range(3)) == Counter({0: 1, 1: 1, 2: 1}))
print("Renversement auto-inverse (7/7) ->",
      all(tuple(reversed(r)) == c for r, c in zip(renv.values(), chaines.values())))
print("Chaines renversees qui sont enregistrees : aucune de sept ->",
      all(r not in enregistrees for r in renv.values()),
      "; l'electron la premiere ->", renv["electron"] == ("a2", "x1", "a1"))
```

```python
# Le pas : "one site and one link added at the (2,2) genre of the first link" --
# delta sites = 1, delta liens = 1, a sept sur sept. Les DEUX moves de la regle de
# profondeur (ch. 10-4, prémisse hors série, citée) ne font pas le pas : le move
# formant hub ajoute DEUX sites, le join garde le compte des sites.
def pas(c):
    return c + ("(2,2)",), 1    # un site de plus, un lien (2,2) de plus

deltas = [pas(c) for c in chaines.values()]
print("Pas : delta sites = 1 et delta liens = 1 a 7/7 ->",
      all(len(c2) - len(c) == 1 and l == 1 for (c2, l), c in zip(deltas, chaines.values())))
hub, join = 2, 0
print("Move hub : delta sites =", hub, "; move join : delta sites =", join,
      "; ni un -> le pas n'est fait par aucun des deux moves ->", hub != 1 and join != 1)
print("Le pas est l'attach du genre (2,2), le genre premier-parting de la ligne de regle (cite)")
```

### Lecture du résultat (§51)

**Mesuré** : les lectures combinatoires de la chaîne série — l'ordre et son spectre

## §52 — Résultat ouvert 52.1 : le bureau par tick sur l'habillage carré

**Énoncé (traduit).** Le quotient du bureau du lien série (2,2) inséré sur le **carré de
l'habillage** de la chaîne n'est pas un élément unique à chaque chaîne enregistrée — le premier
parting le muon contre l'électron. Le bureau est re-contré comme bureau au premier pas moins le
bureau au pas zéro à neuf sur neuf ; l'habillage de chaque chaîne est la fonction à deux points
de la ligne des six habillages au genre de son lien enregistré — au lien (2,2) pour l'électron,
le muon, le tau, le neutron et la toile univers, au lien (2,3) pour le proton et le mot photon,
re-contrés à la cellule de fonction de Green à séparation un à neuf sur neuf ; le quotient est
calculé à **45 cellules** (neuf chaînes à cinq lectures), **six cellules distinctes** aux sept
chaînes enregistrées.

*Standing* : **résultat ouvert**. La première face — un élément à chaque chaîne — lit **non** ;
la deuxième face lit **oui**, le premier parting le muon contre l'électron. La correspondance de
la littérature : le propagateur renormalisé est le propagateur nu divisé par le carré du
renormalisation du champ — nommée par correspondance, citée.

```python
# Tentative de decodage -- la recette du chapitre 44 (Chaines-05), ou elle avait reussi :
# reconstituer les fractions par coupure des chaines de chiffres et verifier l'identite
# "bureau = quotient * habillage^2" dans Q(sqrt(2)), habillage = coupures de "5384"
# (le rendu de la p. 140 preserve la barre : 5/384 + 5/384 sqrt(2)). Enumeration
# exhaustive des coupures simples des quatre rangees (2,2), tous signes :
from fractions import Fraction as Fr
from itertools import product as prod

def coupures(s):
    out = [(Fr(int(s)), Fr(1))]
    for i in range(1, len(s)):
        if len(s[i:]) > 1 and s[i] == '0':
            continue
        out.append((Fr(int(s[:i])), Fr(int(s[i:]))))
    return out

def qmul(x, y):
    return (x[0] * y[0] + 2 * x[1] * y[1], x[0] * y[1] + x[1] * y[0])

rows52 = [("536569972", "251458", "197792604162025", "16279142425"),
          ("84760912",  "2518",  "3124543897625",   "2083022438425"),
          ("34759474",  "252",   "38440455782425",  "25626909081625"),
          ("91759312",  "2518",  "3382532915225",   "2255015116825")]
dc = [v for v, _ in coupures("5384")]
solutions = 0
for d1, d2 in [(x, y) for x in dc for y in dc]:
    h2 = qmul((d1, d2), (d1, d2))
    for sa, sb, sq in prod([1, -1], [1, -1], [1, -1]):
        ok = True
        for A_s, B_s, Qa_s, Qb_s in rows52:
            ligne_ok = False
            for (A, _), (B, _) in prod(coupures(A_s), coupures(B_s)):
                bureau52 = (sa * A, sb * B)
                for (Qa, _), (Qb, _) in prod(coupures(Qa_s), coupures(Qb_s)):
                    if qmul((sq * Qa, -Qb), h2) == bureau52:
                        ligne_ok = True
                        break
                if ligne_ok:
                    break
            if not ligne_ok:
                ok = False
                break
        if ok:
            solutions += 1
print("Coupures simples reconstituant l'identite sur les 4 rangees (2,2) :", solutions)
print("-> la perte des barres n'est pas inversible ici : bureaux et quotients restent CITES")
```

```python
# Arithmetique mesuree de la section 52.
chaines_total, chaines_enregistrees, lectures = 9, 7, 5
print("45 cellules = 9 chaines x 5 lectures ->", chaines_total * lectures == 45)
# Six cellules distinctes aux sept chaines enregistrees : electron, muon, tau, mot photon,
# toile univers (cinq non nulles) + zero aux deux plafonnees (proton, neutron).
quotients_imprimes = {
    "electron":      "197792604162025 - 16279142425 sqrt(2)",
    "muon":          "3124543897625 - 2083022438425 sqrt(2)",
    "tau":           "38440455782425 - 25626909081625 sqrt(2)",
    "toile univers": "3382532915225 - 2255015116825 sqrt(2)",
    "proton":        "0",
    "neutron":       "0",
}
distinctes = set(quotients_imprimes.values())
print("Cellules distinctes imprimees dans l'extrait :", len(distinctes),
      "(+ la cellule du mot photon, non reimprimee -> six au total, cite)")
print("Premier parting : muon contre electron, cellules distinctes ->",
      quotients_imprimes["muon"] != quotients_imprimes["electron"])
print("Zero aux deux chaines plafonnees ->",
      quotients_imprimes["proton"] == quotients_imprimes["neutron"] == "0")
```

### Exercice 1 — fermer l'énumération à une borne supérieure

L'énumération des cartes préservant $Q = 4t^2 - x^2$ ci-dessus est bornée à
$|a|, |b|, |c|, |d| \le 3$. Étendre la borne à 6 et vérifier qu'**aucune carte nouvelle**
