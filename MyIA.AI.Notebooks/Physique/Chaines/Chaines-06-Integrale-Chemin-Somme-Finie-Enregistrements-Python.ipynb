## 3. Le carrefour des ordres : 172

Les ordres de lecture du lien (2, 3) sont les échanges signés survécus. Le comptage par ordre donne le carrefour arithmétique du chapitre — le même espace, compté deux fois.

```python
ordres_lecture = {0: 79, 1: 90, 2: 3}     # enregistrements impairs / a un echange / de cloture

total_lettres = sum(ordres_lecture.values())
print("79 + 90 + 3      =", total_lettres, "(le chapitre annonce 172 lettres) ->", total_lettres == 172)

# Le meme espace, decompose en secteurs de l'etoile (Chaines-01 et chapitre 44) :
secteurs_etoile = {"impairs": 79, "pairs mixtes": 48, "pairs identiques": 42, "cloture": 3}
total_etoile = sum(secteurs_etoile.values())
print("79 + 48 + 42 + 3 =", total_etoile, "->", total_etoile == 172)
print("48 + 42          =", 48 + 42, "= ordre 1 (90) ->", 48 + 42 == 90)

# Le rang support (CITE, non derive ici) : 319 = compte de Born
rang_support = 319
print("rang support     =", rang_support, "(CITE : egal au compte de Born)")
```

### Lecture du résultat
Les deux décompositions comptent le **même espace de 172 enregistrements** : par ordres d'échange signé (79 impairs d'ordre 0, 90 à un échange d'ordre 1, 3 de clôture d'ordre 2), ou par secteurs de l'étoile des carnets précédents (79 + 48 + 42 + 3). La réunion des secteurs pairs — 48 + 42 = 90 — est le carrefour : le lien (2, 3) vit dans le même espace que l'étoile du chapitre 44, lu par ordres d'échange au lieu de secteurs. Le rang support 319, cité égal au compte de Born, dit que la taille de l'espace d'enregistrements **est** la règle de Born.

## 5. Les trois théorèmes conditionnels voisins

Autour de l'objet ouvert, trois théorèmes **conditionnels** (démontrés sous leurs hypothèses de census) :

1. **Ordre de lecture** — à chaque enregistrement, l'ordre (0, 1 ou 2) est égal au nombre de facteurs du compte de gain : vérifié aux **172**.
2. **Lecture à chaque ordre** — la série de lecture **termine à l'ordre deux** ; l'ordre un redonne l'enregistrement de premier ordre du résultat de gain aux 172 ; à la paire tenue, le terme d'ordre k au rang n égale le terme au rang 1/n², aux **huit rangs** (correspondance Rydberg [78]).
3. **Lecture à travers le corps** — deux termes exacts aux **13 corps** ; l'ordre deux est un élément exact du corps Q(√2, √3), non nul à **8 des 9 corps de chaîne** ; au-delà de deux, tout est nul.

```python
# L'echelle de Rydberg aux huit rangs (correspondance [78], CITEE par nom) :
echelle_rydberg = {n: Fraction(1, n * n) for n in range(1, 9)}
for n, valeur in echelle_rydberg.items():
    print(f"rang {n} : 1/{n}^2 = {valeur}")

# La structure d'echelle se rejoue : chaque terme est le CARRE de l'inverse du rang
carre_ok = all(valeur == Fraction(1, n) ** 2 for n, valeur in echelle_rydberg.items())
print("1/n^2 == (1/n)^2 aux huit rangs ->", carre_ok)
print("Terme d'ordre k au rang n = terme au rang 1/n^2 (paire tenue, CITE aux huit rangs)")
```

### Lecture du résultat
L'échelle 1, 1/4, 1/9, 1/16, …, 1/64 est la signature **Rydberg** [78] : la valeur d'un terme varie comme l'inverse **carré** du rang, comme les niveaux de l'atome d'hydrogène. Ce que ce carnet rejoue est la structure d'échelle (chaque terme est le carré de l'inverse du rang) ; l'égalité terme-au-rang-n = terme-au-rang-1/n² à la paire tenue est citée aux huit rangs — la constante devant l'échelle n'est pas dérivée ici.

```python
# Lecture a travers le corps (theoreme 3, CITE) :
corps_chaine = 13
print("Corps avec deux termes exacts          :", corps_chaine)

corps_de_chaines = 9
non_nuls_ordre2 = 8
print("Ordre 2 non nul a 8 des 9 corps de chaines ->", 0 < non_nuls_ordre2 < corps_de_chaines)
print("Ordres > 2 : nuls (la serie s'arrete d'elle-meme, correspondance [81])")

# Le facteur manquant du bureau de masse (resultat ouvert n. 3) :
facteur_manquant = Fraction(1, 3 ** 2)
print("Facteur manquant 1/3^2                 =", facteur_manquant, "->", facteur_manquant == Fraction(1, 9))
print("Bureau de masse des 8 corps non unitaires : ecart de ce facteur a l'invariant de repos debout (OUVERT)")
```

### Lecture du résultat
La lecture à travers le corps s'arrête d'elle-même : deux termes exacts aux 13 corps, un ordre deux qui vit dans Q(√2, √3) non nul à 8 corps de chaîne sur 9, et **zéro au-delà de deux** — la contrepartie exacte de la terminaison du développement perturbatif (correspondance [81]). Le facteur manquant **1/9** relie ce théorème au résultat ouvert n° 3 : le bureau de masse ne retrouve l'invariant d'énergie de repos debout à **aucun** des 9 corps, et l'écart aux 8 corps non unitaires est précisément 1/3².
