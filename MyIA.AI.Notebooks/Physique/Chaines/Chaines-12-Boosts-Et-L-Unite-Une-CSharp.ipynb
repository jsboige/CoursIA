## §56 — Théorème 56.1 : l'ordre aux trois directions

« The twirl of the definition known, run over the order register and the three directions together at the record of every read, carries the order coordinate into no direction coordinate at the pair ket » (p. 150). Les **quarante-huit appariements** admis par la construction — « the twenty-four rotations of the cube on the directions joined with the two signs on the order coordinate » — portent chacun **un bloc nul** de l'ordre vers les directions : aucun enregistrement n'est porté vers un enregistrement dont la composante de direction viendrait de son ordre (0 de 322714 enregistrements à 0 de 3072 paires, sur les 64 toiles cataloguées — cité).

L'appariement à la main qui **échange** l'ordre et une direction porte 319332 de ces enregistrements et efface l'ordre à chacun (cité) — mais il est **hors de la fermeture admise** : 384 cartes contre 48 (cité). Et le premier open result de la release précédente — aucune règle établie ne mélange l'ordre aux directions — se **referme** ici : les neuf règles portent le bloc nul, le signe est propriété du twirl et d'aucune règle, la table décisive re-dérivée en Lean aux quarante-huit appariements (cité).

```
// Les 24 rotations du cube (permutations signees de determinant +1, deja construites
// aux Chaines-10) et les 48 appariements admis (24 x 2 signes sur l'ordre).
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Linq;

// Conserver les comparaisons exécutées pour le relevé final.
var comparaisonsParite = new List<(string Section, string Controle, bool Conforme)>();
string Invariant(FormattableString texte) => texte.ToString(CultureInfo.InvariantCulture);

void VerifierParite<T>(string section, string controle, T calcule, T attendu)
{
    bool conforme = EqualityComparer<T>.Default.Equals(calcule, attendu);
    comparaisonsParite.Add((section, controle, conforme));
    if (!conforme)
        throw new InvalidOperationException(Invariant($"{section} — {controle} : ÉCART, obtenu {calcule}, attendu {attendu}."));
}

int[,] Zeros(int n) => new int[n, n];
var rotations = new List<int[,]>();
int[] signes = { 1, -1 };
const string sectionAppariements = "§56 — Appariements";
const int rotationsAttendues = 24;
const int appariementsAttendus = 48;

for (int p0 = 0; p0 < 3; p0++)
for (int p1 = 0; p1 < 3; p1++)
for (int p2 = 0; p2 < 3; p2++)
{
    if (p0 == p1 || p0 == p2 || p1 == p2)
        continue;
    int parite = (p0, p1, p2) is (0, 1, 2) or (1, 2, 0) or (2, 0, 1) ? 1 : -1;
    foreach (int s0 in signes)
    foreach (int s1 in signes)
    foreach (int s2 in signes)
    {
        if (s0 * s1 * s2 * parite != 1)
            continue;
        var r = Zeros(3);
        r[0, p0] = s0;
        r[1, p1] = s1;
        r[2, p2] = s2;
        rotations.Add(r);
    }
}
VerifierParite(sectionAppariements, "Rotations du cube", rotations.Count, rotationsAttendues);
Console.WriteLine(Invariant($"rotations du cube (determinant +1) : {rotations.Count} (attendu {rotationsAttendues})"));

int[,] Bloc5(int eps, int[,] r)
{
    var m = Zeros(5);
    m[0, 0] = eps;
    for (int i = 0; i < 3; i++)
    for (int j = 0; j < 3; j++)
        m[1 + i, 1 + j] = r[i, j];
    return m;
}

var matchings = new List<int[,]>();
foreach (int eps in signes)
foreach (var r in rotations)
    matchings.Add(Bloc5(eps, r));
VerifierParite(sectionAppariements, "Appariements admis", matchings.Count, appariementsAttendus);
Console.WriteLine(Invariant($"appariements admis (24 rotations x 2 signes) : {matchings.Count} (attendu {appariementsAttendus})"));

bool BlocOrdreDirectionsNul(int[,] m)
{
    for (int j = 1; j <= 4; j++)
        if (m[0, j] != 0 || m[j, 0] != 0)
            return false;
    return true;
}

// Construction, pas mesure : bloc5 POSE les zeros hors blocs (eps en (0,0), R en 3x3,
// zeros partout ailleurs), donc le bloc nul ordre->directions est impose par la forme.
// Ce que la construction porte et ce que le source ferme a l'unite une se rejoignent
// sur les 48 cartes -- la fermeture est le theoreme, la forme est notre realization.
var zeroOk = matchings.Select(m => BlocOrdreDirectionsNul(m)).ToArray();
int blocsNuls = zeroOk.Count(estNul => estNul);
VerifierParite(sectionAppariements, "Appariements à bloc nul", blocsNuls, 48);
VerifierParite(sectionAppariements, "Appariements testés pour le bloc nul", zeroOk.Length, appariementsAttendus);
Console.WriteLine(Invariant($"Bloc nul ordre->directions aux 48 appariements (CONSTRUCTION : bloc5 pose les zeros) : {blocsNuls} / {zeroOk.Length}"));
```

### Lecture du résultat (les appariements)

**Mesuré** : les 24 rotations reconstruites (le groupe de Chaînes-10), jointes aux deux signes de l'ordre, donnent exactement 48 cartes — et **chacune porte le bloc nul** : l'ordre n'entre jamais dans les directions. **Cité** : les comptes du census — 0 de 322714 enregistrements à 0 de 3072 paires, sur les 64 toiles cataloguées, la première toile nommée avec ses 48 appariements comptés, aucun ne portant.

## L'éq. 83 : la même Gram, le même générateur

« From the Hermitian order unit and the three direction steps under the plain trace, by the Gram matrix, we obtain the Lorentzian pairing multiplied by three halves » (p. 151-152) —

$$G = \tfrac32\,\mathrm{diag}(4, -1, -1, -1), \qquad K^2 = 1. \tag{83}$$

C'est **la même Gram que l'éq. 79** du chapitre précédent (le lien qutrit-qubit), et le générateur hyperbolique qu'on en résout est celui de Chaînes-11 : notre boost de plan $(1/2, 2)$, bloc $2\times2$ carré l'identité.

```
// Eq. 83 sur notre modele 4D (t, x, y, z) : G = (3/2) eta, le boost de plan (1/2, 2).
using System.Numerics;

// La réduction par PGCD et le signe au numérateur donnent une représentation unique.
readonly record struct Rationnel
{
    public BigInteger Numerateur { get; }
    private readonly BigInteger denominateurMoinsUn;
    // Ce décalage fait aussi de la valeur par défaut d'une case le rationnel 0/1.
    public BigInteger Denominateur => denominateurMoinsUn + BigInteger.One;

    public Rationnel(BigInteger n, BigInteger d)
    {
        if (d.IsZero)
            throw new DivideByZeroException();
        if (d.Sign < 0)
        {
            n = -n;
            d = -d;
        }
        var pgcd = BigInteger.GreatestCommonDivisor(BigInteger.Abs(n), d);
        Numerateur = n / pgcd;
        denominateurMoinsUn = d / pgcd - BigInteger.One;
    }

    public static implicit operator Rationnel(int n) => new Rationnel(n, 1);
    public static Rationnel operator +(Rationnel a, Rationnel b) =>
        new Rationnel(a.Numerateur * b.Denominateur + b.Numerateur * a.Denominateur,
                      a.Denominateur * b.Denominateur);
    public static Rationnel operator -(Rationnel a, Rationnel b) =>
        new Rationnel(a.Numerateur * b.Denominateur - b.Numerateur * a.Denominateur,
                      a.Denominateur * b.Denominateur);
    public static Rationnel operator *(Rationnel a, Rationnel b) =>
        new Rationnel(a.Numerateur * b.Numerateur, a.Denominateur * b.Denominateur);
    public static Rationnel operator /(Rationnel a, Rationnel b) =>
        new Rationnel(a.Numerateur * b.Denominateur, a.Denominateur * b.Numerateur);

    public override string ToString()
    {
        string n = Numerateur.ToString(CultureInfo.InvariantCulture);
        return Denominateur.IsOne ? n : n + "/" + Denominateur.ToString(CultureInfo.InvariantCulture);
    }
}

// Construire les matrices résultat case par case, sans conversion flottante.
Rationnel[,] Matrice(int lignes, int colonnes, Func<int, int, Rationnel> valeur)
{
    var resultat = new Rationnel[lignes, colonnes];
    for (int i = 0; i < lignes; i++)
    for (int j = 0; j < colonnes; j++)
        resultat[i, j] = valeur(i, j);
    return resultat;
}

Rationnel[,] Mul(Rationnel[,] a, Rationnel[,] b) =>
    Matrice(a.GetLength(0), b.GetLength(1), (i, j) =>
    {
        Rationnel somme = 0;
        for (int k = 0; k < a.GetLength(1); k++)
            somme += a[i, k] * b[k, j];
        return somme;
    });

Rationnel[,] Add(Rationnel[,] a, Rationnel[,] b) =>
    Matrice(a.GetLength(0), a.GetLength(1), (i, j) => a[i, j] + b[i, j]);

bool Egales(Rationnel[,] a, Rationnel[,] b)
{
    if (a.GetLength(0) != b.GetLength(0) || a.GetLength(1) != b.GetLength(1))
        return false;
    for (int i = 0; i < a.GetLength(0); i++)
    for (int j = 0; j < a.GetLength(1); j++)
        if (a[i, j] != b[i, j])
            return false;
    return true;
}

Rationnel[,] Transp(Rationnel[,] a) =>
    Matrice(a.GetLength(1), a.GetLength(0), (i, j) => a[j, i]);
Rationnel[,] Scale(Rationnel[,] a, Rationnel c) =>
    Matrice(a.GetLength(0), a.GetLength(1), (i, j) => c * a[i, j]);
Rationnel[,] Diag(params int[] diagonale) =>
    Matrice(diagonale.Length, diagonale.Length, (i, j) => i == j ? diagonale[i] : 0);

const string section83 = "Éq. 83 — Gram et générateur";
var eta = Diag(4, -1, -1, -1);
var G83 = Scale(eta, new Rationnel(3, 2));
var K1 = new Rationnel[4, 4];
K1[0, 1] = new Rationnel(1, 2);
K1[1, 0] = 2;
var I4 = Diag(1, 1, 1, 1);
var resid = Add(Mul(Transp(K1), G83), Mul(G83, K1));
bool resid0 = Egales(resid, new Rationnel[4, 4]);
VerifierParite(section83, "Résidu d'algèbre nul", resid0, true);
Console.WriteLine($"K1^T G + G K1 = 0 (residu d'algebre) -> {resid0}");

var K1carre = Mul(K1, K1);
bool carre4D = Egales(K1carre, Diag(1, 1, 0, 0));
VerifierParite(section83, "Carré 4D égal à la projection", carre4D, true);
Console.WriteLine($"K1^2 = diag(1, 1, 0, 0) -> {carre4D}");

var Kp = Matrice(2, 2, (i, j) => K1[i, j]);
var I2 = Diag(1, 1);
bool planCarre = Egales(Mul(Kp, Kp), I2);
VerifierParite(section83, "Carré du bloc de plan égal à l'identité", planCarre, true);
Console.WriteLine($"bloc 2x2 K^2 = I -> {planCarre} (le carre identite de l'eq. 83 est celui du plan)");
```

### Lecture du résultat (éq. 83)

**Mesuré** : résidu d'algèbre nul sur $G = \tfrac32\eta$ ; le carré du générateur 4D est la **projection sur le plan** $\mathrm{diag}(1,1,0,0)$ — le « squares to the identity » de l'éq. 83 est le carré du **bloc de plan**, mesuré $I_2$. **Carrefour** : le générateur est exactement celui de Chaînes-11, et la Gram exactement celle de l'éq. 79 — le chapitre 51 travaille avec les objets du chapitre 50, sans en poser de nouveaux.

## L'éq. 84 : l'isométrie de groupe et la coquille

L'exponentielle aux points rationnels de l'hyperbole donne un groupe à un paramètre d'isométries de la forme — le boost appliqué comme carte de cadre aux quatre coordonnées de chaque lecture (le niveau = première puissance de la toile, les trois composantes = fonctionnelle de direction sur la flèche — données de la construction, citées) préserve **l'invariant de coquille** à chaque lecture, le niveau déplacé par la direction et la direction par le niveau dès que l'un est non nul :

$$\exp(\theta K)^\top G \exp(\theta K) = G, \qquad I(\ell, \vec p) = I(\ell', \vec p'). \tag{84}$$

Dans la langue de la littérature (cité par nom) : un boost de vitesse $v$ est la rotation hyperbolique de rapidité $\tanh\eta = v$, et la coquille de masse $E^2 - p^2 = m^2$ est portée à elle-même par tout boost.

```
// Eq. 84 au point hyperbolique rationnel (ch, sh) = (5/4, 3/4).
// exp(th K1) = I + (ch - 1) K1^2 + sh K1  (K1 s'annule sur y, z : identite hors du plan).
const string section84 = "Éq. 84 — Isométrie et coquille";
var ch = new Rationnel(5, 4);
var sh = new Rationnel(3, 4);
var Mexp = Add(Add(I4, Scale(K1carre, ch - 1)), Scale(K1, sh));
var Mtg = Mul(Mul(Transp(Mexp), G83), Mexp);
bool isoGroupe = Egales(Mtg, G83);
VerifierParite(section84, "Condition de groupe en 4D", isoGroupe, true);
Console.WriteLine($"eq.84 : exp^T G exp = G (4D, condition de groupe) -> {isoGroupe}");

Rationnel Coquille(Rationnel[] v) =>
    4 * v[0] * v[0] - v[1] * v[1] - v[2] * v[2] - v[3] * v[3];

Rationnel[] Applique(Rationnel[,] m, Rationnel[] v)
{
    var w = new Rationnel[m.GetLength(0)];
    for (int i = 0; i < w.Length; i++)
    for (int k = 0; k < v.Length; k++)
        w[i] += m[i, k] * v[k];
    return w;
}

Rationnel[][] exemples =
{
    new Rationnel[] { 1, 0, 0, 0 },
    new Rationnel[] { 1, 1, 0, 0 },
    new Rationnel[] { 2, 1, 1, 1 },
    new Rationnel[] { 3, 2, 0, 0 },
    new Rationnel[] { 1, 0, 1, 1 }
};
bool invOk = true;
bool mixOk = true;
foreach (var v in exemples)
{
    var w = Applique(Mexp, v);
    if (Coquille(w) != Coquille(v))
        invOk = false;
    if (v[1] != 0 && w[0] == v[0])
        mixOk = false;
    if (v[0] != 0 && w[1] == v[1])
        mixOk = false;
}
VerifierParite(section84, "Nombre de vecteurs exemples", exemples.Length, 5);
VerifierParite(section84, "Invariance de la coquille sur les exemples", invOk, true);
VerifierParite(section84, "Mélange du niveau et de la direction", mixOk, true);
Console.WriteLine(Invariant($"invariant de coquille preserve sur {exemples.Length} vecteurs exemples -> {invOk}"));
Console.WriteLine($"niveau deplace par la direction et direction par le niveau -> {mixOk}");

var tanh = sh / ch;
var rapportXt = Mexp[1, 0] / Mexp[0, 0];
VerifierParite(section84, "Rapport sh/ch", tanh, new Rationnel(3, 5));
VerifierParite(section84, "Rapport x/t de la première colonne", rapportXt, new Rationnel(6, 5));
Console.WriteLine(Invariant($"tanh = sh/ch = {tanh} ; rapport x/t de la premiere colonne = {rapportXt}"));
// Le facteur 2 entre tanh et x/t est la metrique elle-meme : la coquille 4 t^2 - x^2 a
// pour cone nul x/t = +/- 2 (le (1, +/- 2) mesure aux Chaines-11), et le generateur
// antisymetrique pour G porte ce rapport de part et d'autre : K1[0, 1] = 1/2, K1[1, 0] = 2,
// d'ou premiere colonne (ch, 2 sh) et x/t = 2 tanh.
bool facteurDeux = Coquille(new Rationnel[] { 1, 2, 0, 0 }) == 0 && rapportXt == 2 * tanh;
VerifierParite(section84, "Cône nul et facteur deux", facteurDeux, true);
Console.WriteLine($"facteur 2 = la metrique : (1, 2) est sur le cone nul et x/t = 2 tanh -> {facteurDeux}");
```

### Lecture du résultat (éq. 84)

**Mesuré** : la condition de groupe quadratique tient en 4D ; l'invariant de coquille $4\ell^2 - \vec p\,^2$ est préservé sur les cinq vecteurs d'exemple, et le mélange est réel — le niveau bouge quand la direction est non nulle, la direction bouge quand le niveau est non nul ; et le facteur $2$ entre $\tanh\theta = 3/5$ et $x/t = 6/5$ est la métrique elle-même : le cône nul de $4\ell^2 - \vec p\,^2$ porte $x/t = \pm 2$ (le $(1, \pm 2)$ de Chaînes-11), que le générateur antisymétrique $K_1$ — $1/2$ et $2$ de part et d'autre — transporte. **Cité** : le census applique la porte de boost aux **huit points exacts de l'hyperbole** aux quatre coordonnées de **309 lectures** tabulées, l'invariant évalué avant et après la carte, préservé à chaque lecture — « the mass shell at the tabulated reads is the level set of the invariant ».

## L'éq. 85 et le parting nommé : l'unité manquante

À l'unité debout, les deux ticks sont exacts et leur **rapport au carré deux** ; aux deux alternats admis, il vaut **un**. D'où le parting : les cartes entières préservant la forme lorentzienne sont la **famille de Pell** à l'unité debout, les **cartes de signe seules** aux alternats — le boost issu du rapport de tick **dépend de l'unité**, l'unité une est **l'objet manquant**. Le source le mesure par énumération fraîche aux bornes dix, mille et cent mille :

$$\#\{\text{cartes entieres de } (4, -1)\} = 4 \text{ a toute borne}, \qquad \#\{\text{cartes de mixage}\} = 0. \tag{85}$$

```
// Eq. 85 : cartes entieres de diag(4, -1) aux bornes 10, 1000, 100000.
// Colonnes c1, c2 entieres : c1^T Q c1 = 4, c2^T Q c2 = -1, c1^T Q c2 = 0.
long RacineEntiere(long n)
{
    if (n < 0)
        throw new ArgumentOutOfRangeException(nameof(n));
    if (n < 2)
        return n;

    // Pour m > 0, m <= n/m équivaut à m² <= n, sans risque de débordement.
    long bas = 1;
    long haut = n;
    while (bas < haut)
    {
        long milieu = bas + (haut - bas + 1) / 2;
        if (milieu <= n / milieu)
            bas = milieu;
        else
            haut = milieu - 1;
    }
    return bas;
}

List<(long A, long B)> SolNorm(int q0, int q1, int cible, int borne)
{
    var solutions = new List<(long A, long B)>();
    for (long b = -(long)borne; b <= borne; b++)
    {
        long reste = checked(cible - q1 * b * b);
        if (reste < 0 || reste % q0 != 0)
            continue;
        long a2 = reste / q0;
        long a = RacineEntiere(a2);
        if (a * a != a2 || a > borne)
            continue;
        solutions.Add((a, b));
        if (a != 0)
            solutions.Add((-a, b));
    }
    return solutions;
}

List<((long A, long B) C1, (long A, long B) C2)> Cartes(int q0, int q1, int borne)
{
    // Énumérer une fois chaque ensemble de colonnes, puis tester leur orthogonalité.
    var colonnes1 = SolNorm(q0, q1, q0, borne);
    var colonnes2 = SolNorm(q0, q1, q1, borne);
    var cartes = new List<((long A, long B) C1, (long A, long B) C2)>();
    foreach (var c1 in colonnes1)
    foreach (var c2 in colonnes2)
        if (checked(q0 * c1.A * c2.A + q1 * c1.B * c2.B) == 0)
            cartes.Add((c1, c2));
    return cartes;
}

bool Melange(((long A, long B) C1, (long A, long B) C2) carte) =>
    carte.C1.B != 0 || carte.C2.A != 0;

const string section85 = "Éq. 85 — Cartes de diag(4,-1)";
const int cartes41Attendues = 4;
const int mixages41Attendus = 0;
var comptes41 = new Dictionary<int, (int Total, int Mixage)>();
foreach (int borne in new[] { 10, 1000, 100000 })
{
    var cs = Cartes(4, -1, borne);
    var compte = (Total: cs.Count, Mixage: cs.Count(carte => Melange(carte)));
    comptes41.Add(borne, compte);
    VerifierParite(section85, Invariant($"Nombre de cartes à la borne {borne}"), compte.Total, cartes41Attendues);
    VerifierParite(section85, Invariant($"Nombre de cartes de mixage à la borne {borne}"), compte.Mixage, mixages41Attendus);
    Console.WriteLine(Invariant($"borne {borne} : cartes de (4,-1) = {compte.Total}, de mixage = {compte.Mixage} (attendu {cartes41Attendues} / {mixages41Attendus})"));
}
```

### Lecture du résultat (éq. 85)

**Mesuré** : à chacune des trois bornes du source — dix, mille, cent mille — exactement **quatre cartes** de $\mathrm{diag}(4,-1)$, les cartes de signe, et **aucune** de mixage. La raison se voit dans la norme : $4a^2 - b^2 = -1$ se factorise en $(b-2a)(b+2a) = 1$, qui n'a que les solutions triviales — pas de Pell ici, pas de premier boost.
