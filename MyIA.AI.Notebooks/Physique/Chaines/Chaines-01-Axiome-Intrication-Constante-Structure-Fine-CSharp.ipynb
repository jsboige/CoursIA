## 2. L'axiome et ses cinq définitions

Voici l'unique entrée du modèle, traduite de la Letter :

> **Axiome.** Rien n'existe que des paires finies de systèmes de dimension finie maximalement intriqués, certains partagés, connus seulement en les distillant.

Cinq définitions en déploient les termes (Cité) :

1. **Qudit** : un système quantique de dimension $d \ge 2$ — un système de dimension 1 ne peut pas être intriqué et ne termine donc aucune paire.
2. **Paire (lien)** de type $(d_1, d_2)$ : deux qudits de dimensions $d_1$ et $d_2$ dans l'état maximalement intriqué sur leur support commun de dimension $m = \min(d_1, d_2)$ :
$$|\omega_m\rangle = \frac{1}{\sqrt{m}} \sum_{i=1}^{m} |i\rangle|i\rangle.$$
3. **Partagé** : un qudit qui est un bout de deux paires ou plus — porté comme une copie par paire (façon PEPS), sans jamais violer la monogamie.
4. **Distiller** deux qudits : les projeter sur $P_m = |\omega_m\rangle\langle\omega_m|$ — la généralisation de l'échange d'intrication, ici sans branche d'échec puisque tout état est déjà maximalement intriqué.
5. **Lecture** (*read*) : un acte à un bout d'une paire, qui distille les deux bouts dans une base de ce bout et transmet la base le long du lien. **Le connu** d'un qudit : ce que ses lectures fixent conjointement, sous tout appariement et tout échange.

Commençons par mesurer l'objet le plus basique : l'état de la paire elle-même.

````
#nullable enable
using System;
using System.Globalization;
using System.Linq;
using System.Numerics;

// |omega_m> sur le lien de type (d1, d2) : amplitude 1/sqrt(m) sur chaque |i>|i>, i < m = min(d1, d2).
// Rejeu flottant du carnet Python : le ket est un vecteur de Complex et les matrices reduites sont
// calculees par produits ecrits a la main (la bibliotheque standard n'a pas d'algebre lineaire).
Complex[] OmegaKetC4(int d1, int d2)
{
    int m = Math.Min(d1, d2);
    var v = new Complex[d1 * d2];
    for (int i = 0; i < m; i++)
    {
        v[i * d2 + i] = new Complex(1.0 / Math.Sqrt(m), 0.0);
    }
    return v;
}

Complex[,] ReshapeC4(Complex[] v, int lignes, int colonnes)   // lignes majeures, comme reshape de numpy
{
    var matrice = new Complex[lignes, colonnes];
    for (int i = 0; i < lignes; i++)
    {
        for (int j = 0; j < colonnes; j++)
        {
            matrice[i, j] = v[i * colonnes + j];
        }
    }
    return matrice;
}

Complex[,] TransposeeConjugueeC4(Complex[,] a)   // A^H : transposee conjuguee
{
    int n = a.GetLength(0), p = a.GetLength(1);
    var t = new Complex[p, n];
    for (int i = 0; i < n; i++)
    {
        for (int j = 0; j < p; j++)
        {
            t[j, i] = Complex.Conjugate(a[i, j]);
        }
    }
    return t;
}

Complex[,] ProduitC4(Complex[,] a, Complex[,] b)   // A B, chaque somme cumulee de gauche a droite
{
    int n = a.GetLength(0), k = a.GetLength(1), p = b.GetLength(1);
    var c = new Complex[n, p];
    for (int i = 0; i < n; i++)
    {
        for (int j = 0; j < p; j++)
        {
            Complex s = Complex.Zero;
            for (int r = 0; r < k; r++)
            {
                s += a[i, r] * b[r, j];
            }
            c[i, j] = s;
        }
    }
    return c;
}

Complex[] ProduitVecC4(Complex[,] a, Complex[] v)   // A v, chaque somme cumulee de gauche a droite
{
    int n = a.GetLength(0), k = a.GetLength(1);
    var y = new Complex[n];
    for (int i = 0; i < n; i++)
    {
        Complex s = Complex.Zero;
        for (int r = 0; r < k; r++)
        {
            s += a[i, r] * v[r];
        }
        y[i] = s;
    }
    return y;
}

double NormeC4(Complex[] v)   // sqrt(sum |v_k|^2), termes cumules de gauche a droite
{
    double s = 0.0;
    for (int k = 0; k < v.Length; k++)
    {
        s += v[k].Real * v[k].Real + v[k].Imaginary * v[k].Imaginary;
    }
    return Math.Sqrt(s);
}

// Valeurs propres d'une matrice HERMITIENNE : methode de Jacobi cyclique, ecrite a la main.
// Chaque paire (p, q) est annulee par un changement de base unitaire U^H A U en deux temps :
//   1. une phase diagonale rend A[p, q] reel positif ;
//   2. une rotation reelle d'angle phi dans le plan (p, q), avec tan(2 phi) = 2 A[p, q] / (A[p, p] - A[q, q]).
// Les valeurs propres d'une hermitienne sont reelles ; la sortie est triee par ordre croissant,
// comme numpy.linalg.eigvalsh.
double[] ValeursPropresHermitiennesC4(Complex[,] matrice)
{
    int n = matrice.GetLength(0);
    var a = (Complex[,])matrice.Clone();
    double normeCarree = 0.0;
    for (int i = 0; i < n; i++)
    {
        for (int j = 0; j < n; j++)
        {
            normeCarree += a[i, j].Magnitude * a[i, j].Magnitude;
        }
    }
    for (int balayage = 0; balayage < 100; balayage++)
    {
        double horsDiagCarree = 0.0;
        for (int p = 0; p < n - 1; p++)
        {
            for (int q = p + 1; q < n; q++)
            {
                horsDiagCarree += a[p, q].Magnitude * a[p, q].Magnitude;
                double r = a[p, q].Magnitude;
                if (r == 0.0)
                {
                    continue;
                }
                Complex phase = Complex.Conjugate(a[p, q]) / r;   // de module 1 et d'argument -arg(A[p, q])
                for (int i = 0; i < n; i++)
                {
                    a[i, q] *= phase;
                }
                for (int j = 0; j < n; j++)
                {
                    a[q, j] *= Complex.Conjugate(phase);
                }
                double phi = 0.5 * Math.Atan2(2.0 * r, a[p, p].Real - a[q, q].Real);
                double c = Math.Cos(phi), s = Math.Sin(phi);
                for (int i = 0; i < n; i++)
                {
                    Complex bip = a[i, p], biq = a[i, q];
                    a[i, p] = c * bip + s * biq;
                    a[i, q] = -s * bip + c * biq;
                }
                for (int j = 0; j < n; j++)
                {
                    Complex apj = a[p, j], aqj = a[q, j];
                    a[p, j] = c * apj + s * aqj;
                    a[q, j] = -s * apj + c * aqj;
                }
                a[p, q] = Complex.Zero;
                a[q, p] = Complex.Zero;
                a[p, p] = new Complex(a[p, p].Real, 0.0);
                a[q, q] = new Complex(a[q, q].Real, 0.0);
            }
        }
        if (horsDiagCarree <= normeCarree * 1e-28)
        {
            break;
        }
    }
    var valeurs = new double[n];
    for (int i = 0; i < n; i++)
    {
        valeurs[i] = a[i, i].Real;
    }
    Array.Sort(valeurs);
    return valeurs;
}

// Autotest du solveur sur trois hermitiennes de spectre connu (aucune egalite flottante : tolerance 1e-12).
Complex[,] autotestC4 =   // bloc -sigma_y (valeurs propres -1 et 1) et une direction decouplee de valeur propre 1
{
    { Complex.Zero, Complex.ImaginaryOne, Complex.Zero },
    { -Complex.ImaginaryOne, Complex.Zero, Complex.Zero },
    { Complex.Zero, Complex.Zero, Complex.One },
};
Complex[,] autotest2C4 =
{
    { Complex.One, new Complex(1.0, 1.0) },
    { new Complex(1.0, -1.0), Complex.One },
};
Complex phasePiQuartsC4 = Complex.FromPolarCoordinates(1.0, Math.PI / 4.0);    // e^(i pi/4)
Complex phasePiTiersC4 = Complex.FromPolarCoordinates(1.0, Math.PI / 3.0);     // e^(i pi/3)
Complex[,] autotest3C4 =   // trois niveaux relies en chaine, diagonal nulle : spectre {-sqrt(2), 0, sqrt(2)}
{
    { Complex.Zero, phasePiQuartsC4, Complex.Zero },
    { Complex.Conjugate(phasePiQuartsC4), Complex.Zero, phasePiTiersC4 },
    { Complex.Zero, Complex.Conjugate(phasePiTiersC4), Complex.Zero },
};
double[] spectreAutotestC4 = ValeursPropresHermitiennesC4(autotestC4);
double[] spectreAutotest2C4 = ValeursPropresHermitiennesC4(autotest2C4);
double[] spectreAutotest3C4 = ValeursPropresHermitiennesC4(autotest3C4);
Exige(Math.Abs(spectreAutotestC4[0] + 1.0) < 1e-12 && Math.Abs(spectreAutotestC4[1] - 1.0) < 1e-12
    && Math.Abs(spectreAutotestC4[2] - 1.0) < 1e-12,
    "solveur de valeurs propres : spectre {-1, 1, 1} retrouve, valeur propre double comprise (tolerance 1e-12)");
Exige(Math.Abs(spectreAutotest2C4[0] - (1.0 - Math.Sqrt(2.0))) < 1e-12
    && Math.Abs(spectreAutotest2C4[1] - (1.0 + Math.Sqrt(2.0))) < 1e-12,
    "solveur de valeurs propres : spectre {1 - sqrt(2), 1 + sqrt(2)} retrouve (tolerance 1e-12)");
Exige(Math.Abs(spectreAutotest3C4[0] + Math.Sqrt(2.0)) < 1e-12 && Math.Abs(spectreAutotest3C4[1]) < 1e-12
    && Math.Abs(spectreAutotest3C4[2] - Math.Sqrt(2.0)) < 1e-12,
    "solveur de valeurs propres : spectre {-sqrt(2), 0, sqrt(2)} retrouve, phases complexes comprises (tolerance 1e-12)");

// Affichage a la Python : "1.0", "0.5", "0.0" (jamais "1").
string PythonDoubleC4(double x)
{
    string texte = x.ToString("R", CultureInfo.InvariantCulture);
    if (texte.IndexOfAny(new[] { '.', 'e', 'E', 'N', 'I' }) < 0)
    {
        texte += ".0";
    }
    return texte;
}

// Affichage a la numpy d'un reel : un entier recoit un point, jamais un zero ("0.").
string NumpyDoubleC4(double x)
{
    string texte = x.ToString("R", CultureInfo.InvariantCulture);
    if (texte.IndexOfAny(new[] { 'e', 'E', 'N', 'I' }) < 0 && texte.IndexOf('.') < 0)
    {
        texte += ".";
    }
    return texte;
}

// Affichage d'un tableau a la numpy : les elements sont alignes sur la virgule decimale, separes par une espace.
string NumpyVecteurC4(double[] xs)
{
    string[] elements = xs.Select(NumpyDoubleC4).ToArray();
    int largeur = elements.Max(e => e.Length);
    return "[" + string.Join(" ", elements.Select(e => e.PadRight(largeur))) + "]";
}

// Le lien du candidat : un qutrit (d1 = 3) et un qubit (d2 = 2), support commun m = 2
int d1C4 = 3, d2C4 = 2;
int mC4 = Math.Min(d1C4, d2C4);
Complex[] w2C4 = OmegaKetC4(d1C4, d2C4);
Complex[,] psiC4 = ReshapeC4(w2C4, d1C4, d2C4);
Complex[,] rhoQutritC4 = ProduitC4(psiC4, TransposeeConjugueeC4(psiC4));   // cote qutrit : psi psi^H (3 x 3)
Complex[,] rhoQubitC4 = ProduitC4(TransposeeConjugueeC4(psiC4), psiC4);    // cote qubit : psi^H psi (2 x 2)

Console.WriteLine("|w_2> non nul sur :");
for (int i = 0; i < mC4; i++)
{
    Complex amplitudeC4 = w2C4[i * d2C4 + i];
    Console.WriteLine(Inv($"  |{i}>|{i}>  avec amplitude {amplitudeC4.Real:F6}+{amplitudeC4.Imaginary:F6}j"));
}
double traceC4 = Math.Round(rhoQutritC4[0, 0].Real + rhoQutritC4[1, 1].Real + rhoQutritC4[2, 2].Real, 12);
double[] valeursPropresC4 = ValeursPropresHermitiennesC4(rhoQutritC4);
double[] valeursPropresArrondiesC4 = valeursPropresC4.Select(x => Math.Round(x, 12)).ToArray();
bool qubitEgalIdentiteSurDeuxC4 = true;   // tolerance 1e-9, comme numpy.allclose
for (int i = 0; i < 2; i++)
{
    for (int j = 0; j < 2; j++)
    {
        double attendu = i == j ? 0.5 : 0.0;
        if (Math.Abs(rhoQubitC4[i, j].Real - attendu) > 1e-9 || Math.Abs(rhoQubitC4[i, j].Imaginary) > 1e-9)
        {
            qubitEgalIdentiteSurDeuxC4 = false;
        }
    }
}
Console.WriteLine();
Console.WriteLine("trace rho_qutrit = " + PythonDoubleC4(traceC4));
Console.WriteLine("valeurs propres rho_qutrit = " + NumpyVecteurC4(valeursPropresArrondiesC4));
Console.WriteLine("rho_qubit == I/2 : " + (qubitEgalIdentiteSurDeuxC4 ? "True" : "False"));

// CONTROLE EXACT (entiers) : le ket est n / sqrt(m), ou n porte un 1 par paire (i, i).
// Donc rho_qutrit = (1/m) N N^T et rho_qubit = (1/m) N^T N avec N = reshape(n) : des matrices ENTIERES.
// Les deux conclusions se lisent sur ces matrices ; le flottant ci-dessus n'en est que le rejeu.
int[,] TransposeeEntierC4(int[,] a)
{
    var t = new int[a.GetLength(1), a.GetLength(0)];
    for (int i = 0; i < a.GetLength(0); i++)
    {
        for (int j = 0; j < a.GetLength(1); j++)
        {
            t[j, i] = a[i, j];
        }
    }
    return t;
}

int[,] ProduitEntierC4(int[,] a, int[,] b)
{
    var c = new int[a.GetLength(0), b.GetLength(1)];
    for (int i = 0; i < c.GetLength(0); i++)
    {
        for (int j = 0; j < c.GetLength(1); j++)
        {
            for (int r = 0; r < a.GetLength(1); r++)
            {
                c[i, j] += a[i, r] * b[r, j];
            }
        }
    }
    return c;
}

bool EgalesC4(int[,] a, int[,] b)
{
    for (int i = 0; i < a.GetLength(0); i++)
    {
        for (int j = 0; j < a.GetLength(1); j++)
        {
            if (a[i, j] != b[i, j])
            {
                return false;
            }
        }
    }
    return true;
}

int[,] nC4 = new int[d1C4, d2C4];
for (int i = 0; i < mC4; i++)
{
    nC4[i, i] = 1;
}
int[,] nntC4 = ProduitEntierC4(nC4, TransposeeEntierC4(nC4));   // N N^T : attendu diag(1, 1, 0)
int[,] ntnC4 = ProduitEntierC4(TransposeeEntierC4(nC4), nC4);   // N^T N : attendu I_2
int[,] diagUnUnZeroC4 = { { 1, 0, 0 }, { 0, 1, 0 }, { 0, 0, 0 } };
int[,] identiteDeuxC4 = { { 1, 0 }, { 0, 1 } };
int traceNNTC4 = nntC4[0, 0] + nntC4[1, 1] + nntC4[2, 2];

Exige(Math.Abs(traceC4 - 1.0) < 1e-9, "trace rho_qutrit = 1 (rejeu flottant, tolerance 1e-9)");
Exige(Math.Abs(valeursPropresC4[0]) < 1e-9 && Math.Abs(valeursPropresC4[1] - 0.5) < 1e-9
    && Math.Abs(valeursPropresC4[2] - 0.5) < 1e-9,
    "valeurs propres rho_qutrit = [0, 0.5, 0.5] (rejeu flottant, tolerance 1e-9)");
Exige(qubitEgalIdentiteSurDeuxC4, "rho_qubit == I/2 (rejeu flottant, tolerance 1e-9)");
Exige(EgalesC4(nntC4, diagUnUnZeroC4), "controle exact : N N^T = diag(1, 1, 0), donc rho_qutrit = diag(1/2, 1/2, 0)");
Exige(EgalesC4(ntnC4, identiteDeuxC4), "controle exact : N^T N = I_2, donc rho_qubit = I/2");
Exige(EgalesC4(ProduitEntierC4(nntC4, nntC4), nntC4) && traceNNTC4 == mC4,
    "controle exact : (N N^T)^2 = N N^T et tr(N N^T) = m = 2, donc trace rho_qutrit = 1 et spectre exact {0, 1/2, 1/2}");
````

### Lecture du résultat

Les valeurs propres de la matrice réduite du qutrit sont $\{1/2,\ 1/2,\ 0\}$ : deux degrés de liberté portent toute l'information, le troisième niveau du qutrit reste vide. Le qubit réduit est exactement $I/2$ — la définition même de l'intrication **maximale** sur le support $m = 2$. C'est le point de départ mécanique de toute la construction : chaque lien est un état *pur* maximalement intriqué, et rien dans le modèle n'est jamais moins intriqué.

## 3. Le plafond de lisibilité — pourquoi le type (2, 3)

La question suivante est : **quels types de liens une lecture peut-elle fixer ?** La réponse de la Letter (Cité) passe par les applications positives entre algèbres d'opérateurs. Une relation cohérente entre les deux bouts est une application positive ; les lectures n'en transportent que des **décomposables** (complètement positives, ou complètement positives après transposition). Un type est **lisible** quand toute application positive entre ses bouts est décomposable, **sombre** sinon. Les théorèmes de Størmer (1963), Woronowicz (1976) et Choi (1975) tranchent :

- en $(2,2)$ et $(2,3)$ : toute application positive est décomposable — types **lisibles** ;
- dès que $d_1 d_2 \ge 8$ : il existe des applications positives non décomposables — types **sombres** ;
- et comme aucun produit de deux entiers $\ge 2$ ne vaut 7, le plafond tombe exactement à $d_1 d_2 \le 6$.

D'où la table des types :

````
#nullable enable
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Linq;

// Les neuf types de lien (a, b) pour a, b dans 2..4 ; plafond de lisibilite : a * b <= 6.
var typesC7 = new List<(int a, int b)>();
var lisiblesC7 = new List<(int a, int b)>();
for (int a = 2; a <= 4; a++)
{
    for (int b = 2; b <= 4; b++)
    {
        typesC7.Add((a, b));
    }
}
foreach (var (a, b) in typesC7)
{
    bool lisibleC7 = a * b <= 6;
    Console.WriteLine(Inv($"type ({a},{b}) : produit {a * b,2} -> {(lisibleC7 ? "LISIBLE" : "sombre")}"));
    if (lisibleC7)
    {
        lisiblesC7.Add((a, b));
    }
}

Console.WriteLine();
Console.WriteLine("types lisibles : [" + string.Join(", ", lisiblesC7.Select(t => Inv($"({t.a}, {t.b})"))) + "]");
// bouts inegaux, modulo l'ordre des bouts : chaque type range (min, max), sans doublon
var inegauxC7 = new SortedSet<(int a, int b)>(lisiblesC7.Where(t => t.a != t.b).Select(t => (Math.Min(t.a, t.b), Math.Max(t.a, t.b))));
Console.WriteLine("types lisibles a bouts inegaux, modulo l'ordre des bouts : {"
    + string.Join(", ", inegauxC7.Select(t => Inv($"({t.a}, {t.b})"))) + "}");

// Assertions exactes : sur des entiers, aucune tolerance.
Exige(typesC7.Count == 9, "neuf types de lien (a, b), a et b dans 2..4");
Exige(lisiblesC7.SequenceEqual(new[] { (2, 2), (2, 3), (3, 2) }), "types lisibles = [(2, 2), (2, 3), (3, 2)]");
Exige(inegauxC7.SequenceEqual(new[] { (2, 3) }), "un seul type lisible a bouts inegaux, modulo l'ordre des bouts : (2, 3)");
Exige(typesC7.Count(t => t.a * t.b > 6) == 6 && lisiblesC7.Count(t => t.a == t.b) == 1,
    "six types sombres ; un seul type lisible a bouts egaux : (2, 2)");
````

### Lecture du résultat

Deux types lisibles seulement : $(2,2)$ et $(2,3)$. Le type $(2,3)$ — un qubit et un qutrit — est **le seul avec des bouts inégaux**, donc le seul où une lecture a un « gros bout » où se prendre (la règle du bout : une lecture se prend au bout le plus grand). C'est aussi le seul qui porte une charge centrale agissante — c'est l'objet de la section suivante. Le qualificatif « sombre » pour $d_1 d_2 \ge 8$ vient de la correspondance de Choi–Jamiołkowski : ce que les lectures ne peuvent pas transporter, ce sont les témoins d'intrication à transposition positive, qu'aucune distillation standard n'atteint.

## 4. La charge centrale et le groupe de jauge du Modèle Standard

L'algèbre d'identification d'un lien $(d_1, d_2)$ est $\mathfrak{su}(d_1) \oplus \mathfrak{su}(d_2) \oplus \mathfrak{u}(1)$ — les changements de basis infinitésimaux aux deux bouts. Son centre est engendré par (Cité, éq. (3) de la Letter) :
$$Q = i\,\mathrm{diag}\big(d_2\,\mathbb{1}_{d_1}\ ;\ -d_1\,\mathbb{1}_{d_2}\big),$$
un opérateur sur $\mathbb{C}^{d_1} \oplus \mathbb{C}^{d_2}$. Un élément $\xi = \xi_1 \oplus \xi_2$ agit sur le lien comme $\xi_1 \otimes \mathbb{1} + \mathbb{1} \otimes \xi_2$ — donc $Q$ agit sur le ket de paire comme le scalaire $i(d_2 - d_1)$. En $(2,2)$ ce scalaire est nul ; en $(2,3)$ il vaut $i$. Et la Letter montre que $Q = 6iY$, où $Y$ est le générateur d'hypercharge $Y = \mathrm{diag}(\tfrac12, \tfrac12; -\tfrac13, -\tfrac13, -\tfrac13)$ — l'algèbre est celle de $S(U(3) \times U(2)) \cong (SU(3) \times SU(2) \times U(1))/\mathbb{Z}_6$, **la forme globale du groupe de jauge du Modèle Standard telle qu'elle se réalise dans $SU(5)$**. Vérifions ces deux affirmations numériquement :

````
#nullable enable
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Linq;
using System.Numerics;

// La charge centrale : Q = i diag(d2 1_{d1} ; -d1 1_{d2}) sur C^{d1} (+) C^{d2} (a l'echelle pres).
// OmegaKetC4, ProduitVecC4, NormeC4 et NumpyDoubleC4 : les outils de la cellule « axiome ».
Complex[,] ChargeQC10(int d1, int d2)
{
    var q = new Complex[d1 + d2, d1 + d2];
    for (int i = 0; i < d1; i++)
    {
        q[i, i] = new Complex(0.0, d2);
    }
    for (int j = 0; j < d2; j++)
    {
        q[d1 + j, d1 + j] = new Complex(0.0, -d1);
    }
    return q;
}

Complex[,] IdentiteC10(int n)
{
    var identite = new Complex[n, n];
    for (int i = 0; i < n; i++)
    {
        identite[i, i] = Complex.One;
    }
    return identite;
}

Complex[,] MultipleC10(Complex[,] a, Complex lambda)   // lambda A
{
    var produit = new Complex[a.GetLength(0), a.GetLength(1)];
    for (int i = 0; i < a.GetLength(0); i++)
    {
        for (int j = 0; j < a.GetLength(1); j++)
        {
            produit[i, j] = lambda * a[i, j];
        }
    }
    return produit;
}

Complex[,] SommeC10(Complex[,] a, Complex[,] b)
{
    var somme = new Complex[a.GetLength(0), a.GetLength(1)];
    for (int i = 0; i < a.GetLength(0); i++)
    {
        for (int j = 0; j < a.GetLength(1); j++)
        {
            somme[i, j] = a[i, j] + b[i, j];
        }
    }
    return somme;
}

Complex[,] KroneckerC10(Complex[,] a, Complex[,] b)   // produit de Kronecker, comme numpy.kron
{
    int n = a.GetLength(0), p = a.GetLength(1), q = b.GetLength(0), r = b.GetLength(1);
    var produit = new Complex[n * q, p * r];
    for (int i = 0; i < n; i++)
    {
        for (int j = 0; j < p; j++)
        {
            for (int k = 0; k < q; k++)
            {
                for (int l = 0; l < r; l++)
                {
                    produit[i * q + k, j * r + l] = a[i, j] * b[k, l];
                }
            }
        }
    }
    return produit;
}

Complex[,] BlocPleinC10(int n, Complex valeur)   // matrice carree dont TOUS les coefficients valent valeur
{
    var bloc = new Complex[n, n];
    for (int i = 0; i < n; i++)
    {
        for (int j = 0; j < n; j++)
        {
            bloc[i, j] = valeur;
        }
    }
    return bloc;
}

// Affichage a la Python « .1e » : 0.0e+00
string PythonSci1C10(double x)
{
    if (x == 0.0)
    {
        return "0.0e+00";
    }
    int exposant = (int)Math.Floor(Math.Log10(Math.Abs(x)));
    double mantisse = x / Math.Pow(10.0, exposant);
    string texte = mantisse.ToString("F1", CultureInfo.InvariantCulture);
    if (Math.Abs(double.Parse(texte, CultureInfo.InvariantCulture)) >= 10.0)
    {
        exposant += 1;
        texte = (mantisse / 10.0).ToString("F1", CultureInfo.InvariantCulture);
    }
    return texte + "e" + (exposant < 0 ? "-" : "+") + Math.Abs(exposant).ToString("00", CultureInfo.InvariantCulture);
}

// Affichage a la numpy d'un complexe : 0.+1.j (partie reelle, signe, module de la partie imaginaire, puis j)
string NumpyComplexeC10(Complex z)
{
    double im = z.Imaginary;
    return NumpyDoubleC4(z.Real) + (double.IsNegative(im) ? "-" : "+") + NumpyDoubleC4(Math.Abs(im)) + "j";
}

// Action de jauge sur |omega_m> pour les trois types lisibles : attendu i (b - a) |omega_m>.
// Le ket ENTIER n (un 1 par paire (i, i)) sert de controle exact : coefficients entiers, aucune tolerance.
var residusC10 = new List<(int a, int b, double ResiduFlottant, bool Exact)>();
foreach (var (a, b) in new[] { (2, 2), (2, 3), (3, 2) })
{
    Complex[,] xi1C10 = MultipleC10(IdentiteC10(a), new Complex(0.0, b));    // i b 1_a
    Complex[,] xi2C10 = MultipleC10(IdentiteC10(b), new Complex(0.0, -a));   // -i a 1_b
    Complex[,] actionC10 = SommeC10(KroneckerC10(xi1C10, IdentiteC10(b)), KroneckerC10(IdentiteC10(a), xi2C10));
    Complex[] wC10 = OmegaKetC4(a, b);
    Complex[] residuC10 = ProduitVecC4(actionC10, wC10);
    for (int k = 0; k < residuC10.Length; k++)
    {
        residuC10[k] -= new Complex(0.0, b - a) * wC10[k];
    }
    // controle exact : le meme calcul sur le ket entier n
    Complex[] nC10 = new Complex[a * b];
    for (int i = 0; i < Math.Min(a, b); i++)
    {
        nC10[i * b + i] = Complex.One;
    }
    bool exactC10 = ProduitVecC4(actionC10, nC10).SequenceEqual(nC10.Select(x => new Complex(0.0, b - a) * x));
    residusC10.Add((a, b, NormeC4(residuC10), exactC10));
    Console.WriteLine(Inv($"({a},{b}) : Q |w> = i(b-a) |w> = {(b - a < 0 ? "-" : "+")}{Math.Abs(b - a)}i |w>   (residu {PythonSci1C10(residusC10[^1].ResiduFlottant)})"));
}
foreach ((int a, int b, double residuFlottant, bool exact) in residusC10)
{
    Exige(residuFlottant < 1e-9, Inv($"({a},{b}) : Q |w> = i(b-a) |w>, residu nul (rejeu flottant, tolerance 1e-9)"));
    Exige(exact, Inv($"({a},{b}) : Q n = i(b-a) n sur le ket ENTIER n, egalite exacte (entiers de Gauss)"));
}

// Temoin negatif : les blocs PLEINS ne sont pas l'action propre. Le residu sur le vecteur complet
// reste non nul, alors que le rapport (plein w)[support] / w[support] valait exactement l'action propre.
int temoinAC10 = 2, temoinBC10 = 3;
Complex[,] pleinC10 = SommeC10(
    KroneckerC10(BlocPleinC10(temoinAC10, new Complex(0.0, temoinBC10)), IdentiteC10(temoinBC10)),
    KroneckerC10(IdentiteC10(temoinAC10), BlocPleinC10(temoinBC10, new Complex(0.0, -temoinAC10))));
Complex[] wTemoinC10 = OmegaKetC4(temoinAC10, temoinBC10);
Complex[] yTemoinC10 = ProduitVecC4(pleinC10, wTemoinC10);
Complex[] residuTemoinC10 = yTemoinC10.Select((y, k) => y - new Complex(0.0, temoinBC10 - temoinAC10) * wTemoinC10[k]).ToArray();
double residuPleinC10 = NormeC4(residuTemoinC10);
var rapportsC10 = new List<Complex>();
for (int k = 0; k < wTemoinC10.Length; k++)
{
    if (wTemoinC10[k].Magnitude > 1e-12)
    {
        rapportsC10.Add(yTemoinC10[k] / wTemoinC10[k]);
    }
}
string rapportsTexteC10 = "[" + string.Join(" ", rapportsC10.Select(z =>
    NumpyComplexeC10(new Complex(Math.Round(z.Real, 12), Math.Round(z.Imaginary, 12))))) + "]";
Console.WriteLine(Inv($"temoin negatif (2,3), blocs pleins : residu sur le vecteur complet = {residuPleinC10:F16} (= sqrt(5))"));
Console.WriteLine("  or le rapport sur le support valait " + rapportsTexteC10 + " -> faux positif");
// le nombre affiche est le rejeu flottant du carnet Python (norme du vecteur residu, cumulee de gauche a droite) ;
// il vaut sqrt(5) aux arrondis pres. Le temoin EXACT, sans arrondi, suit : sur le ket entier, norme carree 10.

// CONTROLE EXACT du temoin, sur le ket entier n (un 1 par paire (i, i), donc ici les indices 0 et 4).
// Le bloc plein y renvoie i (1, 1, -2, 1, 1, -2) : SUR LE SUPPORT de n, c'est exactement i (b - a) fois n,
// et le rapport sur le support seul annonce un vecteur propre. Mais hors du support la reponse n'est pas
// nulle : le residu i (0, 1, -2, 1, 0, -2) a pour norme carree 1 + 4 + 1 + 4 = 10, soit 10 / m = 5
// pour le ket normalise (m = 2). C'est le vecteur COMPLET qui tranche : faux positif.
Complex[] nTemoinC10 = new Complex[temoinAC10 * temoinBC10];
for (int i = 0; i < Math.Min(temoinAC10, temoinBC10); i++)
{
    nTemoinC10[i * temoinBC10 + i] = Complex.One;
}
Complex[] reponsePleinC10 = ProduitVecC4(pleinC10, nTemoinC10);
Complex[] residuEntierC10 = reponsePleinC10.Select((y, k) => y - new Complex(0.0, temoinBC10 - temoinAC10) * nTemoinC10[k]).ToArray();
Complex[] attendueReponseC10 = { new Complex(0.0, 1.0), new Complex(0.0, 1.0), new Complex(0.0, -2.0),
                                 new Complex(0.0, 1.0), new Complex(0.0, 1.0), new Complex(0.0, -2.0) };
Complex[] attenduResiduC10 = { Complex.Zero, new Complex(0.0, 1.0), new Complex(0.0, -2.0),
                               new Complex(0.0, 1.0), Complex.Zero, new Complex(0.0, -2.0) };
int normeCarreeEntiereC10 = residuEntierC10.Select(z => (int)(z.Imaginary * z.Imaginary) + (int)(z.Real * z.Real)).Sum();
bool egalSurLeSupportC10 = true;
for (int k = 0; k < reponsePleinC10.Length; k++)
{
    if (nTemoinC10[k] != Complex.Zero && reponsePleinC10[k] != new Complex(0.0, temoinBC10 - temoinAC10) * nTemoinC10[k])
    {
        egalSurLeSupportC10 = false;
    }
}
Exige(reponsePleinC10.SequenceEqual(attendueReponseC10),
    "controle exact : le bloc plein renvoie i (1, 1, -2, 1, 1, -2) sur le ket entier n");
Exige(egalSurLeSupportC10,
    "controle exact : SUR LE SUPPORT du ket, le bloc plein renvoie i (b - a) fois le ket : le rapport sur le support seul annonce un faux vecteur propre");
Exige(residuEntierC10.SequenceEqual(attenduResiduC10) && normeCarreeEntiereC10 == 10,
    "controle exact : residu du bloc plein sur le vecteur complet = i (0, 1, -2, 1, 0, -2), norme carree 10, soit 5 pour le ket normalise : le bloc plein n'est PAS l'action");
Exige(Math.Abs(residuPleinC10 * residuPleinC10 - 5.0) < 1e-12, "rejeu flottant : norme carree du residu = 5 (tolerance 1e-12)");

// Q(2, 3) = 6 i Y, avec Y = diag(1/2, 1/2, -1/3, -1/3, -1/3) : le lien lisible porte la charge de l'hypercharge.
double[] yDiagC10 = { 0.5, 0.5, -1.0 / 3.0, -1.0 / 3.0, -1.0 / 3.0 };
Complex[,] q23C10 = ChargeQC10(2, 3);
bool qEgalSixiYC10 = true;   // tolerance 1e-9, comme numpy.allclose
for (int i = 0; i < 5; i++)
{
    for (int j = 0; j < 5; j++)
    {
        Complex attendu = i == j ? new Complex(0.0, 6.0) * new Complex(yDiagC10[i], 0.0) : Complex.Zero;
        if (Math.Abs(q23C10[i, j].Real - attendu.Real) > 1e-9 || Math.Abs(q23C10[i, j].Imaginary - attendu.Imaginary) > 1e-9)
        {
            qEgalSixiYC10 = false;
        }
    }
}
Console.WriteLine("Q(2,3) == 6 i Y : " + (qEgalSixiYC10 ? "True" : "False"));
Exige(qEgalSixiYC10, "Q(2,3) == 6 i Y (rejeu flottant, tolerance 1e-9)");
// CONTROLE EXACT : 6 Y est diagonal a coefficients ENTIERS (3, 3, -2, -2, -2) : la diagonale entiere de Q.
(int Numerateur, int Denominateur)[] yRationnelC10 = { (1, 2), (1, 2), (-1, 3), (-1, 3), (-1, 3) };
bool qEgalSixiYExactC10 = true;
for (int k = 0; k < 5; k++)
{
    int sixFoisC10 = 6 * yRationnelC10[k].Numerateur;
    if (sixFoisC10 % yRationnelC10[k].Denominateur != 0)
    {
        qEgalSixiYExactC10 = false;
    }
    else
    {
        int coefficientC10 = k < 2 ? 3 : -2;   // d2 = 3 sur les deux premieres entrees, -d1 = -2 sur les trois autres
        if (sixFoisC10 / yRationnelC10[k].Denominateur != coefficientC10 || q23C10[k, k].Imaginary != coefficientC10)
        {
            qEgalSixiYExactC10 = false;
        }
    }
}
Exige(qEgalSixiYExactC10 && q23C10.Cast<Complex>().Where((z, index) => index % 6 != 0).All(z => z == Complex.Zero)
    && q23C10.Cast<Complex>().All(z => z.Real == 0.0),
    "controle exact : 6 Y = diag(3, 3, -2, -2, -2) en rationnels exacts, diagonale entiere de Q(2, 3), hors diagonale nulle");

// Dimension de l'algebre de jauge : su(3) (+) su(2) (+) u(1)
int dimensionAlgebreC10 = (3 * 3 - 1) + (2 * 2 - 1) + 1;
Console.WriteLine("dimension de l'algebre su(3) (+) su(2) (+) u(1) : " + dimensionAlgebreC10.ToString(CultureInfo.InvariantCulture));
Exige(dimensionAlgebreC10 == 12, "dim su(3) + dim su(2) + dim u(1) = 8 + 3 + 1 = 12");
````

## 5. L'objet — un qutrit partagé, trois liens-qubits, deux lus

Reste à dire **quelle configuration est la fonction à deux points** du lien. Dans le modèle, les deux insertions sont deux lectures du lien, et l'état est un troisième lien de même type au même qudit partagé, laissé non lu (le lire serait une troisième insertion). Le théorème d'unicité de la Letter (Cité) :

> **Théorème 1 (Unicité).** Parmi les configurations à un qudit partagé dont les liens lus sont lisibles, exactement une est une fonction à deux points, est lue au bout qutrit, et porte une charge centrale agissante sur chaque lien : un qutrit partagé avec trois liens vers des qubits, deux d'entre eux lus.

La Letter l'appelle **l'objet** :

```mermaid
flowchart LR
    Q(("Q : qutrit d=3<br/>partage"))
    q1(("q1 : qubit d=2<br/>lecture 1"))
    q2(("q2 : qubit d=2<br/>lecture 2"))
    f(("f : qubit d=2<br/>lien NON lu = l'etat"))

    Q ---|"lien lu 2"| q2
    Q -.-|"lien non lu : porte l'etat omega_2"| f
```

La règle du bout impose les lectures au bout qutrit : une lecture au bout qubit n'atteindrait le qutrit que sur le support de dimension 2, manquant sa troisième direction. La **dimension du relevé** suit des deux seules dimensions du lien (Cité, éq. (4)) : la partie sans trace du lien non lu donne $(3^2-1)(2^2-1) = 24$ directions, et l'espace d'opérateurs complet du lien à chaque lecture donne $(3\cdot2)^2 = 36$ :

````
#nullable enable
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Linq;
using System.Numerics;

// L'objet : N = (3^2 - 1)(2^2 - 1) * (3*2)^2 * (3*2)^2, en entiers de taille arbitraire (BigInteger).
BigInteger lienNonLuC13 = BigInteger.Pow(3, 2) - 1;                 // (3^2 - 1)
lienNonLuC13 *= BigInteger.Pow(2, 2) - 1;                           // * (2^2 - 1) : trace nulle sur chaque bout
BigInteger lienLuC13 = BigInteger.Pow(3 * 2, 2);                    // espace complet d'un lien lu : (3 * 2)^2
BigInteger nObjetC13 = lienNonLuC13 * lienLuC13 * lienLuC13;
Console.WriteLine("lien non lu, partie sans trace : (3^2-1)(2^2-1) = " + lienNonLuC13.ToString(CultureInfo.InvariantCulture));
Console.WriteLine("chaque lien lu, espace complet  : (3*2)^2       = " + lienLuC13.ToString(CultureInfo.InvariantCulture));
Console.WriteLine("N = 24 * 36 * 36 = " + nObjetC13.ToString(CultureInfo.InvariantCulture));

// les lettres : bases sans trace des observables de chaque bout
var slotsC13 = new List<(string Nom, int NbLettres)>
{
    ("Q1 (copie qutrit, lecture 1)", 3 * 3 - 1),
    ("q1 (qubit, lecture 1)", 2 * 2 - 1),
    ("Q2 (copie qutrit, lecture 2)", 3 * 3 - 1),
    ("q2 (qubit, lecture 2)", 2 * 2 - 1),
};
string slotsTexteC13 = "{" + string.Join(", ", slotsC13.Select(s => Inv($"'{s.Nom}': {s.NbLettres}"))) + "}";
string valeursTexteC13 = "[" + string.Join(", ", slotsC13.Select(s => s.NbLettres.ToString(CultureInfo.InvariantCulture))) + "]";
Console.WriteLine("slots (nb de lettres) : " + slotsTexteC13 + " -> " + valeursTexteC13);

// Assertions exactes : sur des entiers, aucune tolerance.
Exige(lienNonLuC13 == 24 && lienLuC13 == 36, "lien non lu = (3^2 - 1)(2^2 - 1) = 24 ; lien lu = (3 * 2)^2 = 36");
Exige(nObjetC13 == 31104, "N = 24 * 36 * 36 = 31104");
Exige(slotsC13.Select(s => s.NbLettres).SequenceEqual(new[] { 8, 3, 8, 3 }), "slots : [8, 3, 8, 3] lettres");
````

## 6. Les 172 enregistrements — la table saisie et contrôlée

La construction (Cité ; sa sortie, énumérée par le carnet Python, est **saisie** ci-dessous et **contrôlée** par notre code — jamais recalculée) :

- **169 enregistrements produits** : une lettre à chacun des $\binom{4}{2} = 6$ couples de slots distincts, soit $8\cdot8 + 3\cdot3 + 4\times(8\cdot3) = 64 + 9 + 96 = 169$ ;

$$k_r = 3^{[X_r = -1]}\, 2^{[Y_r = -1]}, \qquad \text{poids} = \sqrt{k_r}.$$

### Les secteurs et la loi de poids

Chaque enregistrement est un vecteur propre de l'échange des lectures $X$ ; l'échange des liens $Y$ agit sur ceux qu'il déplace (les produits à une lettre à chaque bout, et les commutateurs). La règle de survie : un enregistrement **survit** à un échange quand celui-ci le retourne multiplié par son propre signe ($+1$ pour un produit, le signe d'ordre $-1$ pour un commutateur). On peut maintenant compter les quatre secteurs et appliquer la loi de poids — c'est la table I de la Letter, dont les quatre comptes sont recalculés ici depuis la table saisie :

````
#nullable enable
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Linq;

// Les signes (X, Y) de chaque enregistrement, puis les secteurs et les coefficients.
// Port exact de la fonction signes du carnet Python : les regles (deux echanges, survie a son propre signe,
// loi de poids) sont celles de la Letter ; l'entree est la table saisie dans la cellule precedente.
// Un terme : une liste triee de lettres (slot, lettre) et un coefficient. Un enregistrement : ses termes, dans l'ordre de saisie.
readonly struct TermeC18
{
    public TermeC18((int Slot, int Lettre)[] lettres, int coefficient)
    {
        Lettres = lettres;
        Coefficient = coefficient;
    }

    public (int Slot, int Lettre)[] Lettres { get; }
    public int Coefficient { get; }
}

// le qutrit partage est UN qudit : ses deux lectures portent la meme etiquette Q ; les qubits, q1 et q2
var quditC18 = new Dictionary<int, string> { [1] = "Q", [3] = "Q", [2] = "q1", [4] = "q2" };
// l'echange des liens laisse Q et permute q1 et q2
var echangeLiensC18 = new Dictionary<string, string> { ["Q"] = "Q", ["q1"] = "q2", ["q2"] = "q1" };
// l'echange des lectures : slot 1 <-> 3, 2 <-> 4
var echangeSlotC18 = new Dictionary<int, int> { [1] = 3, [3] = 1, [2] = 4, [4] = 2 };

// l'echange des lectures sur une cle : chaque lettre change de slot, la cle est triee
(int Slot, int Lettre)[] CleEchangeeC18((int Slot, int Lettre)[] cle)
{
    var echangee = new (int Slot, int Lettre)[cle.Length];
    for (int k = 0; k < cle.Length; k++)
    {
        echangee[k] = (echangeSlotC18[cle[k].Slot], cle[k].Lettre);
    }
    Array.Sort(echangee);
    return echangee;
}

TermeC18[] EchangeC18(TermeC18[] enregistrement)
{
    var image = new TermeC18[enregistrement.Length];
    for (int k = 0; k < enregistrement.Length; k++)
    {
        image[k] = new TermeC18(CleEchangeeC18(enregistrement[k].Lettres), enregistrement[k].Coefficient);
    }
    return image;
}

// egalite de deux enregistrements, a l'ordre des termes pres (l'egalite des dictionnaires en Python)
bool MemeEnregistrementC18(TermeC18[] a, TermeC18[] b)
{
    if (a.Length != b.Length)
    {
        return false;
    }
    foreach (TermeC18 terme in a)
    {
        bool present = false;
        foreach (TermeC18 autre in b)
        {
            if (autre.Coefficient == terme.Coefficient && autre.Lettres.SequenceEqual(terme.Lettres))
            {
                present = true;
                break;
            }
        }
        if (!present)
        {
            return false;
        }
    }
    return true;
}

TermeC18[] MultiplieC18(TermeC18[] enregistrement, int facteur)   // chaque coefficient est multiplie par facteur
{
    var resultat = new TermeC18[enregistrement.Length];
    for (int k = 0; k < enregistrement.Length; k++)
    {
        resultat[k] = new TermeC18(enregistrement[k].Lettres, facteur * enregistrement[k].Coefficient);
    }
    return resultat;
}

// parite sous l'echange des lectures : +1 (invariant), -1 (change de signe), 0 (ni l'un ni l'autre)
int PariteC18(TermeC18[] enregistrement)
{
    TermeC18[] image = EchangeC18(enregistrement);
    if (MemeEnregistrementC18(image, enregistrement))
    {
        return +1;
    }
    if (MemeEnregistrementC18(image, MultiplieC18(enregistrement, -1)))
    {
        return -1;
    }
    return 0;
}

// cle canonique d'un enregistrement (termes tries) : l'index de la table de reinsertion
string CleEnregistrementC18(TermeC18[] enregistrement)
{
    return string.Join(";", enregistrement
        .Select(t => t.Coefficient.ToString(CultureInfo.InvariantCulture) + "@"
            + string.Join(".", t.Lettres.Select(l => l.Slot.ToString(CultureInfo.InvariantCulture) + "."
                + l.Lettre.ToString(CultureInfo.InvariantCulture))))
        .OrderBy(fragment => fragment, StringComparer.Ordinal));
}

// egalite de deux ensembles d'etiquettes (sans ordre ni doublon)
bool MemeEnsembleC18(IEnumerable<string> a, IEnumerable<string> b)
{
    return a.Distinct().OrderBy(e => e, StringComparer.Ordinal).SequenceEqual(b.Distinct().OrderBy(e => e, StringComparer.Ordinal));
}

// X et Y d'un enregistrement, via les deux echanges et la survie a son propre signe.
(int X, int Y) SignesC18(int genre, int memeGenre, TermeC18[] enregistrement)
{
    int propre = genre == 3 ? -1 : +1;
    TermeC18[] cible = MultiplieC18(enregistrement, propre);
    TermeC18[]? imageX = null;
    if (genre == 3)
    {
        imageX = reinjectionC18.TryGetValue(CleEnregistrementC18(enregistrement), out TermeC18[]? image) ? image : null;
    }
    else if (memeGenre == 1)
    {
        imageX = EchangeC18(enregistrement);   // ses deux lectures sont du meme type : on les echange
    }
    TermeC18 premier = enregistrement[0];   // premier terme, dans l'ordre de saisie
    string[] etiquettes = premier.Lettres.Select(l => quditC18[l.Slot]).ToArray();
    bool deplace = !MemeEnsembleC18(etiquettes.Select(e => echangeLiensC18[e]), etiquettes);
    TermeC18[]? imageY = deplace ? EchangeC18(enregistrement) : null;
    int x = imageX is not null && MemeEnregistrementC18(imageX, cible) ? -1 : +1;
    int y = imageY is not null && MemeEnregistrementC18(imageY, cible) ? -1 : +1;
    return (x, y);
}

// Lecture de la table saisie : chaque ligne entiere devient un enregistrement structure.
TermeC18[] LireEnregistrementC18(int[] ligne)
{
    int k = 3;
    var termes = new List<TermeC18>();
    for (int t = 0; t < ligne[2]; t++)
    {
        int nbLettres = ligne[k++];
        var lettres = new (int Slot, int Lettre)[nbLettres];
        for (int u = 0; u < nbLettres; u++)
        {
            lettres[u] = (ligne[k], ligne[k + 1]);
            k += 2;
        }
        termes.Add(new TermeC18(lettres, ligne[k]));
        k++;
    }
    if (k != ligne.Length)
    {
        throw new InvalidOperationException("ECART : ligne de la table saisie mal formee");
    }
    return termes.ToArray();
}

var tableC18 = new (int Genre, int MemeGenre, TermeC18[] Enregistrement)[enregistrementsC16.Length];
for (int r = 0; r < enregistrementsC16.Length; r++)
{
    tableC18[r] = (enregistrementsC16[r][0], enregistrementsC16[r][1], LireEnregistrementC18(enregistrementsC16[r]));
}

// Table de reinsertion des commutateurs (SAISIE, d'apres la sortie du carnet Python ; non recalculee ici) :
// a chaque commutateur de la table, l'image de signe -1 sous l'echange des lectures.
var reinjectionC18 = new Dictionary<string, TermeC18[]>
{
    [CleEnregistrementC18(new[] { new TermeC18(new[] { (2, 2) }, -2), new TermeC18(new[] { (4, 2) }, +2) })] =
        new[] { new TermeC18(new[] { (2, 2) }, +2), new TermeC18(new[] { (4, 2) }, -2) },
    [CleEnregistrementC18(new[] { new TermeC18(new[] { (2, 1) }, -2), new TermeC18(new[] { (4, 1) }, +2) })] =
        new[] { new TermeC18(new[] { (2, 1) }, +2), new TermeC18(new[] { (4, 1) }, -2) },
    [CleEnregistrementC18(new[] { new TermeC18(new[] { (2, 0) }, -2), new TermeC18(new[] { (4, 0) }, +2) })] =
        new[] { new TermeC18(new[] { (2, 0) }, +2), new TermeC18(new[] { (4, 0) }, -2) },
};

int sansPariteC18 = tableC18.Count(t => PariteC18(t.Enregistrement) == 0);
Exige(sansPariteC18 == 0, "chaque enregistrement est un vecteur propre de l'echange (parite +1 ou -1)");
Exige(tableC18.Where(t => t.Genre == 3).All(t => reinjectionC18.TryGetValue(CleEnregistrementC18(t.Enregistrement), out TermeC18[]? image)
        && MemeEnregistrementC18(image, MultiplieC18(t.Enregistrement, -1))),
    "les trois commutateurs de la table ont chacun une image saisie, egale a l'oppose de l'enregistrement");

var secteursC18 = new Dictionary<(int X, int Y), int>();
var coefficientsC18 = new Dictionary<int, int>();
foreach ((int genre, int memeGenre, TermeC18[] enregistrement) in tableC18)
{
    (int x, int y) = SignesC18(genre, memeGenre, enregistrement);
    int k = (x == -1 ? 3 : 1) * (y == -1 ? 2 : 1);   // loi de poids : k = 3^[X = -1] 2^[Y = -1]
    secteursC18[(x, y)] = secteursC18.TryGetValue((x, y), out int nbSecteur) ? nbSecteur + 1 : 1;
    coefficientsC18[k] = coefficientsC18.TryGetValue(k, out int nbPoids) ? nbPoids + 1 : 1;
}
string secteursTexteC18 = "{" + string.Join(", ", secteursC18.OrderBy(kv => kv.Key.X).ThenBy(kv => kv.Key.Y)
    .Select(kv => Inv($"({kv.Key.X}, {kv.Key.Y}): {kv.Value}"))) + "}";
string coefficientsTexteC18 = "{" + string.Join(", ", coefficientsC18.OrderBy(kv => kv.Key)
    .Select(kv => Inv($"{kv.Key}: {kv.Value}"))) + "}";
Console.WriteLine("secteurs (X, Y) -> nombre : " + secteursTexteC18);
Console.WriteLine("coefficients par poids  k -> nombre : " + coefficientsTexteC18);

// Assertions exactes : sur des entiers, aucune tolerance.
Exige(secteursC18.Values.Sum() == 172, "total 172");
Exige(secteursC18[(+1, +1)] == 79 && secteursC18[(+1, -1)] == 48 && secteursC18[(-1, +1)] == 42 && secteursC18[(-1, -1)] == 3,
    "secteurs (+1,+1) : 79, (+1,-1) : 48, (-1,+1) : 42, (-1,-1) : 3");
Exige(coefficientsC18.Count == 4 && coefficientsC18[1] == 79 && coefficientsC18[2] == 48 && coefficientsC18[3] == 42
    && coefficientsC18[6] == 3,
    "coefficients par poids : 1 -> 79, 2 -> 48, 3 -> 42, 6 -> 3");
Console.WriteLine("total 172, secteurs 79/48/42/3 : conforme a la table I de la Letter");
````

````
#nullable enable
using System;
using System.Globalization;
using System.Linq;
using System.Numerics;

// La forme fermee, a 60 chiffres, depuis NOS coefficients (les secteurs recomptes en cellule precedente).
// System.Decimal n'a que 28-29 chiffres : on calcule en BigInteger a ECHELLE FIXE, declaree ici.
//   ECHELLE             : 72 decimales (unite 10^-72)
//   CHIFFRES DE GARDE   : 10 (les decimales 63 a 72 ne servent qu'a l'arrondi et au temoin d'ambiguite)
//   AFFICHAGE           : 60 chiffres significatifs -- 62 decimales pour alpha_cand (premier chiffre en 10^-3),
//                         57 decimales pour son inverse (trois chiffres avant la virgule)
int echelleC20 = 72;
int decimalesAlphaC20 = 62;
int decimalesInverseC20 = 57;
int echelleInverseC20 = decimalesInverseC20;   // l'inverse arrondi vit a l'echelle 10^-57

// racine carree entiere : le plus grand r tel que r^2 <= n (methode de Newton, en BigInteger)
BigInteger RacineEntiereC20(BigInteger n)
{
    if (n.Sign < 0)
    {
        throw new ArgumentOutOfRangeException(nameof(n));
    }
    if (n.Sign == 0)
    {
        return BigInteger.Zero;
    }
    // depart au-dessus de sqrt(n) : 10^(chiffres/2 + 1) ; Newton decroit vers le plancher
    int chiffres = n.ToString(CultureInfo.InvariantCulture).Length;
    BigInteger r = BigInteger.Pow(10, chiffres / 2 + 1);
    while (true)
    {
        BigInteger suivant = (r + n / r) / 2;
        if (suivant >= r)
        {
            return r;
        }
        r = suivant;
    }
}

// controle d'une racine tronquee : le plus grand s tel que (s 10^-72)^2 <= k
bool PlancherRacineC20(BigInteger racine, int k)
{
    BigInteger cible = k * BigInteger.Pow(10, 2 * echelleC20);
    return racine * racine <= cible && (racine + 1) * (racine + 1) > cible;
}

BigInteger[] racinesC20 = new[] { 2, 3, 6 }.Select(k => RacineEntiereC20(k * BigInteger.Pow(10, 2 * echelleC20))).ToArray();
BigInteger dixEchelleC20 = BigInteger.Pow(10, echelleC20);
BigInteger numerateurC20 = coefficientsC18[1] * dixEchelleC20 + coefficientsC18[2] * racinesC20[0]
    + coefficientsC18[3] * racinesC20[1] + coefficientsC18[6] * racinesC20[2];
BigInteger alphaC20 = numerateurC20 / nObjetC13;                                          // par defaut, a l'echelle
BigInteger inverseC20 = BigInteger.Pow(10, 2 * echelleC20) * nObjetC13 / numerateurC20;  // 10^72 / alpha_cand, par defaut
Exige(PlancherRacineC20(racinesC20[0], 2) && PlancherRacineC20(racinesC20[1], 3) && PlancherRacineC20(racinesC20[2], 6),
    "racines entieres de sqrt(2), sqrt(3), sqrt(6) a l'echelle 10^-72 : plancher exact (s^2 <= k 10^(2E) < (s+1)^2)");

// arrondi au plus pres d'une valeur fixe a l'echelle, a `decimales` chiffres apres la virgule ;
// nonAmbigu : la partie abandonnee n'est ni nulle ni exactement une demi-unite
BigInteger ArrondiC20(BigInteger valeur, int decimales, out bool nonAmbigu)
{
    BigInteger unite = BigInteger.Pow(10, echelleC20 - decimales);
    BigInteger abandonnee = valeur % unite;
    nonAmbigu = !abandonnee.IsZero && abandonnee * 2 != unite;
    return (valeur + unite / 2) / unite;
}

// ecriture d'un entier a l'echelle : point decimal insere `decimales` chiffres avant la fin
string FixeC20(BigInteger valeur, int decimales)
{
    string chiffres = valeur.ToString(CultureInfo.InvariantCulture).PadLeft(decimales + 1, '0');
    return chiffres.Substring(0, chiffres.Length - decimales) + "." + chiffres.Substring(chiffres.Length - decimales);
}

BigInteger alphaArrondiC20 = ArrondiC20(alphaC20, decimalesAlphaC20, out bool temoinAlphaC20);
BigInteger inverseArrondiC20 = ArrondiC20(inverseC20, decimalesInverseC20, out bool temoinInverseC20);
string alphaTexteC20 = FixeC20(alphaArrondiC20, decimalesAlphaC20);
string inverseTexteC20 = FixeC20(inverseArrondiC20, decimalesInverseC20);

Console.WriteLine(Inv($"forme fermee : ({coefficientsC18[1]} + {coefficientsC18[2]}*sqrt2 + {coefficientsC18[3]}*sqrt3 + {coefficientsC18[6]}*sqrt6) / {nObjetC13}"));
Console.WriteLine("alpha_cand   = " + alphaTexteC20);
Console.WriteLine("inverse      = " + inverseTexteC20);

// Temoin amont : sortie du script enumerate_records.py rejouee isolement (python -I) le 2026-11-06 (CITE).
string amontInverseC20 = "137.035999189566155029183744020866799566408684673854265541249";
BigInteger EntierSansPointC20(string texte) => BigInteger.Parse(texte.Replace(".", ""), CultureInfo.InvariantCulture);
bool egalAmontC20 = inverseTexteC20 == amontInverseC20;
BigInteger ecartAmontC20 = BigInteger.Abs(EntierSansPointC20(inverseTexteC20) - EntierSansPointC20(amontInverseC20));
Console.WriteLine("egalite avec le script amont (60 chiffres) : " + (egalAmontC20 ? "True" : "False")
    + " | ecart = " + ecartAmontC20.ToString(CultureInfo.InvariantCulture) + "E-57");

// Le carnet Python affiche alpha_cand par un Decimal de precision 60, qui arrondit APRES CHAQUE operation :
// son numerateur n'a que 57 decimales utiles, donc son 60e chiffre significatif n'est pas determine.
// Le calcul a echelle fixe ci-dessus affiche ce chiffre ; l'ecart est mesure, pas suppose.
string alphaPythonC20 = "0.00729735256366226025479763558800893201385159185699908385661404";
BigInteger ecartAlphaC20 = BigInteger.Abs(EntierSansPointC20(alphaTexteC20) - EntierSansPointC20(alphaPythonC20));
Console.WriteLine("alpha_cand affiche par le carnet Python (Decimal, precision 60) = " + alphaPythonC20);
Console.WriteLine("  ecart avec l'arrondi a echelle fixe : " + ecartAlphaC20.ToString(CultureInfo.InvariantCulture)
    + "E-62 (Decimal arrondit apres chaque operation ; son numerateur n'a que 57 decimales utiles)");

// Assertions exactes : sur des entiers, aucune tolerance.
Exige(egalAmontC20, "l'inverse a 60 chiffres egale la reference citee du script amont : ecart nul");
Exige(temoinAlphaC20 && temoinInverseC20, "arrondis non ambigus : la partie abandonnee n'est ni nulle ni une demi-unite");
Exige(ecartAlphaC20 < 5, "l'arrondi a echelle fixe et la sortie Decimal du carnet Python different de moins de 5 unites du 60e chiffre significatif");
// (chacune se lit par produit en croix : a/b = c/d ssi a d = c b)
Exige(coefficientsC18[2] * 648 == 1 * nObjetC13 && coefficientsC18[3] * 5184 == 7 * nObjetC13
    && coefficientsC18[6] * 10368 == 1 * nObjetC13,
    "coordonnees rationnelles du candidat sur (1, sqrt2, sqrt3, sqrt6) : 48/31104 = 1/648, 42/31104 = 7/5184, 3/31104 = 1/10368");
````

````
#nullable enable
using System;
using System.Globalization;
using System.Linq;
using System.Numerics;

// Figure 3 retracee SANS graphique : le candidat (sans incertitude) contre les quatre determinations.
// CITE — valeurs centrales et incertitudes citees par la Letter (sa figure 3), alpha^-1 en unites de 10^-9.
var mesuresC23 = new (string Nom, long ValeurNano, int IncertitudeNano)[]
{
    ("Moment magnetique electron (Fan 2023)", 137035999166L, 15),
    ("Recul Rb (Morel 2020)", 137035999206L, 11),
    ("Recul Cs (Parker 2018)", 137035999046L, 27),
    ("CODATA 2022", 137035999177L, 21),
};

// rejeu flottant du carnet Python : float(candidat - Decimal(str(valeur))) / incertitude
double DifferenceEnDoubleC23(BigInteger difference, int echelle)
{
    // difference est en unites de 10^-echelle ; la chaine decimale exacte est convertie en double (arrondi au plus pres)
    string signe = difference.Sign < 0 ? "-" : "";
    string chiffres = BigInteger.Abs(difference).ToString(CultureInfo.InvariantCulture).PadLeft(echelle + 1, '0');
    string texte = signe + chiffres.Substring(0, chiffres.Length - echelle) + "." + chiffres.Substring(chiffres.Length - echelle);
    return double.Parse(texte, CultureInfo.InvariantCulture);
}

// quotient n/d arrondi a `decimales` chiffres, moitie loin de zero : un entier en unites de 10^-decimales
BigInteger QuotientArrondiC23(BigInteger n, BigInteger d, int decimales)
{
    BigInteger signe = n.Sign < 0 ? -1 : 1;
    BigInteger numerateur = BigInteger.Abs(n) * BigInteger.Pow(10, decimales);
    return signe * ((2 * numerateur + d) / (2 * d));
}

string SigneFixeC23(BigInteger valeur, int decimales)
{
    return (valeur.Sign < 0 ? "-" : "+") + FixeC20(BigInteger.Abs(valeur), decimales);
}

Console.WriteLine("candidat : alpha^-1 = " + FixeC20(inverseArrondiC20, echelleInverseC20) + " ");
Console.WriteLine();

// le trait pointille de la figure 3 : (candidat - 137.035999) * 10^9, EXACT
BigInteger decalageReferenceC23 = new BigInteger(137035999) * BigInteger.Pow(10, echelleInverseC20 - 6);
BigInteger positionCandidatC23 = inverseArrondiC20 - decalageReferenceC23;      // unites de 10^-57
string positionTexteC23 = FixeC20(positionCandidatC23, echelleInverseC20 - 9);  // en unites de 10^-9
Console.WriteLine("candidat - 137.035999, en unites de 10^-9 (exact) : " + positionTexteC23);
Console.WriteLine();

string[] attendusDeuxC23 = { "+1.57", "-1.49", "+5.32", "+0.60" };
string[] attendusDouzeC23 = { "+1.571077001946", "-1.493985906438", "+5.317265001081", "+0.598388334723" };
for (int m = 0; m < mesuresC23.Length; m++)
{
    (string nom, long valeurNano, int incertitudeNano) = mesuresC23[m];
    BigInteger differenceC23 = inverseArrondiC20 - new BigInteger(valeurNano) * BigInteger.Pow(10, echelleInverseC20 - 9);
    BigInteger diviseurC23 = new BigInteger(incertitudeNano) * BigInteger.Pow(10, echelleInverseC20 - 9);
    double ecartFlottant = DifferenceEnDoubleC23(differenceC23, echelleInverseC20) / (incertitudeNano * 1e-9);
    BigInteger ecartDouze = QuotientArrondiC23(differenceC23, diviseurC23, 12);
    BigInteger ecartDeux = QuotientArrondiC23(differenceC23, diviseurC23, 2);
    Console.WriteLine(Inv($"{nom.PadRight(32)} {valeurNano / 1e9:F9}({incertitudeNano:00})  ecart = {ecartFlottant.ToString("+0.00;-0.00", CultureInfo.InvariantCulture)} sigma   exact {SigneFixeC23(ecartDouze, 12)}"));
    Exige(ecartFlottant.ToString("+0.00;-0.00", CultureInfo.InvariantCulture) == attendusDeuxC23[m]
        && SigneFixeC23(ecartDeux, 2) == attendusDeuxC23[m] && SigneFixeC23(ecartDouze, 12) == attendusDouzeC23[m],
        Inv($"ecart de {nom} : {attendusDeuxC23[m]} sigma (rejeu flottant ET calcul exact), valeur exacte a 12 decimales {attendusDouzeC23[m]}"));
}

// tension preexistante entre rubidium et cesium (independante du candidat)
double rbValeurC23 = 137.035999206, csValeurC23 = 137.035999046;
double rbIncertitudeC23 = 11e-9, csIncertitudeC23 = 27e-9;
double tensionFlottanteC23 = (rbValeurC23 - csValeurC23)
    / Math.Sqrt(rbIncertitudeC23 * rbIncertitudeC23 + csIncertitudeC23 * csIncertitudeC23);
Console.WriteLine(Inv($"\ntension Rb/Cs preexistante : {tensionFlottanteC23:F1} sigma entre elles (incertitudes recombinees en quadrature)"));

// controle exact : Rb - Cs = 160 unites de 10^-9, incertitudes en quadrature = 5 sqrt(34) : la tension vaut 32 / sqrt(34)
int echelleTensionC23 = 60;
BigInteger racine34C23 = RacineEntiereC20(34 * BigInteger.Pow(10, 2 * echelleTensionC23));
BigInteger tensionDouzeC23 = QuotientArrondiC23(32 * BigInteger.Pow(10, echelleTensionC23), racine34C23, 12);
Console.WriteLine("tension Rb/Cs exacte : " + FixeC20(tensionDouzeC23, 12) + " sigma  (32 / sqrt(34) : 160 = 32 * 5 et 850 = 25 * 34)");

// Assertions : valeurs attendues (sortie du carnet Python et calcul exact independant), aucune tolerance sur les chaines.
Exige(positionTexteC23 == "189.566155029183744020866799566408684673854265541249",
    "trait pointille de la figure 3 : candidat - 137.035999 = 189.566155029183744020866799566408684673854265541249 unites de 10^-9 (exact)");
Exige(tensionFlottanteC23.ToString("F1", CultureInfo.InvariantCulture) == "5.5" && FixeC20(tensionDouzeC23, 12) == "5.487954724560",
    "tension Rb/Cs : 5.5 sigma (rejeu flottant du carnet Python) et 32/sqrt(34) = 5.487954724560 a 12 decimales (exact)");
````

**2. L'état $|\omega_2\rangle$ vit sur un qutrit ($d = 3$) : est-il maximalement intriqué, et que devient le troisième niveau ?**

<details>
<summary>Réponse</summary>

Oui — mais sur le **support commun** $m = \min(d_1, d_2) = 2$. La sortie imprime deux amplitudes `0.707107+0.000000j`, sur $|0\rangle|0\rangle$ et $|1\rangle|1\rangle$, et rien d'autre. La réduction sur le qubit vaut exactement $I/2$ (`rho_qubit == I/2 : True`), et celle sur le qutrit a pour valeurs propres `[0. 0.5 0.5]` : la trace vaut $1$, et la valeur propre nulle dit que le troisième niveau **n'est pas peuplé**. « Maximalement intriqué » ne veut donc pas dire « uniforme sur tout l'espace », mais « uniforme sur le support que les deux bouts partagent ».

</details>

**3. La charge $Q$ agit-elle sur l'état de paire ? Que montre le témoin négatif à blocs pleins ?**

<details>
<summary>Réponse</summary>

$Q|\omega\rangle = i(b - a)|\omega\rangle$ avec un résidu **exactement nul** sur les trois types lisibles : `+0i` pour $(2,2)$, `+1i` pour $(2,3)$, `-1i` pour $(3,2)$.
