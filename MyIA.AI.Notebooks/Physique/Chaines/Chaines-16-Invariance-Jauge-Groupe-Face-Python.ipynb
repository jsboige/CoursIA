## Le cadre : la clause connue, le groupe d'identification, la lecture twirlée

La **clause connue** de l'axiome (chapitre 1-1) est, dans les mots du compendium, « the matching of basis directions across pairs » : ce qu'une lecture d'un qudit fixe conjointement, quelle que soit la façon dont les directions de base sont appariées d'une copie à l'autre. Le **groupe d'identification** est l'ensemble de ces appariements, construit comme objets exacts au lien. La **lecture** d'un enregistrement est la trace de son projecteur contre l'état twirlé (éq. 95, p. 173) :

$$\langle P \rangle = \frac{\mathrm{tr}\big(M \, T(P)\big)}{\dim R}, \qquad T(P) = \frac{1}{|G|} \sum_{g \in G} g \, P \, g^{-1}.$$

L'argument du théorème tient en une ligne de la source : le twirl « is fixed by every element it averages, so the sum is re-indexed and every readout is invariant ». Pour $h \in G$, $h \, T(P) \, h^{-1} = \frac{1}{|G|}\sum_g (hg) P (hg)^{-1} = T(P)$, puisque $g \mapsto hg$ ne fait que permuter les termes de la somme ; d'où $T(hPh^{-1}) = T(P)$ et la lecture ne bouge pas.

Le lien est le lien **qutrit-qubit** $\mathbb{C}^3 \otimes \mathbb{C}^2$, de type $(2,3)$. Le groupe d'identification y est « re-met as exact objects at the fourth-root monomial groups of the orders four, thirty-two, three hundred eighty-four and six thousand one hundred forty-four » : les **groupes monomiaux à racines quatrièmes** — une permutation des directions, chaque direction multipliée par $1, i, -1$ ou $-i$.

Dans la langue de la littérature : une théorie de jauge est une théorie des champs dont le lagrangien est invariant sous un groupe agissant indépendamment en chaque point ; le groupe de jauge du Modèle Standard est $SU(3)_c \times SU(2)_L \times U(1)_Y$, le groupe agissant fidèlement sur les représentations de matière étant son quotient par un sous-groupe cyclique d'ordre six du centre [44, 107, 108]. La correspondance est **par nom** (p. 172).

**Mesuré** dans la cellule suivante : les quatre ordres, la clôture de chaque groupe sous le produit, et le sous-groupe à signes seuls.

```python
import itertools
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(64)
I4 = [1, 1j, -1, -1j]  # racines quatriemes de l'unite


def groupe_monomial(n):
    """Permutations des n directions, chaque direction multipliee par une racine quatrieme."""
    mats = []
    for perm in itertools.permutations(range(n)):
        for phases in itertools.product(range(4), repeat=n):
            m = np.zeros((n, n), dtype=complex)
            for r in range(n):
                m[r, perm[r]] = I4[phases[r]]
            mats.append(m)
    return np.array(mats)


def cle(m):
    """Empreinte exacte d'une matrice a coefficients dans {0, +-1, +-i}."""
    return np.round(np.concatenate([m.real.ravel(), m.imag.ravel()])).astype(np.int8).tobytes()


GROUPES = {n: groupe_monomial(n) for n in range(1, 5)}
ORDRES = {n: len(g) for n, g in GROUPES.items()}
distincts = all(len({cle(m) for m in g}) == len(g) for g in GROUPES.values())

CLOTURE = {}
for n in range(1, 4):
    g = GROUPES[n]
    cles = {cle(m) for m in g}
    produits = np.einsum("aij,bjk->abik", g, g).reshape(-1, n, n)
    CLOTURE[n] = (len(produits), all(cle(p) in cles for p in produits))
g4 = GROUPES[4]
cles4 = {cle(m) for m in g4}
tirages = rng.integers(0, len(g4), size=(20000, 2))
CLOTURE[4] = (len(tirages), all(cle(g4[a] @ g4[b]) in cles4 for a, b in tirages))

G3, G2 = GROUPES[3], GROUPES[2]
SIGNES3 = [m for m in G3 if np.all(np.isin(np.round(m.real + m.imag * 1j), [0, 1, -1]))]

print("ordres des groupes monomiaux a racines quatriemes :", ORDRES)
print("matrices deux a deux distinctes :", distincts)
for n, (nb, ok) in CLOTURE.items():
    mode = "tous les produits" if n < 4 else "produits tires au hasard"
    print(f"  n = {n} : {nb} {mode}, tous dans le groupe : {ok}")
print("sous-groupe a signes seuls de G3 (coefficients 0, +1, -1) :", len(SIGNES3), "elements")
```

### Lecture du résultat (les quatre groupes)

Les quatre ordres de la source sont retrouvés : $4 = 1! \cdot 4$, $32 = 2! \cdot 4^2$, $384 = 3! \cdot 4^3$, $6144 = 4! \cdot 4^4$ — une permutation des $n$ directions et une racine quatrième par direction. La clôture est **exacte** pour $n \le 3$ (les $4^2$, $32^2$ et $384^2 = 147456$ produits retombent tous dans le groupe) et **échantillonnée** pour $n = 4$ (20 000 produits tirés) : c'est la seule affirmation de cette cellule qui n'est pas exhaustive.

Le sous-groupe à signes seuls de $G_3$ a **48** éléments, $3! \cdot 2^3$ : ce sont les permutations signées de l'éq. 90 (Chaînes-14) et les quarante-huit appariements admis du chapitre 56 (Chaînes-12), plongés ici dans le groupe à 384 éléments. Les 384 appariements côté qutrit et les 32 appariements côté qubit que la source teste plus bas ont exactement les ordres de $G_3$ et de $G_2$.

## Le twirl et le lemme de re-indexation

**Mesuré** : on construit le twirl sur un produit $A \otimes B$ (le facteur $A$ agit côté qutrit, $B$ côté qubit), et l'on vérifie le lemme de re-indexation sur un sous-groupe où le twirl **n'est pas trivial** — le sous-groupe **diagonal** $D_3 \otimes D_2$ (phases seules, sans permutation), d'ordre $4^3 \cdot 4^2 = 1024$. On y attend un **déphasage** : le twirl ne garde que la diagonale de l'état. Un état générique $P$ du lien et un enregistrement générique $M$ (un projecteur de rang un tiré au hasard) servent de témoins.

```python
def twirl(P, A, B):
    """Twirl de P (6 x 6 sur C3 (x) C2) sur le produit A (x) B : A agit cote qutrit, B cote qubit."""
    T = P.reshape(3, 2, 3, 2)
    T = np.einsum("gbe,aecf,gdf->abcd", B, T, B.conj()) / len(B)
    T = np.einsum("gae,ebfd,gcf->abcd", A, T, A.conj()) / len(A)
    return T.reshape(6, 6)


def etat_aleatoire(d):
    x = rng.normal(size=(d, d)) + 1j * rng.normal(size=(d, d))
    r = x @ x.conj().T
    return r / np.trace(r).real


def projecteur_aleatoire(d):
    v = rng.normal(size=d) + 1j * rng.normal(size=d)
    v = v / np.linalg.norm(v)
    return np.outer(v, v.conj())


def lecture(M, T):
    """Eq. 95 : tr(M T) / dim R, avec dim R = 6 au lien qutrit-qubit."""
    return float(np.real(np.trace(M @ T))) / 6


P = etat_aleatoire(6)          # un etat generique du lien
M = projecteur_aleatoire(6)    # un enregistrement generique

D3 = np.array([m for m in G3 if np.allclose(m, np.diag(np.diag(m)))])
D2 = np.array([m for m in G2 if np.allclose(m, np.diag(np.diag(m)))])
TD = twirl(P, D3, D2)
DEPHASAGE = np.allclose(TD, np.diag(np.diag(P)))

ecarts_fixe, ecarts_lecture = [], []
for _ in range(50):
    h = np.kron(D3[rng.integers(len(D3))], D2[rng.integers(len(D2))])
    ecarts_fixe.append(np.abs(h @ TD @ h.conj().T - TD).max())
    ecarts_lecture.append(abs(lecture(M, twirl(h @ P @ h.conj().T, D3, D2)) - lecture(M, TD)))

print("ordres des sous-groupes diagonaux : D3 =", len(D3), " D2 =", len(D2), " produit =", len(D3) * len(D2))
print("twirl diagonal = diagonale de P (dephasage) :", DEPHASAGE)
print("coefficients hors diagonale de P (max) :", round(float(np.abs(P - np.diag(np.diag(P))).max()), 4))
print("re-indexation, 50 tirages : max |h T h^-1 - T| =", f"{max(ecarts_fixe):.1e}")
print("re-indexation, 50 tirages : max |lecture(T(h P h^-1)) - lecture(T(P))| =", f"{max(ecarts_lecture):.1e}")
```

### Lecture du résultat (déphasage et re-indexation)

Le twirl sur le sous-groupe diagonal efface exactement les coefficients hors diagonale de l'état — non nuls ici, de l'ordre de la valeur affichée — et garde la diagonale : c'est un **déphasage**, un twirl qui fait quelque chose. Sur ce twirl non trivial, le lemme de la source se vérifie à la précision machine sur les cinquante tirages : l'élément $h$ fixe le twirl, et la lecture de l'état transformé $hPh^{-1}$ ne bouge pas. C'est l'argument entier du Théorème 64.1, « the sum is re-indexed » : il ne dépend ni de l'état ni de l'enregistrement, seulement du fait que $h$ appartient au groupe sur lequel on moyenne.

## Le groupe complet est irréductible : son twirl est un scalaire

**Mesuré** : la moyenne $\frac{1}{|G|}\sum_g |\mathrm{tr}\, g|^2$ est la dimension du **commutant** du groupe — l'espace des matrices qui commutent avec tous ses éléments. Elle vaut 1 si et seulement si la représentation est **irréductible** (première relation d'orthogonalité des caractères). Le twirl est la projection orthogonale (pour le produit de Hilbert-Schmidt) sur ce commutant ; si celui-ci se réduit aux multiples de l'identité, le lemme de Schur force $T(P) = \frac{\mathrm{tr}\,P}{6}\,I_6$.

On compare au sous-groupe diagonal, et au **groupe entier** $U(3) \times U(2)$ — le twirl de Haar, estimé par Monte-Carlo, donc à une précision de l'ordre de $1/\sqrt{N}$ seulement.

```python
def moyenne_caractere(G):
    return float(np.mean([abs(np.trace(g)) ** 2 for g in G]))


def unitaire_haar(n):
    x = (rng.normal(size=(n, n)) + 1j * rng.normal(size=(n, n))) / np.sqrt(2)
    q, r = np.linalg.qr(x)
    return q @ np.diag(np.diag(r) / np.abs(np.diag(r)))


MC = {"G3": moyenne_caractere(G3), "G2": moyenne_caractere(G2),
      "D3": moyenne_caractere(D3), "D2": moyenne_caractere(D2)}
T = twirl(P, G3, G2)
ECART_SCALAIRE = float(np.abs(T - np.trace(P).real / 6 * np.eye(6)).max())

N_HAAR = 4000
U3s = np.array([unitaire_haar(3) for _ in range(N_HAAR)])
U2s = np.array([unitaire_haar(2) for _ in range(N_HAAR)])
T_HAAR = sum(np.kron(a, b) @ P @ np.kron(a, b).conj().T for a, b in zip(U3s, U2s)) / N_HAAR
ECART_HAAR = float(np.abs(T_HAAR - np.trace(P).real / 6 * np.eye(6)).max())

for nom, v in MC.items():
    print(f"moyenne de |tr g|^2 sur {nom} : {v:.6f}")
print("commutant de G3 (x) G2 : dimension", round(MC["G3"] * MC["G2"]))
print("commutant de D3 (x) D2 : dimension", round(MC["D3"] * MC["D2"]))
print("twirl fini G3 (x) G2 : max |T(P) - tr(P)/6 I| =", f"{ECART_SCALAIRE:.1e}")
print(f"twirl de Haar U(3) x U(2), {N_HAAR} tirages : max |T(P) - tr(P)/6 I| =", f"{ECART_HAAR:.1e}",
      " (1/sqrt(N) =", f"{1 / np.sqrt(N_HAAR):.1e})")
```

### Lecture du résultat (Schur, et un premier écart avec la source)

La moyenne de caractère vaut **1** pour $G_3$ et pour $G_2$ : chacun agit irréductiblement sur sa copie, et — la moyenne d'un produit tensoriel étant le produit des moyennes — $G_3 \otimes G_2$ agit **irréductiblement** sur $\mathbb{C}^3 \otimes \mathbb{C}^2$. Son twirl est donc le scalaire $\frac{\mathrm{tr}\,P}{6} I_6$, à $10^{-16}$ près. Le sous-groupe diagonal a un commutant de dimension $3 \times 2 = 6$ — les six coefficients diagonaux que garde le déphasage. Le twirl de Haar du groupe entier tombe sur le même scalaire, à une erreur Monte-Carlo du même ordre que $1/\sqrt{N}$ : ce n'est pas une coïncidence, le commutant de $U(3) \otimes U(2)$ est inclus dans celui de $G_3 \otimes G_2$, qui est déjà réduit aux scalaires.

Conséquence, qui sera reprise dans la section des résultats ouverts : **sur ce registre, le twirl du groupe fini et celui du groupe entier sont le même opérateur, et cet opérateur ne retient de l'état que sa trace**. Rien ne peut donc le déplacer. Or la source écrit que « the finite-group twirl is moved by the Lie generators at two records and by the Fourier elements at fourteen ». Ce twirl-là ne peut pas être celui de $G_3 \otimes G_2$ sur $\mathbb{C}^3 \otimes \mathbb{C}^2$ : le carnet ne sait pas construire l'objet que la phrase désigne. Le croisement avec le lac Lean Serre100 (relations d'orthogonalité des caractères) est nommé plus bas.

## Les six familles de tests de la source, rejouées sur trois objets

La source applique à chaque enregistrement et à chaque élément le test « whether the readout is unmoved by the element », pour six familles : les **384 appariements côté qutrit**, les **32 appariements côté qubit**, les **13 générateurs de Lie** (« the derivative of the trace along each one » — une base de $\mathfrak{u}(3) \oplus \mathfrak{u}(2)$, $9 + 4 = 13$), les **13 unitaires de Cayley** $(I + X)^{-1}(I - X)$, les **2 unitaires de chaîne** et les **2 éléments de Fourier**.

**Mesuré** : cinq de ces six familles, sur trois objets — le twirl complet $G_3 \otimes G_2$, le twirl diagonal $D_3 \otimes D_2$, et l'état nu (sans twirl). Les **unitaires de chaîne** ne sont pas définis dans le texte de la section : ils restent **cités**. Attention à l'axe de comptage : la source compte les **enregistrements inchangés par famille** (sur ses 172) ; ce carnet compte, pour **un** enregistrement générique, les **éléments qui déplacent la lecture**. Les deux comptes ne se comparent pas terme à terme.

```python
def base_u(n):
    """Base reelle de u(n) : n generateurs diagonaux i E_jj, puis les paires antisymetrique / i-symetrique."""
    out = []
    for j in range(n):
        e = np.zeros((n, n), dtype=complex)
        e[j, j] = 1j
        out.append(e)
    for j in range(n):
        for k in range(j + 1, n):
            a = np.zeros((n, n), dtype=complex)
            a[j, k], a[k, j] = 1, -1
            s = np.zeros((n, n), dtype=complex)
            s[j, k], s[k, j] = 1j, 1j
            out += [a, s]
    return out


GENS = [np.kron(X, np.eye(2)) for X in base_u(3)] + [np.kron(np.eye(3), Y) for Y in base_u(2)]
CAYLEY = [np.linalg.solve(np.eye(6) + X, np.eye(6) - X) for X in GENS]
F3 = np.array([[np.exp(2j * np.pi * a * b / 3) for b in range(3)] for a in range(3)]) / np.sqrt(3)
F2 = np.array([[1, 1], [1, -1]]) / np.sqrt(2)
FOURIER = [np.kron(F3, np.eye(2)), np.kron(np.eye(3), F2)]
CAYLEY_UNITAIRES = all(np.allclose(C @ C.conj().T, np.eye(6)) for C in CAYLEY)

FAMILLES = {
    "appariements qutrit": [np.kron(g, np.eye(2)) for g in G3],
    "appariements qubit": [np.kron(np.eye(3), g) for g in G2],
    "generateurs de Lie": GENS,
    "unitaires de Cayley": CAYLEY,
    "elements de Fourier": FOURIER,
}
OBJETS = {
    "twirl complet": lambda Q: twirl(Q, G3, G2),
    "twirl diagonal": lambda Q: twirl(Q, D3, D2),
    "etat nu": lambda Q: Q,
}
SEUIL = 1e-9


def deplace(objet, famille, X):
    if famille == "generateurs de Lie":  # derivee de la lecture le long de X, lineaire en [X, P]
        return abs(lecture(M, objet(X @ P - P @ X))) > SEUIL
    return abs(lecture(M, objet(X @ P @ X.conj().T)) - lecture(M, objet(P))) > SEUIL


COMPTES = {o: {f: (sum(deplace(fo, f, X) for X in xs), len(xs)) for f, xs in FAMILLES.items()}
           for o, fo in OBJETS.items()}

print("unitaires de Cayley tous unitaires :", CAYLEY_UNITAIRES)
print(f"{'objet':16s}" + "".join(f"{f:>22s}" for f in FAMILLES))
for o, c in COMPTES.items():
    print(f"{o:16s}" + "".join(f"{f'{c[f][0]} / {c[f][1]}':>22s}" for f in FAMILLES))

# deux contre-tests : un twirl a un seul bout, et l'etat de base |00> lu contre lui-meme
T_UN_BOUT = twirl(P, np.array([np.eye(3)]), G2)
TRACE_PARTIELLE = np.einsum("ajbj->ab", P.reshape(3, 2, 3, 2))
UN_BOUT_OK = np.allclose(T_UN_BOUT, np.kron(TRACE_PARTIELLE, np.eye(2) / 2))
UN_BOUT = sum(abs(lecture(M, twirl(h @ P @ h.conj().T, np.array([np.eye(3)]), G2)) - lecture(M, T_UN_BOUT)) > SEUIL
              for h in FAMILLES["appariements qutrit"])
E00 = np.zeros((6, 6), dtype=complex)
E00[0, 0] = 1
NU_00 = sum(abs(lecture(E00, h @ E00 @ h.conj().T) - lecture(E00, E00)) > SEUIL for h in FAMILLES["appariements qutrit"])
print("twirl cote qubit seul = Tr_qubit(P) (x) I/2 :", UN_BOUT_OK, "; deplace par", UN_BOUT, "/ 384 appariements qutrit")
print("etat |00> lu par l'enregistrement |00> : deplace par", NU_00, "/ 384 appariements qutrit")
```

### Lecture du résultat (cinq familles, trois objets)

**Le twirl complet n'est déplacé par aucun élément d'aucune famille** — c'est le Théorème 64.1 sur ce registre, et c'est attendu : l'opérateur est un scalaire. **Le twirl diagonal**, lui, est déplacé par les appariements qui **permutent** les directions (320 des 384 côté qutrit — tous sauf les $4^3 = 64$ éléments diagonaux ; 16 des 32 côté qubit), par les **8** générateurs de Lie non diagonaux (les 5 générateurs diagonaux commutent avec le déphasage), par les 8 unitaires de Cayley correspondants, et par les **deux** éléments de Fourier. Il est invariant sous **son propre** groupe et sous lui seul : la re-indexation ne protège que contre les éléments sur lesquels on moyenne. **L'état nu** est déplacé par presque tout — tous les appariements sauf les quatre multiples scalaires de l'identité ($380/384$, $28/32$), tous les générateurs, tous les Cayley, tous les Fourier.

Les deux contre-tests montrent que le compte dépend de l'objet et de l'enregistrement. Un twirl à **un seul bout** (côté qubit) vaut $\mathrm{Tr}_{\text{qubit}}(P) \otimes I/2$ et reste déplacé par les appariements de l'autre bout. L'état de base $|00\rangle$, lu par son propre projecteur, n'est déplacé que par les appariements qui envoient la première direction du qutrit ailleurs — $256 = 4 \times 64$ des 384. C'est l'analogue, à l'échelle d'un enregistrement, de la phrase de la source sur l'état nu « moved by the three hundred eighty-four matchings at nine records ».

## Théorème 64.2 : le groupe de face, deux constructions d'une même algèbre

Le **groupe de face** du lien est « the image of the local unitaries U(2) x U(3) on the link » ; la source l'identifie à la **tranche de déterminant** $S(U(2) \times U(3)) = \{\mathrm{diag}(U, V) : U \in U(2),\ V \in U(3),\ \det U \det V = 1\} \subset U(5)$, d'algèbre de Lie de dimension 12, d'algèbre dérivée de dimension 11 et de centre de dimension 1.

**Mesuré** : les trois dimensions sur **deux constructions indépendantes** — l'image tensorielle $\mathfrak{u}(3) \otimes 1 + 1 \otimes \mathfrak{u}(2)$ agissant sur $\mathbb{C}^3 \otimes \mathbb{C}^2$ (matrices $6 \times 6$), et la tranche par blocs dans $\mathfrak{u}(5)$ (matrices $5 \times 5$, combinaisons de trace nulle) —, le **rang** (dimension d'un tore maximal), le générateur du centre, puis la **fidélité** de l'action de la tranche sur le lien.

```python
def vec_reel(m):
    return np.concatenate([m.real.ravel(), m.imag.ravel()])


def rang_de(mats):
    return int(np.linalg.matrix_rank(np.array([vec_reel(m) for m in mats]), tol=1e-9))


def base_engendree(mats):
    A = np.array([vec_reel(m) for m in mats])
    _, s, vt = np.linalg.svd(A)
    r = int((s > 1e-9).sum())
    d = mats[0].shape[0]
    return [(v[: d * d] + 1j * v[d * d:]).reshape(d, d) for v in vt[:r]]


def centralisateur(B, S):
    """Combinaisons de la base B qui commutent avec chaque matrice de S."""
    A = np.array([np.concatenate([vec_reel(Bk @ Sj - Sj @ Bk) for Sj in S]) for Bk in B]).T
    _, s, vt = np.linalg.svd(A)
    return [sum(c * Bk for c, Bk in zip(v, B)) for v in vt[int((s > 1e-9).sum()):]]


def dims_lie(generateurs):
    B = base_engendree(generateurs)
    derivee = rang_de([X @ Y - Y @ X for X in B for Y in B])
    return len(B), derivee, centralisateur(B, B), B


def bloc(a, b):
    m = np.zeros((5, 5), dtype=complex)
    if a is not None:
        m[:2, :2] = a
    if b is not None:
        m[2:, 2:] = b
    return m


# construction 1 : image tensorielle sur C3 (x) C2
dim_t, der_t, cen_t, B_t = dims_lie(GENS)
Z_T = cen_t[0] / cen_t[0][0, 0]
TORE_T = [np.kron(X, np.eye(2)) for X in base_u(3)[:3]] + [np.kron(np.eye(3), Y) for Y in base_u(2)[:2]]
rang_t = rang_de(TORE_T)

# construction 2 : tranche par blocs, combinaisons de trace nulle de u(2) + u(3) dans u(5)
BLOCS = [bloc(Y, None) for Y in base_u(2)] + [bloc(None, X) for X in base_u(3)]
traces = np.array([[np.trace(m).imag for m in BLOCS]])
_, _, vt = np.linalg.svd(traces)
TRANCHE = [sum(c * m for c, m in zip(v, BLOCS)) for v in vt[1:]]
dim_b, der_b, cen_b, B_b = dims_lie(TRANCHE)
Z_B = cen_b[0] / (cen_b[0][0, 0] / 3j)
TORE_B = [bloc(np.diag([1j, 0]), None) - bloc(None, np.diag([1j, 0, 0])),
          bloc(np.diag([0, 1j]), None) - bloc(None, np.diag([1j, 0, 0])),
          bloc(None, np.diag([1j, -1j, 0])), bloc(None, np.diag([0, 1j, -1j]))]
rang_b = rang_de(TORE_B)
tore_maximal = len(centralisateur(B_b, TORE_B)) == rang_b

print(f"image tensorielle : dim {dim_t}, derivee {der_t}, centre {len(cen_t)}, rang {rang_t}")
print("  centre normalise = I6 (fois i) :", np.allclose(Z_T, np.eye(6)))
print(f"tranche S(U(2) x U(3)) : dim {dim_b}, derivee {der_b}, centre {len(cen_b)}, rang {rang_b}")
print("  centre normalise : i * diag", np.round(np.diag(Z_B / 1j).real, 6))
print("  le tore de rang 4 est son propre centralisateur (tore maximal) :", tore_maximal)

# fidelite : (U, V) -> V (x) U ; le noyau sur U(2) x U(3) est le cercle (l I2, l^-1 I3)
cercle = [np.exp(2j * np.pi * k / 360) for k in range(360)]
NOYAU_PRODUIT = [l for l in cercle if np.allclose(np.kron(np.eye(3) / l, l * np.eye(2)), np.eye(6))]
NOYAU_TRANCHE = [l for l in NOYAU_PRODUIT if abs(l ** 2 * l ** -3 - 1) < 1e-12]  # det U det V = l^-1
print("cercle (l I2, l^-1 I3), 360 points : agissent trivialement", len(NOYAU_PRODUIT), "; dans la tranche", len(NOYAU_TRANCHE))
IMAGE_OK = True
for _ in range(200):
    U, V = unitaire_haar(2), unitaire_haar(3)
    mu = np.linalg.det(U) * np.linalg.det(V)
    Us, Vs = mu * U, V / mu
    IMAGE_OK &= abs(np.linalg.det(Us) * np.linalg.det(Vs) - 1) < 1e-9 and np.allclose(np.kron(Vs, Us), np.kron(V, U))
print("toute image V (x) U de U(2) x U(3) est atteinte par la tranche (200 tirages) :", IMAGE_OK)
```

### Lecture du résultat (douze, onze, un, rang quatre)

Les deux constructions, l'une sur $\mathbb{C}^3 \otimes \mathbb{C}^2$ et l'autre dans $\mathfrak{u}(5)$, donnent les mêmes invariants : **dimension 12, dérivée 11, centre 1, rang 4** — les nombres du Théorème 64.2. La dérivée de dimension $11 = 3 + 8$ est $\mathfrak{su}(2) \oplus \mathfrak{su}(3)$ ; le centre de dimension 1 est le facteur abélien. Côté tenseur, il est engendré par $i\,I_6$ — une phase globale du lien. Côté blocs, il est engendré par $i\,\mathrm{diag}(3, 3, -2, -2, -2)$ : une phase de **charge 3** sur le doublet et de **charge $-2$** sur le triplet, seule combinaison de trace nulle des deux identités de bloc. Le tore de rang 4 est son propre centralisateur dans la tranche, donc maximal.

La fidélité se lit en deux temps. Sur $U(2) \times U(3)$ tout entier, l'action $(U, V) \mapsto V \otimes U$ a pour noyau le **cercle** $(\lambda I_2, \lambda^{-1} I_3)$ : les 360 points testés agissent trivialement. La condition de déterminant de la tranche, $\lambda^2 \lambda^{-3} = \lambda^{-1} = 1$, n'en garde que l'identité : **la tranche agit fidèlement sur le lien**. Réciproquement, toute image $V \otimes U$ est atteinte par un couple de la tranche, en remplaçant $(U, V)$ par $(\mu U, \mu^{-1} V)$ avec $\mu = \det U \det V$. Le groupe de face — l'image des unitaires locaux sur le lien — est donc bien une copie de la tranche, ce que la source énonce. L'échantillon du cercle est fini ; l'argument algébrique, lui, est exact.

## Le revêtement explicite et son noyau $\mathbb{Z}_6$

La source écrit la tranche comme « isomorphic to (SU(2) x SU(3) x U(1))/Z6 by an explicit covering whose kernel is the cyclic group of order six ». **Mesuré** : deux revêtements de $SU(2) \times SU(3) \times U(1)$, l'un vers la tranche par blocs, l'autre vers l'image tensorielle,

$$\varphi_{\text{bloc}}(A, B, z) = \mathrm{diag}(z^3 A,\ z^{-2} B), \qquad \varphi_{\text{tenseur}}(A, B, z) = z\,(B \otimes A),$$

dont on vérifie qu'ils sont des **homomorphismes**, qu'ils sont **surjectifs**, et qu'ils ont **le même noyau**, cyclique d'ordre six. Les exposants $3$ et $-2$ sont les charges lues au centre de la cellule précédente.

```python
def unitaire_special(n):
    q = unitaire_haar(n)
    return q / np.linalg.det(q) ** (1 / n)


def phi_bloc(A, B, z):
    return bloc(z ** 3 * A, z ** -2 * B)


def phi_tenseur(A, B, z):
    return z * np.kron(B, A)


HOMO_OK = True
for _ in range(200):
    A1, B1, z1 = unitaire_special(2), unitaire_special(3), np.exp(2j * np.pi * rng.random())
    A2, B2, z2 = unitaire_special(2), unitaire_special(3), np.exp(2j * np.pi * rng.random())
    HOMO_OK &= abs(np.linalg.det(phi_bloc(A1, B1, z1)) - 1) < 1e-9
    HOMO_OK &= np.allclose(phi_bloc(A1 @ A2, B1 @ B2, z1 * z2), phi_bloc(A1, B1, z1) @ phi_bloc(A2, B2, z2))
    HOMO_OK &= np.allclose(phi_tenseur(A1 @ A2, B1 @ B2, z1 * z2), phi_tenseur(A1, B1, z1) @ phi_tenseur(A2, B2, z2))

NOYAU_BLOC = set()
for k in range(6):
    z = np.exp(2j * np.pi * k / 6)
    A, B = z ** -3 * np.eye(2), z ** 2 * np.eye(3)
    assert abs(np.linalg.det(A) - 1) < 1e-12 and abs(np.linalg.det(B) - 1) < 1e-12
    if np.allclose(phi_bloc(A, B, z), np.eye(5)):
        NOYAU_BLOC.add((round(np.real(A[0, 0])), complex(np.round(B[0, 0], 9)), complex(np.round(z, 9))))
NOYAU_TENSEUR = set()
for alpha in (1, -1):                       # centre de SU(2)
    for j in range(3):                      # centre de SU(3)
        beta = np.exp(2j * np.pi * j / 3)
        z = 1 / (alpha * beta)
        if np.allclose(phi_tenseur(alpha * np.eye(2), beta * np.eye(3), z), np.eye(6)):
            NOYAU_TENSEUR.add((alpha, complex(np.round(beta, 9)), complex(np.round(z, 9))))
z1 = np.exp(2j * np.pi / 6)
ORDRE_GEN = next(m for m in range(1, 7) if abs(z1 ** m - 1) < 1e-9)

SURJ_OK = True
for _ in range(200):
    U, V = unitaire_haar(2), unitaire_haar(3)
    V = V / (np.linalg.det(U) * np.linalg.det(V)) ** (1 / 3)   # (U, V) dans la tranche
    z = np.linalg.det(U) ** (1 / 6)
    A, B = z ** -3 * U, z ** 2 * V
    SURJ_OK &= abs(np.linalg.det(A) - 1) < 1e-9 and abs(np.linalg.det(B) - 1) < 1e-9
    SURJ_OK &= np.allclose(phi_bloc(A, B, z), bloc(U, V))
    U2, V2 = unitaire_haar(2), unitaire_haar(3)
    a, b = np.linalg.det(U2) ** 0.5, np.linalg.det(V2) ** (1 / 3)
    SURJ_OK &= np.allclose(phi_tenseur(U2 / a, V2 / b, a * b), np.kron(V2, U2))

print("homomorphismes, 200 tirages (les deux revetements) :", HOMO_OK)
print("noyau du revetement bloc :", len(NOYAU_BLOC), "elements ; noyau du revetement tenseur :", len(NOYAU_TENSEUR))
print("les deux noyaux sont le meme ensemble :", NOYAU_BLOC == NOYAU_TENSEUR)
print("element k = 1 : (z^-3 I2, z^2 I3, z) avec z = exp(2 i pi / 6), ordre", ORDRE_GEN)
print("surjectivite, 200 tirages (les deux revetements) :", SURJ_OK)
```

### Lecture du résultat (le $\mathbb{Z}_6$)

Les deux applications sont des homomorphismes à valeurs dans la tranche et dans l'image tensorielle, et elles sont surjectives sur les 200 tirages. Leur noyau est le **même** ensemble de six triplets $(z^{-3} I_2,\ z^{2} I_3,\ z)$ avec $z^6 = 1$, et l'élément $z = e^{2i\pi/6}$ est d'ordre 6 : le noyau est **cyclique d'ordre six**, $\mathbb{Z}_6$, comme l'énonce la source. On le lit aussi par les centres : $-1 \in SU(2)$ (ordre 2) et $e^{2i\pi/3} \in SU(3)$ (ordre 3) se compensent par une phase $z$, et $2 \times 3 = 6$ avec $\mathrm{pgcd}(2, 3) = 1$ donne un groupe cyclique.

Le quotient $(SU(2) \times SU(3) \times U(1))/\mathbb{Z}_6$ porte, dans la littérature, le nom du groupe agissant fidèlement sur la matière du Modèle Standard [44, 107, 108]. Ce carnet n'établit que la structure de groupe finie-dimensionnelle ; l'identification physique est une **correspondance par nom**, et la source le dit.

## La cellule d'unicité : pourquoi le type $(2,3)$

```python
CELLULE = {}
for a in range(2, 5):
    for b in range(a, 5):
        gens_ab = [np.kron(X, np.eye(b)) for X in base_u(a)] + [np.kron(np.eye(a), Y) for Y in base_u(b)]
        dim_ab, der_ab, cen_ab, _ = dims_lie(gens_ab)
        tore_ab = [np.kron(X, np.eye(b)) for X in base_u(a)[:a]] + [np.kron(np.eye(a), Y) for Y in base_u(b)[:b]]
        CELLULE[(a, b)] = (dim_ab, der_ab, len(cen_ab), rang_de(tore_ab))

FORMULES_OK = all(v[0] == a * a + b * b - 1 and v[3] == a + b - 1 and v[2] == 1 for (a, b), v in CELLULE.items())
print(f"{'type':8s}{'dim':>6s}{'derivee':>9s}{'centre':>8s}{'rang':>6s}")
for t, v in CELLULE.items():
    print(f"{str(t):8s}{v[0]:>6d}{v[1]:>9d}{v[2]:>8d}{v[3]:>6d}")
print("formules a^2 + b^2 - 1 et a + b - 1 verifiees aux six types :", FORMULES_OK)
print("types de dimension 12 :", [t for t, v in CELLULE.items() if v[0] == 12])
print("types de rang 4 :", [t for t, v in CELLULE.items() if v[3] == 4])
```
