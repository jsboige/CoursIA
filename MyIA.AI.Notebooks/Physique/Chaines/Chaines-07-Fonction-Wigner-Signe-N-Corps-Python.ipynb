# Chaînes — Carnet 07 : La fonction de Wigner et le signe à N corps

Série **Chaînes** (distillation du compendium, série #25013). **Chapitre 46 — Théorème 46.1 : « The Wigner function and the many-body sign »** — un **théorème conditionnel**.

> **Standing : CONDITIONNEL.** Le théorème lit ensemble la fonction d'espace des phases et le signe à N corps. Il est conditionnel sur la fonction de Wigner comme cellule de lecture swap — un résultat ouvert qui tient à **quatre des cinq** vérifications : à la cinquième, le recouvrement de deux distributions *part* du recouvrement d'enregistrements, et l'entrée Wigner devient une glose.

Patron de la série : chaque affirmation est **Mesurée** (rejouée ici en `Fraction` exacte), **Cité** (constante du compendium, non dérivée) ou **Ouverte** (énoncée comme ouverte par la source).

## Contexte et prémices

Le théorème repose sur les enregistrements, la mesure du chapitre 42 (**Théorème 42.1**, carnet Chaînes-03) et l'échange du chapitre 43 (**Théorème 43.1**, carnet Chaînes-04) : la négativité des marginales de Wigner au-delà de deux copies est la négativité du **carré swap-face de la somme directe**. Prémices : Définitions 1.1-1.6, Théorèmes 18.1-18.4, Open 18.5-18.6, Thm 42.1, Thm 43.1. Utilisé par le Théorème 47.1.

Dans le langage de la littérature : la fonction de Wigner d'un état est une quasi-distribution réelle sur l'espace des phases, de marginales les distributions de position et d'impulsion ; pour un système fini de dimension d, une fonction sur les d×d cellules, qui **peut être négative** ; le recouvrement de deux états est l'intégrale du produit de leurs fonctions de Wigner **[82]**. L'échange de N copies identiques (bosons ou fermions) entre dans la distribution jointe par les termes d'interférence, au signe des statistiques **[76]**.

## 1. L'énoncé : deux objets, le même signe

Le théorème énonce cinq volets, tous **mesurés dans la source** :

1. Les deux objets — marginales de Wigner et carré swap-face — portent **le même signe** à chacun des cinq comptes de copies, de N = 1 à la borne parente 3.
2. Même **premier N négatif : 3**.
3. Le total de la distribution signée égale le **carré swap-face plus la lecture de clôture** aux quatre toiles jointes tabulées.
4. Une **seconde route**, depuis les liens à N corps, reproduit les grilles parentes **cellule par cellule**.
5. Le premier N auquel les deux objets *partent* : **jamais atteint** (`N part = none`).

## 2. La double paire : la plus petite toile jointe

La plus petite toile jointe de deux copies où se lit le carré swap-face est la **double paire** : deux copies du corps à deux docks du chapitre 42, appariées aux deux docks par deux liens d'appariement de type (2, 2), aucun lien frappé, habillage 9.

```python
from fractions import Fraction

print("Outil unique du carnet : Fraction (arithmetique exacte). Aucune egalite flottante.")
```

### Lecture du résultat
Aucune bibliothèque numérique : tout le carnet raisonne en rationnels exacts (`Fraction`), pour que chaque égalité affichée soit une **vérification**, pas une approximation. Les éléments de l'anneau Q(√2, √3, √6) sont portés par leurs coefficients rationnels.

```python
# Equation 68 : premieres puissances des trois blocs de la double paire.
# Regle de premiere puissance : habillage x somme des mesures des liens non frappes.
habillage = 9
S1_c1 = habillage * (67 + 6 + 6)     # chaque copie, deux liens docks non frappes
S1_c2 = habillage * (67 + 6 + 6)
S1_appariement = habillage * (4 + 4) # les deux liens d'appariement ensemble
S1 = S1_c1 + S1_c2 + S1_appariement

print("S1(c1) = 9 * (67+6+6)  =", S1_c1, "(eq. 68 annonce 711) ->", S1_c1 == 711)
print("S1(c2)                 =", S1_c2, "->", S1_c2 == 711)
print("S1(appariement) = 9*8  =", S1_appariement, "(eq. 68 annonce 72) ->", S1_appariement == 72)
print("S1 total               =", S1, "(eq. 68 annonce 1494) ->", S1 == 1494)
```

### Lecture du résultat
La règle de première puissance (habillage × somme des mesures des liens non frappés) donne aux deux copies 9·79 = 711 chacune — 67 pour le corps, deux fois 6 pour les docks — et 9·8 = 72 aux deux liens d'appariement. Le total 1494 est la première puissance de la toile jointe entière : trois blocs, aucune frappe.

```python
# Equation 69 : somme de paire et carre swap-face.
# Regle de paire : somme de paire = carre de la premiere puissance ;
# carre swap-face = carre lave (somme des carres de blocs)
#                 + termes croises avec les liens d'appariement
#                 - termes croises entre copies (chaque terme croise entre copies au signe -1).
P2 = S1 ** 2
carre_lave = S1_c1 ** 2 + S1_c2 ** 2 + S1_appariement ** 2
croix_appariement = 2 * 2 * S1_appariement * S1_c1     # deux liens, deux copies
croix_copies = 2 * S1_c1 * S1_c2                        # signe -1
Q_swap = carre_lave + croix_appariement - croix_copies

print("P2 = 1494^2                    =", P2, "(eq. 69 annonce 2232036) ->", P2 == 2232036)
print("carre lave = 711^2+711^2+72^2 =", carre_lave)
print("croix appariement = 2*2*72*711 =", croix_appariement)
print("croix copies (signe -1)        =", -croix_copies)
print("Q_swap                         =", Q_swap, "(eq. 69 annonce 209952) ->", Q_swap == 209952)
```

### Lecture du résultat
Le carré swap-face assemble trois morceaux : le **carré lavé** 1016226 (chaque bloc au carré, sommé), les **termes croisés d'appariement** +204768 (les deux liens de type (2, 2) croisés avec les deux copies), et les **termes croisés entre copies** soustraits — 1011042 — chaque terme croisé entre copies portant le signe −1, le signe de l'échange. La soustraction domine : 209952, bien en dessous du carré lavé. C'est cette structure — une somme de carrés moins les croix d'échange — qui fera tourner le signe à trois copies.

## 3. La toile paire : le carrefour avec le chapitre 42

À la toile paire de deux copies — **un seul** lien d'appariement, les liens docks non appariés frappés, blocs 657, 657 et 36 — la même règle donne le carré swap-face de l'équation 71. Les briques de ce calcul sont les comptes de paires du chapitre 42, déjà rencontrés au carnet Chaînes-03.

```python
# Equation 71 : carre swap-face a la toile paire (blocs 657, 657, 36).
b1, b2, b3 = 657, 657, 36

carre_lave_pair = b1 ** 2 + b2 ** 2 + b3 ** 2
croix_appariement_pair = 2 * 2 * b3 * b1
croix_copies_pair = 2 * b1 * b2
Q_swap_pair = carre_lave_pair + croix_appariement_pair - croix_copies_pair

print("carre lave = 657^2+657^2+36^2  =", carre_lave_pair, "(comptes du ch. 42 : 863298 + 1296)")
print("croix appariement = 2*2*36*657 =", croix_appariement_pair, "(compte du ch. 42 : 94608)")
print("croix copies (signe -1)        =", -croix_copies_pair, "(-863298)")
print("Q_swap(paire)                  =", Q_swap_pair, "(eq. 71 annonce 95904) ->", Q_swap_pair == 95904)

# Carrefour avec le carnet Chaines-03 (ch. 42) : les memes entiers servent les deux objets.
comptes_ch42 = {"carre lave (hors 36^2)": 863298, "croix appariement": 94608, "carre de 36": 1296}
print("Carrefour ch. 42 / Chaines-03 : ", comptes_ch42)
```

### Lecture du résultat
Le calcul de la toile paire réutilise **les mêmes entiers que le chapitre 42** — 863298, 94608, 1296, les comptes de paires déjà rejoués au carnet Chaînes-03. C'est le carrefour de ce chapitre : la toile jointe de deux copies n'est pas un objet neuf, elle est la toile de mesure du chapitre 42 lue par la règle de paire. Le résultat 95904 est positif : à deux copies, aucun signe négatif.

## 4. À trois copies : le signe tourne

Aux trois copies, la même règle, lue à la **chaîne de trois copies** (blocs 2133, 1971 et 1971, habillage 27, deux liens d'appariement 216 ensemble) et au **triangle** (blocs 2133, 2133 et 2133, trois liens d'appariement 324 ensemble), donne des carrés swap-face **négatifs** — valeurs citées, reproductibles par la même règle. Le premier N négatif est le même aux deux objets (équation 72) :

`N_neg(Wigner) = 3 = N_neg(swap)`, `N part = none`.

```python
# Equation 72 : le premier N negatif, le meme aux deux objets.
Q_swap_chaines3 = {1: None, 2: 209952, 3: -9595827}     # a N=2 : la double paire ; a N=3 : la chaine (CITES)
Q_swap_triangle3 = -9397539                              # a N=3 : le triangle (CITE)

print("N=2 : Q_swap(double paire) =", Q_swap_chaines3[2], "-> positif ?", Q_swap_chaines3[2] > 0)
print("N=3 : Q_swap(chaine)       =", Q_swap_chaines3[3], "-> negatif ?", Q_swap_chaines3[3] < 0)
print("N=3 : Q_swap(triangle)     =", Q_swap_triangle3, "-> negatif ?", Q_swap_triangle3 < 0)

premier_negatif_swap = min(n for n, q in Q_swap_chaines3.items() if q is not None and q < 0)
premier_negatif_wigner = 3   # marginales de Wigner negatives a N=3, pas a N=2 (CITE)
print("Premier N negatif, swap-face  :", premier_negatif_swap)
print("Premier N negatif, Wigner     :", premier_negatif_wigner)
print("N_neg(Wigner) = N_neg(swap)   ->", premier_negatif_wigner == premier_negatif_swap)
print("Premier N de parting          : aucun (N part = none, eq. 72)")
```

### Lecture du résultat
Entre deux et trois copies, le signe tourne — aux **deux** toiles de trois copies à la fois (chaîne −9595827, triangle −9397539). C'est le cœur du théorème : la négativité des marginales de Wigner au-delà de deux copies **est** la négativité du carré swap-face — même premier N négatif (3), aucun parting sur toute la plage mesurée. La structure l'explique : chaque copie ajoutée multiplie les termes croisés entre copies, tous au signe −1, jusqu'à dominer les carrés lavés.

```python
# Equation 70 : la lecture de cloture, element exact de l'anneau Q(rac2, rac3, rac6).
# Portee par ses coefficients rationnels (la partie a deux echanges, 254 enregistrements de cloture) :
R_cl = {"1": 259966108392949120000, "rac2": 1286225, "rac3": 5977541147456000, "rac6": 0}
enregistrements_cloture = 254

for base, coef in R_cl.items():
    print(f"coefficient de {base:5s} : {coef:+d}")

print("Enregistrements de cloture sommes :", enregistrements_cloture)
print("Marginales de cloture non negatives aux 22 sites (CITE) ; total = Q_swap + R_cl (eq. 71, CITE)")
print("R_cl est une cellule des tables de somme directe, re-computee sur la page not (CITE)")
```

### Lecture du résultat
La lecture de clôture vit dans l'anneau Q(√2, √3, √6) : quatre coefficients rationnels, dont un nul (√6). Elle est la seconde partie de la somme directe — le total de la distribution signée est toujours **Q_swap + R_cl** (équation 71), et ses marginales de clôture restent non négatives aux vingt-deux sites. Le signe négatif du chapitre ne vient donc jamais de la clôture : il vient du carré swap-face, c'est-à-dire de l'échange.

## 5. Le census et le résultat ouvert : quatre checks sur cinq

Le census (cité) : énumération exhaustive des cellules de la distribution signée aux quatre toiles jointes tabulées — **18 cellules à la double paire**, chaque cellule élément exact de l'anneau ; les **deux routes** d'accord cellule par cellule ; totaux de paires égaux aux sommes de paires aux **six comparaisons** ; contre-test : frapper le signe déplace le premier N négatif à la route principale (au signe +, aucun compte négatif) ; **douze contrôles** aux deux moteurs ; second moteur ; Lean énumère « 6 objects ».

Le résultat ouvert tient la fonction de Wigner comme cellule swap-face à **quatre des cinq vérifications** — et pas à la cinquième : le **recouvrement** de deux distributions part du recouvrement d'enregistrements à la somme directe, aux deux paires testées (l'électron avec lui-même, l'électron avec le corps muon). L'entrée de la fonction de Wigner type donc une **glose à sa cinquième étape**.

```python
# Les cinq checks du resultat ouvert : 4 passent, le 5e fait la glose.
checks_wigner = {
    "cellules non negatives, 16 cellules / 3685 enregistrements (electron a deux docks)": True,
    "decomposition en cellules, croix entre copies au signe -1, paire = Q_swap, cloture = R_cl": True,
    "marginales negatives a N=3, pas a N=2 ; negativite hors cellules jointes seules": True,
    "recouvrement de deux distributions = recouvrement d'enregistrements": False,
    "total = Q_swap + R_cl aux quatre toiles": True,
}
for check, ok in checks_wigner.items():
    print(("TIENT  " if ok else "PART   "), check)

nb_tiennent = sum(checks_wigner.values())
print("Checks qui tiennent :", nb_tiennent, "sur", len(checks_wigner), "-> la cellule Wigner est une GLOSSE a sa 5e etape")

# Chiffres du census (CITES) :
print("Census : 18 cellules a la double paire ; 6 comparaisons de totaux ; 12 controles aux deux moteurs ;")
print("         contre-test : frapper le signe deplace le premier N negatif (une cellule) ; Lean : 6 objects enumerated")
```

### Lecture du résultat
Quatre vérifications sur cinq font de la fonction de Wigner la cellule de lecture swap de la somme directe — et la cinquième échoue sur le **recouvrement** : l'intégrale du produit de deux fonctions de Wigner [82] ne reproduit le recouvrement d'enregistrements qu'avant la somme directe (aux deux paires testées : électron–électron, électron–muon). C'est le statut exact du théorème : **conditionnel** sur la fonction de Wigner comme cellule swap — une correspondance qui tient partout sauf au recouvrement, et qui y devient glose. Le théorème lui-même (le signe à N corps) est démontré sous cette hypothèse.
