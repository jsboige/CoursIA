## 1. L'espace des applications lineaires

Une application lineaire `T : V →ₗ[K] W` respecte additions et scalaires. Ces applications forment
elles-memes un espace vectoriel : on les additionne point par point, on les multiplie par un scalaire
point par point - et la premiere chose qu'elles font toutes, c'est envoyer le vecteur nul sur le
vecteur nul (Axler 3.10).

```lean4
#check somme_point_par_point
#check scalaire_point_par_point
#check zero_vers_zero
```

### Lecture du resultat

`somme_point_par_point (S T : V →ₗ[K] W) (v : V) : (S + T) v = S v + T v` et son pendant scalaire :
l'addition et la multiplication de `LinearMap` sont definies ponctuellement - Mathlib en fait une
instance de `Module K (V →ₗ[K] W)` (Axler 3.6) que la marche n'a pas a redefinir.
`zero_vers_zero (T : V →ₗ[K] W) : T 0 = 0` est le theoreme 3.10 d'Axler, delegate a
`LinearMap.map_zero` : toute application lineaire fixe le vecteur nul. C'est le premier geste
d'Axler au chapitre 3 : parler des applications lineaires AVANT de parler des matrices.

## 2. La composition, associative et neutre

Composer des applications lineaires reste lineaire. La composition est associative, les identites
agissent comme neutres a gauche et a droite, et elle commute aux scalaires au sens ou
`(a • S) ∘ T = a • (S ∘ T)`.

```lean4
#check associativite_composition
#check composition_identite_gauche
#check composition_identite_droite
#check scalaire_composition
```

### Lecture du resultat

Ces quatre enonces disent que les endomorphismes `V →ₗ[K] V` forment une **algebre** : un espace
vectoriel muni d'une multiplication associative et unifere. Chaque identite delegue a son organe
Mathlib (`comp_assoc`, `id_comp`, `comp_id`, `smul_comp`) - la structure porte la preuve. C'est
exactement la machinerie que le capstone
escalier (`Lean-11c`) consommera quand il parlera d'algebre d'operateurs.

## 3. Noyau et image : les deux invariants d'une application lineaire

Le noyau mesure ce que `T` ecrase ; l'image mesure ce qu'elle atteint. Chacun donne une
caracterisation complete d'une propriete : noyau nul si et seulement si injective, image pleine si
et seulement si surjective.

```lean4
#check membre_noyau
#check injectif_ssi_noyau_nul
#check membre_image
#check surjectif_ssi_image_pleine
```

### Lecture du resultat

`membre_noyau` et `membre_image` traduisent l'appartenance en equation (`T v = 0`, `T v = w`) - ce
sont les ponts entre la theorie des ensembles et le calcul. Les deux ssi transforment une propriete
fonctionnelle (injective, surjective) en propriete d'un sous-espace (noyau nul, image pleine) :
c'est ce retournement qui rend possible le theoreme du rang, qui suit.

## 4. Le theoreme fondamental de la dimension (Axler 3.21 : theoreme du rang)

Pour `T : V →ₗ[K] W` avec `V` de dimension finie : `dim V = dim ker T + dim range T`. Axler en fait
le theoreme central du chapitre 3 - notre marche le nomme `theoreme_fondamental`, delegue a
`LinearMap.finrank_range_add_finrank_ker` (qui somme dans l'autre ordre ; `omega` realigne).

```lean4
#check theoreme_fondamental
```

### Lecture du resultat

`theoreme_fondamental [Module.Finite K V] (T : V →ₗ[K] W)` : la dimension de l'espace de depart se
coupe exactement entre ce qui est tue et ce qui est atteint. Tout le reste du chapitre en est
corollaire - et la marche en donne deux immediately :

```lean4
#check vers_dimension_inferieure_non_injectif
#check vers_dimension_superieure_non_surjectif
```

### Lecture des deux corollaires

- **Vers une dimension strictement inferieure** : pas d'application lineaire injective - l'egalite
  du theoreme fondamental forcerait `dim range T > dim W`, impossible.
- **Vers une dimension strictement superieure** : pas d'application surjective - symetriquement,
  `dim ker T` devrait etre negatif.

Ce sont les deux intuitions fondamentales du chapitre 3 d'Axler : on ne compresse pas sans perte, on
ne couvre pas plus grand que soi. La marche les prouve chacune en quelques lignes depuis
`theoreme_fondamental` et `dim_sous_espace_le` (marche 1), le tout clos par `omega`.

## 5. Inversibilite : trois caracterisations equivalentes

Quand `dim V = dim W` finies, injectif, surjectif et bijectif s'equivalent ; et sur un espace de
dimension finie, un inverse a gauche suffit, un inverse a droite suffit. Le corollaire final donne
la caracterisation des espaces isomorphes.

```lean4
#check bilaterne_ssi_bijectif
#check injectif_ssi_surjectif_meme_dimension
#check bilaterne_a_droite_ssi_a_gauche
```

### Lecture du resultat

- `bilaterne_ssi_bijectif` : avoir un inverse bilateral equivaut a la bijectivite.
- `injectif_ssi_surjectif_meme_dimension` : a dimensions egales, la moitie suffit - la premiere
  moitie du theoreme fondamental disparait quand `dim ker` et `dim range` se disputent le meme total.
- `bilaterne_a_droite_ssi_a_gauche` : en dimension finie, un inverse unilatéral se complète tout
  seul - consequence directe du ssi precedent.

```lean4
#check isomorphe_ssi_meme_dimension
```

### Lecture du resultat

Le bouquet final de la marche 2 : deux espaces de dimension finie sur `K` sont isomorphes si et
seulement si ils ont meme dimension. La classification complete des espaces vectoriels de dimension
finie tient dans cette ligne - c'est elle que le capstone escalier utilisera pour transporter les
resultats spectraux d'un espace a l'autre.
