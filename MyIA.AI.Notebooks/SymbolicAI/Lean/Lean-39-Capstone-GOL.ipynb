## 5. Mesurer la surface de la sous-série

Même geste que le capstone [Lean-37](Lean-37-Capstone-Serre100.ipynb) : on ne déclare pas la surface, on la **mesure**. Trois questions : quels kernels tournent les neuf carnets, et que dit `ast` de leurs dépendances ; combien de modules le lake consacre à Life, et comment se répartissent les paires i18n FR/`_en` ; quels artefacts d'exécution le lake expose. La cellule suivante lit les carnets, le lake et les artefacts — elle ne suppose rien.

```python
# Dependances de la sous-serie : on lit les carnets, le lake et les
# artefacts, on ne suppose rien.
import ast
import glob
import json
import os
import sys

NB_DIR = os.getcwd()
if not os.path.isdir(os.path.join(NB_DIR, "GOL")):
    # Execution hors Lean/ : on remonte jusqu'au dossier qui porte GOL/
    # et conway_lean/ (le carnet vit dans Lean/).
    for _ in range(4):
        NB_DIR = os.path.dirname(NB_DIR)
        if os.path.isdir(os.path.join(NB_DIR, "GOL")):
            break
assert os.path.isdir(os.path.join(NB_DIR, "GOL")), "dossier GOL introuvable"

STDLIB = set(sys.stdlib_module_names)


def kernel(notebook):
    return notebook["metadata"].get("kernelspec", {}).get("name", "?")


def importations(notebook):
    """Importations de premier niveau + cellules illisibles par ast.

    Le compteur d'illisibles distingue « du Lean » (kernel lean4-wsl) d'une
    absence de dependance : sans lui, un carnet entier passerait pour un
    carnet sans importation.
    """
    trouves = set()
    illisibles = 0
    for cell in notebook["cells"]:
        if cell.get("cell_type") != "code":
            continue
        source = "".join(cell.get("source", []))
        try:
            arbre = ast.parse(source)
        except SyntaxError:
            illisibles += 1
            continue
        for noeud in ast.walk(arbre):
            if isinstance(noeud, ast.Import):
                trouves |= {a.name.split(".")[0] for a in noeud.names}
            elif isinstance(noeud, ast.ImportFrom) and noeud.module:
                trouves.add(noeud.module.split(".")[0])
    return trouves, illisibles


carnets = sorted(glob.glob(os.path.join(NB_DIR, "GOL", "GOL-0*.ipynb")))
print("Surface notebooks : %d carnets" % len(carnets))
for chemin in carnets:
    with open(chemin, encoding="utf-8") as f:
        nb = json.load(f)
    deps, illisibles = importations(nb)
    hors_stdlib = sorted(d for d in deps if d not in STDLIB)
    print("  %-32s kernel %-13s illisibles %2d  hors-stdlib %s"
          % (os.path.basename(chemin), kernel(nb), illisibles, hors_stdlib))

lake = sorted(glob.glob(os.path.join(NB_DIR, "conway_lean", "Conway", "Life", "*.lean")))
fr = [l for l in lake if not l.endswith("_en.lean")]
print("Lake conway_lean/Conway/Life : %d modules .lean (%d FR, %d siblings _en)"
      % (len(lake), len(fr), len(lake) - len(fr)))
for artefact in ("scripts", "patterns"):
    n = glob.glob(os.path.join(NB_DIR, "conway_lean", artefact, "*"))
    print("Artefacts conway_lean/%s : %d fichiers" % (artefact, len(n)))
```

La mesure confirme la marchabilité de la sous-série et **précise** ce que la table du README annonce :

- **Neuf carnets, trois kernels** : `lean4-wsl` pour GOL-03, GOL-05 et GOL-07 — et ce sont exactement les trois qui n'exposent **aucune** importation Python : leurs cellules illisibles par `ast` (10, 9 et 14) sont du Lean. Le compteur d'illisibles est ce qui rend ce « aucune » lisible comme du Lean, et non comme une absence de dépendance — la leçon du carnet 08 de Serre 100 vaut ici mot pour mot.
- `python3-lean` (GOL-01, GOL-06) : le CPython canonique de la série, qui **pilote** Lean par sous-processus — `lean_notebook_utils` dans les importations de GOL-01 — ou calcule côté Python (`numpy` dans GOL-06).
- `python3` (GOL-02, GOL-04, GOL-08, GOL-09) : le compagnon Golly, le carnet canons, la seconde jambe de la Loi II (GOL-08) et le capstone perplexité (GOL-09) ; le seul import `hashlife` de la sous-série est dans GOL-02 — le moteur de référence du lake, chargé depuis `conway_lean/scripts/` —, GOL-09 chargeant l'instrument du dépôt `scripts/hashlife/k_trajectory.py` par chemin (importlib), hors lake.
- **Le versant preuves pèse lourd** : les modules `.lean` de `Conway/Life/`, répartis en FR et siblings `_en` — la convention i18n #4980 y est presque saturée (comptes exacts rendus par la cellule de mesure ci-dessus), signe que ce lake est destiné à une audience externe.
- **Trois instruments, trois témoins versionnés** — les comptes bruts de la mesure (entrées du dossier) incluent aussi les `README` et le cache `__pycache__` : `hashlife.py` et les deux sondes ; `turingmachine.rle`, `otcametapixel.rle` et `p5760unitlifecell.rle` — chacun un témoin d'un barreau de l'échelle (s'y ajoute localement `gemini.rle`, gitignoré).
