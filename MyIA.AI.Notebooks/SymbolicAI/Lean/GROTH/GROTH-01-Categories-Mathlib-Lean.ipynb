## 1. Où vit la sémantique — les organes réels

Ce carnet n'implémente aucune théorie des catégories : il en **exécute** deux organes existants.

| Organe | Apport | Rôle ici |
|---|---|---|
| [`Mathlib.CategoryTheory.*`](https://leanprover-community.github.io/mathlib4_docs/Mathlib/CategoryTheory/Category/Basic.html) | les structures `Category`, `Functor`, `NatTrans`, `Adjunction`, Yoneda | les définitions exécutées, une catégorie à deux objets construite à la main |

La différence avec [Lean-15b](../Lean-15b-Lean-Grothendieck.ipynb) (qui affiche les modules du lake **en texte**) est de nature : ici le kernel charge réellement les environnements compilés — chaque signature ci-dessous est rendue par le moteur Lean, pas recopiée.

```lean4
import Mathlib.CategoryTheory.Category.Basic
import Mathlib.CategoryTheory.Types.Basic
import Mathlib.CategoryTheory.Opposites
import Mathlib.CategoryTheory.Functor.Basic
import Mathlib.CategoryTheory.NatTrans
import Mathlib.CategoryTheory.Discrete.Basic
import Mathlib.CategoryTheory.Adjunction.Basic
import Mathlib.CategoryTheory.Yoneda
import Grothendieck.Adjunction
import Grothendieck.CategoryAndSites

open CategoryTheory

#check CategoryTheory.Category
#check CategoryTheory.Functor
#check CategoryTheory.NatTrans
#check CategoryTheory.Adjunction
#check @CategoryTheory.yoneda
```

### Lecture des signatures

| Structure | Champs de données | Champs de preuve |
|---|---|---|
| `Category C` | `Hom : C → C → Type`, `id`, `comp` | `id_comp`, `comp_id`, `assoc` — la structure **porte ses lois** |
| `Functor C D` | `obj : C → D`, `map` | `map_id`, `map_comp` — un foncteur est un diagramme qui commute |
| `NatTrans F G` | `app` (les composantes) | `naturality` — tous les carrés commutent |
| `Adjunction` | `homEquiv` (une famille d'équivalences Hom) | `unit`/`counit` et les identités triangulaires |

C'est le style Mathlib : pas de théorème décoratif, des **lois dans les structures** — impossible de construire un foncteur sans prouver qu'il respecte la composition. Toute la suite du carnet consiste à habiter ces champs.

## 2. Trois catégories de Mathlib, sans en construire aucune

Avant de bâtir, vérifier que l'environnement en contient déjà : la catégorie des types, un ensemble discret vu comme catégorie, et la catégorie des endofoncteurs de `Type`. Cette dernière permet aussi de distinguer, dès maintenant, `Type ⥤ Type` de la catégorie des préfaisceaux `Cᵒᵖ ⥤ Type`.

```lean4
open CategoryTheory

example : Category (Type) := inferInstance
example : Category (Discrete ℕ) := inferInstance
example : Category (Type ⥤ Type) := inferInstance

#check (fun n : ℕ => n + 1 : ℕ → ℕ)
#check (𝟭 (Type) : Type ⥤ Type)
```

### Interprétation

| Instance | Objets | Flèches | Ce qu'elle enseigne |
|---|---|---|---|
| `Category (Type)` | les types | les fonctions | la catégorie de base, celle où vivent les valeurs des diagrammes et des préfaisceaux |
| `Category (Discrete ℕ)` | les entiers | uniquement les identités | un ensemble « discret » : la catégorie la plus pauvre, filet de sécurité pédagogique |
| `Category (Type ⥤ Type)` | les endofoncteurs de `Type` | les transformations naturelles | une catégorie **d'objets structurés**, distincte des préfaisceaux |

La ligne `#check (fun n => n + 1 : ℕ → ℕ)` donne l'exemple de fonction que `TypeCat.ofHom` promouvra en flèche dans la section 4. Dans `Type ⥤ Type`, le premier `Type` est la catégorie source et le second la catégorie but ; les objets du résultat sont les foncteurs `Type → Type`. Un préfaisceau sur une catégorie `C` est au contraire un foncteur **contravariant** `Cᵒᵖ ⥤ Type` : il renverse les flèches. Ce n'est donc pas un objet de `Type ⥤ Type`.

## 3. Une catégorie à deux objets, construite à la main

La flèche `0 → 1` — la plus petite catégorie qui ne soit pas discrète. Deux objets, trois flèches (`id₀`, `id₁`, et l'unique `a01 : 0 → 1`), et **aucune** flèche `1 → 0` : le type vide `Empty` encode l'impossible. C'est la « flèche générique » (*walking arrow*) : un diagramme de forme `0 → 1` dans une catégorie `C` est exactement un foncteur depuis ici.

```lean4
open CategoryTheory

inductive Ob where
  | zero : Ob
  | one : Ob

def Fleche : Ob → Ob → Type
  | .zero, .zero => Unit
  | .zero, .one  => Unit
  | .one,  .zero => Empty
  | .one,  .one  => Unit

/-- L'unique flèche non triviale `0 → 1`. -/
def a01 : Fleche .zero .one := ()

def idF : (X : Ob) → Fleche X X
  | .zero => ()
  | .one  => ()

def compF : ∀ {X Y Z : Ob}, Fleche X Y → Fleche Y Z → Fleche X Z
  | .zero, .zero, .zero, _, _ => ()
  | .zero, .zero, .one,  _, _ => ()
  | .zero, .one,  .one,  _, _ => ()
  | .one,  .one,  .one,  _, _ => ()
  | .zero, .one,  .zero, _, g => Empty.elim g
  | .one,  .one,  .zero, _, g => Empty.elim g
  | .one,  .zero, .zero, f, _ => Empty.elim f
  | .one,  .zero, .one,  f, _ => Empty.elim f

instance catFleche : Category Ob where
  Hom := Fleche
  id := idF
  comp := @compF
  id_comp := by intro X Y f; cases X <;> cases Y <;> cases f <;> rfl
  comp_id := by intro X Y f; cases X <;> cases Y <;> cases f <;> rfl
  assoc := by
    intro W X Y Z f g h
    cases W <;> cases X <;> cases Y <;> cases Z <;>
      cases f <;> cases g <;> cases h <;> rfl

#check catFleche
#check a01
```

### Interprétation : les trois lois sont vérifiées, pas décrétées

Les clauses `Empty.elim` méritent une lecture : elles sont **inatteignables**. Une composition `(1, 1, 0)` exigerait une flèche `1 → 0` — le type `Empty` n'a aucun habitant, la preuve d'impossibilité est mécanique. Les trois champs de loi (`id_comp`, `comp_id`, `assoc`) se ferment par énumération des cas : sur une catégorie à deux objets, chaque loi se réduit à un nombre fini de compositions toutes réduites à l'identité — c'est la vertu des exemples **petits et finis**.

Ce que la carte #17545 appelle « la marche manquante » commence ici : le lecteur qui verra `Spec ⊣ Γ` dans le companion aura **déjà construit** une catégorie de ses mains, loi par loi.

## 4. Foncteurs : un diagramme covariant sur la flèche

Un foncteur `P : Ob ⥤ Type` est un **diagramme covariant** : à une flèche `f : X ⟶ Y`, il associe une fonction `P.obj X → P.obj Y`, dans le même sens. Ici `P` envoie `0` sur `ℕ`, `1` sur `Bool`, et la flèche `a01` sur la parité.

Un préfaisceau sur `Ob` serait au contraire un foncteur `Obᵒᵖ ⥤ Type` : sa catégorie source renverse les flèches. `P` n'est donc pas un préfaisceau ; il prépare précisément la distinction de variance nécessaire à Yoneda.

```lean4
open CategoryTheory

def PObj : Ob → Type
  | .zero => ℕ
  | .one  => Bool

def PMap (X Y : Ob) (f : Fleche X Y) : PObj X → PObj Y :=
  match X, Y with
  | .zero, .zero => fun x => x
  | .zero, .one  => fun (x : ℕ) => (x % 2 == 0)
  | .one,  .one  => fun x => x
  | .one,  .zero => fun _ => Empty.elim f

/-- Le diagramme covariant de parité : `ℕ` en 0, `Bool` en 1, la flèche `a01` devient `n % 2 == 0`. -/
def P : Ob ⥤ Type where
  obj := PObj
  map {X Y} f := TypeCat.ofHom (PMap X Y f)
  map_id := by
    intro X
    cases X <;> rfl
  map_comp := by
    intro X Y Z f g
    cases X <;> cases Y <;> cases Z <;> cases f <;> cases g <;> rfl

#check P
#eval (show (ℕ → Bool) from P.map a01) 5
#eval (show (ℕ → Bool) from P.map a01) 8
#eval (show (ℕ → ℕ) from P.map (idF .zero)) 4
```

### Interprétation : la parité est une image de flèche

Les deux premiers `#eval` rendent `false` puis `true` : la flèche `a01` **agit** — elle transporte un entier vers sa parité. Le troisième rend `4` : l'identité de `0` agit comme l'identité de `ℕ`, parce que le champ `map_id` l'exigeait. Après les `cases`, les constructions sont ici définitionnellement égales : `rfl` ferme les lois. Cette forme courte est propre à cet exemple entièrement explicite ; elle n'est pas une recette universelle pour tout foncteur.

La catégorie `Type` de ce pin Mathlib enveloppe les fonctions dans `TypeCat.Fun`, puis dans `TypeCat.Hom`. `TypeCat.ofHom` construit donc la flèche native ; une égalité de telles flèches n'est pas directement un but de fonctions sur lequel `funext` s'appliquerait.

Le pont vers la théorie des sites se fera par la catégorie opposée : un préfaisceau `Obᵒᵖ ⥤ Type` et notre diagramme `Ob ⥤ Type` vivent dans des catégories de foncteurs différentes.

## 5. Transformations naturelles : le carré qui commute

Une transformation naturelle `α : F ⟶ G` donne une flèche `α X : F.obj X → G.obj X` pour chaque objet, avec une loi : pour **toute** flèche `f : X ⟶ Y` de la catégorie source, le carré commute. On construit la plus simple — l'identité — à la main, pour voir la loi dans le code.

```lean4
open CategoryTheory

/-- La transformation naturelle identité de P, champ par champ. -/
def idNatP : P ⟶ P where
  app X := TypeCat.ofHom id
  naturality := by
    intro X Y f
    rfl

#check idNatP
#check (𝟙 P : P ⟶ P)
```

### Interprétation

Le champ `naturality` de `α : F ⟶ G` impose `F.map f ≫ α.app Y = α.app X ≫ G.map f` : on parcourt d'abord la flèche `f`, puis la composante en `Y`, ou d'abord la composante en `X`, puis `G.map f`. Pour `F = G = P` et `α` identité, les deux côtés sont définitionnellement égaux : la preuve est `rfl`. La deuxième ligne demande au **produit** la transformation identité `𝟙 P` : Mathlib la construit pour toute catégorie de foncteurs.

Le mot « naturel » de la littérature a ici un sens précis et vérifiable : pas d'inscription manuscrite, un champ de structure que le kernel a vérifié avant de nous rendre la main.

## 6. Adjonctions : lecture dans l'organe du lake

L'adjonction `L ⊣ R` équilibre « construire librement » et « oublier » : une équivalence naturelle `Hom_D(L X, Y) ≃ Hom_C(X, R Y)`. Le lake `grothendieck_lean` possède déjà le parcours complet — module `Grothendieck/Adjunction.lean` (Partie 25) — qui ré-expose Mathlib en parcours organisé. On l'importe et on lit ses énoncés directement.

```lean4
open CategoryTheory

#check @CategoryTheory.Adjunction.homEquiv
#check @CategoryTheory.Adjunction.unit
#check @CategoryTheory.Adjunction.counit

#check @Grothendieck.Adjunction.left_triangle_components_apply
#check @Grothendieck.Adjunction.homEquiv_unit_apply
#check @Grothendieck.Adjunction.rightAdjoint_preserves_limits
```

### Interprétation : trois lectures d'une même structure

| Énoncé du lake | Ce qu'il dit en français |
|---|---|
| `homEquiv` | l'équivalence Hom naturelle — la **définition** de l'adjonction |
| `left_triangle_components_apply` | l'identité triangulaire à l'objet `X` : `L.map (h.unit.app X) ≫ h.counit.app (L.obj X) = 𝟙 (L.obj X)` — composer ces deux flèches redonne l'identité sur `L.obj X` |
| `homEquiv_unit_apply` | `(h.homEquiv X Y) f = h.unit.app X ≫ R.map f` : l'unité détermine le sens direct de l'équivalence Hom |
| `rightAdjoint_preserves_limits` | l'adjoint à droite **préserve les limites** — le théorème qui fait travailler l'adjonction en géométrie |

Le module du lake ajoute à Mathlib la mise en parcours : chaque énoncé est relié à son usage grothendieckien (`Spec ⊣ Γ`, faisceautisation ⊣ inclusion, fibre ⊣ gratte-ciel — cités dans sa docstring). C'est la profondeur que ce carnet n'a pas à re-démontrer : il la **cite par ses énoncés exécutés**.

## 7. Le pont vers les sites : les trois axiomes d'une topologie

Pourquoi ces préparatifs : la Partie 1 du lake (`CategoryAndSites`) définit ce qu'est une **topologie de Grothendieck** — des cribles couvrants sur une catégorie. Les trois axiomes sont des énoncés sur `Hom` et le pullback : sans le vocabulaire des sections 3-5, ils sont illisibles. Avec lui, ils se lisent.

```lean4
open CategoryTheory

#check @Grothendieck.top_covers
#check @Grothendieck.pullback_cover
#check @Grothendieck.transitivity
#check @Grothendieck.trivial_eq_bot
#check @Grothendieck.discrete_eq_top
#check @CategoryTheory.yoneda
```

### Interprétation

Les trois premiers énoncés sont les **trois axiomes** d'une topologie de Grothendieck : le crible maximal couvre toujours ; les cribles couvrants sont stables par pullback (un recouvrement se restreint à tout voisinage) ; la couverture est locale (transitivité). Les deux suivants encadrent l'espace des topologies : la triviale est le **bottom** du treillis, la discrète le **top** — exactement le treillis que [Lean-15b](../Lean-15b-Lean-Grothendieck.ipynb) explore en texte, ici rendu par le kernel.

Et `yoneda` referme la boucle ouverte en section 4 : sa signature `C ⥤ Cᵒᵖ ⥤ Type` montre explicitement que `yoneda.obj X` est un préfaisceau contravariant sur `C`, avec `(yoneda.obj X).obj (op Y) = Y ⟶ X`. Un crible sur `X` sélectionne des flèches dans ces Hom-ensembles. Notre `P : Ob ⥤ Type`, lui, reste covariant : ce n'est ni `yoneda.obj X` ni un préfaisceau, et il ne doit pas être assimilé aux objets que les topologies de Grothendieck organiseront.

Miroir du companion : un théorème Lean ne vaut que par ses axiomes. Demandons au kernel la liste exacte de ceux qu'emportent nos trois constructions.

```lean4
#print axioms catFleche
#print axioms P
#print axioms idNatP
```

### Lecture de la sortie

La hiérarchie est nette : `catFleche` ne dépend **d'aucun axiome** ; `P` dépend de `propext` et `Quot.sound` ; `idNatP` y ajoute `Classical.choice`. Ces trois-là sont les axiomes fondateurs standard du calcul des constructions de Lean — ils portent l'égalité extensionnelle des propositions, le quotient, et la logique classique — et aucun d'eux n'est une dette de preuve : ni `sorryAx` transitif, ni `native_decide`. Les lois de `catFleche`, `P` et `idNatP` restent vérifiées par le kernel ; les énoncés du lake, eux, sont cités depuis leurs modules.

La commande n'imprime ici les axiomes que de nos trois constructions : elle ne constitue pas un audit d'intégrité des modules importés, dont le companion [Lean-15c](../Lean-15c-Lean-Grothendieck-Companion.ipynb) fait son propre inventaire au §10.
