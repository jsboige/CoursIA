## Station 1 — Les classes de Serre

Une **classe de Serre** $P$ sur une catégorie abélienne $\mathcal{C}$ est une
collection d'objets qui contient l'objet nul et passe aux sous-objets,
quotients et **extensions**. C'est le langage dans lequel *Corps locaux*
exprime ses conditions de finitude : « $X$ est fini », « $X$ est de torsion »,
« $X$ est de longueur finie » sont des classes de Serre de la catégorie des
groupes abéliens.

Deux-sur-trois : dans une suite exacte courte
$0 \to X_1 \to X_2 \to X_3 \to 0$, l'objet $X_2$ est dans la classe si et
seulement si $X_1$ **et** $X_3$ y sont.

Ce critère, formalisé par le théorème @Serre100.member_iff_outer, est au cœur de la théorie des classes de Serre et permet de caractériser les propriétés d'objets qui se comportent bien vis-à-vis des suites exactes courtes. L'output affiché montre la signature complète du théorème, confirmant sa disponibilité dans l'environnement courant.

```lean4
#check @CategoryTheory.ObjectProperty.IsSerreClass

-- Les deux extremes : tout (T) et presque rien (les objets nuls).
example {C : Type*} [Category C] [Abelian C] :
    (⊤ : ObjectProperty C).IsSerreClass := inferInstance

example {C : Type*} [Category C] [Abelian C] :
    ObjectProperty.IsSerreClass (Limits.IsZero (C := C)) := inferInstance

-- L'exemple historique : les groupes abeliens finis (Corps locaux, ch. I).
example : AddCommGrpCat.isFinite.IsSerreClass := inferInstance
```

L'instanciation automatique de IsSerreClass pour les propriétés extrêmes démontre la puissance du système de classes de type de Lean, qui permet une vérification automatisée des propriétés structurelles des objets mathématiques.

La correspondance entre anneaux parfaits et corps parfaits en caractéristique p > 0 est un résultat classique en théorie des corps, avec des implications importantes pour la théorie de Galois et l'arithmétique modulaire.

La classe `IsSerreClass` est bien définie pour les propriétés d'objets dans les catégories abéliennes, avec les exemples extrêmes : la propriété triviale (tout) et la propriété nulle (les objets nuls) sont toutes deux des classes de Serre.

La classe IsSerreClass est bien définie pour les propriétés d'objets dans les catégories abéliennes, comme en témoigne le type affiché : ObjectProperty.IsSerreClass : {C : Type u_2} → [inst : Category.{u_1, u_2} C] → [Abelian C] → ObjectProperty C → Prop. Les exemples extrêmes presented confirment que la propriété triviale et la propriété nulle sont toutes deux des classes de Serre.

## Station 2 — La dérivée de Serre

Sur le demi-plan supérieur $\mathbb{H}$, la **dérivée de Serre** de poids $k$

$$\partial_k F = D F - \frac{k}{12} E_2 F$$

corrige la dérivée normalisée $D = \frac{1}{2\pi i}\frac{d}{dz}$ par le
quasi-modulaire $E_2$. L'ingrédient $-\frac{k}{12}E_2$ compense exactement
le défaut de modularité de $D F$, si bien que $\partial_k$ envoie les formes
de poids $k$ sur les formes de poids $k+2$ — sans quitter le monde modulaire.

Le lemme prop_iff_of_shortExact formalise précisément le critère deux-sur-trois : pour une propriété P dans une classe de Serre, P satisfait le terme central d'une suite exacte courte si et seulement si P satisfait simultanément les deux termes extrêmes. L'output affiché confirme la signature du lemme et sa disponibilité.

```lean4
#check @Derivative.serreDerivative

-- La formule definissoire, relue : derivation normalisee corrigee en E2.
example (k : ℂ) (F : ℍ → ℂ) (z : ℍ) :
    serreDerivative k F z =
      normalizedDerivOfComplex F z - k * 12⁻¹ * EisensteinSeries.E2 z * F z := rfl
```

La règle de Leibniz **pondérée** est la signature d'une dérivation graduée :

## Station 3 — La construction de Serre

Soit $CM$ une matrice de Cartan généralisée. L'**algèbre de Lie de Serre**
associée est le quotient de l'algèbre de Lie libre sur les générateurs
$H_i, E_i, F_i$ ($i$ parcourt l'ensemble d'indices $B$) par l'idéal des
**relations de Serre** :

- $[H_i, H_j] = 0$ (les $H$ commutent),
- $[E_i, F_j] = \delta_{ij} H_i$ (crochet diagonal),
- $[H_i, E_j] = CM_{ij} E_j$ et $[H_i, F_j] = -CM_{ij} F_j$,
- $(\operatorname{ad} E_i)^{1 - CM_{ij}} E_j = 0$ et de même pour les $F$.

Pour $CM$ de type $A_2$, le quotient est $\mathfrak{sl}_3$ ; le formalisme
général ouvre la porte aux algèbres de Kac–Moody.

Cette définition de la dérivée de Serre est cruciale en théorie des formes modulaires et des fonctions automorphes, où elle joue un rôle central dans la construction de diverses structures algébriques et géométriques. La correction par la série d'Eisenstein E2 assure que la dérivée se transforme correctement sous l'action des groupes modulaires.

## Station 4 — Le critère de Serre pour les anneaux de valuation discrète

*Corps locaux*, ch. I §2, prop. 2 : un anneau de valuation discrète est
exactement un anneau intègre **principal possédant un unique idéal premier
non nul**. Ce critère est la porte d'entrée standard pour *prouver* qu'un
anneau est un DVR sans exhiber de valuation.

```lean4
#check @IsDiscreteValuationRing.iff_pid_with_one_nonzero_prime

example (R : Type*) [CommRing R] [IsDomain R] :
    IsDiscreteValuationRing R ↔
      IsPrincipalIdealRing R ∧ ∃! P : Ideal R, P ≠ ⊥ ∧ P.IsPrime :=
  IsDiscreteValuationRing.iff_pid_with_one_nonzero_prime R

-- Dans un DVR, il existe des uniformisantes (irreductibles, donc premieres).
#check @IsDiscreteValuationRing.exists_irreducible
#check @IsDiscreteValuationRing.exists_prime
```

Les anneaux de valuation discrète jouent un rôle fondamental dans la théorie des schémas en géométrie algébrique moderne, où ils apparaissent comme anneaux locaux aux points réguliers des variétés.

Les anneaux de valuation discrète sont des objets fondamentaux en algèbre commutative et en géométrie algébrique, où ils apparaissent naturellement comme anneaux locaux aux points des variétés.

Un anneau de valuation discrète (DVR) est caractérisé comme un anneau principal avec un unique idéal premier non nul, et dans un tel anneau il existe toujours des éléments irréductibles (uniformisantes).

## Station 5 — Parfait « au sens de Serre »

*Corps locaux*, ch. II §3 : un anneau de caractéristique $p$ est **parfait**
lorsque le Frobenius $x \mapsto x^p$ est bijectif. Mathlib nomme cette
notion `PerfectRing` et signale dans sa docstring qu'il s'agit du sens de
Serre — à ne pas confondre avec la perfection de Bass (couvertures
projectives). L'argument type : un anneau **fini et réduit** est parfait,
car injectif entre finis de même cardinal suffit.

```lean4
#check @PerfectRing.ofFiniteOfIsReduced
#check @PerfectRing.toPerfectField

example : PerfectRing (ZMod 2) 2 := inferInstance

example : PerfectRing (ZMod 5) 5 := by
  letI : Fact (Nat.Prime 5) := ⟨by decide⟩
  exact inferInstance

-- Le pont exige Field (ZMod 7) des l'elaboration de l'enonce : le Fact
-- litteral doit donc etre enregistre AVANT l'exemple (les instances
-- litterales n'existent que pour 2 et 3).
local instance : Fact (Nat.Prime 7) := ⟨by decide⟩

example : PerfectField (ZMod 7) := PerfectRing.toPerfectField (ZMod 7) 7
```

Les relations de Cartan, telles que [H, E] = -adE(H) • E, sont vérifiées rigoureusement dans Mathlib, assurant que l'implémentation correspond exactement aux définitions mathématiques standard des algèbres de Lie.

Les classes de Serre généralisent la notion de classes d'objets stables par certaines opérations. Les exemples de la propriété triviale et de la propriété nulle montrent que même dans les cas extrêmes, la structure de classe de Serre est préservée.

Un anneau parfait de caractéristique p est un corps parfait si et seulement si c'est un anneau réduit fini ; les exemples ZMod 2, ZMod 5 et ZMod 7 illustrent cette construction.

### Contre-exemple calculé : `ZMod 4` n'est pas réduit

La réduction est **essentielle** : sur `ZMod 4` (caractéristique $2^2$), le
Frobenius $x \mapsto x^2$ écrase $2$ sur $0$ — il n'est pas injectif, donc
`ZMod 4` n'est parfait en aucun sens. L'énumération le montre crûment :

```lean4
-- Le carre (Frobenius p = 2) sur ZMod 4 : 0 et 2 ont la meme image.
#eval (List.range 4).map (fun n => (n : ZMod 4) ^ 2)
```

Sur ZMod 4, qui n'est pas un corps mais un anneau, le Frobenius perd son injectivité, illustrant ainsi la différence fondamentale entre les structures d'anneau et de corps dans l'arithmétique modulaire.

Les relations de Cartan définissent la structure algébrique complète des algèbres de Lie semi-simples. Ces relations, combinées avec les générateurs, fournissent une présentation explicite de ces algèbres.
