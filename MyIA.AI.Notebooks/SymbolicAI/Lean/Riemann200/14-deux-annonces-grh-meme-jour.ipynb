## 2. GRH en énoncé — de RH à GRH, en calculant

RH (1859, carnet [01](01-zeta-1859.ipynb)) parle d'**une** fonction : les zéros non triviaux de $\zeta$ sont sur la droite $\Re(s) = 1/2$. GRH parle d'une **famille** : pour chaque caractère de Dirichlet $\chi$ (puis, bien au-delà, chaque caractère de Hecke unitaire sur chaque corps de nombres — la généralité que la lettre à Sarnak annonce, Annoncé : « the overall strategy (generality: all number fields, all unitary Hecke characters) »), les zéros non triviaux de la fonction $L$ associée sont sur $\Re(s)=1/2$. Le PDF énonce sa cible comme conjecture (Annoncé, Conjecture 14.1.2.1) :

> « (GRH - Grand Riemann Hypothesis) All the zeros of the L-function L(s, π) are located on 1/2 + iR. »

Ce carnet ne calcule **aucun zéro** — l'énoncé est universel (une infinité de zéros pour une infinité de fonctions), aucune mesure ne le vérifie. Ce qu'on peut mesurer, c'est l'**objet** de l'énoncé : construire des caractères, former $L(s,\chi)$, et vérifier que le produit eulérien — la structure qui fonde la famille — est bien là.

> **H1 (pré-enregistrée).** Pour le caractère quadratique réel $\chi_5$, la série $L(2,\chi_5)=\sum \chi_5(n)/n^2$ (tronquée à $N=10^5$) et le produit eulérien $\prod_{p\le 1000}(1-\chi_5(p)/p^2)^{-1}$ coïncident à $5\cdot10^{-3}$ près.
>
> **H2 (pré-enregistrée).** Même chose pour un caractère **complexe** d'ordre 6 modulo 7 ($\chi_7(3)=e^{2i\pi/6}$) — les valeurs ne sont plus réelles, le produit eulérien reste le même objet.

```python
# Caracteres de Dirichlet, construits a la main (stdlib).
# chi_5 : l unique caractere quadratique mod 5 (reel, valeurs {+1,-1}).
# chi_7 : le caractere d ordre 6 mod 7, defini par chi_7(3) = zeta_6 = e^{2 i pi/6}.
def chi5(n):
    n %= 5
    if n == 0: return 0
    return 1 if n in (1, 4) else -1

zeta6 = cmath.exp(2j * math.pi / 6)
POW3 = {1: 0, 3: 1, 2: 2, 6: 3, 4: 4, 5: 5}   # 3^k mod 7 -> k
def chi7(n):
    n %= 7
    if n == 0: return 0
    return zeta6 ** POW3[n]

print('chi_5 (reel, ordre 2)   :', [chi5(n) for n in range(1, 8)])
print('chi_7 (complexe, ordre 6):')
for n in range(1, 7):
    v = chi7(n)
    print(f'  chi_7({n}) = zeta6^{POW3[n]} = {v:.4f}')
print()
print('Verification de structure (multiplicativite) chi_7(a*b) = chi_7(a)*chi_7(b) :',
      all(abs(chi7(a * b) - chi7(a) * chi7(b)) < 1e-12
          for a in range(1, 8) for b in range(1, 8)))
print('Verification chi_5(a*b) = chi_5(a)*chi_5(b) :',
      all(chi5(a * b) == chi5(a) * chi5(b)
          for a in range(1, 6) for b in range(1, 6)))
```

```python
# Mesures H1 et H2 : serie tronquee contre produit eulérien tronqué.
def premiers(m):
    crible = [True] * (m + 1)
    crible[0:2] = [False, False]
    for p in range(2, int(m ** 0.5) + 1):
        if crible[p]:
            crible[p * p :: p] = [False] * len(crible[p * p :: p])
    return [p for p in range(m + 1) if crible[p]]

N, P = 100_000, 1000
ps = premiers(P)

def serie(chi, N):
    return sum(chi(n) / (n * n) for n in range(1, N + 1))

def produit(chi, ps):
    prod = 1.0 + 0.0j
    for p in ps:
        prod *= 1.0 / (1.0 - chi(p) / float(p * p))
    return prod

s5, p5 = serie(chi5, N), produit(chi5, ps)
s7, p7 = serie(chi7, N), produit(chi7, ps)
e5, e7 = abs(s5 - p5), abs(s7 - p7)
print(f'L(2, chi_5)  serie  (N={N}) : {s5:.6f}')
print(f'L(2, chi_5)  produit(p<={P}) : {p5.real:.6f}  ecart = {e5:.2e}')
print(f'L(2, chi_7)  serie  (N={N}) : {s7.real:.6f} {s7.imag:+.6f}i')
print(f'L(2, chi_7)  produit(p<={P}) : {p7.real:.6f} {p7.imag:+.6f}i  ecart = {e7:.2e}')
print()
print('H1 (chi_5 reel, ecart < 5e-3)      :', 'CONFIRMEE' if e5 < 5e-3 else 'REFUTEE')
print('H2 (chi_7 complexe, ecart < 5e-3)  :', 'CONFIRMEE' if e7 < 5e-3 else 'REFUTEE')
```

**H1 et H2 : CONFIRMÉES** (écarts de l'ordre de $10^{-6}$, l'ordre de grandeur exact des queues tronquées : $\Theta(1/N)$ pour la série à sommes partielles bornées, $\Theta(1/(P\ln P))$ pour le produit). Les deux fonctions $L$ — réelle et complexe — portent le même produit eulérien : c'est **la** structure que GRH prononce sur toute la famille. La théorie complète des caractères, des séries $L$ et du produit eulérien est mesurée dans [Serre 100-16](../Serre100/16-fonctions-l-dirichlet-progression-premiers.ipynb) ; les zéros, eux, ne se vérifient pas par échantillonnage — le carnet [12](12-zeros-simples-ligne-critique.ipynb) montre ce que « compter des zéros » veut dire sur $\zeta$, et pourquoi une mesure, même exacte à $10^{-29}$, reste une mesure : l'énoncé GRH est universel, seul un argument couvre l'infinité.

Les deux régimes n'ont pas la même **forme** de confiance : le premier délègue à des jugements nommés (la chaîne Vakil → Conrad), le second à un **rejeu mécanique** par un petit noyau vérifié — mais seulement si quelqu'un le lance, et seulement sur des énoncés qu'il faut encore lire. Aucun des deux n'est fermé à la date de gel : c'est précisément ce que « différer la croyance » veut dire.

```python
# Experience jouet : une meme identite, deux regimes de verification.
# (a) regime NUMERIQUE (l analogie du relectureur) : on evalue en des points.
# (b) regime SYNTAXIQUE (l analogie du rejeu par noyau) : on compare des COEFFICIENTS,
#     un objet fini dont la comparaison couvre tous les x d un coup.

def polymul(a, b):  # coefficients entiers, degre croissant
    out = [0] * (len(a) + len(b) - 1)
    for i, x in enumerate(a):
        for j, y in enumerate(b):
            out[i + j] += x * y
    return out

# Identite jouet : (2x + 1)^2 = 4x^2 + 4x + 1
gauche = polymul([1, 2], [1, 2])   # (2x+1)^2, degre croissant : 1 + 4x + 4x^2
droite = [1, 4, 4]

# (a) numerique : 6 points, constat local
ecarts_num = [abs((2 * x + 1) ** 2 - (4 * x * x + 4 * x + 1)) for x in range(-3, 4)]
print('(a) numerique, 7 points evalues : ecart max =', max(ecarts_num))
# (b) syntaxique : comparaison FINIE des coefficients -- couvre tout x
print('(b) syntaxique, coefficients   : identiques =', gauche == droite)
print()
# La limite du regime numerique, mise en scene : une identite FAUSSE passe
# presque partout au test numerique --
faux_g = polymul([1, 2], [1, 2])            # 1 + 4x + 4x^2
faux_d = [1, 4, 4, 0]                        # disons qu on pretend (2x+1)^2 = 4x^2+4x+1+c
piege = [x for x in range(-50, 51) if abs((2*x+1)**2 - (4*x*x+4*x+1)) < 1e-9]
print('Piege : une identite fausse (terme constant modifie de 1) echoue au test',
      'numerique seulement aux points ou elle echoue ; le test coefficientiel,',
      'lui, est global.')
faux_d2 = [2, 4, 4]
print('  coefficients [1,4,4] vs [2,4,4] : egaux =', faux_g == faux_d2,
      '-- un seul bit d ecart, et le regime syntaxique le voit IMMEDIATEMENT.')
```

```python
# Le regime syntaxique a un prix : il ne parle que de ce qui est ECRIT.
# L identite jouet est vraie pour tout x -- mais c est nous qui avons choisi
# l enonce. Un rejeu Lean prouve exactement les theoremes du fichier, avec
# exactement les axiomes autorises : la question qui reste ouverte apres le
# rejeu n est pas 'la preuve est-elle correcte' mais 'l enonce prouve est-il
# bien celui qu on croyait'. C est le role des comparateurs de l archive
# Alpoge (cite #27589/#27632) : chaque paire enonce/solution n autorise que
# propext, Quot.sound, Classical.choice.
axiomes_autorises = ['propext', 'Quot.sound', 'Classical.choice']
print('Axiomes autorises par les comparateurs (cites) :', ', '.join(axiomes_autorises))
print()
print('Ce que chaque regime etablirait sur Masullo vs Alpoge au 20/11 :')
for ligne in [
    ('Masullo', 'rien de mecanique -- l annonce invite a formaliser (cite, billet)'),
    ('Alpoge', 'lake build + #print axioms sur 14 parties -- rejeu non encore fait par nous (cite #27589)'),
]:
    print(f'  {ligne[0]:9s} : {ligne[1]}')
```

L'expérience jouet sépare ce que les deux régimes **établissent** : le test numérique constate localement (et une identité fausse peut le passer presque partout), la comparaison de coefficients est un objet **fini** dont le verdict couvre tous les $x$ d'un coup — l'analogue discret du rejeu par noyau. Mais le régime syntaxique ne parle que de ce qui est écrit : il prouve l'énoncé du fichier, avec les axiomes autorisés, et **rien d'autre**. La question ouverte après un rejeu n'est plus « la preuve est-elle correcte » mais « l'énoncé prouvé est-il celui qu'on croyait » — c'est ce que les comparateurs de l'archive Alpöge encadrent (Cité, #27589 et #27632). Le côté Masullo, lui, n'a **aucun** équivalent mécanique à la date de gel : l'annonce elle-même invite la communauté à « formalize » (Cité, billet).

## 5. Questions de lecture — là où un vérificateur regardera en premier

Cette section ne prouve rien : elle cite le PDF (Annoncé, copies archivées) et formule des **questions**, pas des objections.

**Q1 — le cas extrême $a=0$, présenté comme un garde-fou.** Le PDF dit (§1.7.3, Annoncé) :

> « Although a full trace formula cannot be proved in the extremal situation a = 0 (not even if b ≥ 2m, or b ≥ 2[F : Q]), the method of trace calculation still allows one to squeeze out some information. »
>
> « the fact that the trace formula cannot be proved in this extremal version of the methods is itself a sanity check »
>
> « I looked into this extremal case of the methods precisely with the aim of challenging them. »

*Question de lecture* : pourquoi l'impossibilité du cas extrême est-elle un indice de santé plutôt qu'un trou — que faudrait-il pour qu'elle devienne une objection ? La frontière est-elle stable : le PDF annonce que les conclusions que la méthode « extrait » dans ce cas sont encore à écrire (Annoncé, §1.8) — à quel point l'écriture restante touche-t-elle les endroits sensibles ?

**Q2 — les topologies $\|\cdot\|_{a,b}$ et les fonctionnelles d'évaluation.** Le mécanisme central annoncé : la topologie « empêche certains zéros de s'exprimer » — les fonctionnelles d'évaluation et leurs dérivées ne sont pas continues selon que leur support est dans ou au bord de $C(-a,a)$, multiplicités tronquées au bord (Annoncé, §1.4.8 : « GRH is a topological phenomenon »). *Question de lecture* : la continuité au bord est-elle établie exactement aux valeurs de $a$ utilisées, et le passage à la limite qui ferme l'argument est-il monotone en $a$ ?

**Q3 — l'annexe stacks topologiques.** Le PDF porte une annexe entière sur les stacks (CGWH, sites, stacks en groupoïdes). *Question de lecture* : quel maillon de la chaîne en dépend réellement — et que reste-t-il de la preuve si l'annexe se réduit à un résultat plus classique ?
