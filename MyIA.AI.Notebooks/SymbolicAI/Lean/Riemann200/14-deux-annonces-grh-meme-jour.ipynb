## 2. GRH en énoncé — de RH à GRH, en calculant

RH (1859, carnet [01](01-zeta-1859.ipynb)) parle d'**une** fonction : les zéros non triviaux de $\zeta$ sont sur la droite $\Re(s) = 1/2$. GRH parle d'une **famille** : pour chaque caractère de Dirichlet $\chi$ (puis, bien au-delà, chaque caractère de Hecke unitaire sur chaque corps de nombres — la généralité que la lettre à Sarnak annonce, Annoncé : « the overall strategy (generality: all number fields, all unitary Hecke characters) »), les zéros non triviaux de la fonction $L$ associée sont sur $\Re(s)=1/2$. Le PDF énonce sa cible comme conjecture (Annoncé, Conjecture 14.1.2.1) :

> « (GRH - Grand Riemann Hypothesis) All the zeros of the L-function L(s, π) are located on 1/2 + iR. »

Ce carnet ne calcule **aucun zéro** — l'énoncé est universel (une infinité de zéros pour une infinité de fonctions), aucune mesure ne le vérifie. Ce qu'on peut mesurer, c'est l'**objet** de l'énoncé : construire des caractères, former $L(s,\chi)$, et vérifier que le produit eulérien — la structure qui fonde la famille — est bien là.

> **H1 (pré-enregistrée).** Pour le caractère quadratique réel $\chi_5$, la série $L(2,\chi_5)=\sum \chi_5(n)/n^2$ (tronquée à $N=10^5$) et le produit eulérien $\prod_{p\le 1000}(1-\chi_5(p)/p^2)^{-1}$ coïncident à $5\cdot10^{-3}$ près.
>
> **H2 (pré-enregistrée).** Même chose pour un caractère **complexe** d'ordre 6 modulo 7 ($\chi_7(3)=e^{2i\pi/6}$) — les valeurs ne sont plus réelles, le produit eulérien reste le même objet.

````python
# Caracteres de Dirichlet, construits a la main (stdlib).
# chi_5 : l unique caractere quadratique mod 5 (reel, valeurs {+1,-1}).
# chi_7 : le caractere d ordre 6 mod 7, defini par chi_7(3) = zeta_6 = e^{2 i pi/6}.
def chi5(n):
    n %= 5
    if n == 0: return 0
    return 1 if n in (1, 4) else -1

zeta6 = cmath.exp(2j * math.pi / 6)
POW3 = {1: 0, 3: 1, 2: 2, 6: 3, 4: 4, 5: 5}   # 3^k mod 7 -> k
def chi7(n):
    n %= 7
    if n == 0: return 0
    return zeta6 ** POW3[n]

print('chi_5 (reel, ordre 2)   :', [chi5(n) for n in range(1, 8)])
print('chi_7 (complexe, ordre 6):')
for n in range(1, 7):
    v = chi7(n)
    print(f'  chi_7({n}) = zeta6^{POW3[n]} = {v:.4f}')
print()
print('Verification de structure (multiplicativite) chi_7(a*b) = chi_7(a)*chi_7(b) :',
      all(abs(chi7(a * b) - chi7(a) * chi7(b)) < 1e-12
          for a in range(1, 8) for b in range(1, 8)))
print('Verification chi_5(a*b) = chi_5(a)*chi_5(b) :',
      all(chi5(a * b) == chi5(a) * chi5(b)
          for a in range(1, 6) for b in range(1, 6)))
````

````python
# Mesures H1 et H2 : serie tronquee contre produit eulérien tronqué.
def premiers(m):
    crible = [True] * (m + 1)
    crible[0:2] = [False, False]
    for p in range(2, int(m ** 0.5) + 1):
        if crible[p]:
            crible[p * p :: p] = [False] * len(crible[p * p :: p])
    return [p for p in range(m + 1) if crible[p]]

N, P = 100_000, 1000
ps = premiers(P)

def serie(chi, N):
    return sum(chi(n) / (n * n) for n in range(1, N + 1))

def produit(chi, ps):
    prod = 1.0 + 0.0j
    for p in ps:
        prod *= 1.0 / (1.0 - chi(p) / float(p * p))
    return prod

s5, p5 = serie(chi5, N), produit(chi5, ps)
s7, p7 = serie(chi7, N), produit(chi7, ps)
e5, e7 = abs(s5 - p5), abs(s7 - p7)
print(f'L(2, chi_5)  serie  (N={N}) : {s5:.6f}')
print(f'L(2, chi_5)  produit(p<={P}) : {p5.real:.6f}  ecart = {e5:.2e}')
print(f'L(2, chi_7)  serie  (N={N}) : {s7.real:.6f} {s7.imag:+.6f}i')
print(f'L(2, chi_7)  produit(p<={P}) : {p7.real:.6f} {p7.imag:+.6f}i  ecart = {e7:.2e}')
print()
print('H1 (chi_5 reel, ecart < 5e-3)      :', 'CONFIRMEE' if e5 < 5e-3 else 'REFUTEE')
print('H2 (chi_7 complexe, ecart < 5e-3)  :', 'CONFIRMEE' if e7 < 5e-3 else 'REFUTEE')
````

**H1 et H2 : CONFIRMÉES** (écarts de l'ordre de $10^{-6}$, l'ordre de grandeur exact des queues tronquées : $\Theta(1/N)$ pour la série à sommes partielles bornées, $\Theta(1/(P\ln P))$ pour le produit). Les deux fonctions $L$ — réelle et complexe — portent le même produit eulérien : c'est **la** structure que GRH prononce sur toute la famille. La théorie complète des caractères, des séries $L$ et du produit eulérien est mesurée dans [Serre 100-16](../Serre100/16-fonctions-l-dirichlet-progression-premiers.ipynb) ; les zéros, eux, ne se vérifient pas par échantillonnage — le carnet [12](12-zeros-simples-ligne-critique.ipynb) montre ce que « compter des zéros » veut dire sur $\zeta$, et pourquoi une mesure, même exacte à $10^{-29}$, reste une mesure : l'énoncé GRH est universel, seul un argument couvre l'infinité.

Les deux régimes n'ont pas la même **forme** de confiance : le premier délègue à des jugements nommés (la chaîne Vakil → Conrad), le second à un **rejeu mécanique** par un petit noyau vérifié — mais seulement si quelqu'un le lance, et seulement sur des énoncés qu'il faut encore lire. Aucun des deux n'est fermé à la date de gel : c'est précisément ce que « différer la croyance » veut dire.

````python
# Experience jouet : une meme identite, deux regimes de verification.
# (a) regime NUMERIQUE (l analogie du relectureur) : on evalue en des points.
# (b) regime SYNTAXIQUE (l analogie du rejeu par noyau) : on compare des COEFFICIENTS,
#     un objet fini dont la comparaison couvre tous les x d un coup.

def polymul(a, b):  # coefficients entiers, degre croissant
    out = [0] * (len(a) + len(b) - 1)
    for i, x in enumerate(a):
        for j, y in enumerate(b):
            out[i + j] += x * y
    return out

# Identite jouet : (2x + 1)^2 = 4x^2 + 4x + 1
gauche = polymul([1, 2], [1, 2])   # (2x+1)^2, degre croissant : 1 + 4x + 4x^2
droite = [1, 4, 4]

# (a) numerique : 6 points, constat local
ecarts_num = [abs((2 * x + 1) ** 2 - (4 * x * x + 4 * x + 1)) for x in range(-3, 4)]
print('(a) numerique, 7 points evalues : ecart max =', max(ecarts_num))
# (b) syntaxique : comparaison FINIE des coefficients -- couvre tout x
print('(b) syntaxique, coefficients   : identiques =', gauche == droite)
print()
# La limite du regime numerique, mise en scene : une identite FAUSSE passe
# presque partout au test numerique --
faux_g = polymul([1, 2], [1, 2])            # 1 + 4x + 4x^2
faux_d = [1, 4, 4, 0]                        # disons qu on pretend (2x+1)^2 = 4x^2+4x+1+c
piege = [x for x in range(-50, 51) if abs((2*x+1)**2 - (4*x*x+4*x+1)) < 1e-9]
print('Piege : une identite fausse (terme constant modifie de 1) echoue au test',
      'numerique seulement aux points ou elle echoue ; le test coefficientiel,',
      'lui, est global.')
faux_d2 = [2, 4, 4]
print('  coefficients [1,4,4] vs [2,4,4] : egaux =', faux_g == faux_d2,
      '-- un seul bit d ecart, et le regime syntaxique le voit IMMEDIATEMENT.')
````

````python
# Le regime syntaxique a un prix : il ne parle que de ce qui est ECRIT.
# L identite jouet est vraie pour tout x -- mais c est nous qui avons choisi
# l enonce. Un rejeu Lean prouve exactement les theoremes du fichier, avec
# exactement les axiomes autorises : la question qui reste ouverte apres le
# rejeu n est pas 'la preuve est-elle correcte' mais 'l enonce prouve est-il
# bien celui qu on croyait'. C est le role des comparateurs de l archive
# Alpoge (cite #27589/#27632) : chaque paire enonce/solution n autorise que
# propext, Quot.sound, Classical.choice.
axiomes_autorises = ['propext', 'Quot.sound', 'Classical.choice']
print('Axiomes autorises par les comparateurs (cites) :', ', '.join(axiomes_autorises))
print()
print('Ce que chaque regime etablirait sur Masullo vs Alpoge au 20/11 :')
for ligne in [
    ('Masullo', 'rien de mecanique -- l annonce invite a formaliser (cite, billet)'),
    ('Alpoge', 'lake build + #print axioms sur 14 parties -- rejeu non encore fait par nous (cite #27589)'),
]:
    print(f'  {ligne[0]:9s} : {ligne[1]}')
````

L'expérience jouet sépare ce que les deux régimes **établissent** : le test numérique constate localement (et une identité fausse peut le passer presque partout), la comparaison de coefficients est un objet **fini** dont le verdict couvre tous les $x$ d'un coup — l'analogue discret du rejeu par noyau. Mais le régime syntaxique ne parle que de ce qui est écrit : il prouve l'énoncé du fichier, avec les axiomes autorisés, et **rien d'autre**. La question ouverte après un rejeu n'est plus « la preuve est-elle correcte » mais « l'énoncé prouvé est-il celui qu'on croyait » — c'est ce que les comparateurs de l'archive Alpöge encadrent (Cité, #27589 et #27632). ## 5. Questions de lecture — là où un vérificateur regardera en premier

Cette section ne prouve rien : elle cite le PDF (Annoncé, copies archivées) et formule des **questions**, pas des objections.

**Q1 — le cas extrême $a=0$, présenté comme un garde-fou.** Le PDF dit (§1.7.3, Annoncé) :

> « Although a full trace formula cannot be proved in the extremal situation a = 0 (not even if b ≥ 2m, or b ≥ 2[F : Q]), the method of trace calculation still allows one to squeeze out some information. »
>
> « the fact that the trace formula cannot be proved in this extremal version of the methods is itself a sanity check »
>
> « I looked into this extremal case of the methods precisely with the aim of challenging them. »

### 9.1 Une réécriture, pas une vérification

Le premier fait est le plus facile à manquer, parce qu'il est écrit dans un fichier annexe. Le `NOTICE` — présent dans les deux éditions — dit dans une seule phrase ce qui vient d'Alpöge et ce qui a été refait :

> This product is a rewriting of Levent Alpöge's Lean proof (see README.md). src/Statements.lean and the names and types of the five theorems in src/Main.lean are taken from that proof unchanged; lakefile.toml is adapted from it.

« Rewriting » n'est pas « vérification ». Dans les documents que le dépôt consacre à son propre statut — ses trois `README` et ce `NOTICE` —, nulle part il ne dit avoir rejoué la preuve d'Alpöge pour la trouver juste : il dit l'avoir **réécrite**. Le `README` de la racine le répète en une ligne plus simple :

> The proof is based on Levent Alpöge's, with Claude (https://alpo.ge/remark.pdf). This repository is a shorter version…

Et les deux éditions précisent, avec les mêmes mots, ce que « based on » recouvre :

> This project is a shorter rewriting of Levent Alpöge's Lean proof (see <https://alpo.ge/remark.pdf>). It follows that proof's method; many steps are proved by other arguments, and parts of it, mostly background material, were ported from there and then rewritten.

Ce qui est repris **sans changement**, le `NOTICE` le dit donc : l'énoncé et son vocabulaire. Pour le reste, c'est l'autre citation qui fait foi — la méthode est suivie, mais plusieurs étapes sont prouvées par d'autres arguments, et une partie, surtout le fond, a été portée puis réécrite. D'où le tableau :

| | Repris d'Alpöge | Réécrit pour ce dépôt |
|---|---|---|

| Poids | l'interface seule, dans deux petits fichiers | la quasi-totalité du texte |
| D'une édition à l'autre | identique, empreintes à l'appui | deux textes différents pour les mêmes cinq théorèmes |

Un lecteur qui veut savoir ce que le dépôt ajoute doit donc regarder `New` — et c'est précisément la partie que ce carnet ne lit pas. Il mesure ce qui est visible : l'interface, les énoncés, la bibliographie, les volumes. Jamais la preuve.

Le `README` de chaque édition dit enfin, avec les mêmes mots, sur quoi porte le résultat :

Le dépôt publie la même preuve deux fois. Le `README` de la racine les présente en deux lignes, et les `README` de chaque dossier donnent les volumes :

L'édition compacte avertit de ce qu'elle échange :

Les deux éditions ne se distinguent donc pas par ce qu'elles prouvent, mais par ce qu'elles coûtent à lire. Et l'interface, elle, est **la même** : `src/Statements.lean`, les noms et les types des cinq théorèmes, la liste des axiomes. La cellule suivante le mesure plutôt que de le supposer.

La différence est donc dans ce que chaque `README` affirme, pas nécessairement dans ce qui a été exécuté. Elle compte tout de même. D'un côté, trois vérificateurs indépendants sont dits avoir accepté un export des cinq théorèmes et de tout ce dont ils dépendent, Mathlib compris. ### Un seul résultat, quatre corollaires

`src/Main.lean` déclare cinq théorèmes et ne contient aucun `sorry`. Trois d'entre eux citent le module `New` ; les deux autres n'y font pas appel. En lisant les termes plutôt que les noms, la chaîne est courte :

| Théorème | Preuve | Ce qu'il est |
|---|---|---|
| `LFunction_ne_zero_of_half_lt_re` | `New.grh` | le résultat lui-même |
| `LFunction_eq_zero_re` | `New.zeros_on_line New.grh` | une déduction |
| `riemannHypothesis` | `New.zeta_zeros_on_line New.grh` | une déduction |
| `riemannZeta_eq_zero_re` | `riemannHypothesis` | un autre nom |
| `LFunction_eq_zero_re_of_isPrimitive` | `LFunction_eq_zero_re` | une ligne |

Le type de `New.grh`, lu sur le seul théorème qu'il prouve à lui seul, est `DirichletLNonvanishingRightOfHalf`. Et `riemannHypothesis` — le premier théorème du fichier — a pour type `RiemannHypothesis`, qui n'est pas défini dans `src/Statements.lean` : c'est l'énoncé de Mathlib, et son `README` le dit (« The first theorem has Mathlib's type `RiemannHypothesis` »). `Challenge.lean`, qui n'importe que `Statements`, l'emploie aussi. L'hypothèse de Riemann affirmée ici est donc celle que Mathlib formalise, pas une variante rédigée par le dépôt.

Dans l'édition compacte, tout passage vers les cinq théorèmes traverse donc **un** terme, `New.grh`, auquel s'ajoutent deux lemmes de déduction, `New.zeros_on_line` et `New.zeta_zeros_on_line`. Cela ne fait pas du dépôt un seul lemme : son `README` range `New.grh` parmi les trois résultats que prouve `src/New`, avec l'inégalité principale `star_inequality` et la borne à un côté `eventually_primeError_le_rpow` ; et à côté s'étend un second dossier, `src/Lib`, qui contient les résultats de fond que Mathlib n'a pas, avec les étapes de leurs preuves, et qui n'importe rien de `src/New`. Mais l'interface à cinq théorèmes n'est que l'habillage de ce terme. C'est aussi pourquoi elle peut être identique alors que les preuves ne se ressemblent pas : ce qui est partagé entre les deux éditions, c'est le contrat, pas l'argument. « Habillage » n'est pas péjoratif ici : le contrat est exactement ce qu'un vérificateur doit comparer.

C'est l'édition lisible qui nomme les choses. Ses docstrings intitulent le troisième théorème « The Generalized Riemann Hypothesis, half-plane form. » et le quatrième « The Generalized Riemann Hypothesis. » : non-annulation dans le demi-plan `re s > 1/2`, puis zéros sur la droite critique — dans les deux cas **pour les fonctions L des caractères de Dirichlet**. La GRH que le dépôt revendique, c'est celle de ces deux énoncés, et pas une autre.

### Ce que les énoncés laissent dehors

Quatre définitions, un seul import (`Mathlib`), aucun `sorry` dans le fichier d'énoncés. Trois des quatre quantifient sur **tout** module `N` — l'instance `[NeZero N]` n'exige que `N` soit non nul —, **tout** caractère `χ` et **tout** `s` ; une seule ajoute l'hypothèse `χ.IsPrimitive`. Deux se restreignent à la bande ouverte `0 < re s < 1`. Ce n'est pas un affaiblissement arbitraire : à gauche de la bande, les fonctions L ont des zéros **hors** de la droite critique — les zéros triviaux, que la cellule suivante trouve dans Mathlib —, et l'énoncé « tout zéro est sur la droite » n'y tient pas. À droite, la non-annulation est l'objet d'un troisième énoncé, et « tout zéro est sur la droite » y est vrai sans rien dire. La bande est donc le lieu propre de la question.

Un seul énoncé porte la clause `(χ ≠ 1 ∨ s ≠ 1)` : `DirichletLNonvanishingRightOfHalf`. C'est le point `s = 1` du caractère principal. La cellule suivante mesure ce que ce point coûte, et pourquoi la clause est exactement posée ainsi.

L'énoncé pour ζ, lui, nomme ses exclusions dans son propre énoncé : `¬∃ n : ℕ, s = -2 * ((n : ℂ) + 1)` écarte une famille de points, `s ≠ 1` en écarte un autre. Aucune convention cachée : ce qui est exclu est écrit, et un lecteur peut le vérifier sans ouvrir la preuve.

Enfin les absents. Les marqueurs `Hecke`, `NumberField`, `AutomorphicForm`, `ModularForm` n'apparaissent nulle part dans le fichier d'énoncés. La portée des énoncés est donc bornée aux fonctions L de Dirichlet — pas aux corps de nombres, pas aux fonctions L de Hecke, pas aux formes automorphes. Ce n'est pas qu'un comptage de mots : les quatre définitions sont là en entier, leurs types ne parlent que de `riemannZeta` et de `DirichletCharacter`, et le `README` le confirme (« All moduli N ≥ 1 and all characters are covered, primitive or not. »). La preuve, elle, mobilise davantage : les premières rangées de `REFERENCES.md`, recopiées plus bas, citent des fichiers de `Lib/Adelic` et de `Lib/Shimura`. Ce qu'une preuve utilise n'est pas ce qu'elle établit.

### Ce que le comparateur ajoute

`Challenge.lean` reprend les cinq théorèmes, dans le même ordre et avec les mêmes types, en mettant `sorry` à la place des preuves ; il importe `Statements.lean`, le fichier même où la solution prend ses définitions. Une définition ne peut donc pas différer entre le fichier-défi et le fichier-réponse : c'est le même import. `comparator.json` ne met en regard que les cinq noms de théorèmes — aucune définition n'est comparée — et n'autorise que trois axiomes : `propext`, `Quot.sound`, `Classical.choice`. Le `README` de l'édition compacte dit le reste :

> `Challenge.lean` states the five theorems with `sorry` in place of proofs, and `comparator.json` sets up the tool [Comparator](https://github.com/leanprover/comparator), which compares them with `src/Main.lean` and checks the proofs again. Nothing imports `Challenge.lean`, and `lake build` does not build it. We ran Comparator v4.35.0 on this folder (Linux, with a stand-in for its sandbox and `ulimit -s unlimited`), and it accepted the proofs; allow 64 GB of memory and a few hours.

````python
# La clause (chi != 1 ou s != 1), en calculant : ce qui se passe a s = 1 (stdlib seule).
# Mathlib, au pin que le depot epingle (56a00585), le dit deja en six endroits : la definition de
# la fonction L, la condition de derivabilite, les deux familles de zeros triviaux, le produit
# eulerien du caractere principal et son residu. Extraits verbatim de
# Mathlib/NumberTheory/LSeries/DirichletContinuation.lean :
MATHLIB = {
    "definition": """/--
The unique meromorphic function `ℂ → ℂ` which agrees with `∑' n : ℕ, χ n / n ^ s` wherever the
latter is convergent. This is constructed as a linear combination of Hurwitz zeta functions.

Note that this is not the same as `LSeries χ`: they agree in the convergence range, but
`LSeries χ s` is defined to be `0` if `re s ≤ 1`.
-/
@[pp_nodot]
noncomputable def LFunction (χ : DirichletCharacter ℂ N) (s : ℂ) : ℂ := ZMod.LFunction χ s""",
    "derivabilite": """/--
The L-function of a Dirichlet character is differentiable, except at `s = 1` if the character is
trivial.
-/
@[fun_prop]
lemma differentiableAt_LFunction (χ : DirichletCharacter ℂ N) (s : ℂ) (hs : s ≠ 1 ∨ χ ≠ 1) :
    DifferentiableAt ℂ (LFunction χ) s :=
  ZMod.differentiableAt_LFunction χ s (hs.imp_right χ.sum_eq_zero_of_ne_one)""",
    "zeros_pairs": """/-- The L-function of an even Dirichlet character vanishes at strictly negative even integers. -/
@[simp]
lemma Even.LFunction_neg_two_mul_nat_add_one {χ : DirichletCharacter ℂ N} (hχ : Even χ) (n : ℕ) :
    LFunction χ (-(2 * (n + 1))) = 0 :=
  ZMod.LFunction_neg_two_mul_nat_add_one hχ.to_fun n""",
    "zeros_impairs": """/-- The L-function of an odd Dirichlet character vanishes at negative odd integers. -/
@[simp] lemma Odd.LFunction_neg_two_mul_nat_sub_one
    {χ : DirichletCharacter ℂ N} (hχ : Odd χ) (n : ℕ) :
    LFunction χ (-(2 * n) - 1) = 0 :=
  ZMod.LFunction_neg_two_mul_nat_sub_one hχ.to_fun n""",
    "produit_du_principal": """/-- The L function of the trivial Dirichlet character mod `N` is obtained from the Riemann
zeta function by multiplying with `∏ p ∈ N.primeFactors, (1 - (p : ℂ) ^ (-s))`. -/
lemma LFunctionTrivChar_eq_mul_riemannZeta {s : ℂ} (hs : s ≠ 1) :
    LFunctionTrivChar N s = (∏ p ∈ N.primeFactors, (1 - (p : ℂ) ^ (-s))) * riemannZeta s := by""",
    "residu": """/-- The L function of the trivial Dirichlet character mod `N` has a simple pole with
residue `∏ p ∈ N.primeFactors, (1 - p⁻¹)` at `s = 1`. -/
lemma LFunctionTrivChar_residue_one :
    Tendsto (fun s ↦ (s - 1) * LFunctionTrivChar N s) (𝓝[≠] 1)
      (𝓝 <| ∏ p ∈ N.primeFactors, (1 - (p : ℂ)⁻¹)) := by""",
}
# Empreinte sha256 de chaque extrait (LF, sans saut de ligne final), mesuree sur le fichier telecharge.
EMPREINTES_MATHLIB = {
    "definition": "f2402702025df2392e16b3192a5d84b2cd9a2b716aa510dfb32f17b0b74acd3c",
    "derivabilite": "eeb742d98139c413bc53fec0ffc490fd2b77a400ab1750e501147609814d1d1f",
    "zeros_pairs": "dba28bc78d24509165f05922ac7cdeaf4e107a97247b66d5b0275d278da82cf0",
    "zeros_impairs": "5b0b4a4220c3609922abce1edfd585f5b7e84c356b0e614bb64adf99a91f9654",
    "produit_du_principal": "bec0772cab418c187d24c254f858b2fe9380ad37dc591850b9c724c101e028b5",
    "residu": "2b96e758df3622736a6859c1ca86dbc9abf29191e85df931a2eeea12db0bbaed",
}
CLAUSE_DEPOT = re.findall(r"\((χ ≠ 1 ∨ s ≠ 1)\)", STATEMENTS)[0]
CLAUSE_MATHLIB = re.findall(r"\(hs : (.+?)\)", MATHLIB["derivabilite"])[0]


def en_ascii(texte: str) -> str:
    """Translitteration pour la sortie : le noyau n'imprime que de l'ASCII, les symboles restent aux textes."""
    return texte.replace("χ", "chi").replace("≠", "!=").replace("∨", "or")


print("1. La clause du depot et celle de Mathlib, des deux cotes du 'ou'")
print("   depot   :", en_ascii(CLAUSE_DEPOT), "| Mathlib :", en_ascii(CLAUSE_MATHLIB))
print("   memes deux cotes, l'ordre pres :", set(CLAUSE_DEPOT.split(" ∨ ")) == set(CLAUSE_MATHLIB.split(" ∨ ")))
print("   extraits embarques identiques aux extraits mesures :")
for nom, texte in MATHLIB.items():
    print(f"      {nom:22s} {hashlib.sha256(texte.encode('utf-8')).hexdigest() == EMPREINTES_MATHLIB[nom]}")

import math


def chi0(n: int, N: int) -> int:
    """Caractere principal modulo N : 1 sur les entiers premiers a N, 0 ailleurs."""
    return 1 if math.gcd(n, N) == 1 else 0


def chi4(n: int) -> int:
    """Caractere quadratique modulo 4 : +1, -1, +1, -1, ... sur 1, 3, 5, 7, ... ; 0 sur les pairs."""
    return (1 if n % 4 == 1 else -1) if n % 2 else 0


def somme_partielle(poids, x: int) -> float:
    """S(x) = somme des chi(n)/n pour n <= x ; fsum n'arrondit qu'une fois, a la fin."""
    return math.fsum(poids(n) / n for n in range(1, x + 1))


def residu_principal(N: int) -> float:
    """Produit des (1 - 1/p) sur les premiers p divisant N : le residu annonce par Mathlib."""
    produit = 1.0
    for p in range(2, N + 1):
        if N % p == 0 and all(p % q for q in range(2, p)):
            produit *= 1 - 1 / p
    return produit


bornes = [10 ** k for k in range(1, 7)]
S0 = {N: [somme_partielle(lambda n: chi0(n, N), x) for x in bornes] for N in (1, 4, 6)}
S4 = [somme_partielle(chi4, x) for x in bornes]

print("\n2. Caractere principal modulo N : S(x) - residu * ln(x), pour x = 10, 100, ..., 10^6")
for N, valeurs in S0.items():
    r = residu_principal(N)
    print(f"   N = {N} (residu {r:.4f}) : " + "  ".join(f"{v - r * math.log(x):9.5f}" for v, x in zip(valeurs, bornes)))

print("\n3. Pente mesuree par decade : [S(10^k) - S(10^(k-1))] / ln(10), k = 5, 6")
for N, valeurs in S0.items():
    pentes = [(b - a) / math.log(10) for a, b in zip(valeurs[-3:-1], valeurs[-2:])]
    print(f"   N = {N} : " + "  ".join(f"{p:.5f}" for p in pentes)
          + f"   | residu de Mathlib : {residu_principal(N):.5f}")

print("\n4. Caractere non principal modulo 4 : S(x), puis S(x) - pi/4")
print("   S(x)         : " + "  ".join(f"{v:9.5f}" for v in S4))
print("   S(x) - pi/4  : " + "  ".join(f"{v - math.pi / 4:9.6f}" for v in S4))
print(f"   pi/4 = {math.pi / 4:.6f}")
````

**Mesuré.** La clause du dépôt et celle de Mathlib portent les **mêmes deux côtés**, dans l'ordre inverse : `(χ ≠ 1 ∨ s ≠ 1)` d'un côté, `(hs : s ≠ 1 ∨ χ ≠ 1)` de l'autre — l'hypothèse de `differentiableAt_LFunction`. Le point écarté est donc le même des deux côtés : `s = 1` du caractère principal. Mathlib l'écarte au titre de la dérivabilité de la fonction L ; le dépôt, dans un énoncé de non-annulation. Les six extraits embarqués sont par ailleurs fidèles : leurs empreintes coïncident avec celles mesurées sur le fichier de Mathlib à l'état épinglé `56a00585`.

### Pourquoi le point `s = 1`, et pas un autre

Mathlib répond en six endroits, et ils se lisent dans cet ordre.

**Les zéros triviaux existent par théorème.** `Even.LFunction_neg_two_mul_nat_add_one` dit que la fonction L d'un caractère **pair** s'annule aux entiers négatifs pairs ; `Odd.LFunction_neg_two_mul_nat_sub_one` dit la même chose pour un caractère **impair** aux négatifs impairs. Ces zéros ne sont pas des accidents de calcul : ils sont dans Mathlib, marqués `@[simp]`. La famille `-2 * (n + 1)` que l'énoncé pour ζ écarte est donc la famille des zéros triviaux, et Mathlib la prouve sous la forme qui vaut pour tout caractère pair.

**Le pôle est un théorème, lui aussi.** `LFunctionTrivChar_eq_mul_riemannZeta` écrit la fonction L du caractère principal mod `N` comme ζ multipliée par un produit eulérien fini, `∏ p ∈ N.primeFactors, (1 - p^(-s))`, et l'énoncé vaut pour `s ≠ 1`. Puis `LFunctionTrivChar_residue_one` dit ce qui se passe au point exclu : `(s - 1) * L(s)` tend vers `∏ (1 - p⁻¹)`, un résidu **non nul**. Ce n'est donc pas un zéro que la clause évite : c'est un pôle. Le `README` du dépôt glose son troisième théorème dans les mêmes termes : « L(s, χ) ≠ 0 for re s > 1/2, except at s = 1 for the principal character of each modulus (the pole). »

**Une fonction totale en un point qui n'en a pas.** Mathlib présente `LFunction` comme « unique meromorphic function `ℂ → ℂ` » qui prolonge la série de Dirichlet, et précise qu'elle est « constructed as a linear combination of Hurwitz zeta functions ». Son type, `ℂ → ℂ`, est celui d'une fonction totale : en Lean, elle est définie en tout point. En `s = 1`, pour le caractère principal, elle prend donc la valeur que lui donne sa construction, et non une valeur de la fonction méromorphe, qui n'en a pas en ce point. La clause retire exactement ce point, et celui-là seul.

**Et le calcul le retrouve.** La cellule précédente calcule les sommes partielles `S(x) = Σ_{n ≤ x} χ(n)/n` et retire `r · ln x`, où `r` est le résidu annoncé par Mathlib :

| Caractère principal | Constante mesurée `S(x) - r·ln x` | Pente par décade, `k = 5` puis `k = 6` | Résidu de Mathlib |
|---|---|---|---|
| mod 1 | 0,57722 | 0,99998 puis 1,00000 | 1,00000 |
| mod 4 | 0,63518 | 0,50000 puis 0,50000 | 0,50000 |
| mod 6 | 0,60656 | 0,33335 puis 0,33333 | 0,33333 |

Deux choses apparaissent. D'abord la **pente** : le coefficient de la croissance logarithmique est le résidu, à l'arrondi près — `0,99998` et `1,00000` contre `1,00000`, `0,33335` et `0,33333` contre `0,33333`. C'est le comportement d'un pôle simple de résidu `r`. Ensuite la **constante** qui reste : pour `N = 1`, elle vaut `0,57722`, soit la constante d'Euler-Mascheroni γ à la cinquième décimale. Le carnet constate cette valeur ; il ne la démontre pas, et il n'identifie pas les deux autres, qui dépendent des facteurs eulériens du module.

**Un caractère non principal montre l'autre côté de la clause.** Modulo 4, la même somme partielle ne diverge pas : elle s'approche de `π/4 = 0,785398`. Les restes `S(x) - π/4` valent successivement `0,049522`, `-0,005000`, `-0,000500`, `-0,000050`, `-0,000005`, `-0,000000` : le premier est positif, tous les suivants sont négatifs — la parité du nombre de termes décide de quel côté tombe la somme partielle — et leur valeur est divisée par dix à chaque décennie. C'est la décroissance d'une série alternée convergente, dont l'erreur est de l'ordre du premier terme négligé, en `1/x`. Il n'y a pas de pôle, et la limite est finie et non nulle — ce qui est cohérent avec la clause `(χ ≠ 1 ∨ s ≠ 1)`, qui conserve ce point pour ce caractère.

La clause n'est donc pas une précaution de rédaction. Elle retire le seul point du demi-plan `re s > 1/2` où l'énoncé de non-annulation perd son sens, parce que la fonction y a un pôle au lieu d'une valeur ; elle le conserve partout ailleurs, où Mathlib prouve que la fonction est dérivable.

Reste une question que ces énoncés ne posent pas : d'où viennent les résultats de fond que le dépôt utilise, et jusqu'où sont-ils vérifiables ? Le fichier de citations y répond, et il répond avec ses propres chiffres.

````python
# Le bilan bibliographique du depot, recompte. Chaque edition publie un REFERENCES.md : un tableau
# ou chaque citation porte un etat, puis la liste des fichiers de la preuve qui s'y appuient.
# Les quatre libelles d'etat sont notes par une lettre.
import hashlib
import re

LIBELLES = {
    "C": "compared, place confirmed",
    "S": "place confirmed at second hand",
    "W": "work found, no place given",
    "N": "not compared, no place given",
}
INVERSE = {libelle: lettre for lettre, libelle in LIBELLES.items()}

# Les premieres rangees du tableau de l'edition compacte, verbatim : on voit d'ou viennent les colonnes.
EXTRAIT = """| citation | state | files |
|---|---|---|
| A. Borel, Introduction to automorphic forms | work found, no place given | `Lib/Adelic/Poincare/BesselSubgroup` |
| A. Borel, Some finiteness properties of adele groups over number fields, Publ. IHES 16 (1963) | compared, place confirmed | `Lib/Adelic/Reduction/Defs`, `Lib/Shimura/Uniformisation/AppellHumbert` |
| A. Knapp, Lie Groups Beyond an Introduction | place confirmed at second hand | `Lib/DiscreteSeries/Bruhat/ComplexPlace` |"""
MODULES_LIBELLES = {"l": "src/Lib seul", "n": "src/New seul", "m": "des deux"}


def rangees(texte: str) -> list:
    """Les rangees du tableau : (etat, fichiers cites), dans l'ordre du fichier."""
    out = []
    for ligne in texte.split("\n"):
        colonnes = [c.strip() for c in ligne.strip("|").split("|")]
        if len(colonnes) >= 3 and colonnes[1] in INVERSE:
            out.append((INVERSE[colonnes[1]], re.findall(r"`([^`]+)`", colonnes[2])))
    return out


def etats(texte: str) -> str:
    """La colonne 'state' : une lettre par rangee."""
    return "".join(etat for etat, _ in rangees(texte))


def modules(texte: str) -> str:
    """La colonne 'files' : une lettre par rangee -- l (src/Lib seul), n (src/New seul), m (des deux)."""
    lettres = ""
    for _, fichiers in rangees(texte):
        racines = {f.split("/")[0] for f in fichiers}
        lettres += "l" if racines == {"Lib"} else "n" if racines == {"New"} else "m"
    return lettres


# La colonne 'state' entiere de chaque edition, telle que la fonction ci-dessus la lit sur la copie.
ETATS = {
    "grh-compact": """WCSCSCCWCSWCWWCWCWWWWWCWWSSWWCWCWNCWWSNWSCWWWWWWWWCCWWWWCSWCCSWW
SWWCSSCWCWCCWSCSWWCCCCCWWWWWWWWSSCWWWWWWSSNWNWNSCWWWCCWCWWCCWWNW
WWWWWCCSWCCCWSWWCCNWWSWCWWCWSWCSCSSCWCWCCWSWWWSCWCCCWWWCCSWSWWWW
WWWW""",
    "grh": """WCNSCSCCWCCSWCWWCWCWWCWWWCWWCCSSWWWCWWCWNCWWSNWSWSSCWWWWWWWWCCWW
WWCWWSWCCSWWSWWCSSCWCWCCWSCSWWNCCCCWWWWWWWWSSCCWWWWWWSSNWNWSCWWW
CCWCWWWWWCCWWNWWWWWWWCCSWCCCWSWWWCCCSWWSWCWWWWWWCWWSWCSSCCSSCWCW
CCWSWWWSSCWCCCCWWWCCSCCWSWWWWWWWNWWW""",
}
# La colonne 'files', lue par la meme regle : une lettre de module par rangee.
MODULES = {
    "grh-compact": """lllllllllmllllmnnnnnnllllnnmllnllllnlllllmnlllllnmllnlllllnlllll
lnnllllnmnmlnnmlnlmllllllllllllllllnllllllnlllnlnnnlllllnmmllnll
lllllllllmmmlllllmnnllnlmnllllllllllllmllnllnnllnlllllllllllllll
llnn""",
    "grh": """lllllllllnllllllmnnnnlnnllllllnnnlllnlllllllllllllllnlllllnlllnl
llllllnllllllnnllllnlnmlllllnlllllllllllllllllllnlllllllllllllnl
lllllllnmlllnlllllnllllllllnlllllllllnllnlnlllllllllllllllllllll
llnlllnlllnlllllllllllllllllllllnnln""",
}
# Totaux annonces en tete de chaque REFERENCES.md, et citations hors du controle (meme paragraphe).
TOTAUX_ANNONCES = {"grh-compact": {"C": 57, "N": 7, "S": 30, "W": 102},
                   "grh": {"C": 66, "N": 8, "S": 35, "W": 119}}
HORS_CONTROLE = {"grh-compact": (76, 43), "grh": (122, 68)}   # autres fichiers, dont avec lieu

# Empreinte sha256 des libelles complets de la colonne 'state', joints par un saut de ligne, et de
# la suite des lettres de module, mesurees sur la copie du depot par un script distinct. Le carnet
# recalcule les deux depuis ce qu'il embarque : elles ne peuvent coincider que si les colonnes
# embarquees sont exactement celles des fichiers.
EMPREINTES_COLONNE = {
    "grh-compact": "cc5edc1838e3f232a1324c928daf87ee456d188dbc2244db8276caeb1e178a42",
    "grh": "309ebf07fd3e0ebd40764541b0a9d2fe3aeeb183a09cb995c53560058a909d46",
}
EMPREINTES_MODULES = {
    "grh-compact": "507d8cb78b484a6955c74baba4da94983eec3ffe1c66a3837336f46624f78d0b",
    "grh": "2a48d629dbe22f1c93a0133d1f8920593f91f6cbe34d89b48a07028c98b9d0f3",
}

COLONNES = {edition: "".join(brut.split()) for edition, brut in ETATS.items()}
MODULES_COLONNES = {edition: "".join(brut.split()) for edition, brut in MODULES.items()}
print("1. Le lecteur de colonnes sur les rangees verbatim : etats", etats(EXTRAIT), "| modules",
      modules(EXTRAIT),
      "| identiques aux premieres lettres embarquees (compacte) :",
      etats(EXTRAIT) == COLONNES["grh-compact"][:3],
      modules(EXTRAIT) == MODULES_COLONNES["grh-compact"][:3])
for numero, (edition, colonne) in enumerate(COLONNES.items(), start=2):
    comptages = {lettre: colonne.count(lettre) for lettre in "CSWN"}
    sans_lieu = comptages["W"] + comptages["N"]
    empreinte = hashlib.sha256("\n".join(LIBELLES[c] for c in colonne).encode("utf-8")).hexdigest()
    hors, avec_lieu = HORS_CONTROLE[edition]
    total = len(colonne)
    print(f"\n{numero}. REFERENCES.md de {edition} : {total} citations dans le tableau")
    for lettre in "CSWN":
        n = comptages[lettre]
        print(f"   {LIBELLES[lettre]:34s} {n:4d}   (annonce : {TOTAUX_ANNONCES[edition][lettre]:4d})"
              f"   {100 * n / total:5.1f} %")
    print("   totaux announces verifies :", comptages == TOTAUX_ANNONCES[edition],
          "| empreinte de la colonne identique a celle de la copie :",
          empreinte == EMPREINTES_COLONNE[edition])
    print(f"   sans lieu verifie (W + N) : {sans_lieu} / {total} = {100 * sans_lieu / total:.1f} %"
          f" | comparee a sa source (C) : {100 * comptages['C'] / total:.1f} %")
    print(f"   hors du controle : {hors} autres fichiers, dont {avec_lieu} donnent un lieu"
          f" ({100 * avec_lieu / hors:.0f} %)")

    # --- la colonne 'files' : la citation appuie-t-elle le fond (src/Lib) ou l'argument (src/New) ? ---
    lettres = MODULES_COLONNES[edition]
    par_module = {lettre: lettres.count(lettre) for lettre in "lnm"}
    empreinte_modules = hashlib.sha256(lettres.encode("utf-8")).hexdigest()
    print(f"   rangees citant au moins un fichier de src/Lib : {par_module['l'] + par_module['m']} / {total}"
          f" = {100 * (par_module['l'] + par_module['m']) / total:.1f} %"
          f" | au moins un de src/New : {par_module['m'] + par_module['n']} / {total}"
          f" = {100 * (par_module['m'] + par_module['n']) / total:.1f} %")
    print("   une lettre de module par rangee, comme d'etats :", len(lettres) == total,
          "| empreinte des lettres identique a celle de la copie :",
          empreinte_modules == EMPREINTES_MODULES[edition])
    for lettre in "CSWN":
        croises = {mod: sum(1 for e, m in zip(colonne, lettres) if e == lettre and m == mod)
                   for mod in "lnm"}
        print(f"      {LIBELLES[lettre]:34s}" + "  ".join(f"{MODULES_LIBELLES[mod]} {croises[mod]:3d}"
                                                         for mod in "lnm"))
````

**Mesuré.** Le fichier de citations est le même genre de document dans les deux éditions, et la cellule précédente le lit ligne à ligne. Elle extrait la colonne `state` — une lettre par rangée, l'une des quatre que le fichier nomme lui-même —, vérifie l'empreinte de la colonne entière contre celle mesurée sur la copie du dépôt, et recompte les quatre totaux annoncés en tête. Elle lit ensuite la colonne `files` avec la même régularité, et range chaque rangée selon les modules qu'elle cite.

### Quatre états, et ce que chacun dit

L'édition compacte ouvre son tableau par un paragraphe qui fixe le vocabulaire :

> Only some of the citations were compared with their sources. The table lists 196 citations. A place is a theorem, section or page number; places that were looked for and could not be confirmed were dropped. At second hand means from a table of contents, a catalogue or a citation in another work.

L'édition lisible ouvre par la même phrase, à une clause près : « The table lists the 228 citations that were part of the comparison. » Le périmètre est donc déclaré, et les quatre états sont définis avant d'être employés :

| État, tel que le fichier le nomme | Ce qu'il dit exactement | Compacte, sur 196 | Lisible, sur 228 |

### Ce que le tableau ne couvre pas

Le même paragraphe dit son périmètre, et une phrase plus bas le borne une seconde fois. Édition compacte :

Édition lisible, même phrase avec d'autres nombres : « The citations of 122 other files were not part of this check, nor were further citations in the files below that stand outside the checked paragraphs; 68 of these files give places. »

|---|---|---|
| `grh-compact/` | 76 | 43, soit 57 % |
| `grh/` | 122 | 68, soit 56 % |

Ces nombres ne comptent pas des citations mais des fichiers de la preuve, qui ne sont pas passés par le contrôle du tableau ; ceux d'entre eux qui donnent un lieu sont listés à la fin du fichier. Le tableau ne décrit donc pas toute la bibliographie du dépôt : il décrit la partie qui a été soumise au contrôle et publiée comme telle, et laisse de côté à la fois les citations des autres fichiers et, dans les fichiers qu'il liste, celles qui vivent hors des paragraphes contrôlés.

### Trois choses que ces chiffres disent, et trois qu'ils ne disent pas

**Un lieu est une adresse.** Les états ne mesurent pas la même chose : `work found, no place given` dit qu'un ouvrage a été identifié, sans dire où y lire quoi ; `place confirmed` dit qu'une adresse a été trouvée, et `compared, place confirmed` que l'ouvrage a été comparé à sa source. Dans l'édition compacte, près de trois rangées sur dix vont jusqu'à l'adresse vérifiée dans la source (29,1 %), une sur six à peu près s'arrêtent à une adresse d'emprunt (15,3 %), et plus d'une sur deux n'ont pas d'adresse du tout (52,0 %). Trouver un ouvrage, trouver une adresse, ouvrir l'ouvrage : trois actes, et le second peut n'être fait qu'à travers un catalogue ou une table des matières.

**L'absence de lieu ne dit pas qu'on n'a pas cherché.** Le paragraphe d'en-tête l'explique sans le vouloir : « places that were looked for and could not be confirmed were dropped ». Une adresse cherchée sans succès disparaît donc de la colonne, et la rangée ne garde qu'un état sans lieu. Deux rangées `work found, no place given` se ressemblent alors à la lecture — l'une après une recherche qui a échoué, l'autre sans recherche —, et rien dans le fichier ne les distingue. Seul `not compared` dit quelque chose de la démarche, et il ne dit qu'une chose : l'ouvrage n'a pas été comparé. Un lecteur qui compte les lignes d'un tableau ne compte donc pas des vérifications ; il compte des déclarations.

**Aucun contrôle machine n'ouvre un livre.** Le fichier ne dit ni qui a comparé, ni comment, ni avec quel instrument — et c'est cohérent avec le partage que le `README` revendique : un noyau qui accepte une dérivation, rien de plus. Les `README` des deux éditions disent, eux, qui a écrit le texte — « The Lean text and the documents in this project were written by Claude (AI). » —, et c'est justement pourquoi la question se pose pour la bibliographie : la comparaison des citations est un acte distinct de l'écriture du texte, et ni les `README` ni l'en-tête de `REFERENCES.md` ne disent qui l'a faite. `ATTRIBUTION.md` est du même genre : une déclaration qui décrit ce qui est repris, pas un contrôle qui le prouve.

### Qui cite quoi

La seconde colonne du tableau — `files` — indique, pour chaque citation, les fichiers de la preuve qui s'en servent. Chaque rangée se range donc selon les modules qu'elle cite : `src/Lib`, les résultats de fond que Mathlib ne contient pas ; `src/New`, l'argument propre au dépôt ; ou les deux.

| Édition | Rangées citant au moins un fichier de `src/Lib` | Rangées citant au moins un fichier de `src/New` |
|---|---|---|
| `grh-compact/`, sur 196 | 157, soit 80,1 % | 56, soit 28,6 % |
| `grh/`, sur 228 | 193, soit 84,6 % | 38, soit 16,7 % |

Plus de huit rangées sur dix s'appuient sur `src/Lib` ; moins de trois sur dix sur `src/New`. Le détail par état confirme la tendance : dans l'édition compacte, sur les 109 rangées sans adresse, 70 ne citent que des fichiers de `src/Lib`. La bibliographie du dépôt est donc d'abord une bibliographie de fond : elle atteste des résultats repris, plus qu'elle ne documente l'argument qui fait la preuve. Ce n'est pas un défaut en soi — une preuve de cette taille repose sur beaucoup de fond —, mais c'est une information sur ce que le tableau couvre, et sur ce qu'il ne couvre pas.

Les deux tables ne sont pas le même document — l'une compte trente-deux rangées de plus que l'autre —, mais elles répartissent leurs rangées de la même manière : les quatre parts tombent à deux dixièmes de point près (29,1 contre 28,9 ; 15,3 contre 15,4 ; 52,0 contre 52,2 ; 3,6 contre 3,5), et la part sans adresse est la même à un dixième (55,6 contre 55,7). La réécriture a donc changé le texte et le nombre de citations, pas la politique de citation. Reste à mesurer ce qu'elle a changé au texte lui-même.

````python
# Les deux editions face a face. Les chiffres viennent de deux sources, toujours distinguees :
#  - ANNONCES : les README, recopies tels ("About 300,000 lines of code; 433,000 ...") ;
#  - RECOMPTE : comptes faits hors carnet sur une copie du depot (26/11/2026, script distinct) :
#    lignes de code = lignes non vides une fois les commentaires de bloc /- ... -/ retires,
#    fins de ligne comprises (une fermeture d'attribut qui suit la fin d'un bloc reste sur la
#    ligne ou le bloc s'est ouvert), et lignes commencant par -- ecartees ; lignes non vides
#    du texte brut ; lignes en tout au sens de splitlines() -- sur tous les fichiers .lean.
# Le carnet compare et calcule les rapports ; il ne recommence pas le decompte (il lui faudrait
# l'arbre du depot).
ORDRE = ("grh", "grh-compact")                     # edition lisible, edition compacte
NOMS = {"grh": "lisible  ", "grh-compact": "compacte"}
MESURES = ("lignes de code", "non vides avec commentaires", "lignes en tout")
ANNONCE = {"grh": (300_000, 433_000, 477_000), "grh-compact": (76_000, 95_000, 113_000)}
RECOMPTE = {"grh": (299_916, 432_547, 477_151), "grh-compact": (76_166, 95_050, 112_854)}
FICHIERS = {"grh": (805, 495, 306), "grh-compact": (796, 573, 219)}   # .lean : total, src/Lib, src/New

print("1. Les trois volumes, annonces contre recomptes")
for ed in ORDRE:
    for mesure, a, r in zip(MESURES, ANNONCE[ed], RECOMPTE[ed]):
        print(f"   {NOMS[ed]} : {mesure:29s} annonce {a:7d} | recompte {r:7d}"
              f" | ecart {100 * (r - a) / a:+5.2f} %")

print("\n2. Fichiers .lean de chaque edition")
for ed in ORDRE:
    tous, lib, nouveau = FICHIERS[ed]
    print(f"   {NOMS[ed]} : {tous} fichiers .lean"
          f" | src/Lib {lib} | src/New {nouveau}"
          f" | {100 * lib / tous:.1f} % et {100 * nouveau / tous:.1f} % des fichiers")

print("\n3. Ce que contient le texte (parts des lignes en tout)")
for ed in ORDRE:
    code, non_vides, total = RECOMPTE[ed]
    commentaires = non_vides - code
    print(f"   {NOMS[ed]} : code {100 * code / total:5.1f} % | commentaires {100 * commentaires / total:5.1f} %"
          f" | lignes vides {100 * (total - non_vides) / total:5.1f} %"
          f" | lignes de commentaire par ligne de code : {commentaires / code:.3f}")

l, c = RECOMPTE["grh"], RECOMPTE["grh-compact"]
commentaire_l, commentaire_c = l[1] - l[0], c[1] - c[0]
print("\n4. Rapports edition lisible / edition compacte, pour les memes cinq theoremes")
print(f"   lignes de code        : {l[0] / c[0]:.2f}")
print(f"   lignes en tout        : {l[2] / c[2]:.2f}")
print(f"   lignes de commentaire : {commentaire_l / commentaire_c:.2f}")
print("   lignes de commentaire par ligne de code, compacte / lisible :",
      f"{(commentaire_c / c[0]) / (commentaire_l / l[0]):.2f}")
print(f"   fichiers de src/Lib   : {FICHIERS['grh'][1] / FICHIERS['grh-compact'][1]:.2f}"
      f" | fichiers de src/New : {FICHIERS['grh'][2] / FICHIERS['grh-compact'][2]:.2f}")
````

**Mesuré.** Les deux éditions sont recomptées hors carnet, sur une copie du dépôt, et confrontées aux volumes que leurs README annoncent. La définition du comptage est celle que la cellule précédente écrit en tête : lignes non vides, commentaires de bloc retirés et lignes de commentaire écartées pour le code ; toutes les lignes pour le total.

| Édition | Lignes de code | Non vides avec commentaires | En tout |
|---|---|---|---|
| `grh/`, annoncées | 300 000 | 433 000 | 477 000 |
| `grh/`, recomptées | 299 916 | 432 547 | 477 151 |
| `grh-compact/`, annoncées | 76 000 | 95 000 | 113 000 |
| `grh-compact/`, recomptées | 76 166 | 95 050 | 112 854 |

Les volumes annoncés sont exacts à l'arrondi qu'ils revendiquent : le plus grand écart est de 0,22 %, et les README disent « about ». Un comptage honnête ne prouve pas une preuve juste ; mais un comptage faux aurait suffi à ne pas lire le reste, et il n'y en a pas ici.

### Que paie-t-on quand on resserre une preuve par quatre ?

C'est la question que pose l'existence de deux éditions pour les **mêmes** cinq théorèmes. Le recompte y répond en trois tableaux.

Ce que contient le texte :

| Édition | Code | Commentaires | Lignes vides |
|---|---|---|---|
| `grh/` | 62,9 % | 27,8 % | 9,3 % |
| `grh-compact/` | 67,5 % | 16,7 % | 15,8 % |

Le rapport de l'édition lisible à l'édition compacte :

| | Rapport |
|---|---|
| Lignes de code | 3,94 |
| Lignes en tout | 4,23 |
| Lignes de commentaire | 7,02 |
| Densité de commentaire (compacte sur lisible) | 0,56 |
| Fichiers de `src/Lib` | 0,86 |
| Fichiers de `src/New` | 1,40 |

La façon de découper le travail :

| Édition | Fichiers `.lean` | dont `src/Lib` | dont `src/New` |
|---|---|---|---|
| `grh/` | 805 | 495 | 306 |
| `grh-compact/` | 796 | 573 | 219 |

Trois choses s'y mesurent.

**Les commentaires paient plus que le code.** Le code est divisé par près de quatre, les commentaires par sept : à volume de code égal, l'édition compacte porte un peu plus de la moitié des commentaires de l'édition lisible. Ce qu'elle échange contre la brièveté, c'est donc d'abord l'explication — exactement ce que son `README` annonce en écrivant que ses preuves sont écrites pour être courtes, pas pour être faciles à lire.

**La compression se fait dans les fichiers, pas entre eux.** Les deux éditions comptent presque le même nombre de fichiers ; ce qui change, c'est ce qu'il y a dedans. Les lignes vides y prennent d'ailleurs une part plus grande, ce que la cellule ne cherche pas à expliquer.

**Le découpage se déplace.** L'édition compacte a plus de fichiers de `src/Lib` — les résultats de fond que Mathlib ne contient pas — et moins de fichiers de `src/New` — l'argument. Elle morcelle davantage le fond, et compte moins de fichiers d'argument. Or c'est `src/Lib` que le tableau de citations couvre, plus de huit rangées sur dix, et c'est aussi la partie dont plus d'une citation sur deux ne donne aucun lieu : la partie la plus recomposée est celle dont la provenance est la moins adressée. Les deux mesures se répondent sans s'expliquer l'une l'autre ; le carnet ne dit pas laquelle cause l'autre, et rien dans le dépôt ne le dit.

Pour mémoire, sans comparaison : la déclaration du 21/11 (section 8) parle de dix-sept millions de lignes pour l'archive d'Alpöge. Rien ne garantit que les deux comptages suivent la même définition, et ce carnet ne rapporte pas l'un à l'autre.

### Ce que le dépôt établit, et ce qu'il n'établit pas

| Ce que les mesures de cette section établissent | Ce qu'elles n'établissent pas |
|---|---|
| l'interface est fidèle au dépôt, et identique dans les deux éditions à une exception près — `src/Main.lean`, dont les docstrings et l'ordre de deux lignes changent | la preuve elle-même : rien n'a été compilé ni rejoué ici |
| les énoncés sur les fonctions L couvrent tout module et tout caractère ; le dernier se restreint aux caractères primitifs, et deux énoncés à la bande critique | l'absence de `sorry` dans tout `src/` : l'affirmation des `README` n'est recoupée que sur les deux fichiers d'interface |
| la clause du pôle est celle de Mathlib, pour la même raison | le corps de `RiemannHypothesis` à l'état que le dépôt épingle : le carnet en lit le nom, et la section 8 donne le corps à un autre état, sans que les deux soient reliés ici |
| les volumes annoncés sont exacts à l'arrondi revendiqué | la justesse de la bibliographie : le carnet recompte des états déclarés, il n'ouvre aucun ouvrage |
| plus d'une citation sur deux ne donne aucun lieu | l'exactitude des citations qui en donnent un |

**La portée, une dernière fois.** Le `README` de la racine l'écrit en une phrase, et c'est la plus précise du dépôt :

> A Lean 4 proof of the Riemann Hypothesis and of the Generalized Riemann Hypothesis for Dirichlet L-functions, built on Mathlib and using only the axioms `propext`, `Classical.choice` and `Quot.sound`.

Les énoncés du dépôt portent donc sur l'hypothèse de Riemann et sa généralisation **aux fonctions L de Dirichlet** — tous les modules, tous les caractères, primitifs ou non, d'après les `README` —, et sur rien d'autre : ils ne nomment ni corps de nombres, ni fonctions L de Hecke, ni formes automorphes. Les deux énoncés sur les zéros ne disent rien à gauche de la bande critique, et le troisième ne dit rien au point `s = 1` du caractère principal. « GRH est prouvée » ne veut donc rien dire seule : il faut dire **quelle** GRH, sous **quels** contrôles, et avec **quelle** relecture. Ici : celle de Dirichlet ; un noyau et un comparateur pour les deux éditions, des vérificateurs indépendants pour l'édition lisible seulement ; et, d'après le `README` de la racine lui-même, aucune relecture mathématicienne de bout en bout. C'est le piège du mot « généralisée » — il ne dit pas jusqu'où il va.

**Et par rapport à la section 8.** La déclaration du 21/11 certifie une preuve de RH — l'hypothèse pour ζ seule — et ne dit rien de GRH. Le dépôt du 25/11 revendique davantage : RH **et** GRH pour les fonctions L de Dirichlet. Mais les deux ne portent pas sur le même texte : d'après son propre `NOTICE`, le dépôt se donne comme une réécriture de la preuve d'Alpöge, et non comme la vérification d'une archive antérieure. La section 8 fermait une question sur un périmètre étroit ; celle-ci inventorie un périmètre plus large sans le fermer. Rien ici ne permet de dire que la certification du 21/11 couvre le dépôt du 25/11 : elle est antérieure, elle porte sur un autre travail, et elle nomme un énoncé plus étroit.
