## 3. La marche : trois marches verifiees en kernel

Premiere marche : la **dependance lineaire**. Le lemme de dependance (Axler 2.19, forme
existence) dit qu'une famille libre finie ne peut pas rester libre apres l'ajout d'un vecteur
du sous-espace qu'elle engendre ; sa consequence de longueur (2.22) borne une famille libre par
toute famille couvrante. Deuxieme marche : **Grassmann** (2.43). Troisieme : le **theoreme du
rang** (3.21).

```lean4
#check @combinaison_non_triviale
#check @longueur_libre_le_couverte
#check @grassmann
```

### Lecture du resultat

- `combinaison_non_triviale` : le lemme de dependance d'Axler 2.19, forme existence - la famille
  couvrante trop longue porte une combinaison non triviale qui vaut zero. La marche delegue le
  canal Mathlib `finrank_span_eq_card`.
- `longueur_libre_le_couverte` : Axler 2.22 - une famille libre n'est jamais plus longue qu'une
  famille couvrante du meme espace.
- `grassmann` : Axler 2.43, **forme additive en finrank** -
  `finrank K (U ⊔ W) + finrank K (U ⊓ W) = finrank K U + finrank K W`, deleguee a l'organe
  Mathlib `Submodule.finrank_sup_add_finrank_inf_eq`.

```lean4
#check @theoreme_fondamental
#check @vers_dimension_inferieure_non_injectif
#check @vers_dimension_superieure_non_surjectif
```

### Lecture du resultat

`theoreme_fondamental` est le theoreme du rang (Axler 3.21) :
`dim V = dim ker T + dim range T`, delegue a `LinearMap.finrank_range_add_finrank_ker`. Ses deux
corollaires de dimension (3.22, 3.24) disent l'impossible compression sans perte et l'impossible
couverture au-dessus de soi - deux intuitions qu'Axler place au centre du chapitre 3, sans
jamais ecrire une matrice.

## 4. Mesurer la surface : l'empreinte des axiomes

Un capstone sert a decider si l'on entre. La mesure qui compte ici : **qu'est-ce qui est prouve,
qu'est-ce qui est cible**. Le lac distingue les enonces du texte d'Axler (preuves completes,
delegation aux organes Mathlib) des **exercices** - quatre sorry-cibles documentees, comptees
dans la baseline CI du depot (`distinct_code_sorry = 4`), chacune portant un niveau prover
vise dans sa docstring. L'empreinte `#print axioms` est le temoin honnete de cette distinction.

```lean4
#print axioms grassmann
#print axioms decomposition_unique_somme_directe
#print axioms theoreme_fondamental
#print axioms isomorphe_ssi_meme_dimension
```

### Lecture du resultat

Les quatre enonces du texte reposent sur les axiomes standards de Mathlib (`propext`,
`Classical.choice`, `Quot.sound`) - aucun `sorry`, aucun `native_decide`. La preuve est la,
deleguee la ou Mathlib a l'organe, ecrite a la main la ou Axler demande une demonstration
(c'est le cas de `decomposition_unique_somme_directe`, Axler 1.45).

```lean4
#print axioms exercice_1B_3
#print axioms exercice_2A_13
#print axioms exercice_2C_12
#print axioms exercice_2C_13
```

### Lecture du resultat

`sorryAx` sur chacune : c'est la **signature d'une preuve non ecrite**, assumee et comptee. Les
quatre cibles vivent dans la baseline CI du depot avec leur niveau prover vise (`easy`,
`medium`...) - un moteur qui les resoudra fera tomber le compteur, et le carnet LINEAR-01 montre
deja cette empreinte en fin de parcours. Rien n'est maquille : l'ascension sait exactement ce
qu'elle a gravi et ce qui reste a grimper.
