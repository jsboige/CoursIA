## 1. Le geste : digérer, pas seulement vérifier

Tao distingue trois usages de l'IA sur les preuves : les **générer**, les **vérifier**, et les **digérer**, c'est-à-dire en tirer une compréhension. Sa déclaration *A Severe Misalignment of AI in Mathematics* (11 septembre 2026, DOI [10.5281/zenodo.22737750](https://doi.org/10.5281/zenodo.22737750)) rappelle la hiérarchie : *« Solving problems is only a tool and proxy for achieving the primary goal of conceptual understanding and insight. »*

Une preuve vérifiée par Lean n'est pas encore une preuve comprise. La sous-série pratique la digestion à travers quatre gestes :

| Geste | Ce qu'il produit | Où |

| **Lire** le lac | l'architecture de la preuve, fichier par fichier | ANALYSE-01, ANALYSE-02, ANALYSE-03 |

Le dernier geste est celui qu'on oublie. Une primitive qui ne se transporte pas n'est pas un échec : c'est un résultat, qui dit où finit le cadre d'origine.

ANALYSE-03 (section 1) donne la version entropique de PFR sous cette forme : pour deux variables X₁ et X₂ à valeurs dans F₂ⁿ, il existe un sous-groupe H tel que

$$d[X_1 ; U_H] + d[X_2 ; U_H] \le 11 \cdot d[X_1 ; X_2],$$

où U_H est la loi uniforme sur H et d la distance entropique de Ruzsa. Avec X₁ = X₂ = X, l'énoncé dit que **si X est presque « compressée » (d[X ; X] petit), alors X est proche d'une loi uniforme sur un sous-groupe**. C'est le sens de la formule « compression ⇒ structure ».

## 4. La marche : deux primitives de PFR à la main

On travaille dans le plus petit groupe où l'énoncé a du sens, F₂³. Ses 8 éléments sont codés comme des entiers de 3 bits, et l'addition est le « ou exclusif » (XOR).

- **Primitive 2 — projection et fibres.** Pour une projection π, par exemple le passage au quotient par un sous-groupe H, on a la règle de chaîne H(X) = H(π(X)) + H(X | π(X)). L'information de X se partage exactement entre la classe de X et sa position dans la fibre. La cellule prend H = {000, 001}, qui découpe F₂³ en 4 classes.
- **Primitive 1 — distance de Ruzsa.** d[X ; Y] = H(X' − Y') − H(X)/2 − H(Y)/2, où X' et Y' sont des copies indépendantes. Si X est uniforme sur un sous-groupe, X' − X'' est uniforme sur le même sous-groupe, donc d[X ; X] = 0. La cellule mélange la loi uniforme sur un sous-groupe à 4 éléments avec la loi uniforme sur F₂³, en proportion ε. Pour chaque mélange, elle cherche, parmi les 16 sous-groupes de F₂³, celui qui rend d[X ; U_H] minimal.

*Copie pédagogique déclarée* : ces fonctions reprennent une petite partie d'ANALYSE-04 pour que la marche tienne en une cellule. La digestion complète, limites comprises, reste dans [ANALYSE-04](Analyse/ANALYSE-04-PFR-Primitives-Python.ipynb).

```
// Deux primitives de PFR, calculees a la main sur le groupe F_2^3.
// Copie pedagogique declaree : la digestion complete vit dans ANALYSE-04.
using System;
using System.Collections.Generic;
using System.Globalization;
using System.Linq;

var elements = Enumerable.Range(0, 8).ToArray(); // F_2^3, un element = un entier de 3 bits, addition = XOR
var culturePfr = CultureInfo.InvariantCulture;
// Quelques ulps autour d'une entropie de 3 bits pour le résidu de la règle de chaîne.
const double toleranceChaine = 2e-15;

// Références citées : sortie de la cellule 5 du carnet Python, aux précisions affichées.
var referenceChaine = (HX: "2.708695", HPi: "1.760964", HFibres: "0.947731", Classes: "4", Ecart: "0.00e+00");
const string referenceSousGroupes = "16";
var referenceRuzsa = new (string Etiquette, string Dxx, string Dxh, string Taille)[]
{
    ("melange eps=0.0", "0.0000", "0.0000", "4"),
    ("melange eps=0.1", "0.1665", "0.1432", "4"),
    ("melange eps=0.3", "0.2093", "0.1951", "8"),
    ("melange eps=0.6", "0.1002", "0.0594", "8"),
    ("melange eps=1.0", "0.0000", "0.0000", "8"),
    ("X de la primitive 2", "0.2520", "0.1457", "8")
};
var comparaisonsPfr = new List<(string Cas, string Calcule, string ReferencePython)>();

string FormatPfr(double valeur, string format) => valeur.ToString(format, culturePfr);

void VerifierPfr(string cas, string calcule, string referencePython)
{
    comparaisonsPfr.Add((cas, calcule, referencePython));
    if (!string.Equals(calcule, referencePython, StringComparison.Ordinal))
        throw new InvalidOperationException($"{cas} : C# = {calcule} ; référence Python citée = {referencePython} (ÉCART).");
}

// Entropie de Shannon (en bits) d'une loi valeur → probabilité.
double Entropie(IReadOnlyDictionary<int, double> loi)
{
    double somme = 0.0;
    foreach (var p in loi.Values)
        if (p > 0.0)
            somme += p * Math.Log2(p);
    return -somme;
}

// Loi de f(X) quand X suit loi.
Dictionary<int, double> LoiImage(IReadOnlyDictionary<int, double> loi, Func<int, int> f)
{
    var image = new Dictionary<int, double>();
    foreach (var (x, p) in loi)
    {
        int c = f(x);
        image.TryGetValue(c, out double masse);
        image[c] = masse + p;
    }
    return image;
}

// H(X | f(X)) : moyenne pondérée des entropies sur les fibres.
double EntropieConditionnelle(IReadOnlyDictionary<int, double> loi, Func<int, int> f)
{
    double total = 0.0;
    foreach (var (c, pc) in LoiImage(loi, f))
    {
        var fibre = loi.Where(kv => f(kv.Key) == c)
                       .ToDictionary(kv => kv.Key, kv => kv.Value / pc);
        total += pc * Entropie(fibre);
    }
    return total;
}

// Loi de X' − Y' pour des copies indépendantes : dans F₂³, la différence est XOR.
Dictionary<int, double> SommeIndependante(IReadOnlyDictionary<int, double> loiX, IReadOnlyDictionary<int, double> loiY)
{
    var somme = new Dictionary<int, double>();
    foreach (var (x, p) in loiX)
        foreach (var (y, q) in loiY)
        {
            int z = x ^ y;
            somme.TryGetValue(z, out double masse);
            somme[z] = masse + p * q;
        }
    return somme;
}

// Distance entropique de Ruzsa.
double DistanceRuzsa(IReadOnlyDictionary<int, double> loiX, IReadOnlyDictionary<int, double> loiY) =>
    Entropie(SommeIndependante(loiX, loiY)) - Entropie(loiX) / 2.0 - Entropie(loiY) / 2.0;

// Comme itertools.combinations, énumérer les familles de taille fixée sans répétition.
IEnumerable<int[]> Familles(int premier, int taille)
{
    if (taille == 0)
    {
        yield return Array.Empty<int>();
        yield break;
    }
    for (int v = premier; v <= elements.Length - taille; v++)
        foreach (var suite in Familles(v + 1, taille - 1))
            yield return new[] { v }.Concat(suite).ToArray();
}

int[][] SousGroupes()
{
    var vus = new Dictionary<int, int[]>();
    for (int k = 0; k <= 3; k++)
        foreach (var famille in Familles(1, k))
        {
            var engendre = new HashSet<int> { 0 };
            foreach (int v in famille)
                engendre.UnionWith(engendre.Select(x => x ^ v).ToArray());
            int masque = engendre.Aggregate(0, (bits, x) => bits | (1 << x));
            vus[masque] = engendre.OrderBy(x => x).ToArray();
        }
    // À taille égale, l'ordre lexicographique est celui des chiffres en base 8.
    return vus.Values.OrderBy(h => h.Length)
                     .ThenBy(h => h.Aggregate(0, (cle, x) => 8 * cle + x))
                     .ToArray();
}

Dictionary<int, double> Uniforme(IReadOnlyCollection<int> ensemble) =>
    ensemble.ToDictionary(x => x, x => 1.0 / ensemble.Count);

// Primitive 2 : projection / fibres. Quotient par le sous-groupe H = {000, 001}.
var X = new Dictionary<int, double>
{
    [0] = 0.30, [1] = 0.20, [2] = 0.15, [3] = 0.05,
    [4] = 0.10, [5] = 0.10, [6] = 0.05, [7] = 0.05
};
int Classe(int x) => x & 0b110; // representant de la classe x + H
var projection = LoiImage(X, Classe);
double hX = Entropie(X);
double hPi = Entropie(projection);
double hFibres = EntropieConditionnelle(X, Classe);
double ecartChainePfr = Math.Abs(hX - (hPi + hFibres));
if (!(ecartChainePfr <= toleranceChaine))
    throw new InvalidOperationException($"Règle de chaîne : résidu {FormatPfr(ecartChainePfr, "G17")} bits supérieur à la tolérance.");

string texteHX = FormatPfr(hX, "F6");
string texteHPi = FormatPfr(hPi, "F6");
string texteHFibres = FormatPfr(hFibres, "F6");
string texteClasses = projection.Count.ToString(culturePfr);
// Le résidu brut reste disponible ; le zéro affiché signifie « nul à la précision machine ».
string texteEcart = FormatPfr(ecartChainePfr <= toleranceChaine ? 0.0 : ecartChainePfr, "0.00e+00");
VerifierPfr("H(X)", texteHX, referenceChaine.HX);
VerifierPfr("H(pi(X))", texteHPi, referenceChaine.HPi);
VerifierPfr("H(X | pi(X))", texteHFibres, referenceChaine.HFibres);
VerifierPfr("Nombre de classes", texteClasses, referenceChaine.Classes);
VerifierPfr("Résidu affiché de la règle de chaîne", texteEcart, referenceChaine.Ecart);
Console.WriteLine("Primitive 2 -- regle de chaine H(X) = H(pi(X)) + H(X | pi(X))");
Console.WriteLine($"  H(X)          = {texteHX} bits");
Console.WriteLine($"  H(pi(X))      = {texteHPi} bits  ({texteClasses} classes)");
Console.WriteLine($"  H(X | pi(X))  = {texteHFibres} bits  (moyenne sur les fibres)");
Console.WriteLine($"  ecart         = {texteEcart}");

// Primitive 1 : distance de Ruzsa, et le signal compression => structure.
var groupes = SousGroupes();
string texteSousGroupes = groupes.Length.ToString(culturePfr);
VerifierPfr("Nombre de sous-groupes", texteSousGroupes, referenceSousGroupes);
Console.WriteLine($"\nPrimitive 1 -- distance de Ruzsa sur {texteSousGroupes} sous-groupes de F_2^3");
Console.WriteLine($"  {"loi de X",-28} {"d[X;X]",10} {"min_H d[X;U_H]",14} {"|H*|",8}");

(string Etiquette, double Dxx, double Dxh, int Taille) MesurerRuzsa(string etiquette, IReadOnlyDictionary<int, double> loi)
{
    double dxx = DistanceRuzsa(loi, loi);
    // Comme min en Python, conserver le premier sous-groupe en cas d'égalité.
    int[] meilleur = groupes[0];
    double minimum = DistanceRuzsa(loi, Uniforme(meilleur));
    foreach (var h in groupes.Skip(1))
    {
        double distance = DistanceRuzsa(loi, Uniforme(h));
        if (distance < minimum)
        {
            meilleur = h;
            minimum = distance;
        }
    }
    return (etiquette, dxx, minimum, meilleur.Length);
}

var uSg = Uniforme(new[] { 0, 1, 2, 3 });
var lignesRuzsa = new List<(string Etiquette, double Dxx, double Dxh, int Taille)>();
foreach (double eps in new[] { 0.0, 0.1, 0.3, 0.6, 1.0 })
{
    var loi = elements.ToDictionary(x => x, x =>
    {
        uSg.TryGetValue(x, out double masse);
        return (1.0 - eps) * masse + eps / 8.0;
    });
    var mesure = MesurerRuzsa($"melange eps={FormatPfr(eps, "F1")}", loi);
    if (eps == 0.0 || eps == 1.0)
    {
        if (mesure.Dxx != 0.0 || mesure.Dxh != 0.0)
            throw new InvalidOperationException($"{mesure.Etiquette} : les deux distances doivent être nulles.");
    }
    else if (!(mesure.Dxx > 0.0 && mesure.Dxh < mesure.Dxx))
        throw new InvalidOperationException($"{mesure.Etiquette} : ordre des distances inattendu.");
    lignesRuzsa.Add(mesure);
}
var mesureX = MesurerRuzsa("X de la primitive 2", X);
if (!(mesureX.Dxh < mesureX.Dxx))
    throw new InvalidOperationException("X de la primitive 2 : ordre des distances inattendu.");
lignesRuzsa.Add(mesureX);
VerifierPfr("Nombre de cas Ruzsa", lignesRuzsa.Count.ToString(culturePfr), referenceRuzsa.Length.ToString(culturePfr));
for (int i = 0; i < lignesRuzsa.Count; i++)
{
    var (etiquette, dxx, dxh, taille) = lignesRuzsa[i];
    var reference = referenceRuzsa[i];
    string texteDxx = FormatPfr(dxx, "F4");
    string texteDxh = FormatPfr(dxh, "F4");
    string texteTaille = taille.ToString(culturePfr);
    VerifierPfr($"Cas Ruzsa {i.ToString(culturePfr)}", etiquette, reference.Etiquette);
    VerifierPfr($"{etiquette} : d[X;X]", texteDxx, reference.Dxx);
    VerifierPfr($"{etiquette} : min_H d[X;U_H]", texteDxh, reference.Dxh);
    VerifierPfr($"{etiquette} : |H*|", texteTaille, reference.Taille);
    Console.WriteLine($"  {etiquette,-28} {texteDxx,10} {texteDxh,14} {texteTaille,8}");
}
```

### Lecture du résultat

- **La règle de chaîne tient exactement** : 2,7087 = 1,7610 + 0,9477 bits, avec un écart nul à la précision machine. Elle vaut pour toute loi et pour toute projection. C'est pourquoi ANALYSE-04 classe cette primitive parmi celles qui se transportent le plus largement.
- **La structure, c'est une distance nulle.** Aux deux extrémités, ε = 0 (loi uniforme sur un sous-groupe à 4 éléments) et ε = 1 (loi uniforme sur F₂³ tout entier, qui est lui-même un sous-groupe), d[X ; X] vaut exactement 0. Entre les deux, la distance est strictement positive, et le sous-groupe le plus proche passe de 4 à 8 éléments entre ε = 0,1 et ε = 0,3.
- **Sur chaque ligne, min_H d[X ; U_H] reste inférieur ou égal à d[X ; X]**, avec égalité aux deux extrémités du mélange. Quand d[X ; X] est strictement positif, le minimum est donc très en deçà de ce que garantit le théorème (2 · d[X ; U_H] ≤ 11 · d[X ; X]). Sur un groupe aussi petit, la borne n'est pas serrée, et on peut essayer les 16 sous-groupes un par un. Tout l'intérêt du théorème est qu'il vaut dans F₂ⁿ pour tout n, là où le nombre de sous-groupes explose et où aucune énumération n'est possible.

## 5. Mesurer la surface de la sous-série

Un capstone sert à décider **si l'on entre**. Les questions du lecteur (quel noyau, combien d'exercices, quels lacs) ont des réponses objectives, écrites dans les carnets eux-mêmes. La cellule suivante lit les carnets du dossier. Pour chacun, elle relève :

- le noyau déclaré ;
- le nombre de cellules de code et combien portent une trace d'exécution ;
- le nombre de titres d'exercice ;
- le nombre de déclarations distinctes passées à `#check` ;
- les lacs `teorth` cités ;
- le premier titre.

Elle rend aussi visible ce que la description ne dit pas.

```
// Surface de la sous-serie : on lit les carnets, on ne suppose rien.
using System.IO;
using System.Text;
using System.Text.Json;
using System.Text.Json.Nodes;
using System.Text.RegularExpressions;

const string racineAnalyse = "Analyse";
const string caracteresMotAnalyse = @"\p{L}\p{N}_";
const string separateursLignesAnalyse = @"\r\n|[\n\r\x0B\x0C\x1C-\x1E\x85\u2028\u2029]";
var ordreAnalyse = Comparer<string>.Create((a, b) =>
    Encoding.UTF8.GetBytes(a).AsSpan().SequenceCompareTo(Encoding.UTF8.GetBytes(b)));
var exosAnalyse = new Regex(@"(?m)^#{3,4} *(?:[\d.]+ *)?Exercice(?![" + caracteresMotAnalyse + "])");
var lacsAnalyse = new Regex(@"teorth/[" + caracteresMotAnalyse + "-]+");
var checksAnalyse = new Regex(@"#check[\s\x1C-\x1F]+@?([" + caracteresMotAnalyse + ".]+)");

// Le découpage et les largeurs Python comptent des points de code, non des unités UTF-16.
string TrancheAnalyse(string texte, int debut, int longueur) =>
    string.Concat(texte.EnumerateRunes().Skip(debut).Take(longueur).Select(r => r.ToString()));
string ColonneAnalyse(string texte, int largeur) =>
    texte + new string(' ', Math.Max(0, largeur - texte.EnumerateRunes().Count()));
string NombreAnalyse(int n) => n.ToString(CultureInfo.InvariantCulture);
string[] TrierDistinctAnalyse(IEnumerable<string> elementsAnalyse) =>
    elementsAnalyse.Distinct(StringComparer.Ordinal).OrderBy(s => s, ordreAnalyse).ToArray();

// Noyau, cellules exécutées, exercices, lacs cités et titre d'un carnet.
string SourceAnalyse(JsonElement cellule)
{
    var source = cellule.GetProperty("source");
    return source.ValueKind == JsonValueKind.String
        ? source.GetString() ?? ""
        : string.Concat(source.EnumerateArray().Select(partie => partie.GetString()));
}

(string Kernel, int Code, int Executees, int Exercices, string[] Lacs, int Checks, string Titre)
Fiche(string chemin)
{
    using var flux = File.OpenRead(chemin);
    using var document = JsonDocument.Parse(flux);
    var carnet = document.RootElement;
    var cellules = carnet.GetProperty("cells").EnumerateArray().ToArray();
    var code = cellules.Where(c => c.GetProperty("cell_type").GetString() == "code").ToArray();
    var texte = string.Join("\n", cellules.Select(SourceAnalyse));
    var titre = cellules.Where(c => c.GetProperty("cell_type").GetString() == "markdown")
                        .SelectMany(c => Regex.Split(SourceAnalyse(c), separateursLignesAnalyse))
                        .FirstOrDefault(ligne => ligne.StartsWith("# ", StringComparison.Ordinal)) ?? "(aucun titre)";
    var metadata = carnet.GetProperty("metadata");
    string kernel = metadata.TryGetProperty("kernelspec", out var spec) && spec.TryGetProperty("name", out var nom)
        ? nom.GetString() ?? "None" : "?";
    return (
        kernel,
        code.Length,
        code.Count(c => c.TryGetProperty("execution_count", out var execution) && execution.ValueKind != JsonValueKind.Null),
        exosAnalyse.Matches(texte).Count,
        TrierDistinctAnalyse(lacsAnalyse.Matches(texte).Select(m => m.Value)),
        checksAnalyse.Matches(texte).Select(m => m.Groups[1].Value).Distinct(StringComparer.Ordinal).Count(),
        TrancheAnalyse(titre, 2, 60));
}

if (!Directory.Exists(racineAnalyse))
    throw new DirectoryNotFoundException($"Dossier Analyse introuvable depuis {Directory.GetCurrentDirectory()}.");
var carnetsAnalyse = Directory.EnumerateFiles(racineAnalyse)
    .Select(chemin => Path.GetFileName(chemin) ?? "")
    .Where(nom => nom.EndsWith(".ipynb", StringComparison.Ordinal))
    .OrderBy(nom => nom, ordreAnalyse).ToArray();
if (carnetsAnalyse.Length == 0)
    throw new InvalidOperationException("Le dossier Analyse ne contient aucun carnet .ipynb.");

var fiches = new Dictionary<string, (string Kernel, int Code, int Executees, int Exercices, string[] Lacs, int Checks, string Titre)>(StringComparer.Ordinal);
foreach (var nomCarnet in carnetsAnalyse)
    fiches.Add(nomCarnet, Fiche(Path.Combine(racineAnalyse, nomCarnet)));
var unionLacsAnalyse = TrierDistinctAnalyse(fiches.Values.SelectMany(f => f.Lacs));

// Contre-lecture locale : énumération séparée, arbre JSON mutable et parcours explicite des cellules.
// Les références viennent du disque ; aucune ligne de l'inventaire historique n'est figée ici.
int nbComparaisonsAnalyse = 0;
int nbEcartsAnalyse = 0;
void VerifierAnalyse(bool conforme, string champ)
{
    nbComparaisonsAnalyse++;
    if (!conforme)
    {
        nbEcartsAnalyse++;
        throw new InvalidOperationException($"Surface d'Analyse — écart de contre-lecture : {champ}.");
    }
}

var nomsControleAnalyse = Directory.GetFileSystemEntries(racineAnalyse)
    .Where(chemin => chemin.EndsWith(".ipynb", StringComparison.Ordinal) && File.Exists(chemin))
    .Select(chemin => Path.GetFileName(chemin) ?? "")
    .OrderBy(nom => nom, ordreAnalyse).ToArray();
VerifierAnalyse(carnetsAnalyse.SequenceEqual(nomsControleAnalyse, StringComparer.Ordinal), "liste triée des carnets");
var unionControleAnalyse = new SortedSet<string>(ordreAnalyse);
foreach (var nomCarnet in nomsControleAnalyse)
{
    var carnet = JsonNode.Parse(File.ReadAllText(Path.Combine(racineAnalyse, nomCarnet)))!.AsObject();
    var spec = carnet["metadata"]!["kernelspec"]?.AsObject();
    string noyauControle = spec != null && spec.ContainsKey("name") ? spec["name"]?.GetValue<string>() ?? "None" : "?";
    int codeControle = 0;
    int execControle = 0;
    string titreControle = "(aucun titre)";
    bool titreTrouve = false;
    var textesControle = new List<string>();
    foreach (var cellule in carnet["cells"]!.AsArray())
    {
        var source = cellule!["source"];
        var texteCellule = source is JsonArray parties
            ? string.Concat(parties.Select(partie => partie!.GetValue<string>()))
            : source!.GetValue<string>();
        textesControle.Add(texteCellule);
        string type = cellule["cell_type"]!.GetValue<string>();
        if (type == "code")
        {
            codeControle++;
            if (cellule["execution_count"] != null)
                execControle++;
        }
        if (type == "markdown" && !titreTrouve)
        {
            // splitlines() reconnaît les séparateurs Python, y compris hors CR/LF.
            var premierTitre = Regex.Match(texteCellule,
                @"(?:\A|[\n\r\x0B\x0C\x1C-\x1E\x85\u2028\u2029])(# [^\n\r\x0B\x0C\x1C-\x1E\x85\u2028\u2029]*)");
            if (premierTitre.Success)
            {
                titreControle = premierTitre.Groups[1].Value;
                titreTrouve = true;
            }
        }
    }
    string texteControle = string.Join("\n", textesControle);
    // En mode multiligne, ^ dans re ne reconnaît que le début du texte et l'après-LF.
    int exosControle = texteControle.Split('\n').Count(ligne => exosAnalyse.IsMatch(ligne));
    var checksControle = new HashSet<string>(StringComparer.Ordinal);
    foreach (Match m in checksAnalyse.Matches(texteControle))
        checksControle.Add(m.Groups[1].Value);
    var lacsControle = new SortedSet<string>(ordreAnalyse);
    foreach (Match m in lacsAnalyse.Matches(texteControle))
        lacsControle.Add(m.Value);
    unionControleAnalyse.UnionWith(lacsControle);

    var f = fiches[nomCarnet];
    VerifierAnalyse(f.Kernel == noyauControle, $"{nomCarnet} / noyau");
    VerifierAnalyse(f.Code == codeControle, $"{nomCarnet} / code");
    VerifierAnalyse(f.Executees == execControle, $"{nomCarnet} / exec");
    VerifierAnalyse(f.Exercices == exosControle, $"{nomCarnet} / exos");
    VerifierAnalyse(f.Checks == checksControle.Count, $"{nomCarnet} / #check");
    VerifierAnalyse(f.Lacs.SequenceEqual(lacsControle, StringComparer.Ordinal), $"{nomCarnet} / lacs");
    string titreAttendu = string.Concat(titreControle.EnumerateRunes().Take(62).Skip(2).Select(r => r.ToString()));
    VerifierAnalyse(f.Titre == titreAttendu, $"{nomCarnet} / premier titre");
}
VerifierAnalyse(unionLacsAnalyse.SequenceEqual(unionControleAnalyse, StringComparer.Ordinal), "union triée des lacs");

Console.WriteLine($"{ColonneAnalyse("carnet", 40)} {ColonneAnalyse("kernel", 10)} {"code",6} {"exec",6} {"exos",5} {"#check",6}");
foreach (var nomCarnet in carnetsAnalyse)
{
    var f = fiches[nomCarnet];
    Console.WriteLine($"{ColonneAnalyse(TrancheAnalyse(nomCarnet, 0, 40), 40)} {ColonneAnalyse(f.Kernel, 10)} {NombreAnalyse(f.Code),6} {NombreAnalyse(f.Executees),6} {NombreAnalyse(f.Exercices),5} {NombreAnalyse(f.Checks),6}");
}
Console.WriteLine("\nLacs externes cites : [" + string.Join(", ", unionLacsAnalyse.Select(lac => "'" + lac + "'")) + "]");
Console.WriteLine("\nPremier titre de chaque carnet :");
foreach (var nomCarnet in carnetsAnalyse)
    Console.WriteLine($"  {ColonneAnalyse(TrancheAnalyse(nomCarnet, 0, 11), 12)} -> {fiches[nomCarnet].Titre}");

// Attestation : égalité au format publié pour PFR, contrôle sur les fichiers courants pour Analyse.
if (comparaisonsPfr.Count == 0)
    throw new InvalidOperationException("Les comparaisons PFR doivent précéder le bilan.");
Console.WriteLine("\nBilan des comparaisons exécutées");
void BilanPfr(string section, bool regleDeChaine)
{
    var cas = comparaisonsPfr.Where(c =>
        (c.Cas.StartsWith("H(", StringComparison.Ordinal) || c.Cas == "Nombre de classes" ||
         c.Cas == "Résidu affiché de la règle de chaîne") == regleDeChaine).ToArray();
    if (cas.Length == 0)
        throw new InvalidOperationException($"Aucune donnée de parité pour {section}.");
    int ecarts = cas.Count(c => c.Calcule != c.ReferencePython);
    Console.WriteLine($"{section} — référence Python citée : {NombreAnalyse(ecarts)} écart(s) sur {NombreAnalyse(cas.Length)} comparaisons.");
    if (ecarts != 0)
        throw new InvalidOperationException($"Écart de parité pour {section}.");
}
BilanPfr("Règle de chaîne", true);
BilanPfr("Distance de Ruzsa", false);
Console.WriteLine($"Surface d'Analyse — contre-lecture locale : {NombreAnalyse(nbEcartsAnalyse)} écart(s) sur {NombreAnalyse(nbComparaisonsAnalyse)} contrôles, pour {NombreAnalyse(fiches.Count)} carnets lus.");
```
