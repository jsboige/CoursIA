## 1. Le problème du vote electronique

Le vote electronique doit satisfaire des proprietes **contradictoires en apparence** :

| Propriete | Description | Tension |
|-----------|-------------|--------|
| **Anonymat** | Personne ne sait ce que j'ai vote | vs Verifiabilite |
| **Verifiabilite individuelle** | Je peux verifier que mon vote a ete compte | vs Anonymat |
| **Verifiabilite universelle** | N'importe qui peut verifier le decompte | vs Anonymat |
| **Eligibilite** | Seuls les electeurs autorises votent | vs Anonymat |
| **Unicite** | Un electeur = un vote | vs Anonymat |
| **Resistance a la coercition** | Impossible de prouver pour qui on a vote | vs Verifiabilite |

La solution : le **chiffrement homomorphique** permet de compter les votes
sans les dechiffrer individuellement, et les **preuves a divulgation nulle** (ZKP)
permettent de verifier qu'un bulletin est valide sans reveler son contenu.

### Historique

| Annee | Système | Contribution |
|-------|---------|-------------|
| 2008 | **Helios** (Ben Adida) | Premier système E2E verifiable grand public |
| 2012+ | **Belenios** (INRIA) | Preuves formelles en Coq, plus fort garanties |
| 2019 | **ElectionGuard** (Microsoft) | SDK Python production-ready, threshold decryption |

## 2. Chiffrement des bulletins (Paillier)

Chaque electeur chiffre son vote avec Paillier (additively homomorphic).
Un vote pour le candidat i est represente comme un vecteur one-hot :
- Candidat A : `[1, 0, 0]`
- Candidat B : `[0, 1, 0]`
- Candidat C : `[0, 0, 1]`

Chaque composante est chiffree individuellement.

````python
try:
    from phe import paillier
    PHE_AVAILABLE = True
except ImportError:
    PHE_AVAILABLE = False
    print("phe non installe. Installez avec: pip install phe")
import hashlib
import json

if PHE_AVAILABLE:
    # Autorite electorale : genere les cles
    public_key, private_key = paillier.generate_paillier_keypair(n_length=1024)
    print("AUTORITE ELECTORALE")
    print("=" * 60)
    print(f"Cle publique (n) : {str(public_key.n)[:40]}...")
    print(f"Cle publique distribuee a tous les electeurs")
    print(f"Cle privee conservee par l'autorite (ou repartie via threshold)")
    print()

    # Candidats
    candidates = ["Alice Dupont", "Bob Martin", "Charlie Durand"]
    print(f"Candidats : {candidates}")
    print()
else:
    print("Section sautee : phe non disponible")
````

### Interpretation : Generation des clés

L'autorite electorale genere une paire de clés Paillier :

| Élément | Rôle | Distribution |
|---------|------|-------------|
| **Cle publique** (n) | Chiffrer les bulletins | Distribuee a tous les electeurs |
| **Cle privee** | Dechiffrer les résultats agreges | Conservee par l'autorite (ou k-of-n gardiens) |

**Representation des votes** : chaque vote est un vecteur one-hot (un 1 et des 0). Par exemple avec 3 candidats, voter pour Alice = `[1, 0, 0]`. Cette representation permet l'addition homomorphique composante par composante pour obtenir le decompte sans dechiffrer les bulletins individuels.

**Point cle** : en production, la cle privee est repartie via **Shamir secret sharing** entre N gardiens (threshold k-of-n), eliminant le point de defaillance unique.


Simulation du processus de vote pour 7 électeurs avec chiffrement homomorphique de chaque bulletin.

````python
# Simulation de 7 electeurs
# Note sur la reproductibilite : `phe` tire ses aleas d'un CSPRNG
# (SystemRandom, qui s'appuie sur /dev/urandom sous Linux et sur
# BCryptGenRandom sous Windows). `random.seed(42)` ne rend donc PAS le
# chiffrement reproductible : deux exécutions de ce carnet produisent des
# ciphertexts et des `ballot_hash` différents. C'est le comportement attendu
# d'un système de vote -- l'alea fraiche protege le secret du vote. La cle
# publique est elle aussi re-tiree a chaque exécution (cf. cellule 3).
# Un test qui a besoin de reexecuter doit FIXER les aleas lui-même
# (`r_value=` de `encrypt`), pas s'attendre a une reproductibilite du
# bibliotheque.

if PHE_AVAILABLE:
    voters = [
        {"name": "Electeur 1", "choice": 0},  # vote Alice
        {"name": "Electeur 2", "choice": 1},  # vote Bob
        {"name": "Electeur 3", "choice": 0},  # vote Alice
        {"name": "Electeur 4", "choice": 2},  # vote Charlie
        {"name": "Electeur 5", "choice": 0},  # vote Alice
        {"name": "Electeur 6", "choice": 1},  # vote Bob
        {"name": "Electeur 7", "choice": 2},  # vote Charlie
    ]

    # Chaque electeur chiffre son bulletin
    encrypted_ballots = []
    print("CHIFFREMENT DES BULLETINS")
    print("=" * 60)

    for voter in voters:
        # Creer le vecteur one-hot
        plaintext = [0] * len(candidates)
        plaintext[voter["choice"]] = 1

        # Chiffrer chaque composante
        encrypted = [public_key.encrypt(v) for v in plaintext]
        encrypted_ballots.append(encrypted)

        # Le bulletin chiffre ne revele rien
        print(f"  {voter['name']} : bulletin chiffre (3 ciphertexts de ~{len(str(encrypted[0].ciphertext()))} digits)")

    print(f"\n-> {len(encrypted_ballots)} bulletins chiffres, aucun vote individuel visible")
else:
    print("Section sautee : phe non disponible")
````

### Interpretation : securite sémantique des bulletins

Chaque bulletin chiffre est un vecteur de 3 ciphertexts Paillier.
Même l'autorite ne peut pas deviner le contenu d'un bulletin individuel
sans utiliser la cle privee. Le chiffrement est **semantiquement securise** :
deux chiffrements du même message sont différents (grace a l'aléatoire).

## 3. Decompte homomorphique

Grace a la propriete additive de Paillier, on peut **additionner les bulletins chiffres**
sans les dechiffrer. Seul le **total** est ensuite dechiffre.

````python
# Addition homomorphique des bulletins
# (SOTA Prong-A #3801 : la barre ASCII devisualisation est remplacee par un vrai
#  graphique matplotlib ; le decompte homomorphique Paillier, lui, est inchange.)
if PHE_AVAILABLE:
    print("DECOMPTE HOMOMORPHIQUE")
    print("=" * 60)

    # Initialiser les totaux chiffres a 0
    tally_encrypted = [public_key.encrypt(0) for _ in candidates]

    # Additionner tous les bulletins
    for ballot in encrypted_ballots:
        for i in range(len(candidates)):
            tally_encrypted[i] = tally_encrypted[i] + ballot[i]

    print("Totaux chiffres calcules (sans dechiffrer aucun bulletin individuel)")
    print()

    # Seul le total est dechiffre par l'autorite
    results = [private_key.decrypt(t) for t in tally_encrypted]

    print("RESULTATS :")
    print("-" * 40)
    for candidate, votes in zip(candidates, results):
        print(f"  {candidate:20s} : {votes} votes")
    print(f"  {'Total':20s} : {sum(results)} votes")
    print()

    # Verification manuelle
    expected = [0] * len(candidates)
    for v in voters:
        expected[v["choice"]] += 1
    assert results == expected, "Le decompte homomorphique doit correspondre au decompte manuel"
    print("-> Verification : le decompte homomorphique correspond au decompte en clair")

    # Visualisation : repartition des voix dechiffrees (matplotlib, remplace l'ASCII).
    # Rendu via savefig + IPython.display.Image (backend Agg headless : plt.show()
    # n'affiche rien en sortie et emet un avertissement ; savefig garantit un PNG).
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    import io
    from IPython.display import Image, display

    fig, ax = plt.subplots(figsize=(7, 3))
    palette = ["#2980b9", "#27ae60", "#c0392b", "#8e44ad", "#d35400", "#16a085"]
    ax.barh(range(len(candidates)), results, color=palette[:len(candidates)])
    ax.set_yticks(range(len(candidates)))
    ax.set_yticklabels(candidates)
    ax.invert_yaxis()
    ax.set_xlabel("Nombre de voix")
    ax.set_title("Decompte homomorphique Paillier - resultats dechiffres")
    ax.set_xlim(0, max(results) + 1)
    for i, v in enumerate(results):
        ax.text(v + 0.05, i, str(v), va="center", fontweight="bold")
    plt.tight_layout()
    buf = io.BytesIO()
    fig.savefig(buf, format="png", dpi=110, bbox_inches="tight")
    buf.seek(0)
    display(Image(data=buf.getvalue()))
    plt.close(fig)
else:
    print("Section sautee : phe non disponible")
````

### Interpretation : decompte sur données chiffrees

Le decompte s'est fait **entierement sur des données chiffrees** :
- Aucun bulletin individuel n'a ete dechiffre
- Seul le résultat agrege a ete revele
- L'anonymat de chaque electeur est preserve

Mais comment garantir que les bulletins sont **valides** (exactement un 1 et des 0) ?
C'est la que les ZKP interviennent.

## 4. Preuve de validite du bulletin (ZKP simplifiee)

Chaque electeur doit prouver que son bulletin chiffre contient un **vote valide**
(exactement un 1 parmi les composantes) **sans reveler pour qui il a vote**.

Nous implementons ici une version simplifiee basee sur la **somme des composantes**.
Un bulletin valide a une somme de composantes = 1 (un seul candidat choisi).

````python
import hashlib

def compute_ballot_hash(encrypted_ballot):
    """Calculer un identifiant unique pour un bulletin chiffre."""
    data = "|".join(str(e.ciphertext()) for e in encrypted_ballot)
    return hashlib.sha256(data.encode()).hexdigest()

def prove_valid_ballot(public_key, encrypted_ballot, plaintext_vote, num_candidates):
    """Constater qu'un bulletin est un vote unique (un seul candidat choisi).

    Pourquoi pas seulement la somme ? Paillier est additif modulo n : le
    triplet [1000, -999, 0] a une somme de 1. Un contrôle de la somme seule
    declare donc valide un bulletin qui porte 1000 voix pour un candidat et
    -999 pour un autre. La contre-preuve est executee plus bas.

    Le contrôle porte donc sur chaque composante : elle doit dechiffrer en 0
    ou en 1, leur somme doit valoir 1, et le dechiffre doit correspondre au
    bulletin que l'électeur a annonce. C'est un contrôle d'AUTORITE (la cle
    privee est necessaire) : la vérification publique d'un bulletin exige une
    preuve a divulgation nulle, presentee en section 6.

    `plaintext_vote` et `num_candidates` sont enfin lus : dans la version
    initiale, ils etaient recus puis ignores.
    """
    if len(encrypted_ballot) != num_candidates:
        raise ValueError("Le bulletin doit avoir une composante par candidat")
    if len(plaintext_vote) != num_candidates:
        raise ValueError("Le bulletin doit avoir une composante par candidat")
    if sum(plaintext_vote) != 1 or any(v not in (0, 1) for v in plaintext_vote):
        raise ValueError("Le bulletin annonce doit etre un vecteur one-hot")

    # La somme chiffree des composantes (addition homomorphique)
    sum_encrypted = encrypted_ballot[0]
    for e in encrypted_ballot[1:]:
        sum_encrypted = sum_encrypted + e

    # Le hash du bulletin, pour identification (et non comme commitment)
    ballot_hash = compute_ballot_hash(encrypted_ballot)

    return {
        "ballot_hash": ballot_hash,
        "sum_encrypted": sum_encrypted,
        "num_candidates": num_candidates,
        "components": encrypted_ballot,
        "declared_vote": plaintext_vote,
    }

def verify_ballot_proof(private_key, proof):
    """Verifier, avec la cle privee, qu'un bulletin est un vote unique.

    Trois conditions, toutes necessaires :
    1. chaque composante dechiffre en 0 ou en 1 ;
    2. leur somme vaut 1 (un seul candidat choisi) ;
    3. le dechiffre correspond au bulletin annonce par l'électeur.
    """
    components = proof["components"]
    if len(components) != proof["num_candidates"]:
        return False

    plaintexts = [private_key.decrypt(c) for c in components]
    if any(v not in (0, 1) for v in plaintexts):
        return False
    if plaintexts != proof["declared_vote"]:
        return False
    return sum(plaintexts) == 1


# Generer et verifier les preuves
print("PREUVES DE VALIDITE DES BULLETINS")
print("=" * 60)

all_valid = True
proofs = []
for i, (voter, ballot) in enumerate(zip(voters, encrypted_ballots)):
    # Le bulletin one-hot que l'électeur a annonce
    declared = [0] * len(candidates)
    declared[voter["choice"]] = 1
    proof = prove_valid_ballot(public_key, ballot, declared, len(candidates))
    valid = verify_ballot_proof(private_key, proof)
    proofs.append(proof)
    status = "VALIDE" if valid else "INVALIDE"
    print(f"  {voter['name']} : {status} (one-hot, ballot_hash={proof['ballot_hash'][:16]}...)")
    if not valid:
        all_valid = False

print(f"\n-> Tous les bulletins valides : {all_valid}")

# Contre-preuve : le bulletin [1000, -999, 0] a une somme de 1.
# Un contrôle de la somme seule le declarerait VALIDE ; le contrôle one-hot
# le rejette, car ses composantes ne valent pas 0 ou 1.
print("\nCONTRE-PREUVE : bulletin a somme 1 mais non one-hot")
forge_ballot = [
    public_key.encrypt(1000),
    public_key.encrypt(-999),
    public_key.encrypt(0),
]
forge_sum = forge_ballot[0]
for e in forge_ballot[1:]:
    forge_sum = forge_sum + e
print(f"  Somme dechiffree du bulletin forge : {private_key.decrypt(forge_sum)}")
forge_valid = None
try:
    forge_proof = prove_valid_ballot(public_key, forge_ballot, [1, 0, 0], len(candidates))
    forge_valid = verify_ballot_proof(private_key, forge_proof)
except ValueError as exc:
    print(f"  Bulletin forge rejete a la production de la preuve : {exc}")
if forge_valid is not None:
    print(f"  Bulletin forge : {'VALIDE' if forge_valid else 'REJETE'}")
print("  -> La somme vaut 1, mais ce bulletin n'est pas un vote.")
print("  -> contrôler la somme seule ne prouve rien : il faut contrôler")
print("     que CHAQUE composante vaut 0 ou 1.")

# démonstration : un bulletin qui en contient deux est rejete.
# Le fraudeur ANNONCE un vote unique mais SOUMET deux votes : c'est l'ecart
# entre l'annonce et le contenu chiffre que le contrôle detecte.
print("\nDEMONSTRATION : bulletin avec deux votes")
fake_ballot = [public_key.encrypt(1), public_key.encrypt(1), public_key.encrypt(0)]
fake_proof = prove_valid_ballot(public_key, fake_ballot, [1, 0, 0], len(candidates))
fake_valid = verify_ballot_proof(private_key, fake_proof)
print(f"  Bulletin avec 2 votes : {'VALIDE' if fake_valid else 'REJETE'}")
print("  -> Un electeur ne peut pas voter deux fois dans un bulletin")
````

### Interpretation : ZKP simplifiee vs preuves non-interactives

Dans notre version simplifiee, l'autorite verifie les preuves avec sa cle privee.
Dans un vrai système (Helios, ElectionGuard), les preuves sont des **ZKP non-interactives**
(Chaum-Pedersen / Sigma protocols) verifiables par **n'importe qui** sans cle privee.

C'est la différence entre notre demo pedagogique et un système production-ready.

## 5. Bulletin board public

Le **bulletin board** est un registre public ou tous les bulletins chiffres
et leurs preuves sont publies. N'importe qui peut :
- Verifier que son bulletin est present
- Verifier les preuves de validite de tous les bulletins
- Recompter le total homomorphique

````python
# Construction du bulletin board
if PHE_AVAILABLE:
    bulletin_board = []
    for i, (voter, ballot, proof) in enumerate(zip(voters, encrypted_ballots, proofs)):
        entry = {
            "voter_id": hashlib.sha256(voter["name"].encode()).hexdigest()[:16],
            "ballot_hash": proof["ballot_hash"],
            # Artefacts de vérification PUBLIES : un tiers recalcule la somme
            # a partir des ciphertexts et la compare a la somme publiee.
            # (Le booléen "proof_valid" de la version initiale etait un
            # constat ecrit par l'autorite : il ne prouvait rien a un tiers.)
            "proof_sum_encrypted": proof["sum_encrypted"],
            "encrypted_ballot": ballot,  # les ciphertexts
        }
        bulletin_board.append(entry)

    print("BULLETIN BOARD PUBLIC")
    print("=" * 60)
    print("  Voter ID           | Ballot Hash      | Somme publiee")
    print("-" * 60)
    for entry in bulletin_board:
        vid = entry["voter_id"]
        bh = entry["ballot_hash"][:16]
        sum_ct = entry["proof_sum_encrypted"].ciphertext(be_secure=False)
        print(f"  {vid:16s} | {bh}... | {str(sum_ct)[:12]}...")

    print(f"\nTotal : {len(bulletin_board)} bulletins publies")
    print("-> Chaque entree publie ses ciphertexts ET la somme chiffree :")
    print("   un tiers recalcule la somme et vérifie la coherence, sans cle privee.")

    # Contre-preuve 1 : le `voter_id` publie se defait par dictionary.
    # Le nom de l'électeur est dans un espace enumerable : SHA-256 ne le
    # protege pas, il le masque seulement.
    print("\nCONTRE-PREUVE : le voter_id public se inverse par dictionary")
    for voter, entry in zip(voters, bulletin_board):
        recomputed = hashlib.sha256(voter["name"].encode()).hexdigest()[:16]
        print(f"  {voter['name']} -> {recomputed} (publie : {entry['voter_id']})")
    print("  -> Le board public associe donc chaque identite a son bulletin.")

    # Contre-preuve 2 : un bulletin falsifie est detectable par un tiers.
    # L'addition homomorphique de Paillier est le produit des ciphertexts
    # modulo n^2 : si un tiers modifie un ciphertext, la somme publiee ne
    # correspond plus.
    print("\nCONTRE-PREUVE : un bulletin falsifie est detecte sans cle privee")
    honest = bulletin_board[0]
    product = 1
    for c in honest["encrypted_ballot"]:
        product = (product * c.ciphertext(be_secure=False)) % public_key.nsquare
    published = honest["proof_sum_encrypted"].ciphertext(be_secure=False)
    print(f"  Bulletin honnete : produit des composantes == somme publiee : "
          f"{product == published}")
    forged_ballot = [public_key.encrypt(9)] + list(honest["encrypted_ballot"][1:])
    forged_product = 1
    for c in forged_ballot:
        forged_product = (forged_product * c.ciphertext(be_secure=False)) % public_key.nsquare
    print(f"  Bulletin falsifie : produit des composantes == somme publiee : "
          f"{forged_product == published}")
    print("  -> Le contrôle public rejette la falsification, sans cle privee.")
else:
    print("Section sautee : phe non disponible")
````

### Interpretation : Bulletin board

Le bulletin board est le **registre public** de l'election :

| Information publiee | Forme | Ce qu'un tiers peut en faire |
|---------------------|-------|-------------------------------|
| Voter ID | SHA-256 du nom, tronque a 64 bits, **non sale** | **Reconstitue le nom** : l'espace des noms est enumerable, le hachage est reversible par dictionary |
| Ballot hash | SHA-256 des ciphertexts | L'electeur retrouve son bulletin |
| Preuve de validite | contrôle one-hot par composante, cote autorite | L'autorite certifie ; un tiers ne peut pas rejouer ce contrôle sans la cle |
| Ciphertexts | 3 valeurs Paillier par bulletin | Quiconque peut recompter |

**Le `voter_id` n'est pas une pseudonymisation.** C'est un SHA-256 non sale
tronque a 16 caractères hexadecimaux (64 bits) d'un nom issu d'un espace
enumerable : le hachage se defait par dictionary, comme le montre la
démonstration de la cellule qui construit le board. Comme le même board
public associe chaque `voter_id` a un `ballot_hash`, il etablit publiquement
le lien **électeur <-> bulletin depose** -- exactement le lien que la
propriete d'anonymat pretend masquer. Un bulletin board qui ne publie pas
d'identite d'électeur rompt ce lien.

**Point cle** : le bulletin board est **append-only** (on ajoute, on ne retire jamais). Sur une blockchain, c'est garanti par l'immutabilite de la chaîne. Sur un serveur web, c'est garanti par des signatures numériques et un log public.


Vérification individuelle permettant à chaque électeur de confirmer que son bulletin est bien enregistré sur le tableau d'affichage public.

````python
# Verification individuelle : un electeur verifie que son bulletin est present
if PHE_AVAILABLE:
    print("VERIFICATION INDIVIDUELLE")
    print("=" * 60)

    # L'electeur 3 verifie
    my_voter = voters[2]
    my_ballot = encrypted_ballots[2]
    my_hash = compute_ballot_hash(my_ballot)

    # Chercher dans le bulletin board
    found = False
    for entry in bulletin_board:
        if entry["ballot_hash"] == my_hash:
            found = True
            break

    print(f"Electeur : {my_voter['name']}")
    print(f"Mon ballot hash : {my_hash[:32]}...")
    print(f"Present dans le bulletin board : {'OUI' if found else 'NON'}")
    print()
    print("-> L'electeur peut verifier que son vote a bien ete enregistre")

    # Contre-preuve : l'électeur PEUT prouver son vote a un tiers.
    # Paillier utilise g = n + 1 : si l'électeur choisit lui-même son alea r
    # (l'API `encrypt(..., r_value=...)` le permet), un tiers retrouve le
    # dechiffre m par c * r^-n = 1 + m*n (mod n^2).
    print()
    print("CONTRE-PREUVE : l'électeur peut-il prouver son vote ?")
    my_r = public_key.get_random_lt_n()
    my_ciphertext = public_key.encrypt(1, r_value=my_r)
    raw = my_ciphertext.ciphertext(be_secure=False)
    n = public_key.n
    nsquare = n * n
    stripped = (raw * pow(my_r, -n, nsquare)) % nsquare
    recovered = ((stripped - 1) // n) % n
    print(f"  Un tiers retrouve le dechiffre a partir de l'alea publie : {recovered}")
    print("  -> Le carnet ne resiste donc PAS a la coerction.")
    print("  -> C'est pourquoi aucun alea n'est publie dans le board : le publier")
    print("     transformerait la verifiabilite individuelle en preuve de vote.")
    print("  -> Un système resistant a la coerction exige d'autres constructions :")
    print("     c'est le travail de Juels, Catalano et Jakobsson (2005).")
else:
    print("Section sautee : phe non disponible")
````

### Interpretation : les proprietes du bulletin board

Le bulletin board fournit la **transparence** :

| Propriete | mécanisme | état mesure dans ce carnet |
|-----------|-----------|------------------------------|
| Verifiabilite individuelle | L'électeur conserve le hash de son bulletin | Tient : la vérification individuelle retrouve le bulletin dans le board |
| Verifiabilite universelle | Quiconque peut reverifier le decompte homomorphique | **Partielle** : les ciphertexts sont publics et recomptables, mais le total exige la cle privee |
| Anonymat | Le chiffrement est randomise | **rompu** : le `voter_id` public est le nom en clair (cf. le tableau du board) |
| Resistance a la coerition | L'électeur ne peut pas prouver son vote a un tiers | **Non** : le contre-temoin de la vérification individuelle le montre |

**Resistance a la coerition : absente ici.** Paillier utilise `g = n + 1` :
si l'électeur publie son alea `r`, un tiers retrouve le dechiffre `m` par
`c * r^-n = 1 + m*n (mod n^2)`. La cellule de vérification individuelle
execute ce calcul : un tiers sait alors pour qui l'électeur a vote. C'est
pourquoi le board ne publie aucun alea -- et pourquoi la preuve de validite
reste un contrôle d'autorite plutot qu'une preuve a divulgation nulle.

Helios (2008) n'est pas davantage coerce-resistant : il est concu pour des
elections a **faible coerition** (Cortier, Gaudry, Yang, ESORICS 2014).
C'est précisément cette faiblesse qui a motive le schema de Juels, Catalano
et Jakobsson (2005), reference de la littérature sur ce point.

## 6. ElectionGuard - Le SOTA

**ElectionGuard** (Microsoft, 2019) est le système de vote E2E verifiable
le plus avance, disponible en open-source avec un SDK Python.

### Différences avec notre construction a la main

| Aspect | Notre demo | ElectionGuard |
|--------|-----------|---------------|
| Chiffrement | Paillier (additif) | ElGamal exponentiel (additif + ZKP) |
| ZKP bulletins | Verification avec cle privee | Chaum-Pedersen non-interactive |
| Dechiffrement | Cle privee unique | Threshold (k-of-n gardiens) |
| Bulletin board | Liste Python | Artefacts JSON standardises |
| Audit | Manuel | Outils de verification automatises |

### Architecture ElectionGuard

```
1. Configuration : Manifest (candidats, règles) + Gardiens (threshold keys)
2. Chiffrement   : Chaque bulletin chiffre avec ElGamal + ZKP Chaum-Pedersen
3. Publication    : Bulletin board avec tous les ciphertexts + preuves
4. Decompte      : Addition homomorphique des ciphertexts
5. Dechiffrement : k-of-n gardiens revelent le total (threshold decryption)
6. Verification  : Quiconque peut reverifier avec les artefacts publics
```

### Threshold Decryption

La cle privee est **repartie entre N gardiens** (ex: 5 commissaires electoraux).
Il faut au moins **k gardiens** (ex: 3) pour dechiffrer le résultat.
Aucun gardien seul ne peut dechiffrer quoi que ce soit.

## 7. Exemple guide : Vote preferentiel verifiable

*Solution proposee par Clovis Lefebvre, Evariste Balvay.*

### Principe du vote preferentiel

Au lieu de voter pour un seul candidat, chaque electeur **classe** les candidats
par ordre de préférence. Si aucun candidat n'a la majorite absolue au premier tour,
le candidat avec le moins de votes de premier choix est elimine, et ses votes
sont redistribues selon les seconds choix, etc.

### Representation cryptographique

Le bulletin preferentiel est encode comme une **matrice de permutation** N x N
ou `entry[i][j] = 1` si le candidat i est classe en position j.
Chaque ligne a exactement un 1 (chaque candidat a un rang) et chaque colonne
aussi (chaque rang est attribue a un seul candidat).

La classe `RankedBallot` ci-dessous construit cette matrice et chiffre
chaque élément avec Paillier.

````python
# Exemple guide : Vote preferentiel chiffre

try:
    from phe import paillier as _paillier_check
    PHE_AVAILABLE_EX = True
except ImportError:
    PHE_AVAILABLE_EX = False

import hashlib

if PHE_AVAILABLE_EX:
    from phe import paillier

    # Configuration
    candidates_ex = ["Alice", "Bob", "Charlie"]
    pk, sk = paillier.generate_paillier_keypair(n_length=1024)

    class RankedBallot:
        """Bulletin de vote preferentiel chiffre.

        Representation : matrice N x N ou entry[i][j] = 1 si le candidat i
        est classe en position j (0-indexed), 0 sinon.
        Chaque ligne a exactement un 1, chaque colonne aussi (permutation).
        """

        def __init__(self, public_key, ranking):
            """Creer un bulletin chiffre a partir d'un classement.

            Args:
                public_key: cle publique Paillier
                ranking: liste d'indices de candidats par preference
                         ex: [2, 0, 1] = Charlie 1er, Alice 2eme, Bob 3eme
            """
            self.n_candidates = len(ranking)
            # Construire la matrice de permutation
            matrix = [[0] * self.n_candidates for _ in range(self.n_candidates)]
            for rank, candidate_idx in enumerate(ranking):
                matrix[candidate_idx][rank] = 1
            self.encrypted_matrix = [[public_key.encrypt(matrix[i][j])
                                      for j in range(self.n_candidates)]
                                     for i in range(self.n_candidates)]

    # --- Validation ---
    ranking = [2, 0, 1]  # Charlie 1er, Alice 2eme, Bob 3eme
    ballot = RankedBallot(pk, ranking)

    # Verifier la matrice dechiffree
    print("VALIDATION DU BULLETIN PREFERENTIEL")
    print("=" * 60)
    print(f"Classement : {ranking}")
    print(f"  = {candidates_ex[ranking[0]]} 1er, {candidates_ex[ranking[1]]} 2e, {candidates_ex[ranking[2]]} 3e")
    print()

    # Dechiffrer et afficher la matrice
    print("Matrice de permutation dechiffree :")
    print("        Rang1  Rang2  Rang3")
    for i in range(ballot.n_candidates):
        row = [sk.decrypt(ballot.encrypted_matrix[i][j]) for j in range(ballot.n_candidates)]
        print(f"  {candidates_ex[i]:8s}  {row[0]}      {row[1]}      {row[2]}")

    # Verifier les contraintes de permutation
    valid = True
    for i in range(ballot.n_candidates):
        row_sum = sum(sk.decrypt(ballot.encrypted_matrix[i][j]) for j in range(ballot.n_candidates))
        if row_sum != 1:
            valid = False
    for j in range(ballot.n_candidates):
        col_sum = sum(sk.decrypt(ballot.encrypted_matrix[i][j]) for i in range(ballot.n_candidates))
        if col_sum != 1:
            valid = False

    print(f"\nMatrice de permutation valide (sommes lignes/colonnes = 1) : {valid}")
else:
    print("Exemple guide non executable (phe non installe)")
````

### Interpretation de l'exemple guide

| Élément | Description |
|---------|-------------|
| **Matrice N x N** | Represente le classement complet de chaque candidat |
| **Ligne i** | Position du candidat i (exactement un 1 par ligne) |
| **Colonne j** | Quel candidat est en position j (exactement un 1 par colonne) |
| **Chiffrement Paillier** | Chaque élément est chiffre independamment |

**Validation** : la matrice dechiffree verifie les contraintes de permutation (somme de chaque ligne et chaque colonne = 1). En production, cette verification se fait via des preuves ZKP (Chaum-Pedersen) sans dechiffrer.
