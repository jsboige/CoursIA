# Sonde de rendu audio (#11179)

Trois sorties portant **le meme** WAV, publiees sous trois bundles MIME differents.
La question mesuree : lesquelles donnent un lecteur audio jouable chez GitHub, nbviewer et JupyterLab.


```python
import base64, io, math, wave
from IPython.display import display

# Sinus 440 Hz, 0.4 s, mono 8 kHz -- assez court pour que la sonde reste legere.
RATE, DUREE = 8000, 0.4
frames = b"".join(
    int(12000 * math.sin(2 * math.pi * 440 * n / RATE)).to_bytes(2, "little", signed=True)
    for n in range(int(RATE * DUREE))
)
buf = io.BytesIO()
with wave.open(buf, "wb") as w:
    w.setnchannels(1); w.setsampwidth(2); w.setframerate(RATE); w.writeframes(frames)
WAV = buf.getvalue()
B64 = base64.b64encode(WAV).decode("ascii")
HTML = (
    '<audio controls="controls">'
    '<source src="data:audio/wav;base64,' + B64 + '" type="audio/wav" />'
    "Your browser does not support the audio element.</audio>"
)
print("WAV", len(WAV), "octets | base64", len(B64), "caracteres")
```

## Variante A — audio/wav + text/plain (SANS text/html)


```python
display({"audio/wav": B64, "text/plain": "<audio A: audio/wav seul>"}, raw=True)
```

## Variante B — text/html data-URI + text/plain (SANS audio/*)


```python
display({"text/html": HTML, "text/plain": "<audio B: text/html seul>"}, raw=True)
```

## Variante C — audio/wav + text/html + text/plain (helper actuel, octets en double)


```python
display({"audio/wav": B64, "text/html": HTML, "text/plain": "<audio C: les deux>"}, raw=True)
```
