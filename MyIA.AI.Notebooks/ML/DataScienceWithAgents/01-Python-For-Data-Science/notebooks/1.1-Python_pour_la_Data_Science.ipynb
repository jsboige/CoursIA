## 1. Types de base : ce que Python manipule

Une valeur en Python a toujours un **type**. Le connaître n'est pas académique : `3` et `3.0` ne se comportent pas identiquement, et un nombre stocké comme texte (`"3.5"`) ne se calcule pas. En data science, une colonne entière peut être piégée par ce dernier cas — une virgule décimale française dans un CSV et l'import produit du texte là où vous attendiez des mesures.


```python
# Les quatre types que nous croiserons en boucle
un_entier = 42          # int   : comptages, identifiants, quantites
un_reel = 1.75          # float : mesures, moyennes, tout ce qui vient d'un calcul
un_booleen = True       # bool  : le resultat d'une comparaison
un_texte = "Adelie"     # str   : categories, etiquettes, identifiants textuels

for valeur in [un_entier, un_reel, un_booleen, un_texte]:
    print(type(valeur).__name__, "->", valeur)
```

Deux pièges reviennent sans cesse en data science. Le premier : **la division réelle**. En Python, `/` renvoie toujours un `float`, même sur des entiers — c'est le comportement que l'on veut pour une moyenne. Le second : **la représentation binaire des flottants**. `0.1 + 0.2` ne fait pas exactement `0.3` ; l'ordinateur approche. Pour comparer, on tolère un écart (`abs(a - b) < 1e-9`), on n'exige jamais l'égalité stricte de deux flottants issus de calculs.


```python
division = 7 / 2          # float, meme si la reponse "tombe juste"
division_entiere = 7 // 2  # quotient entier (nous n'en aurons presque jamais besoin)

somme_flottante = 0.1 + 0.2
print("division :", division, type(division).__name__)
print("division entiere :", division_entiere)
print("0.1 + 0.2 =", somme_flottante)
print("egalite stricte avec 0.3 ?", somme_flottante == 0.3)
print("comparaison tolérante   ?", abs(somme_flottante - 0.3) < 1e-9)
```

```python
# Convertir : str -> float est LE geste d'import de donnees
prix_saisi = "12.50"          # ce que lit un fichier texte
# prix_saisi * 2 serait "12.5012.50" (concatenation de texte !)
prix = float(prix_saisi)
print(prix * 2, type(prix).__name__)

entier_depuis_reel = int(7.9)   # tronque, n'arrondit PAS
texte_depuis_reel = f"{7.2531:.2f}"   # arrondi a l'affichage, pour les sorties propres
print("int(7.9) =", entier_depuis_reel, "| affichage a 2 chiffres :", texte_depuis_reel)
```

## 2. Structures : listes, tuples, dictionnaires, ensembles

Le choix d'une structure n'est pas cosmétique — il encode ce que la donnée *est*. Une **liste** est une séquence ordonnée que l'on fait grandir (des mesures dans l'ordre du prélèvement). Un **tuple** est une séquence figée (une ligne de coordonnées). Un **dictionnaire** associe des clés à des valeurs (des ventes par jour). Un **ensemble** déduplique (les espèces distinctes d'un échantillon). Pandas en 1.3 construit ses tables à partir de dictionnaires de listes — la colonne est la clé, la liste est la colonne.


```python
# Liste : ordonnee, modifiable, indexee a partir de 0
releves = [12.4, 11.9, 13.2, 12.0]
releves.append(12.8)              # ajout en fin
print("releves  :", releves)
print("premier  :", releves[0], "| dernier :", releves[-1])
print("tranche 2-4 :", releves[2:4])   # bornes : inclus / exclue
print("longueur :", len(releves), "| 12.0 present ?", 12.0 in releves)
```

```python
# Tuple : sequence figee — pratique pour une ligne (x, y) ou une paire (min, max)
point = (3.5, -0.2)
x, y = point            # deballage direct
bornes = (min(releves), max(releves))
print("point :", point, "| x =", x, "| y =", y)
print("bornes (min, max) :", bornes)
```

```python
# Dictionnaire : cle -> valeur. La cle est un nom lisible, la valeur est la donnee.
ventes = {"lundi": 3, "mardi": 5, "mercredi": 2, "jeudi": 5, "vendredi": 8}
print("ventes de lundi :", ventes["lundi"])
print("samedi (absent, valeur par defaut) :", ventes.get("samedi", 0))
print("total de la semaine :", sum(ventes.values()))

# iterer proprement : cle ET valeur ensemble
for jour, nombre in ventes.items():
    print(f"{jour} : {nombre} vente(s)")
```

```python
# Ensemble : deduplication en un geste — les valeurs uniques d'un echantillon
especes = ["Adelie", "Gentoo", "Adelie", "Chinstrap", "Adelie", "Gentoo"]
uniques = set(especes)
print("liste brute :", especes)
print("especes distinctes :", uniques, "| nombre :", len(uniques))
```

**Comment choisir.** Liste pour une série homogène qui grandit ; tuple pour une ligne figée ; dictionnaire dès qu'un *nom* porte la valeur (une colonne nommée) ; ensemble pour répondre « combien de valeurs distinctes ? ». Ces quatre réflexes couvrent l'essentiel de ce que font les notebooks suivants.


## 3. Fonctions : écrire une fois, réutiliser

Une fonction empaquette un calcul sous un nom. En data science, elle sert moins à la théorie des langages qu'à la **discipline** : un calcul nommé est un calcul vérifiable, réutilisable, et dont on peut changer l'implémentation sans reprendre dix cellules. La **docstring** — la chaîne de documentation placée juste sous la ligne `def`, entre triples guillemets — dit *ce que fait* la fonction ; le `return` dit *ce qu'elle rend* — sans `return`, une fonction rend `None`.


```python
def amplitude(valeurs):
    """Ecart entre la plus grande et la plus petite valeur d'une liste."""
    return max(valeurs) - min(valeurs)

def rapport(valeurs, seuil=0.0):
    """Proportion de valeurs strictement au-dessus du seuil (parametre par defaut)."""
    au_dessus = sum(1 for v in valeurs if v > seuil)
    return au_dessus / len(valeurs)

print("amplitude des releves :", amplitude(releves))
print("part > 12.1          :", rapport(releves, seuil=12.1))
print("sans return explicite, une fonction rend :", print("bonjour"))
```

La dernière ligne mérite une lecture attentive : `print` affiche « bonjour » mais rend `None` — afficher et rendre sont deux choses distinctes. Une cellule de notebook affiche sa dernière expression ; une fonction qui doit *fournir* un résultat le `return`, elle ne se contente pas de l'imprimer. Nous croiserons cette distinction dès 1.2 (`np.sum` rend une valeur, elle ne l'affiche pas).


## 4. Compréhensions de liste

La syntaxe `[expression for element in source]` construit une liste en une ligne. C'est la construction la plus dense du Python de data science — elle apparaît dans presque chaque notebook de la formation. Le réflexe de lecture : **la boucle se lit de droite à gauche**, comme son équivalent explicite.


```python
temperatures_f = [98.6, 99.1, 97.9, 100.4]

# 1. Boucle classique que vous ecririez "a la main"
temperatures_c = []
for t in temperatures_f:
    temperatures_c.append((t - 32) * 5 / 9)

# 2. La comprehension equivalente — meme resultat, une ligne
temperatures_c_comp = [(t - 32) * 5 / 9 for t in temperatures_f]

print("boucle    :", [round(t, 1) for t in temperatures_c])
print("comprehension :", [round(t, 1) for t in temperatures_c_comp])

# 3. Avec un filtre : garder les temperatures febriles (> 38 C)
febriles = [round(t, 1) for t in temperatures_c if t > 38.0]
print("febriles :", febriles)
```

La version avec filtre se lit : « la liste des `t` arrondis, **pour chaque** `t` dans les températures Celsius, **telles que** `t > 38` ». Quand la compréhension dépasse un niveau de filtre ou devient difficile à lire d'un coup d'œil, revenez à la boucle explicite — la concision n'est jamais un but en soi.


## 5. Lire et écrire des fichiers

Les données vivent dans des fichiers. Le gestionnaire de contexte `with` ouvre le fichier, l'utilise, et le referme **même si une erreur survient** — on ne l'écrit jamais autrement. Après quoi un CSV n'est qu'un fichier texte où chaque ligne porte des champs séparés par une virgule ; l'analyser à la main une fois, ici, c'est comprendre ce que `pd.read_csv` fera pour vous en 1.3.


```python
chemin = "ventes_mini.csv"

# Ecriture : mode "w" (ecriture), encodage explicite — jamais d'encoding implicite
with open(chemin, "w", encoding="utf-8") as f:
    f.write("jour,ventes\n")
    for jour, nombre in ventes.items():
        f.write(f"{jour},{nombre}\n")

print("fichier ecrit :", chemin)
```

```python
# Lecture : mode "r" (lecture). readlines() rend les lignes, fins de ligne comprises.
with open(chemin, "r", encoding="utf-8") as f:
    lignes = f.readlines()

print("lignes brutes :", lignes)

en_tete = lignes[0].strip()
donnees = []
for ligne in lignes[1:]:
    jour, nombre = ligne.strip().split(",")
    donnees.append({"jour": jour, "ventes": int(nombre)})

print("en-tete :", en_tete)
print("premiere ligne parsee :", donnees[0])
print("meilleur jour :", max(donnees, key=lambda d: d["ventes"]))
```

Trois observations qui resserviront. `encoding="utf-8"` est explicite : sous Windows, l'encodage implicite dépend de la machine, et un accent mal encodé casse silencieusement un import. `strip()` retire la fin de ligne avant `split(",")` — le piège classique du « 8\n » qui ne se convertit pas en entier. Et le dictionnaire par ligne est exactement la forme que Pandas généralisera : `read_csv` produit une table où chaque colonne a un nom d'en-tête et chaque ligne une valeur. En 1.5, nous traiterons un vrai CSV de terrain, avec ses cases vides et ses catégories bavardes.

### Déchiffrer un traceback

Un traceback se lit **de bas en haut** : la dernière ligne nomme l'erreur (`ValueError: could not convert string to float: 'abc'`), la ligne juste au-dessus montre l'appel fautif, et les lignes encore au-dessus retracent le chemin. Devant un notebook qui échoue, ce sont ces deux dernières lignes qui disent *quoi* et *où* — le reste est le contexte. Le bloc `try / except` permet d'attraper une erreur attendue pour continuer (un fichier absent, une ligne malformée) ; une erreur inattendue, lui, ne se masque jamais.


```python
# Convertir une valeur peut echouer : la donnee reelle contient du texte inattendu.
# try/except attrape l'erreur ATTENDUE et poursuit avec une valeur explicite.

def pour_float(texte):
    """Convertit en float ; rend None (plutot qu'echouer) si le texte n'est pas un nombre."""
    try:
        return float(texte)
    except ValueError:
        return None

saisies = ["12.5", "7", "n/a", "3.2", ""]
converties = [pour_float(s) for s in saisies]
valides = [v for v in converties if v is not None]
print("saisies    :", saisies)
print("converties :", converties)
print("valides    :", valides)
```
