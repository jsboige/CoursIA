## 1. Types de base : ce que Python manipule

Une valeur en Python a toujours un **type**. Le connaître n'est pas académique : `3` et `3.0` ne se comportent pas identiquement, et un nombre stocké comme texte (`"3.5"`) ne se calcule pas. En data science, une colonne entière peut être piégée par ce dernier cas — une virgule décimale française dans un CSV et l'import produit du texte là où vous attendiez des mesures.


```python
# Les quatre types que nous croiserons en boucle
un_entier = 42          # int   : comptages, identifiants, quantites
un_reel = 1.75          # float : mesures, moyennes, tout ce qui vient d'un calcul
un_booleen = True       # bool  : le resultat d'une comparaison
un_texte = "Adelie"     # str   : categories, etiquettes, identifiants textuels

for valeur in [un_entier, un_reel, un_booleen, un_texte]:
    print(type(valeur).__name__, "->", valeur)
```

### Lecture — quatre types, quatre usages

La sortie aligne type et valeur sur quatre lignes : `int -> 42`, `float -> 1.75`, `bool -> True`, `str -> Adelie`. La colonne de gauche n'est pas décorative — elle dit ce que la valeur **est**, donc ce qu'on peut en faire : `42` se compte, `1.75` se moyenne, `True` se teste, `Adelie` se compare. Le commentaire du code donnait la correspondance métier (comptages, mesures, résultats de comparaison, étiquettes) ; la sortie la rend exécutable. Réflexe d'import de données : avant tout calcul, demandez le `type` — un `str` là où vous attendiez un `float` est l'erreur de degré n°1 (la virgule française du §1), et elle ne se voit **jamais** dans la valeur affichée, seulement dans le type.

Deux pièges reviennent sans cesse en data science. Le premier : **la division réelle**. En Python, `/` renvoie toujours un `float`, même sur des entiers — c'est le comportement que l'on veut pour une moyenne. Le second : **la représentation binaire des flottants**. `0.1 + 0.2` ne fait pas exactement `0.3` ; l'ordinateur approche. Pour comparer, on tolère un écart (`abs(a - b) < 1e-9`), on n'exige jamais l'égalité stricte de deux flottants issus de calculs.


```python
division = 7 / 2          # float, meme si la reponse "tombe juste"
division_entiere = 7 // 2  # quotient entier (nous n'en aurons presque jamais besoin)

somme_flottante = 0.1 + 0.2
print("division :", division, type(division).__name__)
print("division entiere :", division_entiere)
print("0.1 + 0.2 =", somme_flottante)
print("egalite stricte avec 0.3 ?", somme_flottante == 0.3)
print("comparaison tolérante   ?", abs(somme_flottante - 0.3) < 1e-9)
```

### Lecture — la sortie dit le piège

Quatre lignes, deux leçons. Ligne 1 : `division : 3.5 float` — même sur des entiers, Python rend un `float` ; c'est le comportement voulu pour une moyenne. Lignes 3 à 5 : `0.1 + 0.2 = 0.30000000000000004`, puis `False` à l'égalité stricte et `True` à la comparaison tolérante. Le `...0004` n'est pas un bug mais la traduction décimale d'un binaire : `0.1` n'a pas d'écriture exacte en base 2, l'écart apparaît vers la 17e décimale et se propage dans toute somme. D'où la règle de la cellule : **jamais d'égalité stricte sur des flottants issus de calcul** — toujours `abs(a - b) < tolérance`. En 1.3, `np.isclose` industrialisera ce geste.

```python
# Convertir : str -> float est LE geste d'import de donnees
prix_saisi = "12.50"          # ce que lit un fichier texte
# prix_saisi * 2 serait "12.5012.50" (concatenation de texte !)
prix = float(prix_saisi)
print(prix * 2, type(prix).__name__)

entier_depuis_reel = int(7.9)   # tronque, n'arrondit PAS
texte_depuis_reel = f"{7.2531:.2f}"   # arrondi a l'affichage, pour les sorties propres
print("int(7.9) =", entier_depuis_reel, "| affichage a 2 chiffres :", texte_depuis_reel)
```

### Lecture — convertir n'est pas arrondir

Trois gestes distincts dans trois lignes. `float("12.50")` rend `12.5` : la conversion texte vers nombre, LE geste d'import — sans elle, `"12.50" * 2` serait la concaténation `"12.5012.50"`, comme le rappelle le commentaire. `int(7.9)` rend `7` : la conversion **tronque**, elle n'arrondit pas — `int(7.1)` rendrait aussi `7`, et `int(-7.9)` rendrait `-7` (vers zéro). Enfin `f"{7.2531:.2f}"` rend `'7.25'` : l'arrondi **à l'affichage**, qui ne touche pas la valeur stockée — `7.2531` reste `7.2531` en mémoire, seul le texte montré est formaté. Confondre les trois produit le rapport qui affiche 7.25 pendant que le calcul continue avec 7.2531 : arrondir tard (à l'affichage), convertir tôt (à l'import).

## 2. Structures : listes, tuples, dictionnaires, ensembles

Le choix d'une structure n'est pas cosmétique — il encode ce que la donnée *est*. Une **liste** est une séquence ordonnée que l'on fait grandir (des mesures dans l'ordre du prélèvement). Un **tuple** est une séquence figée (une ligne de coordonnées). Un **dictionnaire** associe des clés à des valeurs (des ventes par jour). Un **ensemble** déduplique (les espèces distinctes d'un échantillon). Pandas en 1.3 construit ses tables à partir de dictionnaires de listes — la colonne est la clé, la liste est la colonne.

Le fil de lecture des quatre cellules qui suivent : chaque structure est montrée sur une **donnée** du même monde (des relevés de température, un point, des ventes, des espèces), pas sur des exemples abstraits — et chaque sortie se lit comme la réponse à une question d'analyse : « quelle est la dernière mesure ? », « quel jour a-t-on vendu le plus ? », « combien d'espèces distinctes ? ».

```python
# Liste : ordonnee, modifiable, indexee a partir de 0
releves = [12.4, 11.9, 13.2, 12.0]
releves.append(12.8)              # ajout en fin
print("releves  :", releves)
print("premier  :", releves[0], "| dernier :", releves[-1])
print("tranche 2-4 :", releves[2:4])   # bornes : inclus / exclue
print("longueur :", len(releves), "| 12.0 present ?", 12.0 in releves)
```

### Lecture — une liste, cinq questions

La sortie montre les cinq gestes qui couvrent l'essentiel de l'usage d'une liste. `append` fait grandir (`12.8` arrive **en fin** : l'ordre d'insertion est préservé, crucial pour des mesures séquentielles). `releves[0]` et `releves[-1]` lisent les extrémités — l'index négatif part de la droite, `-1` désigne le dernier. La tranche `releves[2:4]` rend `[13.2, 12.0]` : **la borne de gauche est incluse, celle de droite exclue** — deux éléments pour les indices 2 et 3. Cette convention (celle de `range`) déroute d'abord puis devient un réflexe : `releves[:3]` = les trois premiers, `releves[-2:]` = les deux derniers. Enfin `in` teste l'appartenance (`True`) et `len` compte (`5`, après l'ajout).

```python
# Tuple : sequence figee — pratique pour une ligne (x, y) ou une paire (min, max)
point = (3.5, -0.2)
x, y = point            # deballage direct
bornes = (min(releves), max(releves))
print("point :", point, "| x =", x, "| y =", y)
print("bornes (min, max) :", bornes)
```

```python
# Dictionnaire : cle -> valeur. La cle est un nom lisible, la valeur est la donnee.
ventes = {"lundi": 3, "mardi": 5, "mercredi": 2, "jeudi": 5, "vendredi": 8}
print("ventes de lundi :", ventes["lundi"])
print("samedi (absent, valeur par defaut) :", ventes.get("samedi", 0))
print("total de la semaine :", sum(ventes.values()))

# iterer proprement : cle ET valeur ensemble
for jour, nombre in ventes.items():
    print(f"{jour} : {nombre} vente(s)")
```

### Lecture — le dictionnaire répond par son nom

La différence avec la liste saute aux sorties : on n'accède pas par position mais par **clé** — `ventes["lundi"]` rend `3`. Deux points de lecture. `ventes.get("samedi", 0)` rend `0` sans échouer : `.get` est l'accès à clé absente avec valeur par défaut, là où `ventes["samedi"]` lèverait une `KeyError` (le modèle du traceback du §6). Et l'itération par `.items()` rend les **couples** clé-valeur dans l'ordre d'insertion — le `f"{jour} : {nombre} vente(s)"` de la boucle prépare exactement les lignes que le §5 écrira dans le CSV : le dictionnaire est déjà la table, il ne manque que la virgule.

```python
# Ensemble : deduplication en un geste — les valeurs uniques d'un echantillon
especes = ["Adelie", "Gentoo", "Adelie", "Chinstrap", "Adelie", "Gentoo"]
uniques = set(especes)
print("liste brute :", especes)
print("especes distinctes :", uniques, "| nombre :", len(uniques))
```

### Lecture — dédupliquer, puis compter

Six entrées dans la liste brute, trois dans l'ensemble : `Adelie` apparaît trois fois, `Gentoo` deux, `Chinstrap` une — l'ensemble ne retient que les **présences**. Deux lectures fines. L'ordre d'affichage (`Gentoo`, puis `Adelie`, puis `Chinstrap`) n'est pas l'ordre d'insertion et n'est pas garanti : un ensemble n'est pas une séquence, il répond à « est-ce que X est dedans ? » et « combien ? », pas à « quel est le deuxième élément ? ». Et `len(uniques)` rendant `3` est la réponse à la question analytique (« combien d'espèces distinctes dans l'échantillon ? ») — le geste que 1.3 généralisera en `df["espece"].nunique()`.

**Comment choisir.** Liste pour une série homogène qui grandit ; tuple pour une ligne figée ; dictionnaire dès qu'un *nom* porte la valeur (une colonne nommée) ; ensemble pour répondre « combien de valeurs distinctes ? ». Ces quatre réflexes couvrent l'essentiel de ce que font les notebooks suivants.


## 3. Fonctions : écrire une fois, réutiliser

Une fonction empaquette un calcul sous un nom. En data science, elle sert moins à la théorie des langages qu'à la **discipline** : un calcul nommé est un calcul vérifiable, réutilisable, et dont on peut changer l'implémentation sans reprendre dix cellules. La **docstring** — la chaîne de documentation placée juste sous la ligne `def`, entre triples guillemets — dit *ce que fait* la fonction ; le `return` dit *ce qu'elle rend* — sans `return`, une fonction rend `None`.

Deux mots de vocabulaire avant les deux fonctions qui suivent. Le **paramètre par défaut** (`seuil=0.0`) rend l'argument optionnel : `rapport(releves)` s'écrit sans seuil. Et l'**expression génératrice** dans `sum(1 for v in valeurs if v > seuil)` est une compréhension sans crochets — elle produit les valeurs au fil de l'eau sans construire la liste intermédiaire ; le réflexe de lecture est celui de la section 4, en version économe.

```python
def amplitude(valeurs):
    """Ecart entre la plus grande et la plus petite valeur d'une liste."""
    return max(valeurs) - min(valeurs)

def rapport(valeurs, seuil=0.0):
    """Proportion de valeurs strictement au-dessus du seuil (parametre par defaut)."""
    au_dessus = sum(1 for v in valeurs if v > seuil)
    return au_dessus / len(valeurs)

print("amplitude des releves :", amplitude(releves))
print("part > 12.1          :", rapport(releves, seuil=12.1))
print("sans return explicite, une fonction rend :", print("bonjour"))
```

La dernière ligne mérite une lecture attentive : `print` affiche « bonjour » mais rend `None` — afficher et rendre sont deux choses distinctes. Une cellule de notebook affiche sa dernière expression ; une fonction qui doit *fournir* un résultat le `return`, elle ne se contente pas de l'imprimer. Nous croiserons cette distinction dès 1.2 (`np.sum` rend une valeur, elle ne l'affiche pas).


## 4. Compréhensions de liste

La syntaxe `[expression for element in source]` construit une liste en une ligne. C'est la construction la plus dense du Python de data science — elle revient dans la plupart des carnets Python de la formation. Les notebooks 1.2 et 1.3, eux, la remplacent par des opérations vectorisées NumPy/Pandas : même idée, une seule instruction pour toute la table. Le réflexe de lecture : **la boucle se lit de droite à gauche**, comme son équivalent explicite.

Pourquoi tant d'insistance sur une simple syntaxe : en 1.2 et 1.3, les opérations vectorisées (`x * 2`, `df["col"] > seuil`) sont des compréhensions déguisées — appliquées à toute la table d'un coup, en code compilé. Savoir lire « transforme chaque élément, garde ceux qui passent le filtre » ici, c'est savoir lire la moitié du code des notebooks suivants.

```python
temperatures_f = [98.6, 99.1, 97.9, 100.4]

# 1. Boucle classique que vous ecririez "a la main"
temperatures_c = []
for t in temperatures_f:
    temperatures_c.append((t - 32) * 5 / 9)

# 2. La comprehension equivalente — meme resultat, une ligne
temperatures_c_comp = [(t - 32) * 5 / 9 for t in temperatures_f]

print("boucle    :", [round(t, 1) for t in temperatures_c])
print("comprehension :", [round(t, 1) for t in temperatures_c_comp])

# 3. Avec un filtre : garder les temperatures febriles (> 38 C)
febriles = [round(t, 1) for t in temperatures_c if t > 38.0]
print("febriles :", febriles)
```

La version avec filtre se lit : « la liste des `t` arrondis, **pour chaque** `t` dans les températures Celsius, **telles que** `t > 38` ». Le résultat est **vide**, et il se lit finement : la quatrième température, `100.4 °F`, vaut **exactement `38.0 °C`** — et le filtre est **strict** (`>`), donc la frontière est exclue ; avec `>=`, la liste contiendrait `38.0`. Un filtre vide sur un exemple d'illustration est un piège de lecture : avant de conclure « j'ai mal compris », vérifiez où tombent les valeurs par rapport au seuil. Quand la compréhension dépasse un niveau de filtre ou devient difficile à lire d'un coup d'œil, revenez à la boucle explicite — la concision n'est jamais un but en soi.


## 5. Lire et écrire des fichiers

Les données vivent dans des fichiers. Le gestionnaire de contexte `with` ouvre le fichier, l'utilise, et le referme **même si une erreur survient** — on ne l'écrit jamais autrement. Après quoi un CSV n'est qu'un fichier texte où chaque ligne porte des champs séparés par une virgule ; l'analyser à la main une fois, ici, c'est comprendre ce que `pd.read_csv` fera pour vous en 1.3.

L'ordre des deux cellules est celui du travail réel : on **écrit** d'abord (produire un résultat que d'autres pourront relire), on **relit** ensuite (consommer ce qu'un autre — ou soi-même il y a un an — a écrit). Le round-trip complet, écrit-relu-reconstruit, est exactement le sujet de l'exercice 5.

```python
chemin = "ventes_mini.csv"

# Ecriture : mode "w" (ecriture), encodage explicite — jamais d'encoding implicite
with open(chemin, "w", encoding="utf-8") as f:
    f.write("jour,ventes\n")
    for jour, nombre in ventes.items():
        f.write(f"{jour},{nombre}\n")

print("fichier ecrit :", chemin)
```

### Lecture — ce que le mode "w" implique

Une ligne de sortie (`fichier ecrit : ventes_mini.csv`) pour un geste qui mérite un arrêt : en mode `"w"`, `open` **crée le fichier s'il n'existe pas, et l'écrase s'il existe** — le contenu précédent est perdu sans avertissement. C'est le mode voulu ici (on régénère le CSV depuis le dictionnaire), mais le réflexe de prudence est de connaître `"a"` (append : ajoute à la fin sans écraser) pour les journaux qu'on fait grandir. Le `encoding="utf-8"` explicite est le second arrêt : la lecture de la cellule suivante en dépend — un fichier écrit avec l'encodage implicite de la machine puis relu autrement peut rendre des accents illisibles, et l'erreur est silencieuse.

```python
# Lecture : mode "r" (lecture). readlines() rend les lignes, fins de ligne comprises.
with open(chemin, "r", encoding="utf-8") as f:
    lignes = f.readlines()

print("lignes brutes :", lignes)

en_tete = lignes[0].strip()
donnees = []
for ligne in lignes[1:]:
    jour, nombre = ligne.strip().split(",")
    donnees.append({"jour": jour, "ventes": int(nombre)})

print("en-tete :", en_tete)
print("premiere ligne parsee :", donnees[0])
print("meilleur jour :", max(donnees, key=lambda d: d["ventes"]))
```

Quatre observations qui resserviront. `encoding="utf-8"` est explicite : sous Windows, l'encodage implicite dépend de la machine, et un accent mal encodé casse silencieusement un import. `strip()` retire la fin de ligne avant `split(",")` — le piège classique du « 8\n » qui ne se convertit pas en entier. Enfin, `max(donnees, key=lambda d: d["ventes"])` introduit `lambda` : une fonction anonyme d'une ligne — `lambda d: d["ventes"]` se lit « la fonction qui à `d` associe `d["ventes"]` ». Passée à `key=`, elle dit à `max` sur quoi comparer les dictionnaires : leur champ `ventes`, pas leur contenu entier. L'indice de l'exercice 2 la réemploie telle quelle. Et le dictionnaire par ligne est exactement la forme que Pandas généralisera : `read_csv` produit une table où chaque colonne a un nom d'en-tête et chaque ligne une valeur. En 1.5, nous traiterons un vrai CSV de terrain, avec ses cases vides et ses catégories bavardes.

### Déchiffrer un traceback

Un traceback se lit **de bas en haut** : la dernière ligne nomme l'erreur (`ValueError: could not convert string to float: 'abc'`), la ligne juste au-dessus montre l'appel fautif, et les lignes encore au-dessus retracent le chemin. Devant un notebook qui échoue, ce sont ces deux dernières lignes qui disent *quoi* et *où* — le reste est le contexte. Le bloc `try / except` permet d'attraper une erreur attendue pour continuer (un fichier absent, une ligne malformée) ; une erreur inattendue, lui, ne se masque jamais.


```python
# Convertir une valeur peut echouer : la donnee reelle contient du texte inattendu.
# try/except attrape l'erreur ATTENDUE et poursuit avec une valeur explicite.

def pour_float(texte):
    """Convertit en float ; rend None (plutot qu'echouer) si le texte n'est pas un nombre."""
    try:
        return float(texte)
    except ValueError:
        return None

saisies = ["12.5", "7", "n/a", "3.2", ""]
converties = [pour_float(s) for s in saisies]
valides = [v for v in converties if v is not None]
print("saisies    :", saisies)
print("converties :", converties)
print("valides    :", valides)
```

### Lecture — None comme valeur explicite

La sortie raconte le nettoyage : cinq saisies, dont `"n/a"` et `""` (la chaîne vide) ne sont pas des nombres. `pour_float` rend `None` pour elles plutôt que d'échouer — et la ligne `converties` montre le résultat : `[12.5, 7.0, None, 3.2, None]`. La dernière ligne filtre avec `v is not None` (comparaison d'identité, la forme canonique pour `None`). Ce trio convertir-tolérer-filtrer est le premier geste de tout import réel ; 1.5 en fera tout un notebook, avec les cases vides et les catégories bavardes d'un CSV de terrain. La compréhension de filtrage utilisée ici est exactement celle du §4 — le filtre « garder les valides » — appliquée au résultat de la conversion.

### Exercice 1 : types et conversions — le panier du stand

Un fichier de caisse a livré ses lignes en **texte**. Convertissez et calculez sans `pandas` : le total du panier, puis le prix moyen arrondi à 2 chiffres après la virgule.


**Lecture du contrôle — exercice 1.**

Avec le stub intact, la sortie reste `Exercice a completer : total, moyenne`. ## Vérifiez votre compréhension

**Question 1 — Pourquoi `0.1 + 0.2 == 0.3` rend-il `False` alors que l'affichage semble presque juste ?**

La section 1 affiche `0.1 + 0.2 = 0.30000000000000004` puis `egalite stricte avec 0.3 ? False` et `comparaison tolérante ? True`.

<details><summary>Réponse</summary>

Les flottants sont stockés en base 2 : `0.1` et `0.2` n'y sont pas exacts, et leur somme non plus — elle vaut `0.30000000000000004`, pas `0.3`. L'égalité stricte compare les bits et échoue ; la comparaison tolérante teste si l'écart reste sous un seuil fixé, et réussit. En data science, on compare donc toujours des flottants avec une tolérance, jamais avec `==`.

</details>

**Question 2 — Une fonction se termine par `print("bonjour")`. Qu'affiche-t-elle, et que rend-elle ?**

La section 3 montre une fonction `amplitude` qui rend `1.299999999999999`, puis la sortie : `bonjour` suivi de « sans return explicite, une fonction rend : None ».

<details><summary>Réponse</summary>

Elle affiche « bonjour » et rend `None`. Afficher écrit à l'écran ; rendre transmet une valeur au code appelant via `return`. Sans `return` explicite, Python rend `None` — c'est pourquoi la fonction utile (celle dont on peut réutiliser le résultat) se termine par un `return`, et que le `print` y est un outil de contrôle, pas de livraison.

</details>

**Question 3 — Le filtre fièvre rend `febriles : []`. Pourquoi, alors que `38.0` figure dans les températures ?**

La section 4 part de `temperatures_f = [98.6, 99.1, 97.9, 100.4]` (Fahrenheit) converties en Celsius `[37.0, 37.3, 36.6, 38.0]`, et la sortie de la compréhension filtrée est une liste vide.

<details><summary>Réponse</summary>

Parce que le test est strict (`> 38`) : `38.0` n'est pas strictement supérieur à `38`. La valeur exactement à la borne est exclue — c'est le piège classique de la comparaison stricte, et la raison pour laquelle la sortie se lit au lieu de se présumer : une liste vide veut dire « aucune valeur ne passe le test », pas « le code est cassé ».

</details>
