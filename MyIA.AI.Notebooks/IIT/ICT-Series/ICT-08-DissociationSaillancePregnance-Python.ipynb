## 1. La thèse, énoncée dans sa version honnête

L'organe répond par un **verdict**, pas par une intuition. `case_verdict` rejoue la batterie complète et rend une chaîne parmi quatre. Les seuils qu'elle applique sont **lus dans l'organe** (`case_verdict`, seuils partiels et verdict) : au niveau total comme au niveau de la décision, la dissociation exige que la corrélation partielle de la prégnance conditionnée à la salience dépasse le seuil haut, et que celle de la salience conditionnée à la prégnance reste sous le seuil bas ; le témoin négatif exige en plus que le motif s'inverse chez l'animat réactif.

### Lecture 1 — le canal décision se dissocie, le canal total non, et le témoin négatif est faible

Trois choses sont lisibles dans la sortie ci-dessus.

**Au niveau de la décision**, les deux moitiés de la thèse tiennent : la décision d'approche suit la valence conditionnée à la salience — au-dessus du seuil haut de l'organe — et elle ne suit pas la salience conditionnée à la valence, qui reste sous le seuil bas. C'est exactement la thèse : *ce qui est vu* ne gouvernait pas *ce qui est choisi*.

**Au niveau total**, la dissociation **ne** tient pas : l'engagement total reste corrélé aux deux canaux. C'est attendu et la thèse ne le prétend pas — l'engagement total **inclut** la détection, il porte donc la salience par construction.

**Le témoin négatif, lui, est faible à cette graine.** Pour rendre `DISSOCIATED-AT-DECISION`, l'organe exige que le motif s'inverse chez l'animat réactif ; or la corrélation partielle de la prégnance conditionnée à la salience y reste au-dessus du seuil bas. L'organe rend donc `DISSOCIATED-DECISION-NULL-WEAK` : **la dissociation est établie, son témoin négatif est faible sur cette graine**. Le carnet le dit ainsi plutôt que de choisir la graine qui arrange.

### Lecture 2 — la dissociation tient sur toutes les graines, le témoin négatif sur trois

Sur les quatre graines enregistrées par l'organe, le canal décision est dissocié à chaque fois. Trois d'entre elles rendent le verdict plein — `DISSOCIATED-AT-DECISION` — et une seule (`GRAINE`, celle du carnet) rend `DISSOCIATED-DECISION-NULL-WEAK` : c'est celle où le témoin négatif reste au-dessus du seuil bas.

L'organe qualifie l'ensemble de **robuste**, parce que sa règle de robustesse compte les verdicts qui commencent par `DISSOCIATED`, pas ceux qui rendent le libellé plein. La lecture honnête est donc à deux étages : **la dissociation elle-même est robuste** ; son **témoin négatif** est faible sur une graine sur quatre. La section 3 reviendra à ce témoin, en champ.

## 2. Le transport dans le champ de valence

La batterie de la section 1 vit dans un espace **abstrait** : la valence y vient d'un apprentissage associatif. La question de la section 2 est de savoir si la dissociation survit quand la valence vient d'un **champ spatial** — d'une bosse attractive réelle, celle que `ict.valence` construit.

Le protocole reprend les fonctions de l'organe, sans les réécrire : on construit le profil de valence du champ le long d'un rayon, on en tire une valence par stimulus, puis on rejoue la batterie (décision, engagement, corrélations partielles) sur cette valence. Seule l'origine de `V` change — tout le reste est l'organe.

Deux cellules suivent : la première balaie les quatre régimes de source, la seconde balaie le gain de la réponse d'approche.

```python
def profil_valence(kind="erratique", graine=GRAINE, n_points=24, rayon=12.0):
    """Le profil de valence du champ, le long d'un rayon depuis le centre."""
    r = np.random.default_rng(graine)
    src = V.source_trajectory(kind, N_PAS, TAILLE, rng=r, start=V.center_start(TAILLE))
    centre = V.center_start(TAILLE)
    return np.array([V.valence_at(centre + np.array([d, 0.0]), src[-1])
                     for d in np.linspace(0.0, rayon, n_points)])


def batterie(V_decision, V_reference, s, gain=3.0, reactif=False,
              n_trials=N_TRIALS, graine=GRAINE):
    """La batterie de l'organe, appliquee a une valence donnee.

    `V_decision` est la valence dont l'animat dispose (celle qui pilote sa
    decision) ; `V_reference` est celle contre laquelle on correlate.
    """
    r = np.random.default_rng(graine)
    if reactif:
        p = SVD.approach_probability_reactive(s)
    else:
        p = SVD.approach_probability_valence(V_decision, gain=gain)
    decision = SVD.measure_decision_given_detected(s, p, n_trials=n_trials, rng=r)
    engagement = SVD.measure_engagement(s, p, n_trials=n_trials, rng=r)
    return {
        "partial_V_donne_s": SVD.partial_spearman(decision, V_reference, [s]),
        "partial_s_donne_V": SVD.partial_spearman(decision, s, [V_reference]),
        "engagement_moyen": float(np.mean(engagement)),
        "decision_moyenne": float(np.mean(decision)),
    }


s_champ = np.random.default_rng(GRAINE).uniform(0.1, 1.0, N_STIMULI)

print(f"{'regime':11s} {'bande de valence':>22s} {'partial(V|s)':>13s} {'partial(s|V)':>13s}")
print("-" * 64)
for kind in ("statique", "balistique", "erratique", "bruite"):
    profil = profil_valence(kind)
    V_champ = np.random.default_rng(GRAINE + 1).choice(profil, size=N_STIMULI)
    b = batterie(V_champ, V_champ, s_champ)
    print(f"{kind:11s} {profil.min():9.4f} -> {profil.max():.4f} "
          f"{b['partial_V_donne_s']:+13.3f} {b['partial_s_donne_V']:+13.3f}")
```

```python
profil = profil_valence("erratique")
V_champ = np.random.default_rng(GRAINE + 1).choice(profil, size=N_STIMULI)

print(f"{'gain':>7s} {'partial(V|s)':>13s} {'partial(s|V)':>13s}")
print("-" * 38)
for gain in (3.0, 10.0, 30.0, 60.0, 100.0, 200.0):
    b = batterie(V_champ, V_champ, s_champ, gain=gain)
    print(f"{gain:7.1f} {b['partial_V_donne_s']:+13.3f} {b['partial_s_donne_V']:+13.3f}")

print("\nLe gain par defaut de l'organe est celui de la batterie abstraite ;")
print("la bande de valence du champ est d'une autre ampleur. Le tableau le montre.")
```

### Lecture 3 — la dissociation se transporte, et ce qui la fait porter est la bande de valence

Deux lectures, à lire ensemble.

**La moitié salience se transporte intacte.** Dans les quatre régimes, la corrélation partielle de la décision avec la salience conditionnée à la valence reste très en dessous du seuil bas de l'organe : une fois la valence connue, savoir ce qui est *saillant* n'explique plus la décision. C'est le cœur de la thèse, porté au champ.

**La moitié valence dépend de l'échelle du champ.** Trois des quatre régimes franchissent le seuil haut de l'organe ; le régime `erratique`, lui, reste juste en dessous. La raison est visible dans la colonne de gauche du tableau : le régime erratique n'occupe qu'une bande de valence étroite — la bosse du champ y est presque plate sur la distance, parce que la source change de place à chaque pas. Les régimes `statique` et `bruite` occupent une bande large et franchissent le seuil sans difficulté ; le régime `balistique` se situe entre les deux.

**Ce que la seconde cellule établit.** Le gain est un paramètre de l'organe, pas une constante cachée : `approach_probability_valence` l'expose. Balayer ce gain montre que l'écart du régime `erratique` n'est pas un défaut de mesure mais une question d'**échelle** — la réponse d'approche de l'organe est calibrée sur l'amplitude de la valence apprise, et la valence d'un champ erratique est d'un ordre de grandeur plus petite. On ne « corrige » donc pas la mesure : on **mesure** où le seuil de l'organe est franchi, et on le dit.

## 3. W1 — le témoin négatif nommé

Un témoin qui ne rougit pas ne prouve rien. Le témoin négatif de cette thèse est **nommé** : c'est l'animat **réactif**, celui dont la décision ne dépend que de la salience — `approach_probability_reactive` de l'organe, qui remplace la valence par la salience dans la réponse d'approche.

Si l'instrument est sain, le motif doit **s'inverser** : chez l'animat réactif, la décision doit suivre la salience conditionnée à la valence, et **cesser** de suivre la valence conditionnée à la salience. C'est la mesure que fait la cellule ci-dessous, sur le même champ que la section 2, avec la même salience exogène.

La cellule referme aussi par les portes de l'**étude pré-enregistrée** (`ict.adjonction_saillance_pregnance`) : quatre portes, comptées sur cinq graines, dont deux portent précisément ce témoin négatif — l'une exclut la prégnance quand la salience est annulée, l'autre exclut la salience quand la prégnance est annulée.

```python
profil = profil_valence("erratique")
V_champ = np.random.default_rng(GRAINE + 1).choice(profil, size=N_STIMULI)

print("== animat guide par la valence (la these) ==")
for cle, val in batterie(V_champ, V_champ, s_champ).items():
    print(f"  {cle:22s} {val:+.3f}")

print("\n== animat reactif : decision sur la salience seule (temoin W1) ==")
for cle, val in batterie(V_champ, V_champ, s_champ, reactif=True).items():
    print(f"  {cle:22s} {val:+.3f}")

print("\n== etude pre-enregistree : portes et verdiicts ==")
etude = run_preregistered_study(seeds=(0, 1, 7, 42, 99))
print("  verdicts :", etude["verdicts"])
print("  pass_counts :", etude["pass_counts"])
ligne = [r for r in etude["rows"] if r["seed"] == GRAINE][0]
print("  partials (graine du carnet) :", ligne["partials"])
print("  gates (graine du carnet) :", ligne["gates"])
```

### Lecture 4 — le motif s'inverse : l'instrument n'est pas suspect

Les deux blocs se répondent.

Chez l'animat **guidé par la valence**, la décision suit la valence conditionnée à la salience, et la salience conditionnée à la valence n'explique rien. Chez l'animat **réactif**, l'inverse se produit : la corrélation partielle de la valence passe sous le seuil bas, et celle de la salience monte au-dessus du seuil haut. Le motif est **inversé**, comme le exige un témoin négatif valide : ce n'est pas l'instrument qui produit la dissociation, c'est la règle de décision qui la produit.

L'étude pré-enregistrée confirme le même résultat par une voie indépendante : la spécification d'origine est **falsifiée**, le canal opérationnel est **soutenu**, et les deux portes de témoin négatif tiennent sur les graines enregistrées — dont celle du carnet, où les quatre portes sont au vert.

Le témoin négatif de la section 1, faible sur la graine du carnet, l'est donc **dans la batterie abstraite seulement** : au champ, il est net. C'est une information sur la **puissance** du témoin, pas sur la thèse.

## 4. W2 — la valence brouillée : la dissociation se dégrade

Le second témoin ne cherche pas l'inversion du motif mais son **effacement**. On bruite la valence dont l'animat dispose (bruit gaussien additif, d'amplitude proportionnelle à la plage de valence du champ), et on regarde ce que devient la décision, **corrélée à la valence vraie** — celle qu'on a bruitée.

C'est le point de méthode qui fait toute la mesure : corréler la décision à la valence **bruitée** ne dégrade jamais la mesure (la décision suit alors la valence bruitée par construction) ; il faut corréler à la valence **vraie**. La cellule ci-dessous le fait, et ajoute le seuil de brouillage — le niveau de bruit où la corrélation partielle passe sous le seuil haut de l'organe.

Le gain employé est celui que le balayage de la section 2 a identifié pour la bande de valence du champ : la mesure de dégradation se fait dans la zone où la dissociation tient.

```python
GAIN_CHAMP = 10.0  # gain identifie par le balayage de la section 2

profil = profil_valence("erratique")
V_champ = np.random.default_rng(GRAINE + 1).choice(profil, size=N_STIMULI)
plage = float(V_champ.max() - V_champ.min())

print(f"{'sigma x plage':>14s} {'partial(V_vraie|s)':>19s} {'partial(s|V_vraie)':>18s}")
print("-" * 56)
for sigma in (0.0, 0.5, 1.0, 1.5, 2.0, 3.0):
    r = np.random.default_rng(GRAINE + 2)
    V_bruit = V_champ + sigma * plage * r.normal(0.0, 1.0, N_STIMULI)
    b = batterie(V_bruit, V_champ, s_champ, gain=GAIN_CHAMP)
    print(f"{sigma:14.2f} {b['partial_V_donne_s']:+19.3f} {b['partial_s_donne_V']:+18.3f}")

# bissection du seuil de brouillage : ou la dissociation cesse de tenir
bas, haut = 0.0, 4.0
for _ in range(30):
    milieu = (bas + haut) / 2
    r = np.random.default_rng(GRAINE + 2)
    V_bruit = V_champ + milieu * plage * r.normal(0.0, 1.0, N_STIMULI)
    b = batterie(V_bruit, V_champ, s_champ, gain=GAIN_CHAMP)
    if abs(b["partial_V_donne_s"]) > 0.5:
        bas = milieu
    else:
        haut = milieu
print(f"\nseuil de brouillage : sigma = {bas:.3f} x plage de valence du champ")
```

### Lecture 5 — la dégradation est monotone, et le seuil se mesure

La corrélation partielle de la décision avec la **valence vraie** décroît **monotone** avec le bruit : elle part de la valeur mesurée sans bruit et tend vers zéro à mesure que la valence de l'animat devient du bruit. Pendant toute la traversée, la moitié salience reste collée au canal salience — le bruit n'ouvre pas de brèche de ce côté.

La bissection donne le **seuil de brouillage** : le niveau de bruit, exprimé en fraction de la plage de valence du champ, où la corrélation partielle passe sous le seuil haut de l'organe. Ce n'est pas une constante universelle : c'est une propriété du **substrat** (la bande de valence du champ, l'amplitude du gain) et elle se mesure, elle ne se suppose pas.

C'est ce que le carnet retient du second témoin : la dissociation **n'est pas une propriété de la corrélation** — elle tient tant que la valence qui la porte reste plus informative que son propre bruit.

## 5. Le crochet `gate` : ce qu'il change, et pourquoi il est dans la série source

Jusqu'ici, la dissociation a été mesurée sur une décision abstraite. Le crochet `gate` est l'objet qui permet de la **porter dans le champ** : `simulate_reactive` et `simulate_phat` l'acceptent en paramètre optionnel, et la porte **multiplie le pas de déplacement** de l'animat.

Le défaut est `None`, c'est-à-dire le comportement d'origine : les carnets existants de la série ne changent pas d'un bit. Quand une porte est fournie, le pas est raccourci — proportionnellement à la porte — ce qui permet de rejouer dans le champ la condition « détecté » que l'organe mesure au niveau batterie.

Ce que la cellule mesure : le nombre de pas **réellement déplacés**, la distance moyenne à la source, la cadence de capture à plusieurs rayons, et l'identité bit pour bit du chemin par défaut.

```python
r = np.random.default_rng(GRAINE)
src = V.source_trajectory("erratique", N_PAS, TAILLE, rng=r, start=V.center_start(TAILLE))
depart = V.center_start(TAILLE) + np.array([11.0, 0.0])

# salience de detection : elle depend de la distance a la source, lue sur la
# trajectoire de reference (sans porte), puis la porte est un tirage de Bernoulli.
traj_ref = V.simulate_phat(src, TAILLE, depart, N_PAS)
distance_ref = np.linalg.norm(traj_ref - src, axis=1)
s_t = 1.0 / (1.0 + distance_ref)
porte = (r.random(N_PAS) < s_t).astype(float)

print(f"{'rayon':>6s} {'porte':>6s} {'capture':>8s} {'distance moy.':>14s} {'pas deplaces':>13s}")
print("-" * 54)
for rayon in (0.5, 1.0, 1.5):
    for nom, p in (("sans", None), ("avec", porte)):
        traj = V.simulate_phat(src, TAILLE, depart, N_PAS, gate=p)
        distance = np.linalg.norm(traj - src, axis=1)
        deplace = int(np.sum(np.linalg.norm(np.diff(traj, axis=0), axis=1) > 1e-9))
        capture = V.capture_rate(traj, src, radius=rayon)
        print(f"{rayon:6.1f} {nom:>6s} {capture:8.3f} {distance.mean():14.3f} "
              f"{deplace:8d}/{N_PAS - 1}")

identique = np.array_equal(
    V.simulate_phat(src, TAILLE, depart, N_PAS),
    V.simulate_phat(src, TAILLE, depart, N_PAS, gate=None),
)
print(f"\nretrocompatibilite : gate=None est identique au defaut, bit pour bit : {identique}")
```

### Lecture 6 — la porte agit sur la perception, pas sur le choix

Trois faits, lus dans la sortie ci-dessus.

**La porte coupe le mouvement, pas la direction.** Sans porte, l'animat se déplace à chaque pas ; avec la porte de détection, il ne se déplace que sur les pas où la source est détectée. Le nombre de pas effectivement déplacés tombe de la totalité à une fraction, et la distance moyenne à la source double : l'animat **rattrape moins**.

**La capture baisse, et baisse d'autant plus que le rayon est large.** C'est la signature attendue d'une porte qui agit sur la *poursuite* : à rayon serré, la capture reste comparable ; à rayon large, elle s'effondre. Une porte qui aurait agi sur la décision d'approche aurait fait l'inverse.

**Le chemin par défaut est intact.** `gate=None` redonne bit pour bit le résultat sans paramètre — le carnet peut l'affirmer parce qu'il le vérifie, pas parce qu'il le suppose.

Ce que la valve **ne** fait pas : elle ne change pas la règle de choix. La valence reste le canal qui décide ; la porte ne fait que dire à l'animat **ce qu'il a le droit de percevoir**. C'est exactement la thèse, écrite dans le code de l'organe.

## 6. Ce que le capstone ne conclut pas

La règle de crédit de ce carnet tient en une phrase : **ce qui est vu et ce qui est choisi sont dissociés, et c'est un verdict négatif assumé**.

Le carnet n'affirme pas que la saillance est un canal « plus faible » que la valence. Il n'affirme pas non plus que la prégnance est « plus réelle » que la salience, ni que l'une des deux dérive de l'autre. Les deux canaux sont **deux descriptions** d'un même acte, et la mesure montre qu'ils ne co-varient pas une fois l'un d'eux connu. Ce que la dissociation **dit**, c'est que la saillance compte pour **voir** et la valence pour **agir** — et que confondre les deux, dans un modèle comme dans un protocole, produit des résultats qui n'existent pas.

Trois limites sont assumées :

- la dissociation est établie au **niveau décision**, pas au niveau total — l'engagement total porte la détection par construction ;
- son **témoin négatif** est net au champ mais faible sur une graine de la batterie abstraite ;
- le **seuil** de chaque mesure dépend de l'échelle du substrat (bande de valence du champ, gain), et se mesure à chaque fois.

Ce qui n'est **pas** établi : que ces canaux ont un support biologique, ni que la saillance « précède » la valence. Le carnet est un **modèle** de dissociation, pas une thèse sur le vivant.
