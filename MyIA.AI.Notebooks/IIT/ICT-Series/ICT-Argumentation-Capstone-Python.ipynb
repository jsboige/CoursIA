## Ce que la trajectoire dit, et ce qu'elle ne dit pas

La trajectoire ci-dessus est **croissante en arguments disponibles** : à chaque pas, un énoncé de plus entre dans le débat, et le labeling grounded est recalculé sur le sous-AF induit. Trois choses s'y lisent directement.

**L'englobement est visible dans le comptage.** Le nombre de `in` ne décroît pas de façon monotone : les réfutations de l'Agneau font basculer des thèses du Loup, la clause finale `A4` les rattrape toutes d'un coup. C'est ce mouvement en trois temps (thèse, réfutation, récupération) que le carnet de graine appelle l'englobement.

**Le contre-cas est la seconde fable.** Le Loup et le Chien a la même structure de surface — un locuteur dominant, un locuteur qui résiste — mais la clause finale n'englobe pas : elle **abandonne**. Un AF où le dernier énoncé attaque sans être attaqué produit une trajectoire de forme différente. Les deux specimens sont donc un couple experimental, pas deux exemples.

**Ce que la trajectoire ne dit pas encore.** Elle donne une forme par pas, pas une mesure du débat entier. Le triplet assemblé (`ict.triplet`, livrable 1/2, PR #23851) prend cette trajectoire et en tire trois lectures numériques — σ, K, ordre causal. Les sections qui suivent les calculent, puis **les testent contre la dégénérescence** avant de les appliquer à quoi que ce soit d'historique. C'est l'ordre de la note de conception (§7) : mesurer d'abord si l'instrument distingue, interpréter ensuite.

### Le couple experimental

Les deux specimens ne sont pas deux illustrations d'une meme chose : ce sont **les deux issues** d'une meme forme rhétorique.

| | Le Loup et l'Agneau | Le Loup et le Chien |
|---|---|---|
| Locuteur dominant | Le Loup | Le Chien |
| Locuteur qui résiste | L'Agneau | Le Loup |
| Clause finale | `A4` attaque les trois réfutations | `L3` attaque le compromis et la récompense |
| Effet | **englobement** : la résistance est reprise, pas niée | **rupture** : la résistance quitte le jeu |
| Trajectoire | `in` partout à la fin | `in` pour la sortie, `out` pour le compromis |

Le second cas est le témoin : là où la fable 1 montre une domination qui **passe par** l'argumentation, la fable 2 montre une domination qui **échoue** et se retire. Si un instrument ne distingue pas ces deux issues, il ne mesure pas la rhétorique, il mesure l'AF.

C'est précisément ce que la section suivante vérifie — et le résultat n'est pas celui que la note de conception espérait.


## 1. Le triplet assemblé sur chaque specimen

`ict.triplet.assembler_triplet` (PR #23851) prend un AF et un ordre d'arrivée, reconstruit la trajectoire, estime la matrice de transition, et rend les trois jambes **côte à côte, sans jamais les agréger en un scalaire** :

- **σ** — production d'entropie du discours : la distance entre la matrice de transition et sa réversée. σ > 0 = le discours ne se rejoue pas à l'envers.
- **K** — code en deux parties (MDL) de la suite d'états : combien de bits pour décrire la trajectoire observée.
- **ordre causal** — l'ordre de Markov qui minimise le MDL, comparé à l'ordre qui minimise la log-vraisemblance hors échantillon.

Aucune des trois n'est « le » chiffre. La note de conception est explicite : elles se lisent ensemble ou pas du tout.

On les calcule sur les deux specimens, à leur ordre narratif.


```python
triplets = {}
for nom, af, ordre in (('Loup-Agneau', af_loup, ordre_loup),
                       ('Loup-Chien', af_chien, ordre_chien)):
    r = triplet.assembler_triplet(af, ordre)
    triplets[nom] = r
    print('=' * 72)
    print('%s | n = %d | M = 1 ordre' % (nom, af.n()))
    print('  sigma             = %.17g' % r['sigma']['irreversibilite']['sigma'])
    print('  P_rev_distance    = %.6f' % r['sigma']['irreversibilite']['P_rev_distance'])
    print('  K tpm_bits        = %.3f   (compression %.0f bits)'
          % (r['K']['tpm_bits'], r['K']['bits_compression']))
    print('  etats / transitions = %d / %d' % (r['K']['n_etats'], r['K']['n_transitions']))
    m = r['ordre']['markov']
    print('  ordre MDL = %d | ordre NLL = %d | etats Markov = %d'
          % (m['best_mdl_order'], m['best_nll_order'], m['k_states']))
    for row in m['rows']:
        print('     o=%d  ctx=%2d  mdl=%8.3f  nll_heldout=%8.3f  echappes=%d/%d'
              % (row['order'], row['n_contexts'], row['mdl_bits'],
                 row['nll_heldout'], row['n_escaped'], row['n_contexts']))
    print()

# Verification croisee : les trois jambes restent separees dans le dict rendu.
print('Cles de premier niveau rendues par assembler_triplet :', sorted(triplets['Loup-Agneau']))
```

### Trois observations, dont une qui invalide la lecture naïve

**1. σ est au niveau du bruit flottant.** `4.6e-09` sur le Loup et l'Agneau, `6.0e-09` sur le Loup et le Chien. Ce n'est pas « petit mais non nul par construction » : la section 2 montrera que sur un graphe vide il vaut **exactement** la même valeur. La production d'entropie mesurée ici n'est pas portée par la structure argumentative.

**2. K distingue les deux specimens.** `24.0` bits pour le Loup et l'Agneau, `33.2` bits pour le Loup et le Chien. La différence est cohérente avec la longueur (7 énoncés contre 9) **et** avec la forme : le specimen 2 parcourt dix états distincts là où le specimen 1 en parcourt huit.

**3. L'ordre causal n'achète rien ici, et le notebook doit le dire.** À partir de l'ordre 1, la log-vraisemblance hors échantillon est **rigoureusement constante** (`18.575` pour le specimen 1, `25.850` pour le specimen 2, identiques à tous les ordres), et le MDL plafonne à `8.000` bits (resp. `11.610`). Le nombre de contextes disponibles explose (`1, 4, 16, 64` puis `1, 5, 25, 125, 625`) alors que le nombre d'états suivants distincts reste figé (`4`, resp. `5`) : **aucun contexte n'est jamais réobservé**, donc retenir un ordre supérieur n'achète aucune prédiction.

Le MDL choisit alors l'ordre le plus économique — `2` pour le specimen 1, `1` pour le specimen 2 — ce qui n'est **pas** une découverte sur le discours mais une conséquence du nombre d'états. Le chiffre `ordre MDL = 2` ne dit **pas** que le discours a une mémoire d'ordre deux. Une lecture rapide du tableau le prendrait pour un résultat : c'en est un sur la trajectoire finie, pas sur la rhétorique.

**La question que ces trois observations posent, et qui commande toute la suite :** le triplet distingue-t-il le graphe d'attaque, ou seulement la longueur de la trajectoire ? La section suivante y répond par un test plutôt que par un argument.


## 2. Test de non-dégénérescence : le triplet voit-il le graphe ?

La note de conception (§5.4) impose de vérifier la non-dégénérescence **avant** tout cas historique. Un instrument qui rend le même chiffre sur un anneau bidirectionnel et sur un graphe vide ne mesure rien de rhétorique, et toute lecture « ce débat est irréversibiliste » qu'on en tirerait serait un artefact.

Le test est direct : **cinq graphes structurellement opposés, même nombre d'arguments, même ordre narratif unique**, et on lit le triplet sur chacun.

En tête de section, deux constructeurs repris du carnet de graine (§5) — ce sont les **régimes de référence** :

- `deliberative_af(n)` : anneau où chaque argument attaque son successeur et son prédécesseur (délibération symétrique) ;
- `populist_af(n)` : un argument hub attaque tous les autres, plus une chaîne dirigée (domination sans contradiction).

Le choix de les recopier plutôt que de les importer est **déclaré** (cellule 3) : `ict/` ne porte pas de constructeur de régime, et en extraire un pour deux notebooks serait du couplage sans consommateur.


```python
def deliberative_af(n):
    """Anneau de deliberation : chaque argument attaque ses deux voisins."""
    attaques = []
    for i in range(n):
        attaques.append((i, (i + 1) % n))
        attaques.append(((i + 1) % n, i))
    return arg.DungAF(list(range(n)), attaques)


def populist_af(n):
    """Hub 0 attaque tout le monde, puis chaine dirigee 1 -> 2 -> ... -> n-1."""
    attaques = [(0, i) for i in range(1, n)]
    attaques += [(i, i + 1) for i in range(1, n - 1)]
    return arg.DungAF(list(range(n)), attaques)


def etats_et_transitions(af, ordre, encoding='summary'):
    """Trajectoire -> (suite d'etats, liste de transitions consecutives)."""
    traj = arg.belief_trajectory(af, ordre, include_empty=True)
    if encoding == 'signature':
        etats = [arg.state_signature(lab, n_args=af.n()) for lab in traj]
    else:
        etats = [arg.belief_summary(lab) for lab in traj]
    return etats, list(zip(etats, etats[1:]))


n_ref = 7
ordre_ref = list(range(n_ref))

graphes = {
    'anneau bidirectionnel': deliberative_af(n_ref),
    'hub + chaine (populiste)': populist_af(n_ref),
    'aucune arete': arg.DungAF(list(range(n_ref)), []),
    'arete unique (1,0)': arg.DungAF(list(range(n_ref)), [(1, 0)]),
    'fable Loup-Agneau': af_loup,
}

print('%-26s %22s %16s %8s %12s' % ('graphe', 'sigma', 'P_rev_distance', 'etats', 'K (2 parties)'))
print('-' * 88)
mesures_ordre_unique = {}
for label, af in graphes.items():
    try:
        r = triplet.assembler_triplet(af, ordre_ref)
        etats, _ = etats_et_transitions(af, ordre_ref)
        k2 = mdl.two_part_code(etats, split=0.5)['total_bits']
        s = r['sigma']['irreversibilite']['sigma']
        mesures_ordre_unique[label] = (s, r['sigma']['irreversibilite']['P_rev_distance'], k2)
        print('%-26s %22.17g %16.6f %8d %12.4f'
              % (label, s, r['sigma']['irreversibilite']['P_rev_distance'],
                 r['K']['n_etats'], k2))
    except Exception as exc:
        print('%-26s REFUS %s: %s' % (label, type(exc).__name__, exc))

sigmas = {v[0] for v in mesures_ordre_unique.values()}
ks = {round(v[2], 9) for v in mesures_ordre_unique.values()}
print()
print('valeurs distinctes de sigma :', len(sigmas))
print('valeurs distinctes de K     :', len(ks))
```

### Le résultat, énoncé sans détour

**Une seule valeur distincte de σ et une seule de K, sur cinq graphes qui n'ont en commun que le nombre d'arguments.** Le graphe vide et l'anneau bidirectionnel rendent le même nombre au bit près.

Ce n'est pas un bug de `assembler_triplet` : c'est une propriété de ce qu'on lui donne. Sur **un** ordre narratif, la trajectoire ne revient jamais sur un état déjà visité — chaque pas ajoute un argument, donc l'ensemble des énoncés disponibles croît strictement et la signature de labeling ne peut pas se répéter. La matrice de transition estimée est alors **déterministe et acyclique** : chaque état a exactement un successeur, aucun n'est réobservé.

Or sur une chaîne déterministe acyclique :

- **σ dépend du nombre de pas, pas des étiquettes** — la production d'entropie d'un flux irréversible vaut son taux multiplié par la durée, et le taux ne dépend que du fait qu'il n'y a pas de retour ;
- **K ne dépend que du nombre d'états distincts et de la longueur** — un code en deux parties paie une déclaration par état et zéro bit pour la suite, puisque la suite est forcée.

Autrement dit : **sur un ordre unique, le triplet mesure la longueur du texte.** Les graphes structurellement extrêmes donnent le même chiffre parce qu'ils produisent tous une trajectoire monotone de même longueur. Le test de non-dégénérescence échoue, et il échoue pour tous les graphes à la fois.

### Ce que cela condamne, précisément

La note de conception (§5.3) traitait la jambe σ comme « bruitée mais porteuse ». La mesure dit autre chose : **au format d'un ordre unique, la jambe σ est réfutée**, pas bruitée. Et la jambe K, telle que lue sur un ordre unique, mesure la longueur. Il reste une seule voie honnête pour que le triplet parle du graphe — lui donner **plusieurs ordres d'arrivée**, de sorte que les états soient réobservés et que la matrice cesse d'être déterministe.

C'est ce que fait la section suivante, et c'est aussi le point où la seconde jambe de la note (§5.2.2, `attack_adjacency_symmetric()` efface la direction de l'attaque avant le spectre) trouve sa réponse : l'information directionnelle doit survivre **quelque part** dans le pipeline, sinon aucun encodage ne la retrouvera.


## 3. Le protocole agrégé — et la découverte que l'encodage « signature » est aveugle

Le carnet de graine retient le **niveau signature** comme encodage de σ : on remplace le résumé `(n_in, n_out, n_undec)` par la signature ternaire complète du labeling (`state_signature`). L'argument était que le résumé perd l'identité des arguments encore indécis.

On mesure cet argument plutôt que de le croire. **Neuf graphes au même `n`, trois cents ordres d'arrivée tirés des permutations**, et on compare σ au niveau signature entre graphes. Si l'encodage signature porte de l'information sur le graphe, les neuf valeurs doivent différer.


```python
def regime_sigma(af, ordres, encoding='summary'):
    """Sigma agrege sur plusieurs ordres : transitions mises en commun, TPM unique."""
    transitions = []
    for ordre in ordres:
        _, tr = etats_et_transitions(af, ordre, encoding=encoding)
        transitions.extend(tr)
    P, mapping = tpm_estimation.tpm_from_transitions(transitions, unseen='self')
    return arg.discourse_irreversibility(P)['sigma'], len(mapping)


n_sig = 7
M_ORDRES = 300
_perms = list(itertools.permutations(range(n_sig)))
_rng_ordres = np.random.default_rng(2024)
_choix = _rng_ordres.choice(len(_perms), size=M_ORDRES, replace=False)
ORDRES = [list(_perms[i]) for i in _choix]

graphes_sig = {
    'anneau bidirectionnel': deliberative_af(n_sig),
    'hub + chaine (populiste)': populist_af(n_sig),
    'aucune arete': arg.DungAF(list(range(n_sig)), []),
    'arete unique (1,0)': arg.DungAF(list(range(n_sig)), [(1, 0)]),
    'fable Loup-Agneau': af_loup,
}
_rng_g = np.random.default_rng(7)
for k in range(4):
    m = int(_rng_g.integers(1, 12))
    aretes = set()
    while len(aretes) < m:
        a, b = _rng_g.integers(0, n_sig, size=2)
        if a != b:
            aretes.add((int(a), int(b)))
    graphes_sig['aleatoire #%d (%d aretes)' % (k, m)] = arg.DungAF(list(range(n_sig)), sorted(aretes))

print('%d ordres d\'arrivee, n = %d' % (M_ORDRES, n_sig))
print()
print('%-28s %26s %26s' % ('graphe', 'sigma(signature)', 'sigma(resume)'))
print('-' * 82)
sig_vals = []
for label, af in graphes_sig.items():
    s_sig, n_etats_sig = regime_sigma(af, ORDRES, encoding='signature')
    s_res, _ = regime_sigma(af, ORDRES, encoding='summary')
    sig_vals.append(s_sig)
    print('%-28s %26.17g %26.6e   (%d etats)' % (label, s_sig, s_res, n_etats_sig))

print()
print('valeurs distinctes de sigma(signature) :', len(set(sig_vals)))
```

### Une seule valeur, neuf graphes : l'encodage fin est le plus aveugle

**σ au niveau signature rend exactement le même nombre sur les neuf graphes**, y compris le graphe vide et le graphe aléatoire à onze arêtes. Ce n'est pas « à peu près » : c'est identique à la précision machine.

L'observation contre-intuitive est la suivante : **c'est l'encodage le plus grossier qui porte le signal**, pas le plus fin. σ au niveau résumé `(n_in, n_out, n_undec)` varie d'un régime à l'autre ; σ au niveau signature ne varie pas du tout. Le raisonnement qui avait fait préférer la signature au résumé s'inverse à la mesure.

### Le mécanisme, et pourquoi il faut le dire comme une explication

L'explication tient en une phrase, et je la donne comme telle — **c'est un raisonnement sur le protocole, pas une seconde mesure** : les trois cents ordres sont tirés **uniformément** parmi les permutations. Partant d'un préfixe donné, l'argument suivant est donc uniformément réparti sur les arguments restants, **quel que soit le graphe**. Le noyau de transition agrégé encode surtout ce tirage uniforme, pas la structure d'attaque. Le graphe décide *quels* états apparaissent — et de fait les graphes produisent des ensembles d'états différents — mais pas *avec quelle probabilité* on passe de l'un à l'autre. σ, qui est une fonction du noyau, reste donc invariant.

Le résumé, plus grossier, écrase des états distincts en un même triplet : le noyau agrégé y dépend alors de la **fréquence** des états, qui elle dépend du graphe. De là le peu de signal qui subsiste — et de là aussi sa faiblesse : il ne mesure pas la structure, il mesure la distribution des formes de labeling.

**Conséquence pour la note de conception.** Le « σ retenu = niveau signature » du carnet de graine est structurellement aveugle au graphe. Ce n'est pas une nuance de réglage : à ce format, la jambe σ ne peut pas départager deux structures argumentatives. Elle peut seulement dire qu'une trajectoire avance — ce que la section 2 avait déjà établi.


## 4. Ce qui reste : K, à taille égale, sur des régimes de référence

Si σ est aveugle à ce format, reste la jambe K. On la mesure de la seule façon qui permette une comparaison honnête : **même nombre d'arguments, même nombre d'ordres, même nombre total d'états** entre le specimen et les deux régimes de référence. Sans cette égalisation, une différence de K pourrait n'être qu'une différence de taille.

Les deux specimens ont des tailles différentes (7 et 9 arguments) : chacun est donc comparé à des régimes construits **à son** `n`, et les deux blocs ne se comparent pas entre eux.

Lecture attendue si K mesure la compressibilité rhétorique : le régime populiste — une seule voix qui attaque tout le monde, puis une chaîne — devrait être **le plus compressible** des trois (peu de formes de labeling distinctes), et le délibératif le moins. C'est une prédiction falsifiable, posée avant la mesure.


```python
def regime_complet(af, ordres):
    """Sigma aux deux encodages + K conjoint, sur le meme jeu d'ordres."""
    P_res, _, etats_res = regime_aggregated(af, ordres, encoding='summary')
    P_sig, _ = regime_sigma(af, ordres, encoding='signature')
    return {
        'sigma_resume': arg.discourse_irreversibility(P_res)['sigma'],
        'sigma_signature': P_sig,
        'K': mdl.two_part_code(etats_res, split=0.5)['total_bits'],
        'n_etats': len(etats_res),
    }


def regime_aggregated(af, ordres, encoding='summary'):
    toutes, transitions = [], []
    for ordre in ordres:
        etats, tr = etats_et_transitions(af, ordre, encoding=encoding)
        toutes.extend(etats)
        transitions.extend(tr)
    P, mapping = tpm_estimation.tpm_from_transitions(transitions, unseen='self')
    return P, mapping, toutes


resultats_agreges = {}
for n_bloc, af_specimen, nom_specimen in (
        (7, af_loup, 'Loup-Agneau'),
        (9, af_chien, 'Loup-Chien')):
    perms_bloc = list(itertools.permutations(range(n_bloc)))
    rng_bloc = np.random.default_rng(2024)
    idx_bloc = rng_bloc.choice(len(perms_bloc), size=M_ORDRES, replace=False)
    ordres_bloc = [list(perms_bloc[i]) for i in idx_bloc]

    print('=' * 78)
    print('n = %d | M = %d ordres | %d pas par ordre' % (n_bloc, M_ORDRES, n_bloc + 1))
    print('%-22s %14s %14s %12s %9s'
          % ('graphe', 'sigma(resume)', 'sigma(signat.)', 'K (bits)', 'n_etats'))
    print('-' * 78)
    bloc = {}
    for label, af in (('Fable ' + nom_specimen, af_specimen),
                      ('Deliberatif(%d)' % n_bloc, deliberative_af(n_bloc)),
                      ('Populiste(%d)' % n_bloc, populist_af(n_bloc))):
        r = regime_complet(af, ordres_bloc)
        bloc[label] = r
        print('%-22s %14.3e %14.3e %12.2f %9d'
              % (label, r['sigma_resume'], r['sigma_signature'], r['K'], r['n_etats']))
    resultats_agreges[n_bloc] = bloc
    print()
```

```python
# Detail qui explique les ecarts de K : combien d'etats DISTINCTS chaque regime produit.
# K en deux parties paie une declaration par etat distinct ; le total d'etats est egal par construction.
print('%-22s %10s %10s %10s' % ('graphe', 'n total', 'n distinct', 'K (bits)'))
print('-' * 56)
for n_bloc, bloc in resultats_agreges.items():
    perms_bloc = list(itertools.permutations(range(n_bloc)))
    rng_bloc = np.random.default_rng(2024)
    idx_bloc = rng_bloc.choice(len(perms_bloc), size=M_ORDRES, replace=False)
    ordres_bloc = [list(perms_bloc[i]) for i in idx_bloc]
    for label, af in ((list(bloc)[0], None),):
        pass
    for label in bloc:
        af = {'Fable Loup-Agneau': af_loup, 'Fable Loup-Chien': af_chien}.get(label)
        if af is None:
            af = deliberative_af(n_bloc) if label.startswith('Deliberatif') else populist_af(n_bloc)
        toutes = []
        for ordre in ordres_bloc:
            etats, _ = etats_et_transitions(af, ordre, encoding='summary')
            toutes.extend(etats)
        print('%-22s %10d %10d %10.2f'
              % (label, len(toutes), len(set(toutes)), bloc[label]['K']))
```

### K n'est pas dégénéré, et son ordre est stable

**Ce qui est acquis :** à taille égale, K **varie** entre les trois graphes dans chaque bloc, et **le même ordre apparaît dans les deux blocs** :

| Régime | K à `n = 7` | K à `n = 9` |
|---|---|---|
| Populiste | **1711.11** | **2548.72** |
| Délibératif | 1804.45 | 3104.87 |
| Fable | 2033.77 | 3510.66 |

Populiste < délibératif < fable, aux deux tailles. Contrairement à σ à tous les formats testés, **K n'est donc pas dégénéré** : il distingue des structures qui n'ont que leur nombre d'arguments en commun, et il les ordonne de la même façon quand `n` change. C'est la seule des trois jambes qui survive aux sections 2, 3 et 5.

**Ce qui est falsifié :** la prédiction posée en tête de section avait deux moitiés. « Le populiste sera le plus compressible » est **vérifiée** deux fois. « Le délibératif sera le moins compressible » est **fausse** deux fois : c'est la **fable** qui l'est. Un régime délibératif — symétrique, sans direction privilégiée — produit donc moins de formes de labeling qu'un récit où chaque réplique attaque la précédente.

**Ce qu'il faut retirer comme mécanisme :** le détail ci-dessus montre que le nombre d'états distincts **suit** en partie le K (la fable est la plus riche aux deux tailles : 17 puis 27). Mais il ne l'explique pas seul : à `n = 9`, délibératif et populiste produisent **22 états distincts chacun** et diffèrent pourtant de 556 bits. Le code en deux parties paie aussi la description de la **suite** sous le modèle ajusté, et c'est là que les deux régimes divergent. **K mesure la richesse en formes de labeling, pas une étiquette rhétorique** — et son ordre, stable, n'est pas celui qu'on avait prédit.

### Bilan des trois jambes, à ce point

| Jambe | Statut après mesures | Ce qu'elle peut porter |
|---|---|---|
| σ, ordre unique | **réfutée** — invariante au graphe (section 2) | rien de structurel |
| σ, signature agrégée | **réfutée** — invariante au graphe (section 3) | rien de structurel |
| σ, résumé agrégé | bruit flottant (`1e-8`), ordre instable | rien d'interprétable en l'état |
| K | **non dégénéré**, ordre stable entre les deux blocs | la richesse en formes de labeling |

La note de conception espérait que σ porte la thèse (l'irréversibilité du discours dominateur) et que K la nuance. La mesure donne l'inverse exact : **σ ne porte rien à ces formats, K porte une propriété structurelle réelle mais qui n'est pas celle qu'on avait nommée.**

Ce désaccord entre la note et la mesure n'est pas un échec du carnet — c'est son résultat. La suite le traite comme tel : elle teste la dispersion d'un nul apparié, puis les quatre lectures indépendantes, avant de conclure.


## 5. Le nul apparié, et ce qu'on fait quand sa dispersion est nulle

Un chiffre de σ ou de K ne vaut rien sans un nul : quelle valeur obtiendrait-on sur un graphe qui **conserve les degrés d'attaque** mais détruit la structure ? Le nul apparié est un double échange d'arêtes (`(a,b),(c,d) -> (a,d),(c,b)`), répété, qui garde le nombre d'attaques reçues et émises par chaque argument et casse les motifs.

La note de conception (§5.2.1) signale un piège précis : si l'écart-type du nul est nul, un z-score vaut `+infini`, et le rapporter comme « significatif » est une fabrication. **La dispersion se mesure, elle ne se suppose pas** — et si elle vaut zéro, le verdict correct est **non résolu**, pas significatif.

On tire 200 nuls par specimen, on garde σ et K de chacun, et on lit la distribution.


```python
def nul_degres_conserves(af, rng, n_echanges=None):
    """Nul apparié : doubles échanges d'arêtes, degrés entrants/sortants conservés.

    Ne garantit pas l'absence de doublons si le tirage est défavorable : c'est un nul,
    pas une énumération exhaustive de l'espace à degrés fixes.
    """
    arguments = list(af.arguments)
    aretes = [(a, b) for a, b in af.attacks if a != b]
    if n_echanges is None:
        n_echanges = 8 * max(1, len(aretes))
    aretes = aretes[:]
    for _ in range(n_echanges):
        i, j = rng.integers(0, len(aretes), size=2)
        if i == j:
            continue
        (a, b), (c, d) = aretes[i], aretes[j]
        if b == d or a == c:
            continue
        candidat = (a, d)
        if candidat in aretes or a == d:
            continue
        aretes[i], aretes[j] = (a, d), (c, b)
    return arg.DungAF(arguments, aretes)


# Plancher RELATIF d'ecart-type : en dessous, la dispersion mesure est de la
# poussiere de virgule flottante, pas une variabilite du nul. Un ecart-type nul
# rendrait un z infini ; un ecart-type a 1e-25 rendrait un z de 1.0 trompeur.
# Les deux cas doivent sortir NON RESOLU, pas un nombre.
PLANCHER_RELATIF = 1e-12

N_NULS = 200
rng_nul = np.random.default_rng(20261031)
nuls = {}
for nom, af, ordre in (('Loup-Agneau', af_loup, ordre_loup),
                       ('Loup-Chien', af_chien, ordre_chien)):
    sigmas_nul, ks_nul = [], []
    for _ in range(N_NULS):
        g = nul_degres_conserves(af, rng_nul)
        try:
            rr = triplet.assembler_triplet(g, ordre)
            sigmas_nul.append(rr['sigma']['irreversibilite']['sigma'])
            ks_nul.append(rr['K']['tpm_bits'])
        except Exception:
            pass
    s = np.array(sigmas_nul)
    k = np.array(ks_nul)
    nuls[nom] = (s, k)
    observe = triplets[nom]['sigma']['irreversibilite']['sigma']
    echelle = max(abs(observe), abs(float(s.mean())), 1.0)
    print('=' * 72)
    print('%s  |  %d nuls retenus sur %d' % (nom, len(s), N_NULS))
    print('  sigma observe      = %.17g' % observe)
    print('  sigma nul : moyenne %.17g' % s.mean())
    print('              ecart-type %.3e  |  min %.17g  |  max %.17g'
          % (s.std(), s.min(), s.max()))
    print('  K nul     : moyenne %.9f  |  ecart-type %.3e  |  min %.9f  |  max %.9f'
          % (k.mean(), k.std(), k.min(), k.max()))
    if s.std() <= PLANCHER_RELATIF * echelle:
        print('  verdict sigma : NON RESOLU -- dispersion sous le plancher relatif (%.0e) ;'
              % PLANCHER_RELATIF)
        print('                  aucun z-score definissable, l\'ecart observe est dans le bruit')
    else:
        print('  verdict sigma : z = %.3f' % ((observe - s.mean()) / s.std()))
    print()
```

### Dispersion nulle : le verdict est « non résolu », et il n'y a pas d'autre honnêteté

Sur le specimen 1, l'écart-type des σ du nul vaut **exactement `0.0`**. Sur le specimen 2, il vaut `8.3e-25` — une poussière de virgule flottante, pas une dispersion. Le K du nul a un écart-type de `3.6e-15` et `7.1e-15` : même chose, un ordre de grandeur sous tout ce qui serait interprétable.

**Ces deux cas doivent produire le même verdict**, et c'est pourquoi le code applique un **plancher relatif** au lieu de tester `sd == 0`. Sans lui, le specimen 2 aurait rendu un `z = 1.0` parfaitement trompeur : l'écart observé y vaut exactement un écart-type, parce que les deux quantités sont la même poussière. Un z de 1.0 calculé sur `1e-25` n'est pas une mesure, c'est un artefact de représentation — et il aurait l'air d'un résultat.

Avec le plancher, les deux specimens sortent **`NON RESOLU`** : l'écart observé est dans le bruit, aucun z-score n'est définissable.

**Ce que cela ne veut pas dire.** Ce n'est pas « le specimen est significativement différent du nul », ni « le specimen est identique au nul ». C'est que **le nul ne se distingue pas du specimen par construction** : le double échange d'arêtes conserve les degrés, et à ordre narratif unique, on a vu en section 2 que σ et K ne dépendent que du nombre de pas et du nombre d'états distincts. Or ces deux quantités sont invariantes sous échange d'arêtes. Le nul rend donc la même valeur que l'observé, **toujours**.

**Conséquence méthodologique, et c'est le point que la note avait vu venir :** rapporter un z-score ici serait une fabrication. Le code refuse explicitement de le faire. Un notebook qui écrirait « z = infini, donc significatif » mentirait sur ce qu'il a mesuré — et il mentirait d'autant plus facilement que le chiffre aurait l'air d'un résultat.

**Ce que le nul apprend quand même, et qui n'est pas rien :** il confirme, par une voie indépendante, la conclusion de la section 2. Un instrument dont le nul contrôlé par degrés est indiscernable de l'observé ne mesure pas la structure d'attaque — il mesure sa propre construction. Les sections 2, 3 et 5 convergent donc sur le même constat, par trois protocoles différents.

**La voie qui reste** est celle que la note appelle les **quatre lectures** : ne pas demander au triplet de porter seul un verdict, mais faire lire la même trajectoire par quatre appareils indépendants et regarder où ils divergent. C'est l'objet de la section suivante — et c'est aussi le quatrième témoin négatif du §5.4, dont le résultat ne sera pas celui qu'on attend.


## 6. Les quatre lectures et le collage hors échantillon

Protocole #13041 (carnet `Argumentation-Obs-04-Recollement-Strate6-Python`) : faire lire **la même** situation par quatre appareils hétérogènes, puis mesurer ce qu'un **collage** apporte par rapport au meilleur spécialiste.

Les quatre appareils, ici, sont quatre façons de prédire ce que la sémantique grounded retient :

| Lecture | Appareil | Règle |
|---|---|---|
| `STRUCT` | structurel | un énoncé que personne n'attaque tient |
| `GRAPH` | propagation | un énoncé tient si aucun de ses attaquants n'est lui-même incontesté |
| `RDF` | rôle AIF | une thèse tenue est un énoncé qui attaque sans être contesté |
| `LEX` | surface lexicale | les marqueurs de négation (`ne`, `n'`, `pas`, `point`) annoncent un rejet |

**La cible est le labeling grounded** — la sémantique formelle, pas une annotation humaine : sur ces deux fables, personne n'a annoté argument par argument. Les quatre lectures sont donc des **approximations bon marché** de cette sémantique, et ce qui nous intéresse est **où elles se contredisent**.

Le collage est un vote majoritaire pondéré par la précision **hors échantillon** (validation croisée laissant un énoncé de côté) : chaque lecture pèse ce qu'elle a prouvé sur les autres énoncés. C'est la définition de `GLU` dans Obs-04.

**Le témoin négatif §5.4** exige que sur la **zone disputée** — les énoncés où les lectures ne sont pas d'accord — le collage batte le meilleur spécialiste. Sinon, le collage n'apporte rien là où il servirait.


```python
MARQUEURS_NEGATION = ('ne ', "n'", ' pas ', 'point', "n'est", "n'avait", "n'en", "n'etais", 'nullement')


def degre_entrant(af, i):
    return sum(1 for a, b in af.attacks if b == i)


def degre_sortant(af, i):
    return sum(1 for a, b in af.attacks if a == i)


def lecture_struct(af, i):
    return 1 if degre_entrant(af, i) == 0 else 0


def lecture_graph(af, i):
    attaquants = [a for a, b in af.attacks if b == i]
    if not attaquants:
        return 1
    return 0 if any(degre_entrant(af, a) == 0 for a in attaquants) else 1


def lecture_rdf(af, i):
    return 1 if (degre_sortant(af, i) > 0 and degre_entrant(af, i) == 0) else 0


def lecture_lex(texte):
    minuscule = texte.lower()
    return 0 if any(m in minuscule for m in MARQUEURS_NEGATION) else 1


def quatre_lectures(af, enonces):
    return {
        'STRUCT': [lecture_struct(af, i) for i in range(len(enonces))],
        'GRAPH': [lecture_graph(af, i) for i in range(len(enonces))],
        'RDF': [lecture_rdf(af, i) for i in range(len(enonces))],
        'LEX': [lecture_lex(t) for _, t in enonces],
    }


def collage_loo(lectures, oracle):
    """Vote pondere par la precision hors echantillon (une observation laissee de cote)."""
    noms = list(lectures)
    n = len(oracle)
    predictions = []
    for i in range(n):
        poids = {}
        for nom in noms:
            autres = [j for j in range(n) if j != i]
            acc = float(np.mean([lectures[nom][j] == oracle[j] for j in autres])) if autres else 0.0
            poids[nom] = max(acc, 1e-6)
        score = sum(poids[nom] * (1 if lectures[nom][i] else -1) for nom in noms)
        predictions.append(1 if score > 0 else 0)
    return predictions


def exactitude(pred, oracle, indices=None):
    idx = range(len(oracle)) if indices is None else indices
    idx = list(idx)
    if not idx:
        return float('nan')
    return float(np.mean([pred[i] == oracle[i] for i in idx]))


lectures_par_specimen = {}
for nom, af, enonces in (('Loup-Agneau', af_loup, ENONCES_LOUP),
                         ('Loup-Chien', af_chien, ENONCES_CHIEN)):
    lab = arg.grounded_labeling(af)
    oracle = [1 if lab.get(i) == 'in' else 0 for i in range(len(enonces))]
    lectures = quatre_lectures(af, enonces)
    glue = collage_loo(lectures, oracle)
    n = len(oracle)

    unanimite = [len({lectures[k][i] for k in lectures}) == 1 for i in range(n)]
    dispute = [i for i in range(n) if not unanimite[i]]

    print('=' * 74)
    print('%s | n = %d | grounded retient %d enonces' % (nom, n, sum(oracle)))
    print('  oracle : %s' % oracle)
    for k, v in lectures.items():
        print('  %-7s acc = %.3f  %s' % (k, exactitude(v, oracle), v))
    acc_glue = exactitude(glue, oracle)
    print('  %-7s acc = %.3f  %s' % ('GLU', acc_glue, glue))
    print('  zone disputee : %s  (%d/%d)' % (dispute, len(dispute), n))
    if dispute:
        meilleur = max(lectures, key=lambda k: exactitude(lectures[k], oracle))
        print('  sur la zone disputee : meilleur specialiste (%s) = %.3f | collage = %.3f'
              % (meilleur, exactitude(lectures[meilleur], oracle, dispute),
                 exactitude(glue, oracle, dispute)))
        if exactitude(glue, oracle, dispute) > exactitude(lectures[meilleur], oracle, dispute):
            print('  TEMOIN 4 : OK -- le collage bat le meilleur specialiste sur la zone disputee')
        else:
            print('  TEMOIN 4 : ROUGE -- le collage n\'apporte rien sur la zone disputee')
    else:
        print('  TEMOIN 4 : non concluant -- aucune zone disputee')
    lectures_par_specimen[nom] = (oracle, lectures, glue, dispute)
    print()
```

### Le témoin 4 est rouge — et c'est un défaut de prémisse, pas un verdict sur le collage

**Les chiffres, tels qu'ils sont sortis.**

| | Loup-Agneau (n=7) | Loup-Chien (n=9) |
|---|---|---|
| `STRUCT` | 0.571 | 0.889 |
| `GRAPH` | **1.000** | **1.000** |
| `RDF` | 0.571 | 0.667 |
| `LEX` | 0.857 | **0.333** |
| `GLU` | 0.857 | 0.889 |
| Zone disputée | 3/7 | 7/9 |
| Meilleur spécialiste sur la zone | `GRAPH`, 1.000 | `GRAPH`, 1.000 |
| Collage sur la zone | 0.667 | 0.857 |

Sur les deux specimens, **le collage perd sur la zone disputée**. Le témoin §5.4 est donc **rouge**, et le carnet doit le rapporter ainsi plutôt que de chercher une variante qui le verdirait.

**Pourquoi il est rouge — et ce n'est pas la faute du collage.** La lecture `GRAPH` est une propagation à deux niveaux sur le graphe d'attaque ; la sémantique grounded, sur un AF de profondeur 2 à 3, fait exactement la même chose. `GRAPH` **coïncide donc avec la cible par construction** sur ces deux graphes. Or un collage ne peut battre un spécialiste qui vaut déjà 1.000 : il n'y a pas de marge, la comparaison est vide avant d'avoir tourné. Ce que le témoin mesure ici, c'est que **ces deux fables ne sont pas un terrain d'essai pour le collage** — pas que le collage n'apporte rien.

C'est exactement la leçon de la note (§5.4, quatrième témoin) : il faut un corpus où la meilleure lecture spécialiste est **en dessous de la cible**, sinon le contrôle est satisfait par construction. Le protocole Obs-04, lui, dispose de 344 entrées annotées et d'un désaccord réel entre lectures ; deux fables n'en sont pas un substitut.

**Deux observations secondaires, qui sont des résultats à part entière :**

1. **`STRUCT` et `RDF` rendent la même prédiction sur les deux specimens.** Les deux appareils sont donc **redondants** ici : `RDF` est une règle plus stricte que `STRUCT` sur ces graphes et ne capture aucun énoncé supplémentaire. Un panel de quatre lectures qui en contient deux identiques n'est pas un panel de quatre.

2. **`LEX` s'inverse d'un specimen à l'autre** : 0.857 sur la fable 1, 0.333 sur la fable 2 — pire que le hasard. La raison est lisible dans les textes : dans le Loup et l'Agneau, la négation marque la réfutation (`je ne puis troubler`, `je n'en ai point`) ; dans le Loup et le Chien, elle marque au contraire les énoncés **qui l'emportent** (`Un Loup n'avait que les os`, `vous ne courez donc pas où vous voulez`, `je ne veux en aucune sorte`). **Un marqueur de surface ne porte pas de polarité stable** : il porte ce que le texte en fait.

**Conclusion de section :** le protocole des quatre lectures est applicable et il a produit ses chiffres, mais il ne peut pas conclure sur ces deux specimens. Le témoin 4 reste ouvert, avec sa prémisse nommée.


## 7. Les quatre témoins négatifs du §5.4

La note (§5.4) impose quatre contrôles négatifs **avant** tout cas historique. Un instrument qui ne sait pas refuser de mesurer produira un nombre sur n'importe quoi, et ce nombre aura l'air d'un résultat.

**Témoin 1 — Hynkel (canaux contradictoires).** Un matériau dont le canal principal est incodable et dont l'autre canal se présente comme sa traduction ne se mesure pas : la mesure lirait la traduction. Le verdict attendu n'est pas un seuil, c'est un **type de sortie** : `REFUS:CANAUX_CONTRADICTOIRES`.

**Témoin 2 — Chaplin (même structure, signe inverse).** Le discours final du *Dictateur* est, à l'inverse, du langage ordinaire parfaitement codable : le même verdict doit rendre `MESURABLE`. Le couple (1, 2) montre que le refus dépend du **matériau**, pas de l'appareil. Le texte n'est **jamais** recopié ici : il est sous droits (Roy Export S.A.S.), il se récupère à l'exécution, et s'il est inaccessible le carnet déclare `SOURCE_INDISPONIBLE` plutôt que de fabriquer un extrait.

**Témoin 3 — graphes nuls.** Fait en section 5 : dispersion nulle ⇒ verdict « non résolu ». Le témoin exige les deux — un nul à σ exactement 0 **et** un nul avec dispersion ; sans le second, on ne saurait pas distinguer « pas de signal » de « instrument aveugle ».

**Témoin 4 — collage.** Fait en section 6 : le collage doit battre le meilleur spécialiste **sur la zone disputée**.

## 9. Conclusion — ce que ce carnet établit, et ce qu'il laisse ouvert

**Ce qui est établi par la mesure, dans ce carnet :**

1. À **ordre narratif unique**, le triplet σ/K est **invariant au graphe d'attaque** : cinq graphes structurellement opposés — anneau, hub, graphe vide, arête unique, fable — rendent le même σ et le même K. La jambe σ y dépend du nombre de pas, la jambe K du nombre d'états ; aucune ne porte la structure.
2. Au **niveau signature agrégé**, σ est **exactement invariante** sur neuf graphes à trois cents ordres, dont le graphe vide et quatre graphes aléatoires. L'encodage le plus fin est le plus aveugle, l'inverse de ce que le carnet de graine supposait.
3. Le **nul apparié par degrés** a une dispersion **sous le plancher relatif** sur les deux specimens : aucun z-score n'est définissable, et le seul verdict honnête est « non résolu ».
4. La jambe **K** est la seule **non dégénérée** à taille égale, et son **ordre est stable entre les deux blocs** (populiste < délibératif < fable). Il contredit pourtant la prédiction dans sa seconde moitié : ce n'est pas le délibératif qui est le moins compressible, c'est la fable. K mesure la **richesse en formes de labeling**, pas l'irréversibilité d'un discours.
5. Le **témoin 4** (§5.4) est **rouge sur les deux specimens**, et le carnet nomme la raison : la lecture `GRAPH` coïncide avec la sémantique cible sur des AF aussi peu profonds, donc la comparaison n'a pas de marge. C'est un défaut de **prémisse**, pas un verdict sur le collage.
6. Deux des quatre lectures sont **redondantes** sur ces specimens (`STRUCT` et `RDF` prédisent la même chose), et la lecture lexicale **s'inverse** d'une fable à l'autre : un marqueur de surface ne porte pas de polarité stable.
