Deux auteurs, une seule question, et un désaccord assez précis pour se mesurer.

En 1952, Turing publie *The Chemical Basis of Morphogenesis* : des substances qui réagissent et diffusent peuvent, à elles seules, faire basculer un embryon homogène en motif, et la longueur d'onde de ce motif est **chimique**. En 1991, Thom répond, dans l'*Esquisse d'une sémiophysique*, que cette famille d'explications rate l'essentiel : « la vie est essentiellement canalisation, endiguement et lutte contre la diffusion ».

Ce carnet ne résume pas les deux textes côte à côte. Il prend **une phrase précise de chacun**, et il construit un banc qui mesure laquelle des deux décrit ce qui se passe quand une géométrie est imposée à une instabilité chimique. Le verdict est en deux morceaux, et aucun des deux n'est un acquittement : la chimie fournit la **densité** des lobes, le canal fournit la **grille** des modes possibles, et aucun des deux ne suffit seul à produire la forme.

## 1. Ce que Turing affirme (1952)

Le programme est énoncé sans détour : « The purpose of this paper is to discuss a possible mechanism by which the genes of a zygote may determine the anatomical structure of the resulting organism. » (§1, p. 37.) Puis vient la phrase que tout modèle doit garder en tête : « This model will be a simplification and an idealization, and consequently a falsification. » (p. 38.)

Le mécanisme, en une ligne : un état homogène devient instable, et une perturbation aléatoire est amplifiée sélectivement en un motif stationnaire. Trois ingrédients suffisent — deux substances (un activateur à diffusion courte, un inhibiteur à diffusion longue), une cinétique non linéaire, et une **asymétrie des coefficients de diffusion**.

Turing ne prétend pas que la forme soit dans les gènes ; il prétend que les gènes règlent les constantes et que les lois physiques font le reste. C'est exactement ce que Thom contestera.

## 2. Turing lui-même : la longueur d'onde est chimique, mais avec une marge

Le §8 est le passage qui rend le désaccord mesurable. Turing y traite le cas d'un anneau de cellules et écrit (p. 51) :

> « The wave-lengths of the patterns on the ring do not depend only on the chemical data a, b, c, d, μ', ν' but on the circumference of the ring, since they must be submultiples of the latter. »

Et il ajoute la nuance qui porte tout ce carnet :

> « One may picture the situation by supposing that the chemical wave-length is true wave-length which is achieved whenever possible, but that on a ring it is necessary to 'make do' with an approximation which divides exactly into the circumference. »

La « longueur d'onde chimique » n'est donc **pas** une prescription : c'est une valeur de référence, atteinte quand le contenant le permet, approchée sinon. Turing chiffre même la largeur de l'approximation, dans le sous-cas (d) — celui qu'il désigne comme « the case which is of greatest interest, and has most biological application » (p. 52) :

> « If, however, the radius is chosen so that [the chemical wave-length] cannot hold with an integral s, then (in this example) the actual number of lobes will be one of the two integers nearest to the (non-integral) solutions of this equation, and usually the nearest. Examples can, however, be constructed where this simple rule does not apply. » (p. 53.)

*(L'exposant parasite que le scan place après « two » a été retiré ; le reste est verbatim.)*

C'est une prédiction **falsifiable** : le nombre de lobes réalisé doit tomber dans un ensemble de deux entiers, construit à partir de la solution non entière `s_c = L · k_c / π`. Le §8 enchaîne sur six sous-cas — trois régimes de longueur d'onde croisés avec deux comportements temporels :

| Sous-cas | Régime | Description de Turing |
|---|---|---|
| (a) | stationnaire, extrêmement longue | toutes les cellules en phase, comme isolées |
| (b) | oscillatoire, extrêmement longue | idem, mais la sortie d'équilibre oscille |
| (c) | stationnaire, extrêmement courte | chaque cellule ressemble à sa voisine de rang 2 (N pair) |
| (d) | stationnaire, longueur finie | « greatest interest, most biological application » |
| (e) | oscillatoire, longueur finie | ondes progressives, exige trois morphogènes |
| (f) | métabolique | cellules voisines en opposition de phase |

Deux des six sous-cas exigent trois morphogènes ou plus : le modèle à deux substances de ce carnet ne peut donc pas tous les produire, et nous le dirons plutôt que de le maquiller.

## 3. Ce que Thom objecte (1991)

Dans l'*Esquisse d'une sémiophysique*, au chapitre 3, Thom vise nommément cette famille de modèles (p. 60 de la copie archivée) :

> « Beaucoup de théoriciens modernes — à la suite de Turing — ont évoqué les équations de réaction-diffusion de la forme dx/dt = F(x) + k Δx pour expliquer la morphogénèse. C'est oublier que la vie est essentiellement canalisation, endiguement et lutte contre la diffusion. »

Ce n'est pas une dénégation technique : Thom ne dit pas que ces équations n'ont pas de solutions. Il dit qu'elles sont **le mauvais facteur causal**. Pour lui la forme vient d'une géométrie de contraintes déjà là — le canal, la paroi, l'endiguement — et la dynamique ne fait que remplir un contenant. Son exemple n'est pas emprunté à l'embryologie mais à la coagulation : un liquide déjà canalisé qui bouche lui-même les lacunes accidentelles de sa paroi.

Deux thèses, donc, et elles sont symétriques :

| | Turing 1952 | Thom 1991 |
|---|---|---|
| Ce qui produit la forme | l'instabilité chimique | le canal |
| Statut de la diffusion | moteur, si les coefficients diffèrent | ennemi à combattre |
| Statut de la géométrie | choisit parmi les longueurs d'onde admissibles | produit la forme |
| Ce que la mesure peut trancher | beaucoup | la priorité explicative : non |

## 4. Le point exact à trancher, et ce qu'un banc 1-D peut en décider

Le désaccord n'est pas métaphysique : il porte sur **qui fixe la forme quand les deux facteurs sont présents**. Un banc unidimensionnel peut décider quatre choses, et il ne peut pas décider la cinquième.

| Question | Décidable ici ? | Où |
|---|---|---|
| La chimie produit-elle un motif sans aucun canal ? | oui | §5, §7 |
| Le canal suffit-il, sans instabilité chimique ? | oui | §7 |
| Le canal fixe-t-il le mode exact, ou une fenêtre de modes ? | oui | §8 |
| Le seuil est-il une catastrophe au sens de Thom (exposant universel) ? | oui, en codimension 1 | §6 |
| Laquelle des deux priorités est la bonne explication du vivant ? | **non** | — |

Le modèle est celui de Schnakenberg (`u` s'auto-active, `v` l'inhibe), pas le système linéaire à deux morphogènes du §6 de Turing ; il en porte la même instabilité, avec des coefficients choisis pour que le seuil tombe dans une plage confortable. Tout est en 1-D, sans mécanique, sans croissance, sans mouvement cellulaire — trois absences qui pèsent lourd, et que le §10 nomme.

```python
# --- Modele de Schnakenberg : u s'auto-active, v l'inhibe.
# Deux especes suffisent a la condition de Turing ; ce n'est PAS le systeme
# lineaire a deux morphogenes de 1952, mais il en porte la meme instabilite.
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import dct

A, B = 0.1, 0.9
DU = 1.0
U0, V0 = A + B, B / (A + B) ** 2

# Jacobien au point fixe homogene : fu, fv, gu, gv.
FU = -1.0 + 2.0 * U0 * V0
FV = U0 ** 2
GU = -2.0 * U0 * V0
GV = -U0 ** 2
print(f"point fixe   u0 = {U0:.4f}   v0 = {V0:.4f}")
print(f"jacobien     fu = {FU:+.2f}  fv = {FV:+.2f}  gu = {GU:+.2f}  gv = {GV:+.2f}")


def sigma(k, dv=40.0):
    """Taux de croissance du mode de nombre d'onde k : partie reelle de la plus
    grande valeur propre de [[fu - Du k^2, fv], [gu, gv - Dv k^2]]."""
    tr = FU + GV - (DU + dv) * k ** 2
    det = (FU - DU * k ** 2) * (GV - dv * k ** 2) - FV * GU
    disc = tr * tr - 4.0 * det
    if disc >= 0.0:
        return 0.5 * (tr + np.sqrt(disc))
    return 0.5 * tr


DV = 40.0
ks = np.linspace(1e-6, 1.5, 200001)
sig = np.array([sigma(k, DV) for k in ks])
KC = float(ks[int(np.argmax(sig))])
LAMBDA_C = float(2.0 * np.pi / KC)
SIGMA_MAX = float(sig.max())
instables = ks[sig > 0.0]
LAM_MIN = float(2.0 * np.pi / instables.max())
LAM_MAX = float(2.0 * np.pi / instables.min())

print()
print(f"Dv = {DV:.1f}  ->  kc = {KC:.5f}   lambda_c = {LAMBDA_C:.4f}   sigma_max = {SIGMA_MAX:+.5f}")
print(f"bande instable   k in [{instables.min():.5f}, {instables.max():.5f}]")
print(f"                 lambda in [{LAM_MIN:.3f}, {LAM_MAX:.3f}]   soit un facteur {LAM_MAX / LAM_MIN:.2f}")

fig, ax = plt.subplots(figsize=(7.6, 3.6))
ax.plot(ks, sig, color="C0", lw=2)
ax.axhline(0.0, color="0.4", lw=0.8)
ax.axvspan(instables.min(), instables.max(), color="C1", alpha=0.18,
           label="bande instable (sigma > 0)")
ax.plot([KC], [SIGMA_MAX], "o", color="C3", ms=7)
ax.annotate(f"kc = {KC:.4f}\nsigma = {SIGMA_MAX:+.4f}", (KC, SIGMA_MAX),
            textcoords="offset points", xytext=(14, 6), fontsize=9, color="C3")
ax.set_xlabel("nombre d'onde k")
ax.set_ylabel("sigma(k)")
ax.set_title("Dispersion lineaire : la chimie designe une BANDE, pas une longueur d'onde")
ax.legend(loc="lower left", fontsize=9)
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

**Lecture.** La dispersion dit deux choses, et la seconde est celle qui compte pour la suite.

La première est la condition de Turing : `σ(k) > 0` n'existe que parce que `Dv = 40` dépasse largement `Du = 1`. Sans cette asymétrie, toutes les valeurs propres restent négatives et rien ne pousse — le §7 le mesurera.

La seconde est que la chimie ne désigne **pas** une longueur d'onde : elle désigne une bande. Ici `σ > 0` pour `k` de 0,184 à 0,861, soit des longueurs d'onde de **7,3 à 34,2 unités de grille** — un facteur 4,7. Le maximum `k_c ≈ 0,4205` (`λ_c ≈ 14,94`) n'est pas « la » longueur d'onde : c'est le lieu du taux de croissance maximal, celui que Turing appelle la longueur d'onde chimique et qu'il définit lui-même comme la limite quand l'anneau grandit indéfiniment.

Cette largeur est ce qui rend la dispute mesurable : si la bande est large, le mode exact réalisé ne peut pas être déduit de la seule croissance linéaire, et un autre facteur doit trancher. Turing nomme ce facteur — la circonférence — et Thom nomme le sien — le canal. Dans ce banc, les deux désignent le même objet géométrique.

## 5. Mesure 1 — l'instabilité produit un motif, sans aucun canal

Premier contrôle : un anneau libre de `N = 400` cellules, sans paroi, sans géométrie imposée. On part du bruit (0,01 d'amplitude) et on laisse l'instabilité choisir.

La prédiction de Turing pour ce cas est un entier : le nombre de lobes `m` doit être le plus proche de `N · k_c / 2π`. Attention à ce que cela implique — un anneau est **déjà** un contenant périodique, donc la quantification n'y est jamais abolie. Il n'existe pas, en 1-D fermée, de cas « sans grille » : c'est une limite du banc, pas de la théorie.

```python
# Outils de mesure : Laplacien 1-D cyclique (lien = 0 pose une paroi de Neumann,
# aucun flux), Euler explicite seme par un bruit de 0.01, et lecture du mode
# dominant par FFT sur une fenetre de Hanning.
def laplacien(champ, lien):
    """Laplacien 1-D cyclique ; lien[i] = 1 autorise le flux entre i et i+1,
    lien[i] = 0 pose une paroi (condition de Neumann, aucun flux)."""
    droite = np.roll(champ, -1) - champ
    gauche = champ - np.roll(champ, 1)
    return lien * droite - np.roll(lien, 1) * gauche


def simuler(n_cellules, dv, paroi_tous_les=None, pas=20000, dt=0.005, graine=0):
    """Euler explicite. Le bruit initial de 0,01 est la graine : l'instabilite
    a besoin d'une perturbation, exactement la 'random disturbance' de l'abstrait."""
    rng = np.random.default_rng(graine)
    u = np.full(n_cellules, U0) + 0.01 * rng.standard_normal(n_cellules)
    v = np.full(n_cellules, V0) + 0.01 * rng.standard_normal(n_cellules)
    lien = np.ones(n_cellules)
    if paroi_tous_les:
        lien[::paroi_tous_les] = 0.0
    for _ in range(pas):
        uv2 = u * u * v
        u += dt * (A - u + uv2 + DU * laplacien(u, lien))
        v += dt * (B - uv2 + dv * laplacien(v, lien))
    return u, v, lien


def mode_dominant(segment):
    """Rang du mode dominant, en nombre de lobes sur le segment."""
    x = segment - segment.mean()
    spectre = np.abs(np.fft.rfft(x * np.hanning(len(x))))
    return int(np.argmax(spectre[1:]) + 1)


# Mesure 1 : anneau libre de N = 400 cellules, sans paroi. Le mode realise est
# compare a m* = N kc / 2 pi, l'entier le plus proche que predit Turing 1952
# (section 8) ; l'ecart de longueur d'onde a lambda_c est imprime en dessous.
N = 400
u_anneau, _, _ = simuler(N, DV)
M_MES = mode_dominant(u_anneau)
M_PRED = N * KC / (2.0 * np.pi)
print(f"anneau libre, N = {N} cellules")
print(f"  prediction  m* = N kc / 2 pi = {M_PRED:.3f}  -> entier le plus proche {round(M_PRED)}")
print(f"  mesure      m  = {M_MES} lobes   k = {2 * np.pi * M_MES / N:.5f}   lambda = {N / M_MES:.4f}")
print(f"  ecart de longueur d'onde vs lambda_c : {100 * (N / M_MES / LAMBDA_C - 1):+.2f} %")
```

**Lecture.** La prédiction tombe : `N · k_c / 2π = 26,77`, et le mode réalisé a **27 lobes**. La longueur d'onde mesurée (14,81 unités de grille) est à −0,84 % de `λ_c`. À cette résolution, la règle de l'entier le plus proche est vérifiée.

Ce que cette mesure ne dit **pas** : elle ne montre pas que la chimie impose son mode exact, puisqu'elle ne pouvait pas en choisir un autre — sur un anneau de 400 unités de grille les modes sont des entiers, et le plus proche gagne parce que la dispersion y est maximale. Pour trancher entre « la chimie décide » et « la géométrie décide », il faut une géométrie qui contredise la chimie, c'est-à-dire une paroi.

## 6. Mesure 2 — le seuil est une catastrophe, et son exposant ne dépend pas de la chimie

C'est le point où Thom est le plus solide, et il ne parle pas de biologie : il parle de **classification des transitions**. Une forme change qualitativement en traversant une valeur critique du paramètre, et au voisinage de cette valeur l'amplitude de la nouvelle forme obéit à une loi de puissance dont l'exposant ne dépend que de la codimension de la catastrophe, pas des détails du modèle. Pour une fourche (codimension 1), l'exposant est 1/2.

Le banc le vérifie sans peine : on balaie `Dv` au-dessus du seuil `Dv_c`, on mesure l'amplitude de saturation, et on lit la pente en log-log. Si l'exposant vaut 1/2, alors le seuil de Turing **est** une catastrophe au sens de Thom — et Thom a raison sur ce point précis, contre les lectures qui réduisent sa théorie à une métaphore.

Le nombre de pas est adapté à chaque point (`10 / σ_max / dt`) : on ne compare que des états saturés, jamais une transition en cours.

```python
def sigma_max(dv, n=4000):
    """Maximum de sigma sur les k, et l'abscisse ou il est atteint."""
    kk = np.linspace(1e-6, 1.5, n)
    ss = np.array([sigma(k, dv) for k in kk])
    return float(ss.max()), float(kk[int(np.argmax(ss))])


# Seuil : dichotomie sur Dv. Sur [1, 60], sigma_max est croissante en Dv.
lo, hi = 1.0, 60.0
for _ in range(40):
    milieu = 0.5 * (lo + hi)
    if sigma_max(milieu)[0] > 0.0:
        hi = milieu
    else:
        lo = milieu
DV_C = hi
KC_SEUIL = sigma_max(DV_C)[1]
print(f"Dv critique = {DV_C:.5f}   k au seuil = {KC_SEUIL:.5f}   lambda = {2 * np.pi / KC_SEUIL:.3f}")
print(f"avec Dv = {DV:.1f} on est a eps = {DV / DV_C - 1:.2f} au-dessus du seuil")


def amplitude_saturee(n, dv, dt=0.01, graine=1):
    """Amplitude = ecart-type de u une fois le motif installe."""
    rng = np.random.default_rng(graine)
    u = U0 + 0.02 * rng.standard_normal(n)
    v = V0 + 0.02 * rng.standard_normal(n)
    lien = np.ones(n)
    s, _ = sigma_max(dv, n=20000)
    pas = int(min(400000, max(20000, 10.0 / max(s, 1e-6) / dt)))
    for _ in range(pas):
        uv2 = u * u * v
        u += dt * (A - u + uv2 + DU * laplacien(u, lien))
        v += dt * (B - uv2 + dv * laplacien(v, lien))
    return float(u.std()), pas


EPS = [0.05, 0.10, 0.20, 0.40, 0.80, 1.60]
amps = []
for eps in EPS:
    a, p = amplitude_saturee(400, DV_C * (1.0 + eps))
    amps.append(a)
    print(f"eps = {eps:.2f}  Dv = {DV_C * (1.0 + eps):6.3f}  pas = {p:6d}  amplitude = {a:.6f}")

log_eps = np.log(np.array(EPS))
log_amp = np.log(np.array(amps))
pente_globale = float(np.polyfit(log_eps, log_amp, 1)[0])
pente_proche = float(np.polyfit(log_eps[:3], log_amp[:3], 1)[0])
# Stabilite de l'exposant : deux segments independants de la fenetre asymptotique.
pente_seg1 = float(np.polyfit(log_eps[0:2], log_amp[0:2], 1)[0])
pente_seg2 = float(np.polyfit(log_eps[1:3], log_amp[1:3], 1)[0])
print()
print(f"pente log-log, toute la plage 0.05-1.60      : {pente_globale:.4f}")
print(f"pente log-log, fenetre asymptotique 0.05-0.20: {pente_proche:.4f}   (attendu 0.5)")
print(f"pente des deux segments 0.05-0.10 / 0.10-0.20: {pente_seg1:.4f} / {pente_seg2:.4f}")

fig, ax = plt.subplots(figsize=(6.6, 3.9))
ax.loglog(EPS, amps, "o", color="C0", ms=7, label="amplitude mesuree")
xx = np.linspace(min(EPS), max(EPS), 50)
ax.loglog(xx, amps[0] * (xx / EPS[0]) ** 0.5, "--", color="C3", lw=1.6,
          label="pente 1/2 (fourche)")
ax.loglog(xx, amps[0] * (xx / EPS[0]) ** pente_proche, ":", color="C2", lw=1.6,
          label=f"pente mesuree {pente_proche:.3f}")
ax.set_xlabel("eps = Dv / Dv_c - 1")
ax.set_ylabel("amplitude de u")
ax.set_title("L'onset est une bifurcation fourche : A proportionnel a eps^(1/2)")
ax.legend(fontsize=9)
ax.grid(alpha=0.25, which="both")
plt.tight_layout()
plt.show()
```

**Lecture.** Deux régimes, deux chiffres, et il faut les donner tous les deux.

Dans la fenêtre asymptotique (`ε` de 0,05 à 0,20) la pente mesurée est **0,466**, et elle est stable : les deux segments indépendants donnent 0,4662 et 0,4658. Sur toute la plage, jusqu'à `ε = 1,6`, la pente chute à **0,404** — ce n'est pas un désaccord avec la théorie, c'est la saturation, le terme cubique de la forme normale qui redresse la courbe loin du seuil. L'écart de 7 % à 0,5 dans la fenêtre proche est lui aussi attendu, et de signe prévisible : à `ε = 0,05` on n'est pas infinitésimal.

Ce que cela établit, et qui vaut contre les deux camps : **la nature du seuil est universelle**. Rien dans l'exposant ne dépend de Schnakenberg, des valeurs de `A` et `B`, du nombre de morphogènes. Thom a raison de dire qu'une morphogenèse se décrit comme une traversée de catastrophe ; Turing a raison de dire que l'instabilité linéaire est ce qui fait traverser. Ce sont deux descriptions du même seuil, pas deux mécanismes concurrents.

Ce que cela n'établit pas : le banc ne balaie qu'**un** paramètre. Il ne peut donc montrer qu'une codimension 1, et ne dit rien de la classification plus riche que Thom déploie — plis, fronces, ombilics.

## 7. Mesure 3 — le témoin croisé : le canal suffit-il ?

Voici le test qui tranche la lecture littérale de Thom. Quatre cellules, deux paramètres croisés : l'asymétrie de diffusion (présente ou abolie) et la présence de parois (libre ou paroi tous les 60).

L'hypothèse nulle est posée **avant** la mesure, et elle est double :

- si la canalisation est le moteur, alors `Dv = 1` avec parois doit produire un motif ;
- si la chimie est seule responsable, alors `Dv = 40` avec parois doit donner exactement le même motif que sans parois.

```python
# Mesure 3 : quatre conditions croisees, Dv = 1 ou 40 selon la ligne, libre ou
# parois tous les 60 cellules. Amplitude = ecart-type de u apres 60000 pas ; le
# seuil de motif a 0.01 est deux ordres de grandeur sous l'amplitude attendue
# (0.87) et laisse un etat homogene a 0.000000.
print("temoins croises : deux valeurs de Dv, avec et sans parois (60000 pas chacune)")
print(f"{'condition':<24} {'sigma_max':>10} {'amplitude':>12}   motif ?")
print("-" * 60)
resultats = {}
for dv, paroi in ((1.0, None), (1.0, 60), (40.0, None), (40.0, 60)):
    n = 720 if paroi else 400
    u_t, _, _ = simuler(n, dv, paroi, 60000)
    s, _ = sigma_max(dv)
    amp = float(u_t.std())
    resultats[(dv, paroi)] = amp
    nom = f"Dv={dv:.0f} " + ("+ parois/60" if paroi else "+ libre")
    print(f"{nom:<24} {s:>+10.5f} {amp:>12.6f}   {'oui' if amp > 0.01 else 'non'}")

print()
# Effet des parois a chimie egale (Dv = 40) : rapport des amplitudes, en %.
print(f"effet des parois a Dv = 40 : {100 * (resultats[(40.0, 60)] / resultats[(40.0, None)] - 1):+.2f} % d'amplitude")
```

**Lecture.** Les deux branches de l'hypothèse nulle tombent, et le verdict n'est ni « Turing » ni « Thom ».

Avec `Dv = 1`, l'amplitude est **0,000000** dans les deux cas — avec et sans parois. Le canal, seul, ne produit rien : les parois n'ont aucune prise sur un état stable. C'est la réfutation de la lecture forte de Thom, celle qui ferait de la géométrie une cause suffisante.

Avec `Dv = 40`, le motif apparaît **sans aucune paroi** (amplitude 0,898), et les parois ne changent l'amplitude que de 3 % (0,871). C'est la réfutation de la lecture inverse, celle qui ferait du canal une condition nécessaire.

Reste ce que les parois font réellement : elles **quantifient** — elles imposent un ensemble discret de modes, et c'est l'objet du §8. Thom a donc raison sur un point que ce test ne mesure pas encore : la géométrie est ce qui rend le motif commensurable à une frontière, donc ce qui le rend transmissible d'un tissu à l'autre.

## 8. Mesure 4 — le canal pose la grille, la chimie pose la densité

Protocole : `N = 6L` cellules, une paroi tous les `L`. Chaque compartiment occupe les cellules 1 à `L` — longueur `L`, deux bords de Neumann. Les modes propres d'un tel compartiment sont des cosinus, `cos(π p x / L)` : la base propre est exactement la DCT-II, et le rang lu par la DCT **est** le nombre de demi-ondes, sans conversion.

La prédiction de Turing (§2) est alors d'une précision inhabituelle : la solution chimique `s_c = L · k_c / π` n'est en général pas entière, et le nombre de lobes réalisé doit être **l'un des deux entiers qui l'encadrent** — « usually the nearest ». Le test est donc une appartenance à un ensemble de deux éléments, pas une comparaison de flottants.

```python
# Mesure 4 : la grille du canal contre la densite de la chimie. Balayage de la
# largeur L (30 a 120) avec une paroi tous les L cellules. La DCT-II est la base
# propre du Laplacien de Neumann, donc le rang lu EST le nombre de demi-ondes p.
# Turing 1952 (section 8) predit p dans {floor(s_c), ceil(s_c)} avec
# s_c = L kc / pi, et non necessairement l'argmax de la croissance sigma.
def pic_dct(segment):
    """Rang du mode dominant en base cosinus (DCT-II) : c'est la base propre du
    Laplacien de Neumann, donc le rang lu EST le nombre de demi-ondes."""
    coef = dct(segment - segment.mean(), type=2)
    rangs = np.arange(len(coef))
    return int(rangs[1 + np.argmax(np.abs(coef[1:]))])


# Une ligne par largeur : mode mesure, fenetre des deux entiers, sigma du mode
# mesure et sigma de l'argmax lineaire, verdict d'appartenance a la fenetre.
print(f"{'L':>4} {'s_c':>7} {'fenetre':>9} {'p':>4} {'sigma(p)':>10} {'sigma(argmax)':>14}   verdict")
print("-" * 66)
lignes = []
for L in (30, 40, 50, 60, 70, 80, 100, 120):
    u_l, _, _ = simuler(L * 6, DV, L, 60000)
    p_mes = pic_dct(u_l[1 : L + 1])
    s_c = L * KC / np.pi
    bas, haut = int(np.floor(s_c)), int(np.ceil(s_c))
    sig_mes = sigma(np.pi * p_mes / L, DV)
    p_top = max(range(1, L // 2 + 1), key=lambda q: sigma(np.pi * q / L, DV))
    sig_top = sigma(np.pi * p_top / L, DV)
    dans = bas <= p_mes <= haut
    lignes.append((L, s_c, bas, haut, p_mes, sig_mes, p_top, sig_top, dans))
    fen = f"{bas}..{haut}"
    print(f"{L:>4} {s_c:>7.3f} {fen:>9} {p_mes:>4} {sig_mes:>+10.5f} {sig_top:>+14.5f}   {'dans' if dans else 'HORS'}")

# Bilan : appartenance a la fenetre contre choix de l'argmax, et perte de
# croissance du mode realise quand il n'est pas l'argmax lineaire.
nb_dans = sum(1 for l in lignes if l[8])
nb_arg = sum(1 for l in lignes if l[4] == l[6])
print()
print(f"modes dans la fenetre des deux entiers : {nb_dans} / {len(lignes)}")
print(f"modes sur le maximum lineaire          : {nb_arg} / {len(lignes)}")
for l in lignes:
    if l[4] != l[6]:
        perte = 100 * (1.0 - l[5] / l[7])
        print(f"  L={l[0]:>3} : le mode mesure perd {perte:.1f}% de croissance lineaire sur l'argmax (p={l[6]})")

ecarts = [100.0 * (np.pi * l[4] / l[0] / KC - 1.0) for l in lignes]
print()
print(f"ecart du k mesure a kc : de {min(ecarts):+.1f} % a {max(ecarts):+.1f} %")

# Figure : s_c(L), fenetre des deux entiers, modes mesures.
fig, ax = plt.subplots(figsize=(6.8, 3.9))
Ls = np.array([l[0] for l in lignes])
pmes = np.array([l[4] for l in lignes])
ax.plot(Ls, Ls * KC / np.pi, "-", color="C3", lw=1.6, label="s_c = L kc / pi (chimie)")
ax.fill_between(Ls, Ls * KC / np.pi - 1.0, Ls * KC / np.pi + 1.0, color="C3",
                alpha=0.15, label="fenetre des deux entiers")
ax.plot(Ls, pmes, "o", color="C0", ms=7, label="mode mesure")
ax.set_xlabel("largeur du compartiment L (cellules)")
ax.set_ylabel("nombre de demi-ondes p")
ax.set_title("Le canal pose la grille, la chimie pose la densite")
ax.legend(fontsize=9)
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

**Lecture.** Le résultat est net, et il valide la prédiction de Turing **dans sa forme exacte**, pas dans sa forme relâchée.

- **8 compartiments sur 8** placent leur mode dans la fenêtre des deux entiers encadrant `s_c`. La règle de Turing, lue comme « l'un des deux entiers les plus proches », n'est démentie par aucune mesure.
- **3 sur 8** seulement tombent sur l'argmax linéaire, c'est-à-dire sur « l'entier le plus proche ». Les 5 autres prennent **l'autre** entier de la fenêtre. Turing avait prévu cette latitude, et jusqu'à sa mise en garde : « Examples can, however, be constructed where this simple rule does not apply. »
- La raison de ces 5 cas est mesurable : entre les deux entiers voisins, `σ` ne diffère que de **0,2 % à 2,0 %**. Une croissance linéaire qui sépare aussi peu ne décide rien ; ce qui tranche est la compétition non linéaire entre modes, un objet que la dispersion ne contient pas.

Le `k` mesuré, lui, dépasse `k_c` de −0,4 % à +12,1 % selon la largeur : quand le canal force un mode, il force plutôt un motif **trop serré** que trop large.

Autrement dit — la géométrie fixe la **grille** (`p` entier, pas `π/L`), la chimie fixe la **densité** de lobes sur cette grille (`s_c = L k_c / π`, respecté à ±1), et aucune des deux ne fixe le mode exact. C'est le mot de Turing — « make do » — et c'est aussi, en creux, la thèse de Thom : le canal est ce qui rend le motif discret, donc transmissible.

## 9. Synthèse — qui a raison, et de quoi

| Question | Turing 1952 | Thom 1991 | Mesure de ce carnet |
|---|---|---|---|
| D'où vient la longueur d'onde ? | de la chimie (`k_c`) | du canal | la chimie fixe la densité (`s_c = L k_c/π`, à ±1) ; le canal fixe la grille |
| Le canal suffit-il ? | non dit | oui, implicitement | non ici : amplitude 0,000000 avec `Dv = 1` et parois |
| La chimie suffit-elle ? | oui | non | oui sur anneau libre : amplitude 0,898 |
| Quel mode exact ? | l'un des deux entiers voisins | non spécifié | 8/8 dans la fenêtre, 5/8 prennent le plus éloigné |
| Nature du seuil | instabilité linéaire | catastrophe | exposant 0,466 mesuré près du seuil, 0,5 attendu |
| Priorité explicative | la chimie | la vie comme canalisation | hors de portée d'un banc 1-D |

Le bilan n'est pas un partage d'égalité. Sur le **mécanisme**, le banc donne raison à Turing : avec une asymétrie de diffusion suffisante, un motif naît sans aucune géométrie imposée, et le canal n'est ni nécessaire ni suffisant. Sur le **cadre**, le banc donne raison à Thom : un motif doit être commensurable à une frontière pour être transmissible, et c'est la géométrie, pas la chimie, qui le rend discret. Les deux énoncés ne se contredisent pas; ils ne parlent pas de la même question.

Une nuance que Turing lui-même accorde, et qui rapproche les deux textes plus que la polémique ne le laisse croire. Sa coda (p. 71-72) concède que « Most of an organism, most of the time, is developing from one pattern into another, rather than from homogeneity into a pattern », et que pour ce régime plus général « one cannot hope to have any very embracing theory of such processes, beyond the statement of the equations ». C'est exactement le régime où la canalisation de Thom travaille : l'organisme est déjà structuré, et la vie consiste à tenir cette structure contre la diffusion. Le désaccord porte donc en partie sur un **domaine**, pas sur une contradiction.

## 10. Limites de la lecture, dites avant qu'on les cherche

1. **Le modèle n'est pas celui de 1952.** Schnakenberg n'a pas la structure linéaire du §6 de Turing ; il en porte la même instabilité, mais ses coefficients ne sont pas les « chemical data a, b, c, d, μ', ν' » des citations. Rien ici ne se lit comme une vérification du texte de Turing, seulement comme une vérification de sa **règle de quantification**, qui ne dépend pas du modèle.
2. **Deux substances, donc quatre sous-cas sur six.** Les sous-cas (e) et l'oscillatoire à trois morphogènes sont hors de portée du banc.
3. **Aucune mécanique, aucune croissance, aucun mouvement cellulaire.** Ce sont précisément les ingrédients sur lesquels porte l'objection de Thom ; le banc tranche sur la chimie et la géométrie, pas sur la biomécanique.
4. **Schéma numérique explicite** (`dt = 0,005` ou `0,01`, `dx = 1`). L'exposant 0,466 est une estimation à `ε` fini, pas une asymptotique rigoureuse ; la valeur 0,5 n'est pas mesurée, elle est approchée. Il est de plus mesuré à **graines fixées** — `np.random.default_rng(graine)` avec une seule valeur — : sa stabilité au tirage n'est pas établie ici.
5. **Les parois sont parfaites** (Neumann, flux nul). Un tissu réel fuit ; l'exercice 2 ouvre le canal.
6. **Aucune biologie.** Rien ici ne dit quel mécanisme opère dans un embryon donné. Turing lui-même concède que « the biological examples which it has been possible to give in the present paper are very limited » (p. 72).
7. **Sources.** Les citations sont verbatim des copies archivées ; la pagination est celle des tirés à part. Le scan JSTOR de 1952 est sous droits : il reste au gisement privé, jamais dans le dépôt.
