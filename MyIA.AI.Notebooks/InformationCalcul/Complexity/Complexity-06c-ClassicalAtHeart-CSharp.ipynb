## 1. La question — préparer un état vs implémenter une opération

Le problème de la **synthèse unitaire** (Aaronson-Kuperberg 2007) tient en une asymétrie. On sait depuis longtemps : avec l'accès à la bonne fonction classique $f$ (un oracle), un ordinateur quantique peut **préparer efficacement n'importe quel état quantique** $|0\rangle \mapsto \sum_x \alpha_x |x\rangle$. La question posée demande **plus** : le même type d'accès permet-il aussi d'**implémenter efficacement n'importe quelle opération unitaire** $U$ — pas seulement préparer un état, mais *agir* sur un état arbitraire ?

Formellement : existe-t-il un circuit efficace qui, donné accès à un oracle classique $f$, implémente $U$ pour un $U$ cible — quand $f$ est choisie judicieusement ? La conjecture **négative** de Lombardi-Ma-Wright disait non (il faudrait toujours de la dureté intrinsèque) ; Khurana a cru l'inverse pendant trois ans — minoritaire, « unpopular » — et le manuscrit d'OpenAI (Cité : « manuscript *claiming* a positive solution », vérification en cours) lui donne enfin raison, *si* la preuve tient.

## 2. Le récit — trois ans, une conviction, et les machines

L'article est le récit d'une obsession. Khurana écrit : « *For the past three years, the unitary synthesis problem has occupied a majority of my research time, a considerable part of the rest of my waking life, and more than a few sleepless nights.* »

Elle s'y est intéressée **par la cryptographie** (voir section 3) : avec son étudiant **Kabir Tomer**, elle avait montré que plusieurs formes de *quantum one-wayness* suffisent à une crypto intéressante, et qu'elles peuvent reposer sur des problèmes crus extrêmement durs. La question suivante était naturelle : la crypto quantique peut-elle reposer sur quelque chose d'encore plus faible — sans dureté classique du tout ? La conjecture Lombardi-Ma-Wright disait oui — et impliquait une réponse **négative** à la synthèse unitaire.

Khurana a cru l'inverse pour deux raisons, une mathématique et une esthétique :

- **Les permutations de base sont déjà puissantes** : l'accès classique à un oracle permet déjà de préparer des états arbitraires et d'implémenter des permutations de base arbitraires. Elle ne voyait pas pourquoi des permutations ne pourraient pas servir à construire des unitaires arbitraires — jusqu'au travail récent, on croyait même en avoir *besoin* pour construire des unitaires pseudo-aléatoires.
- **Symétrie du monde** : une réponse positive mettrait les *opérations* sur le même pied que les *états* — « the world seemed more beautiful, more symmetric that way ».

Le point le plus frappant du récit : **« Over the past year, I worked extensively with the robots too, asking agents to explore ideas and attack the obstacles we were hitting. We generated many intermediate results. »** C'est le même fil que le [07c](Complexity-07c-LeBilanMathocalypse-Python.ipynb) — les agents dans la recherche — vu cette fois côté humain *assisté* : les machines explorent, l'humaine garde l'intuition et signe le récit. ## 3. Conséquence pour la cryptographie quantique — la dureté doit vivre quelque part

Si la preuve d'OpenAI tient, ce qu'elle change pour le champ :

- **La dureté de la crypto quantique doit vivre dans un problème classique.** Si un oracle classique (gratuit) implémente toute unitaire, alors « l'unitaire difficile » n'existe plus comme ressource de dureté quantique autonome : la sécurité doit venir d'une dureté *classique*.
- **Mais la crypto quantique peut en exiger bien moins que la classique** (Cité : voir Kretschmer, Kretschmer-Qian-Sinha-Tal, Kretschmer-Qian-Tal) — « quantum cryptography may still need far less of that hardness than classical cryptography does ». Le champ ne s'effondre pas : il se *recale*.

La conclusion de Khurana tient le titre de l'article : « *I trained as a classical cryptographer... Perhaps I am classical at heart. So, it seems, is the field I love.* »

## 5. Jouet exécutable — la même machine que le 06b, l'autre question

Le socle du [06b — Aaronson-Gottesman](Complexity-06b-AaronsonGottesman-Dequantification-Python.ipynb) exécute le formalisme stabilisateur : Gottesman-Knill montre qu'un circuit **Clifford** se *simule* classiquement en $O(n^2)$ — une *simulation*, pas une *synthèse* : on reconnaît l'état final d'un circuit donné, on ne *construit* pas un circuit pour une unitaire cible. Le jouet ci-dessous distille la différence sur un exemple mesuré : préparer un état cible par oracle classique est trivial (une permutation de base) ; implémenter une unitaire arbitraire l'exige d'une structure.

```
using System;
using System.Globalization;
using System.Linq;

CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;  // sorties en notation invariante (pas de virgule decimale)
CultureInfo.CurrentUICulture = CultureInfo.InvariantCulture;

// Un oracle classique = une permutation des etats de base.
// Preparer un etat cible arbitraire : une SEULE permutation suffit (Mesure).
int n = 3;  // 3 qubits, 8 etats de base
double[] cible = { 0, 0, 0, 0.5, 0, 0, 0, Math.Sqrt(0.75) };  // etat cible (norme 1)
Console.WriteLine($"Etat cible sur {n} qubits : [{string.Join(", ", cible)}] (Mesure)");

// Preparation : trouver le support (indices non nuls) + ajuster les amplitudes :
// operation TRIVIALE par oracle classique.
var support = Enumerable.Range(0, cible.Length).Where(i => cible[i] != 0).ToArray();
Console.WriteLine($"Support de la cible : etats de base [{string.Join(", ", support)}] (Mesure)");
Console.WriteLine("-> Preparation : 1 permutation de base + ajustement d'amplitudes (1 requete) (Mesure)");

// Implementer une unitaire ARBITRAIRE : beaucoup plus exigeant.
// Une unitaire U x U complexe a ~ U^2 parametres reels libres.
int U = 1 << n;                    // 2^n etats de base
int paramLibres = U * U;           // ~ U^2 parametres reels
long fact = 1;                     // U! arrangements
for (int i = 2; i <= U; i++) fact *= i;
Console.WriteLine($"Une unitaire {U}x{U} a ~{paramLibres} parametres reels libres (Mesure)");
Console.WriteLine($"Une permutation n'en encode que {U}! = {fact} arrangements (Mesure)");
Console.WriteLine("-> Synthese : la permutation seule NE SUFFIT PAS pour une unitaire arbitraire");
Console.WriteLine("   -> il faut de la STRUCTURE dans l'oracle (borne Huang-Ramkumar-Wright) (Cite)");
```

## Lecture du résultat — un décompte n'est pas une dimension

Les deux lignes de la sortie « 64 paramètres réels » et « 8! arrangements » se lisent ensemble, et le contraste est un **piège de lecture** si on le prend pour un argument de cardinal : $8^2 = 64$ paramètres réels pour une unitaire $8\times 8$, contre $8! = 40\,320$ permutations — donc « la permutation aurait plus d'objets ». Ce n'est pas le propos, parce que les deux nombres ne comptent pas la même chose. $64$ est une **dimension** : le nombre de paramètres réels continus qu'il faut pour décrire l'objet. $8!$ est un **cardinal** : le nombre d'objets discrets disponibles. Une permutation est un point isolé de l'espace des unitaires ; aussi nombreux soient ces points, ils ne couvrent pas un espace de dimension $64$ — de même qu'une infinité de points ne remplit pas un segment. Ce qui manque à l'oracle classique est donc la **structure** (la borne Huang-Ramkumar-Wright, en fin de sortie), pas le nombre d'arrangements.
