## 1. L'ETH, exactement

Impagliazzo et Paturi posent $s_k$ comme l'infimum des $\delta \ge 0$ tels qu'un algorithme déterministe décide le $k$-SAT sur $n$ variables en temps $O(2^{\delta n})$ (**Cité** : [IP01], archivé au gisement — *On the Complexity of k-SAT*, J. Comput. Syst. Sci. 62(2), 367-375, 2001). L'**hypothèse du temps exponentiel** dit $s_3 > 0$ : il existe un $\delta > 0$ tel qu'aucun algorithme ne décide 3-SAT en $O(2^{\delta n})$. Le papier d'Anthropic en donne la forme équivalente $s_k > 0$ pour tout $k \ge 3$ (Définition 6, lue dans la copie archivée).

La réfutation annoncée est donc l'énoncé $s_3 = 0$ : **pour tout** $\delta > 0$, **il existe** un algorithme en $O(2^{\delta n})$. Ce quantificateur universel sur $\delta$ est tout le contenu du résultat — et c'est aussi ce qui le rend compatible avec un monde où aucune formule praticable n'est jamais résolue plus vite (§5b).

```
// --- 1a. Ce que 2^(d·n) veut dire, pour d et n donnes ---
// Culture invariante re-epinglee : la cellule peut tourner sur un autre thread.
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;
const double PAS_PAR_SECONDE = 1e8;

// Sci : mantisse et exposant en notation scientifique a la Python (1.1e+07),
// .NET n'offrant pas ce format a deux chiffres d'exposant.
string Sci(double v, int dec) {
    int e = (int)Math.Floor(Math.Log10(v));
    double m = v / Math.Pow(10, e);
    if (m >= 10.0) { m /= 10.0; e += 1; }
    if (m < 1.0) { m *= 10.0; e -= 1; }
    string signe = e >= 0 ? "+" : "-";
    return $"{m.ToString("F" + dec)}e{signe}{Math.Abs(e).ToString("D2")}";
}
string Expo10(double d) => $"2^({d.ToString("F1")}·n)";

Console.WriteLine("Echelle des budgets exponentiels (nombre de pas de calcul) :");
Console.WriteLine();
Console.WriteLine($"{'n',5} | " + string.Join(" | ", new[] { 1.0, 0.5, 0.1 }.Select(d => $"{Expo10(d),9}")));
foreach (int n in new[] { 25, 50, 100 }) {
    // Chaque case : 10^log10(2^(d*n)), arrondi a une decimale -- l'ordre de grandeur suffit.
    var cases = new[] { 1.0, 0.5, 0.1 }.Select(d => $"10^{Math.Log10(Math.Pow(2.0, d * n)),5:F1}");
    Console.WriteLine($"{n,5} | " + string.Join(" | ", cases.Select(c => $"{c,9}")));
}
Console.WriteLine();
Console.WriteLine($"A {Sci(PAS_PAR_SECONDE, 0)} pas/s (l'echelle de budget du carnet 01) :");
double t50 = Math.Pow(2.0, 50) / PAS_PAR_SECONDE;
Console.WriteLine($"  n = 50,  d = 1.0 : {Sci(t50, 1)} s (~{t50 / 86400:0} jours)");
Console.WriteLine($"  n = 50,  d = 0.5 : {Math.Pow(2.0, 25) / PAS_PAR_SECONDE:F2} s");
Console.WriteLine($"  n = 100, d = 0.1 : {Sci(Math.Pow(2.0, 10) / PAS_PAR_SECONDE, 2)} s");
Console.WriteLine();
Console.WriteLine("s_3 = 0 exige : POUR TOUT d > 0, un algorithme en O(2^(d·n)) -- la contrainte");
Console.WriteLine("porte sur l'exposant seulement, jamais sur la praticabilite (voir 5b).");
```

### Lecture du résultat — l'échelle des exponentielles

Diviser l'exposant par 2 ($\delta = 1 \to 0{,}5$) fait passer, à $n = 50$, d'un budget de ~130 jours à un tiers de seconde ; descendre encore ($\delta = 0{,}1$ à $n = 100$) ne laisse que mille pas. L'ETH ne dit pas « 3-SAT est exponentiel » dans l'absolu : elle dit que l'exposant $\delta$ ne descend pas jusqu'à 0. La réfutation annoncée dit le contraire — **quel que soit** $\delta > 0$, un algorithme existe. Rien dans cet énoncé ne promet qu'il soit utilisable : c'est la distance entre $2^{o(n)}$ et « pratique », que la mesure du §5 quantifiera.

```
// --- 5a. Mesure : compter exactement #3-SAT par enumeration, n = 12..21 ---
// Culture invariante re-epinglee : la cellule peut tourner sur un autre thread.
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;

// 3-CNF aleatoire a densite 4.27 (le seuil dur mesure par le carnet 02b).
// Generateur .NET : les instances different de celles du jumeau Python, le
// protocole (densite, tailles, graines) est le meme.
List<(int Var, bool Pos)[]> Instance3Sat(int n, double densite = 4.27, int graine = 0) {
    var rng = new Random(graine);
    int m = (int)Math.Round(densite * n);
    var clauses = new List<(int, bool)[]>();
    for (int c = 0; c < m; c++) {
        var pool = Enumerable.Range(0, n).ToArray();   // tirage sans remise de 3 variables
        var clause = new (int, bool)[3];
        for (int j = 0; j < 3; j++) {
            int t = rng.Next(j, n);
            (pool[j], pool[t]) = (pool[t], pool[j]);
            clause[j] = (pool[j], rng.Next(0, 2) == 1);
        }
        clauses.Add(clause);
    }
    return clauses;
}

// Par clause : (masqueVariables, masqueSignes). Le literal (i, pos) est
// satisfait ssi le bit i de l'affectation vaut pos ; la clause est satisfaite
// ssi (a XOR signes) ET variables != 0 -- au moins un literal satisfait.
List<(long Mv, long Ms)> Masques(IEnumerable<(int Var, bool Pos)[]> clauses) {
    var sortie = new List<(long, long)>();
    foreach (var clause in clauses) {
        long mv = 0, ms = 0;
        foreach (var (i, pos) in clause) {
            mv |= 1L << i;
            if (!pos) ms |= 1L << i;
        }
        sortie.Add((mv, ms));
    }
    return sortie;
}

// Enumeration bit-parallele : un long par affectation, comme le tableau int64
// de numpy dans le jumeau Python -- la table coute 8 octets par entree, c'est
// elle qui change de regime (cache L2 puis memoire) au fil de n.
long CompteVectorise(IEnumerable<(int Var, bool Pos)[]> clauses, int n) {
    int taille = 1 << n;
    var affectations = new long[taille];
    for (int i = 0; i < taille; i++) affectations[i] = i;
    var sat = new bool[taille];
    Array.Fill(sat, true);
    var temporaire = new long[taille];
    foreach (var (mv, ms) in Masques(clauses)) {
        for (int i = 0; i < taille; i++) temporaire[i] = affectations[i] ^ ms;
        for (int i = 0; i < taille; i++) sat[i] &= (temporaire[i] & mv) != 0;
    }
    long compte = 0;
    for (int i = 0; i < taille; i++) if (sat[i]) compte++;
    return compte;
}

// Enumeration pure, temoin d'accord pour petits n.
long CompteEnumerationPure(IEnumerable<(int Var, bool Pos)[]> clauses, int n) {
    long total = 0;
    for (long a = 0; a < (1L << n); a++) {
        bool ok = true;
        foreach (var clause in clauses) {
            bool satisfaite = false;
            foreach (var (i, pos) in clause) {
                if ((((a >> i) & 1) == 1) == pos) { satisfaite = true; break; }
            }
            if (!satisfaite) { ok = false; break; }
        }
        if (ok) total++;
    }
    return total;
}

// Controles canoniques : (x0) compte 1 ; (x0)(non x0) compte 0 ; (x0)(x1) compte 1
var controles = new[] {
    (Attendu: 1L, Rendu: CompteEnumerationPure(new[] { new[] { (0, true) } }, 1)),
    (Attendu: 0L, Rendu: CompteEnumerationPure(new[] { new[] { (0, true) }, new[] { (0, false) } }, 1)),
    (Attendu: 1L, Rendu: CompteEnumerationPure(new[] { new[] { (0, true) }, new[] { (1, true) } }, 2)),
};
Console.WriteLine($"Controles canoniques : {controles.Count(c => c.Rendu == c.Attendu)}/3 OK");

// Temoin d'accord : enumeration bit-parallele contre enumeration pure, n = 12
var cl12 = Instance3Sat(12, graine: SEED);
Console.WriteLine($"Accord vectorise / enumeration pure (n = 12) : {CompteVectorise(cl12, 12) == CompteEnumerationPure(cl12, 12)}");

Console.WriteLine();
Console.WriteLine($"{'n',3} {'m',4} {"#sat",8} {"temps (ms)",11} {"t_n/t_(n-1)",12}");
var mesures = new List<(int N, double T)>();
var chrono = new Stopwatch();
for (int n = 12; n <= 21; n++) {
    var clauses = Instance3Sat(n, graine: SEED + n);
    long compte = 0;
    double meilleur = double.MaxValue;
    for (int essai = 0; essai < 2; essai++) {   // deux passages, on garde le meilleur
        chrono.Restart();
        compte = CompteVectorise(clauses, n);
        chrono.Stop();
        meilleur = Math.Min(meilleur, chrono.Elapsed.TotalSeconds);
    }
    double precedent = mesures.Count > 0 ? mesures[^1].T : double.NaN;
    string facteur = double.IsNaN(precedent) ? "           -" : $"{meilleur / precedent,10:F3}";
    Console.WriteLine($"{n,3} {clauses.Count,4} {compte,8} {meilleur * 1000,11:F1} {facteur,12}");
    mesures.Add((n, meilleur));
}

// Pente de log2(temps) en fonction de n, ajustement lineaire aux moindres carres.
double PenteLog2(IEnumerable<(int N, double T)> paires) {
    var xs = paires.Select(p => (double)p.N).ToList();
    var ys = paires.Select(p => Math.Log2(p.T)).ToList();
    double mx = xs.Average(), my = ys.Average();
    double numer = xs.Zip(ys, (x, y) => (x - mx) * (y - my)).Sum();
    double denom = xs.Sum(x => (x - mx) * (x - mx));
    return numer / denom;
}

var tempsParN = mesures.ToDictionary(m => m.N, m => m.T);
double pente = PenteLog2(mesures);
Console.WriteLine();
Console.WriteLine($"Pente globale log2(temps) par variable (n = 12..21) : {pente:F3}");
string verdictPre = 0.85 <= pente && pente <= 1.15 ? "CONFIRME" : "A REEXAMINER";
Console.WriteLine($"Verdict H1 au critere pre-enregistre (pente dans 0.85..1.15) : {verdictPre}");

// Re-examen : DEUX REGIMES stables d'une execution a l'autre -- table en cache
// (n <= 15, ~0,3 Mo) puis table en memoire (n >= 17, ~1 Mo). Le point n = 16
// chevauche la transition : on l'exclut des deux ajustements, et on mesure
// l'amplitude du saut par le rapport stable t(17)/t(15).
double penteAvant = PenteLog2(mesures.Where(p => p.N <= 15));
double penteApres = PenteLog2(mesures.Where(p => p.N >= 17));
double amplitude = tempsParN[17] / tempsParN[15];
Console.WriteLine($"  Saut entre les regimes : t(n=17)/t(n=15) = {amplitude:F0}x pour un doublement");
Console.WriteLine("  attendu de 4x -- la table passe de ~0,3 a 1 Mo et quitte le cache L2 :");
Console.WriteLine("  hierarchie memoire, pas un exposant qui change.");
Console.WriteLine($"  Pentes par regime : {penteAvant:F2} (n = 12..15, en cache),");
Console.WriteLine($"  {penteApres:F2} (n = 17..21, en memoire).");
Console.WriteLine("VERDICT H1 FINAL : CONFIRME APRES EXAMEN -- dans chaque regime la pente");
Console.WriteLine($"est ~1 ({penteAvant:F2} / {penteApres:F2}) ; le saut de cache ne fait");
Console.WriteLine("qu'AGGRAVER l'enumeration. Pour cette machine, s_3 = 1 : mur exact 2^n.");
```

### Lecture du résultat — le mur de l'énumération, et un saut de cache

Deux régimes, stables d'une exécution à l'autre : la table d'affectations tient en cache ($n \leq 15$, ~0,3 Mo), puis vit en mémoire ($n \geq 17$, ~1 Mo) — $n = 16$ chevauche la transition et l'ajustement l'exclut. Dans chaque régime la pente est ≈ 1 ; le saut entre les deux (5× pour un doublement attendu de 4×) porte la pente **globale** à ≈ 1.07 — dans la fenêtre pré-enregistrée 0,85..1,15, mais comme un **mélange** de deux régimes, pas comme un exposant mesuré proprement : c'est le re-examen par régimes qui le montre. Sur la machine du jumeau Python, la même pente globale tombait hors fenêtre (≈ 1,32) ; le même re-examen y réparait le verdict — ici il le confirme, en identifiant le saut de cache (la table passe de ~0,3 à 1 Mo et quitte le cache L2) comme de la hiérarchie mémoire, pas un exposant qui change. Dans les deux cas le diagnostic aggrave le mur : la mémoire est plus lente que le cache, l'énumération reste au-dessus de $2^n$ par variable. Pour **cette** machine, $s_3 = 1$ exactement. La question du théorème est de savoir s'il existe une **autre** machine pour laquelle $s_3 = 0$ — la mesure suivante demande non pas si elle existe, mais **à partir de quand elle gagnerait**.

Sur cette exécution C#, les pentes par régime valent ≈ 0.89 (cache, $n = 12..15$) et ≈ 1.09 (mémoire, $n = 17..21$), pour un saut $t(17)/t(15)$ de 5× — le jumeau Python mesure la même structure avec ses propres valeurs (sorties du carnet Python : pentes ≈ 0,92 en cache, ≈ 1,11 en mémoire).

```
// --- 5b. L'exposant du budget annonce, normalise a C = 1, k = 3 ---
// Culture invariante re-epinglee : la cellule peut tourner sur un autre thread.
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;
// On travaille en rapport a n, qui ne depend QUE de L = log2(n) : aucun
// debordement possible, meme pour L = 20000 (n = 2^20000 serait irreprésentable).

// MainTheorem (Anthropic, Th. 1) : 2^{C·min{ k·m/sqrt(log m), k^2·n·loglog n/sqrt(log n) }}
// Theoreme 1.1 (OpenAI) : (n+m)^{O_k(1)} · 2^{O_k(n·loglog n/sqrt(log n))} -- meme forme en n
// Resume CONTENTS.md (OpenAI) : 2^{n/(log n)^(1/2+o(1))} -- le o(1) absorbe le k^2

double RatioM(double L, double k = 3.0, double densite = 4.27) => k * densite / Math.Sqrt(L + Math.Log2(densite));
double RatioN(double L, double k = 3.0) => k * k * Math.Log2(L) / Math.Sqrt(L);
double RatioResume(double L) => 1.0 / Math.Sqrt(L);

// Plus petit L = log2(n) tel que ratio(L) < 1, par dichotomie.
double? Croise(Func<double, double> ratio, double lmax = 20000.0) {
    if (ratio(4.0) < 1.0) return 4.0;   // deja sous le seuil des les petites tailles
    if (ratio(lmax) >= 1.0) return null;   // pas de croisement dans la plage
    double lo = 4.0, hi = lmax;
    while (hi - lo > 1e-3) {
        double mid = (lo + hi) / 2;
        if (ratio(mid) >= 1.0) lo = mid; else hi = mid;
    }
    return hi;
}

Console.WriteLine("Rapport a n de l'exposant (C = 1, k = 3, m = 4.27·n, log en base 2) :");
Console.WriteLine($"{'n',8} {"min(Th.1)/n",12} {"m-cote/n",10} {"n-cote/n",10} {"resume/n",10}");
foreach (int expo in new[] { 10, 15, 20, 25, 30, 40, 60, 100 }) {
    double rm = RatioM(expo), rn = RatioN(expo), rr = RatioResume(expo);
    Console.WriteLine($"{"2^" + expo,8} {Math.Min(rm, rn),12:F3} {rm,10:F3} {rn,10:F3} {rr,10:F3}");
}
Console.WriteLine();
double? cM = Croise(L => RatioM(L));
double? cN = Croise(L => RatioN(L));
double? cR = Croise(L => RatioResume(L));
Console.WriteLine($"Croisement cote m  (k·m/sqrt(log m) < n)      : n ~ 2^{cM:F1}");
Console.WriteLine($"Croisement cote n  (k^2·n·loglog/sqrt(log) < n): n ~ 2^{cN:F0}");
Console.WriteLine($"Forme resume       (n/sqrt(log n) < n)         : n ~ 2^{cR:F0} -- mais voir ci-dessous");
Console.WriteLine();
double r25 = Math.Min(RatioM(25.0), RatioN(25.0));
Console.WriteLine("VERDICT H2 : CONFIRME, ET PLUS FORT QUE PRE-ENREGISTRE.");
Console.WriteLine($"  Le rapport min(Th.1)/n vaut encore ~{r25:F2} a n = 2^25 (> 1 : le budget ne");
Console.WriteLine($"  rejoint meme pas l'enumeration) ; il ne descend sous 1 qu'a partir de");
Console.WriteLine($"  n ~ 2^{Math.Min(cM.Value, cN.Value):F0}, et C : ℕ dans timeMin (C >= 1), non explicite,");
Console.WriteLine("  ne peut que RECULER ce seuil. La forme resume du depot passe sous 1 tres tot");
Console.WriteLine("  seulement parce que son o(1) absorbe le k^2 et le log log -- c'est la loupe");
Console.WriteLine("  de l'asymptotique, pas un budget praticable.");
```

### Lecture du résultat — la distance entre $2^{o(n)}$ et un algorithme utile

La mesure rend deux verdicts nets. **H1 confirmée après examen** : l'énumération coûte $2^n$ par variable dans chaque régime (pentes ≈ 0.89 en cache, ≈ 1.09 en mémoire, saut de cache diagnostiqué entre les deux). **H2 confirmée et renforcée** : à $C = 1$ — le meilleur cas, puisque $C$ est un naturel dans `timeMin` — l'exposant du budget annoncé reste **supérieur** à $n$ sur tout l'univers atteignable ; le croisement se joue au-delà de $2^{160}$ variables ($2^{162}$ côté $m$, $2^{15746}$ côté $n$), et le $C$ réel, non explicite, ne peut que le repousser. Autrement dit : même si les preuves tiennent, l'algorithme annoncé ne bat la force brute **pour aucune instance représentable** — ce que le README des auteurs dit lui-même (« purely asymptotic »), et dont le [07c](Complexity-07c-LeBilanMathocalypse-CSharp.ipynb) a déjà mesuré la loi générale : le matmul en $n^{2.25}$ ne gagne qu'à partir de $n \sim 10^{16}$ si sa constante vaut 100× (section 4 du 07c).

Ce que cette mesure **ne** dit pas : elle n'exécute pas leur machine, ne dit rien de la validité des preuves (§4), et ne prétend pas qu'aucun algorithme ne fait pas mieux que l'énumération en pratique — le seuil de phase du [02b](Complexity-02b-SAT-PhaseTransition-CSharp.ipynb) rend déjà certaines formules faciles sans casser l'ETH. Elle situe l'ordre de grandeur : la victoire annoncée est réelle **en exposant**, invisible **en instance**.

## À retenir

1. **L'ETH, c'est $0 < s_3$** : une constante $s_3$ telle qu'aucun algorithme ne décide le 3-SAT en moins de $2^{\Omega(n)}$ opérations. La réfutation annoncée (`ETHIsFalse`) ne dit pas « P = NP » : elle dit qu'il existe **une** machine et **une** constante $C$, choisies **avant** $k$ et la formule.
2. **Les budgets $2^{\delta n}$ sont hors de portée avant 160 bits.** À $10^8$ pas/s : $n = 50$ avec $\delta = 1{,}0$ → 1,1e+07 s (~130 jours) ; $\delta = 0{,}5$ → 0,34 s ; $n = 100$ avec $\delta = 0{,}1$ → 1,02e-05 s. La contrainte porte sur l'exposant, jamais sur la praticabilité.
3. **Les trois surfaces de vérité sont lues, pas compilées.** `Challenge.lean` finit en `sorry` pour `ethIsFalse` et `mainTheorem` ; les axiomes publiés sont `propext, Classical.choice, Quot.sound` ; le dépôt openai finit en `sorry` pour `exact_count_and_sat`. Aucun `lake build` n'est lancé : ce carnet lit, il ne certifie rien.
4. **H1 : le mur d'énumération est confirmé, après examen.** Pente $\log_2$(temps) par variable : 1.07 sur $n = 12..21$ — mais **0.89** en cache ($n = 12..15$) et **1.09** en mémoire ($n = 17..21$) ; le saut de **5×** entre $n = 15$ et $n = 17$ est un saut de cache (la table passe de ~0,3 à 1 Mo), pas un changement d'exposant. Dans chaque régime, la pente est ≈1.
5. **H2 : le budget annoncé ne rejoint l'énumération qu'au-delà de $2^{162}$.** Le rapport min(Th.1)/$n$ vaut encore **2.461** à $n = 2^{25}$ (> 1) et ne descend sous 1 qu'à $n \approx 2^{162}$ ; côté $n$, le croisement est à $n \approx 2^{15746}$. La forme résumée du dépôt passe sous 1 très tôt seulement parce que son $o(1)$ absorbe le $k^2$ et le $\log\log$ — « la loupe de l'asymptotique, pas un budget praticable ».

<details><summary>Réponses</summary>

1. Parce que l'hypothèse est **asymptotique** : elle porte sur la forme $2^{\Omega(n)}$ et sur la constante $s_3$, pas sur le temps de calcul d'une instance concrète. On ne peut donc pas en déduire qu'un 3-SAT de taille donnée est insoluble en un temps donné : un algorithme sous-exponentiel peut être parfaitement utilisable sans violer l'ETH, et un algorithme $2^{\delta n}$ avec $\delta$ petit reste impraticable.
2. En comparant le rapport de temps successif au doublement attendu (5× au lieu de 4×) et en localisant le saut par la taille de la table (~0,3 Mo → 1 Mo, sortie du cache L2). Un changement d'exposant aurait donné des pentes croissantes dans **chaque** régime ; ici elles valent 0.89 puis 1.09, proches de 1 chacune — la pente globale mélange deux régimes, elle ne mesure aucun exposant propre. Sur la machine du carnet Python, la même pente globale tombait hors fenêtre (≈ 1,32) et le même re-examen ramenait le verdict à « confirmé » : l'écart venait de la mémoire, pas de la théorie.
3. Le rapport min(Th.1)/$n$ : il vaut 2.461 à $n = 2^{25}$, c'est-à-dire que le budget annoncé reste **2,46 fois** supérieur au coût de l'énumération sur tout l'univers atteignable. Aucun régime observable ne voit l'avantage sous-exponentiel — c'est plus fort que le seuil 0,85 demandé.
